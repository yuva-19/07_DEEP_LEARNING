# 🧪 Experiment Design

## 1. What is Experiment Design?

In Deep Learning, **experiment design** means deciding:

- What you want to test
- Which variable you will change
- Which variables you will keep constant
- What metric you will measure
- How you will compare the results

The goal is to determine **why** a model performed better or worse.

For example:

> Does increasing the learning rate improve validation accuracy?

A properly designed experiment changes the **learning rate** while keeping other important conditions unchanged.

---

## 2. The Core Principle

The most important rule in this syllabus is:

> ### 🔥 Change ONE important variable at a time.

Suppose your baseline model uses:

```text
Learning Rate = 0.001
Batch Size    = 32
Optimizer     = Adam
Dropout       = 0.2
Epochs        = 20
Architecture  = 3 layers
```

You want to investigate learning rate.

### Experiment

| Experiment | Learning Rate | Batch Size | Optimizer | Dropout |
|---|---:|---:|---|---:|
| Baseline | 0.001 | 32 | Adam | 0.2 |
| Experiment 1 | **0.0001** | 32 | Adam | 0.2 |
| Experiment 2 | **0.001** | 32 | Adam | 0.2 |
| Experiment 3 | **0.01** | 32 | Adam | 0.2 |

Only the **learning rate** changes.

This allows you to reasonably attribute differences in performance to the learning rate.

---

# 3. Variables in a Deep Learning Experiment

A controlled experiment usually has three important types of variables.

[Wikimedia Commons — Independent, dependent and control variables](https://commons.wikimedia.org/wiki/File%3AGroups_and_Variables.jpg?utm_source=chatgpt.com)

![Independent, dependent and control variables](https://upload.wikimedia.org/wikipedia/commons/1/15/Groups_and_Variables.jpg)

### Independent Variable

The variable that **you intentionally change**.

Examples:

- Learning rate
- Batch size
- Dropout
- Number of layers
- Number of hidden units
- Optimizer

---

### Dependent Variable

The result that you **measure**.

Examples:

- Validation accuracy
- Validation loss
- Training loss
- Training time
- F1-score
- Precision
- Recall

For example:

```text
Change:
Learning Rate

Measure:
Validation Accuracy
```

---

### Control Variables

Variables that you **keep constant**.

For example:

```text
Dataset          → SAME
Train/Validation split → SAME
Architecture     → SAME
Optimizer        → SAME
Batch size       → SAME
Epochs           → SAME
Random seed      → SAME
Data augmentation → SAME
```

This makes the comparison fair.

---

# 4. Why Change Only One Variable?

Imagine you perform this:

### Experiment A

```text
Learning rate = 0.001
Batch size    = 32
Dropout       = 0.2
```

### Experiment B

```text
Learning rate = 0.01
Batch size    = 64
Dropout       = 0.5
```

And Experiment B gets better accuracy.

Now ask:

> **Why did it improve?** 🤔

Was it because of:

- Higher learning rate?
- Larger batch size?
- Higher dropout?
- Interaction between them?

You **cannot know**.

That experiment changed too many variables simultaneously.

---

# 5. Proper Experimental Design

Instead:

### Baseline

```text
Learning rate = 0.001
Batch size    = 32
Dropout       = 0.2
```

Then change only learning rate:

```text
Experiment 1 → LR = 0.0001
Experiment 2 → LR = 0.001
Experiment 3 → LR = 0.01
```

Everything else stays fixed.

Then choose the best learning rate.

After that, you can investigate another variable.

```text
Step 1 → Tune Learning Rate
           ↓
Step 2 → Fix best Learning Rate
           ↓
Step 3 → Tune Batch Size
           ↓
Step 4 → Fix best Batch Size
           ↓
Step 5 → Tune Dropout
```

This creates a much clearer experimental process.

---

# 6. Baseline Experiment

Before changing anything, establish a **baseline**.

A baseline is your reference experiment.

Example:

```text
Model:
3-layer MLP

Optimizer:
Adam

Learning Rate:
0.001

Batch Size:
32

Dropout:
0.2

Epochs:
20

Validation Accuracy:
91.2%
```

Now every new experiment can be compared against:

> **Baseline validation accuracy = 91.2%**

---

# 7. Example: Learning Rate Experiment

Suppose we test:

| Experiment | LR | Validation Loss | Validation Accuracy |
|---|---:|---:|---:|
| Baseline | 0.001 | 0.32 | 91.2% |
| LR-1 | 0.0001 | 0.41 | 88.7% |
| LR-2 | 0.005 | 0.28 | 92.5% |
| LR-3 | 0.01 | 0.36 | 90.4% |

### Observation

```text
0.0001 → Learning too slowly
0.001  → Good
0.005  → Best
0.01   → Performance decreases
```

Therefore:

```text
Best learning rate ≈ 0.005
```

But don't immediately conclude:

> "0.005 is universally the best learning rate."

Instead conclude:

> **0.005 performed best under this experimental setup.**

That's a much more scientifically correct conclusion. 🧠

---

# 8. Compare the Right Metrics

Never compare experiments using only one number blindly.

For classification, you might track:

```text
Training Loss
Validation Loss
Training Accuracy
Validation Accuracy
Training Time
```

Example:

| Experiment | Train Acc | Val Acc | Train Loss | Val Loss |
|---|---:|---:|---:|---:|
| A | 99.8% | 90.2% | 0.05 | 0.45 |
| B | 97.5% | **93.1%** | 0.12 | **0.29** |

Experiment B is probably preferable because it generalizes better.

---

# 9. Training vs Validation Performance

This is extremely important when comparing experiments.

### Case 1 — Both improve

```text
Training accuracy ↑
Validation accuracy ↑
```

Usually a good sign.

---

### Case 2 — Training improves but validation gets worse

```text
Training accuracy ↑↑
Validation accuracy ↓
```

🚨 Possible **overfitting**.

---

### Case 3 — Neither improves

```text
Training accuracy → low
Validation accuracy → low
```

Possible **underfitting** or optimization problems.

# 11. Keep the Dataset Split Consistent

This is a commonly overlooked issue.

Suppose:

```text
Experiment A
Train → 80%
Validation → 20%
```

and:

```text
Experiment B
Train → 80%
Validation → 20%
```

But the actual samples in the validation sets are different.

Your comparison can become noisy.

Therefore, whenever possible, use the **same train/validation/test split** across experiments.

---

# 12. Keep the Random Seed Consistent

Deep learning training can contain randomness from:

- Weight initialization
- Data shuffling
- Dropout
- GPU operations

Therefore:

```python
import tensorflow as tf
import numpy as np

tf.random.set_seed(42)
np.random.seed(42)
```

This can make experiments more reproducible.

⚠️ A fixed seed does not guarantee perfect reproducibility across every hardware/software configuration, but it helps control one important source of variation.

# 13. One Variable at a Time — BUT There Is an Important Limitation

Here's the deeper point. 👇

The **One-Factor-at-a-Time (OFAT)** approach is easy to understand, but it has a limitation:

> It may fail to reveal **interactions between variables**.

NIST explains that OFAT can miss cases where the effect of one input depends on the value of another input. [NIST](https://www.itl.nist.gov/div898/handbook/pri/section2/pri212.htm?utm_source=chatgpt.com)

For example:

```text
Learning Rate
      +
Batch Size
```

may interact.

A learning rate that works well with:

```text
Batch Size = 32
```

may not work equally well with:

```text
Batch Size = 512
```

This is why more advanced experimentation eventually uses methods such as:

- Grid search
- Random search
- Bayesian optimization
- Factorial experiments

# 14. Design of Experiments (DOE)

Once you have more experience, you can deliberately test combinations of variables.

[Wikimedia Commons — Design of Experiments example](https://commons.wikimedia.org/wiki/File%3ADesign_Of_Experiments.png?utm_source=chatgpt.com)

![Design of Experiments](https://upload.wikimedia.org/wikipedia/commons/a/ac/Design_Of_Experiments.png)

For example:

```text
Learning Rate:
0.001
0.005

Batch Size:
32
64
```

You could test:

| Experiment | LR | Batch |
|---|---:|---:|
| 1 | 0.001 | 32 |
| 2 | 0.005 | 32 |
| 3 | 0.001 | 64 |
| 4 | 0.005 | 64 |

Now you're investigating not only individual effects but potentially **interactions**.

This is closer to formal **Design of Experiments (DOE)**.

---

# 15. Practical Deep Learning Experiment Workflow

A professional workflow can look like this:

```text
                BASELINE
                   │
                   ↓
          Define evaluation metric
                   │
                   ↓
       Change ONE important variable
                   │
                   ↓
             Train model
                   │
                   ↓
          Record the results
                   │
                   ↓
          Compare with baseline
                   │
                   ↓
        Select the better setting
                   │
                   ↓
       Freeze that variable
                   │
                   ↓
       Move to the next variable
```

Keep:

# 17. Bad vs Good Experimentation 🚨

### ❌ Bad

```text
Experiment 1
LR = 0.001
Batch = 32
Dropout = 0.2

Experiment 2
LR = 0.01
Batch = 64
Dropout = 0.5
```

You changed three things.

You don't know which caused the improvement.

---

### ✅ Good

```text
Experiment 1
LR = 0.001
Batch = 32
Dropout = 0.2

Experiment 2
LR = 0.01
Batch = 32
Dropout = 0.2
```

Only LR changed.

Therefore the experiment gives you much stronger evidence about the effect of learning rate.

---

# 18. The Golden Rules 🏆

### Rule 1
**Always establish a baseline.**

### Rule 2
**Change one important variable at a time** when you want a simple, interpretable experiment.

### Rule 3
**Keep everything else constant.**

### Rule 4
**Use the same dataset split.**

### Rule 5
**Keep randomness controlled when possible.**

### Rule 6
**Track both training and validation metrics.**

### Rule 7
**Record every experiment.**

### Rule 8
**Don't judge a model from training accuracy alone.**

### Rule 9
**Compare experiments using the same evaluation protocol.**

### Rule 10
**Remember that variables can interact.**

# 🧠 Industry-Level Understanding

The real purpose of experiment design is **not simply finding the highest accuracy**.

It is answering:

> **"What caused this model to perform better?"**

A good ML engineer should be able to say:

```text
Baseline:
91.2% validation accuracy

Changed:
Learning rate only

Result:
92.5% validation accuracy

Conclusion:
Increasing LR from 0.001 → 0.005
improved validation performance under
the current experimental conditions.
```

That's much stronger than:

> "I tried some values and 0.005 worked." 😭😂

# 40.1 Hyperparameters

## What are Hyperparameters?

**Hyperparameters** are configuration values that control how a machine-learning model is trained.

They are **not learned automatically from the training data**.

Instead, **we choose them before or during training experiments**.

Examples:

- Learning rate
- Batch size
- Number of hidden units
- Number of layers
- Dropout rate
- Weight decay
- Optimizer

---

## Parameters vs Hyperparameters

This distinction is extremely important.

### Parameters

Parameters are values that the neural network **learns from data**.

Examples:

- Weights
- Biases

During training:

$$
\theta_{new}
=
\theta_{old}
-
\eta\nabla J(\theta)
$$

Here:

- $\theta$ = model parameters
- $\eta$ = learning rate
- $\nabla J(\theta)$ = gradient of loss

### Hyperparameters

Hyperparameters control **how those parameters are learned**.

For example:

```text
Learning rate = 0.001
Batch size = 32
Hidden units = 128
Number of layers = 4
Dropout = 0.3
Weight decay = 0.0001
Optimizer = Adam
```

---

# 🧠 Big Picture

Think of training like teaching a student.

```text
Training Data
     ↓
Neural Network
     ↓
Prediction
     ↓
Loss
     ↓
Gradient
     ↓
Optimizer
     ↓
Update Parameters
     ↓
Repeat
```

Hyperparameters decide **how this entire learning process behaves**.

---

# 1️⃣ Learning Rate

## What is Learning Rate?

The **learning rate** controls how large a step the optimizer takes when updating the model parameters.

The basic gradient descent update is:

$$
\theta_{new}
=
\theta_{old}
-
\eta\nabla J(\theta)
$$

Where:

- $\theta$ = model parameter
- $\eta$ = learning rate
- $\nabla J(\theta)$ = gradient
- $J(\theta)$ = loss function

---

## 🖼️ Gradient Descent Visualization

```markdown
![Gradient Descent](https://commons.wikimedia.org/wiki/Special:Redirect/file/Gradient_descent.png)
```

![Gradient Descent](https://commons.wikimedia.org/wiki/Special:Redirect/file/Gradient_descent.png)

The diagram shows the optimization process moving toward a lower-loss region. [Wikimedia Commons](https://commons.wikimedia.org/wiki/File%3AGradient_descent.png?utm_source=chatgpt.com)

---

## Intuition

Imagine you are walking down a mountain.

The **gradient** tells you:

> "Which direction is uphill?"

So you move in the **opposite direction**.

The learning rate tells you:

> "How big of a step should I take?"

---

## Small Learning Rate

Example:

$$
\eta = 0.00001
$$

Steps become very small.

```text
Loss
 ↓
5.0
4.9
4.8
4.7
4.6
...
```

### Problems

- Training becomes very slow
- May require many epochs
- Can waste computational resources

---

## Large Learning Rate

Example:

$$
\eta = 1.0
$$

The optimizer may take huge steps.

Instead of:

```text
       ↓
     minimum
```

it may behave like:

```text
↙       ↘
   ↙ ↘
↗       ↖
```

The optimizer may **overshoot the minimum**.

### Problems

- Loss may oscillate
- Training may become unstable
- Loss may explode
- Model may fail to converge

---

## Good Learning Rate

A good learning rate allows the model to:

```text
        ↓
      ↓
    ↓
   minimum
```

with reasonably large but controlled steps.

---

## Typical Values

For many deep-learning models, commonly tested values include:

```text
0.1
0.01
0.001
0.0001
```

But there is **no universally correct learning rate**.

It depends on:

- Model architecture
- Dataset
- Optimizer
- Batch size
- Loss landscape

---

## 🔥 Key Point

Learning rate is often one of the **most important hyperparameters**.

A great architecture with a bad learning rate can train terribly.

# 2️⃣ Batch Size

## What is Batch Size?

Batch size is the number of training examples processed before the model performs one parameter update.

Suppose:

```text
Dataset = 10,000 samples
Batch size = 100
```

Then approximately:

$$
\frac{10000}{100}=100
$$

updates occur in one epoch.

---

# Types of Gradient Descent

## Full-Batch Gradient Descent

Uses the **entire dataset** for one update.

```text
Dataset
████████████████████

       ↓

Gradient

       ↓

Update
```

If there are 100,000 samples:

```text
100,000 samples
      ↓
1 update
```

### Advantages

- Stable gradient estimate
- Deterministic for a fixed setup

### Disadvantages

- High memory requirement
- Slow updates for large datasets

---

# Stochastic Gradient Descent

Uses **one sample** at a time.

```text
Sample 1 → Update
Sample 2 → Update
Sample 3 → Update
Sample 4 → Update
...
```

For:

```text
100,000 samples
```

approximately:

```text
100,000 updates / epoch
```

### Advantage

Very frequent updates.

### Disadvantage

Gradient can be noisy.

---

# Mini-Batch Gradient Descent

Uses a small subset of the dataset.

Example:

```text
Dataset = 10,000

Batch size = 32
```

Approximately:

$$
\frac{10000}{32}\approx313
$$

updates per epoch.

This is the approach commonly used in modern deep learning.

---

# Common Batch Sizes

Typical choices:

```text
16
32
64
128
256
512
```

The appropriate value depends on:

- GPU memory
- Dataset
- Model
- Training stability
- Computational efficiency

---

# Batch Size Trade-Off

### Small Batch

```text
Batch = 16
```

Usually:

- More noisy gradients
- More updates
- Lower memory usage

### Large Batch

```text
Batch = 512
```

Usually:

- Smoother gradient estimates
- Fewer updates
- Higher memory usage

---

# 🔥 Important Relationship

Batch size and learning rate are often related.

Changing:

```text
Batch size
```

can change the behavior of optimization.

Therefore, when experimenting, don't blindly change many things simultaneously.

# 3️⃣ Number of Hidden Units

## What are Hidden Units?

A **hidden unit** is a neuron inside a hidden layer.

Example:

```text
Input Layer       Hidden Layer       Output

 ○                  ○
 ○                  ○
 ○                  ○
 ○                  ○                  ○
 ○                  ○
```

The number of neurons in a hidden layer determines its representational capacity.

# More Hidden Units

Suppose:

```text
Layer 1 = 32 neurons
Layer 2 = 32 neurons
```

Now increase to:

```text
Layer 1 = 256 neurons
Layer 2 = 256 neurons
```

The second network has substantially more capacity.

---

## More Capacity Can Help

If the model is too small:

```text
Underfitting
```

The model may fail to learn important patterns.

Increasing hidden units can improve its ability to represent complex relationships.

---

## But Too Many Can Hurt

Very large networks may:

- Overfit
- Require more memory
- Require more computation
- Take longer to train

Therefore:

```text
Too few neurons
      ↓
Underfitting

Good capacity
      ↓
Good generalization

Too many neurons
      ↓
Potential overfitting
```

---

# Example

```text
Input
 ↓
64 neurons
 ↓
Output
```

Experiment:

```text
Experiment A → 32 neurons
Experiment B → 64 neurons
Experiment C → 128 neurons
Experiment D → 256 neurons
```

Compare validation performance.

This is **hyperparameter experimentation**.

The number of hidden layers determines the **depth** of a neural network.

Example:

### Shallow Network

```text
Input
 ↓
Hidden
 ↓
Output
```

### Deeper Network

```text
Input
 ↓
Hidden 1
 ↓
Hidden 2
 ↓
Hidden 3
 ↓
Hidden 4
 ↓
Output
```

# Why Add More Layers?

Deep networks can learn **hierarchical representations**.

For example, in image recognition:

```text
Layer 1
Edges
 ↓
Layer 2
Shapes
 ↓
Layer 3
Parts
 ↓
Layer 4
Objects
```

The deeper layers can build upon representations learned by earlier layers.

---

# Too Few Layers

The model may not have enough capacity to represent a complex function.

Potential result:

```text
Underfitting
```

---

# Too Many Layers

A very deep network can introduce:

- More parameters
- More computation
- Optimization difficulty
- Greater risk of overfitting in some settings

Modern architectures often solve these problems using techniques such as:

- Residual connections
- Normalization
- Better initialization
- Advanced optimizers

---

This distinction is important.

### Hidden Units

Controls:

> **Width**

### Number of Layers

Controls:

> **Depth**

Example:

```text
Width = 128 neurons

Depth = 4 hidden layers
```

---

Increasing neurons:

```text
128 → 256
```

increases **width**.

Adding layers:

```text
4 → 6
```

increases **depth**.

# 5️⃣ Dropout

## What is Dropout?

**Dropout** is a regularization technique used to reduce overfitting.

During training, dropout randomly disables a fraction of neurons.

For example:

```text
Before dropout:

○ ○ ○ ○ ○ ○
│ │ │ │ │ │
○ ○ ○ ○ ○ ○
```

After dropout:

```text
○ ✕ ○ ○ ✕ ○
│   │ │   │
○ ○ ✕ ○ ○ ○
```

Some neurons are temporarily removed from the computation.

---

## 🖼️ Dropout Visualization

```markdown
![Neural Network Dropout](https://commons.wikimedia.org/wiki/Special:Redirect/file/Neural_Network_Dropout.svg)
```

![Neural Network Dropout](https://commons.wikimedia.org/wiki/Special:Redirect/file/Neural_Network_Dropout.svg)

This diagram illustrates dropout regularization in a neural network. [Wikimedia Commons](https://commons.wikimedia.org/wiki/File%3ANeural_Network_Dropout.svg?utm_source=chatgpt.com)

---

# Why Does Dropout Help?

Imagine a neural network always relying heavily on one particular neuron.

```text
Neuron A
   ↓
Prediction
```

The model may become overly dependent on that feature.

Dropout forces the network to work with different subsets of neurons.

Conceptually:

```text
Training iteration 1

A B C ✕ E
```

```text
Training iteration 2

✕ B C D E
```

```text
Training iteration 3

A ✕ C D ✕
```

This encourages more robust representations.

---

# Dropout Rate

Suppose:

$$
p = 0.5
$$

This means roughly **50% of the selected units are dropped during a training pass** under that dropout configuration.

Other common values might be:

```text
0.1
0.2
0.3
0.5
```

---

# Important

Dropout is primarily applied during **training**.

During inference/evaluation:

```text
Dropout OFF
```

The complete network is used.

Modern deep-learning frameworks handle the appropriate scaling/behavior automatically.

---

# Too Much Dropout

Suppose:

```text
Dropout = 0.8
```

The model may lose too much information during training.

Potential result:

```text
Underfitting
```

Therefore:

```text
Too little dropout
      ↓
Potential overfitting

Appropriate dropout
      ↓
Better generalization

Too much dropout
      ↓
Potential underfitting
```

# 6️⃣ Weight Decay

## What is Weight Decay?

**Weight decay** is a regularization technique that discourages model weights from becoming excessively large.

It is closely related to **L2 regularization**.

The regularized loss can be written as:

$$
\mathcal{L}_{total}
=
\mathcal{L}_{data}
+
\frac{\lambda}{2}
\|\theta\|_2^2
$$

Where:

- $\mathcal{L}_{data}$ = original training loss
- $\theta$ = model parameters
- $\lambda$ = regularization strength
- $\|\theta\|_2^2$ = squared L2 norm

---

A useful geometric visualization of L2 regularization/weight decay shows the regularizer pulling parameters toward smaller magnitudes. [Deep Learning Book](https://www.deeplearningbook.org/contents/regularization.html?utm_source=chatgpt.com)

For Jupyter, you can use the verified lecture PDF directly as a visual source:

```markdown
[![L2 Regularization / Weight Decay](https://cw.fel.cvut.cz/wiki/_media/courses/bev033dle/regularizers5.pdf)](https://cw.fel.cvut.cz/wiki/_media/courses/bev033dle/regularizers5.pdf)
```

**Important:** this is a PDF rather than a direct PNG/SVG image, so Jupyter may treat it as a clickable document rather than display it inline. I am deliberately not inventing a direct image URL for it.

---

# Intuition

Imagine the model has weights:

```text
10
8
6
5
```

Weight decay encourages the optimization process to prefer smaller weights when doing so does not significantly hurt the data loss.

Conceptually:

```text
Large weights
     ↓
Regularization pressure
     ↓
Smaller weights
```

---

# Weight Decay Update

A simplified update can be written as:

$$
\theta_{t+1}
=
(1-\eta\lambda)\theta_t
-
\eta\nabla\mathcal{L}(\theta_t)
$$

Where:

- $\eta$ = learning rate
- $\lambda$ = weight-decay coefficient
- $\nabla\mathcal{L}$ = gradient of the data loss

The factor:

$$
(1-\eta\lambda)
$$

causes the parameter magnitude to shrink.

This is why it is called **weight decay**. [CW Fel CVUT](https://cw.fel.cvut.cz/wiki/_media/courses/bev033dle/regularizers5.pdf?utm_source=chatgpt.com)

---

# Why Use Weight Decay?

It can help prevent the model from relying on excessively large parameters and can improve generalization.

Conceptually:

```text
Without regularization

Complex / large weights
        ↓
Potential overfitting
```

versus:

```text
With weight decay

Controlled weights
        ↓
Potentially better generalization
```

---

# Weight Decay vs Dropout

Both can help reduce overfitting, but they work differently.

| Technique | Main idea |
|---|---|
| Dropout | Randomly disables units during training |
| Weight decay | Penalizes large parameter values |

# 7️⃣ Optimizer

## What is an Optimizer?

An **optimizer** determines how model parameters are updated using the gradients.

The simplest optimizer is **Gradient Descent**.

The basic update is:

$$
\theta_{new}
=
\theta_{old}
-
\eta\nabla J(\theta)
$$

---

![Gradient Descent Optimization](https://commons.wikimedia.org/wiki/Special:Redirect/file/Gradient_descent.svg)

The visualization shows the optimization path toward a lower objective value. [Wikimedia Commons](https://commons.wikimedia.org/wiki/File%3AGradient_descent.svg?utm_source=chatgpt.com)

# Common Optimizers

Some important optimizers are:

```text
SGD
Momentum
RMSprop
Adam
AdamW
```

---

# SGD

Stochastic Gradient Descent updates parameters using gradients estimated from training data.

Basic form:

$$
\theta_{t+1}
=
\theta_t
-
\eta g_t
$$

where:

$$
g_t
=
\nabla_\theta\mathcal{L}(\theta_t)
$$

---

# Momentum

Momentum incorporates information from previous updates.

Conceptually:

```text
Current gradient
      +
Previous movement
      ↓
New update
```

This can help optimization move more smoothly through difficult loss landscapes.

---

# RMSprop

RMSprop adapts the effective learning rate based on a moving average of squared gradients.

Conceptually:

```text
Large historical gradients
        ↓
Smaller effective step

Small historical gradients
        ↓
Larger effective step
```

---

# Adam

**Adam = Adaptive Moment Estimation**

Adam combines ideas related to:

- Momentum / first moment
- Adaptive scaling / second moment

Conceptually:

```text
Gradient
   ↓
First-moment estimate
   +
Second-moment estimate
   ↓
Adaptive parameter update
```

Adam is a very common starting optimizer for many deep-learning experiments.

---

# AdamW

AdamW separates weight decay from the adaptive gradient update.

This distinction is important when studying **regularization and optimization**.

A common modern workflow is to compare:

```text
Adam
vs
AdamW
```

especially when experimenting with weight decay.

---

# 🔥 Hyperparameter Summary

| Hyperparameter | Controls | Too Low / Small | Too High / Large |
|---|---|---|---|
| Learning rate | Step size | Slow training | Instability / divergence |
| Batch size | Samples per update | Noisy gradients | High memory usage |
| Hidden units | Width | Underfitting | More computation / overfitting risk |
| Number of layers | Depth | Insufficient capacity | More complexity / optimization difficulty |
| Dropout | Regularization | Possible overfitting | Possible underfitting |
| Weight decay | Weight regularization | Weak regularization | Excessive regularization |
| Optimizer | Update strategy | Depends on choice | Depends on choice |

---

# 🧠 The Most Important Relationship

These hyperparameters **do not operate independently**.

For example:

```text
Learning Rate
      ↕
Batch Size
      ↕
Optimizer
      ↕
Weight Decay
      ↕
Network Architecture
      ↕
Dropout
```

Changing one can affect how the others behave.

---

# 🔥 Example Experiment

Suppose we have:

```text
Model:

Input
 ↓
128 neurons
 ↓
128 neurons
 ↓
Output
```

Initial configuration:

```python
learning_rate = 0.001
batch_size = 32
dropout = 0.2
weight_decay = 0.0001
optimizer = Adam
```

We could create:

### Experiment A

```text
LR = 0.001
Batch = 32
Dropout = 0.2
Weight decay = 0.0001
Optimizer = Adam
```

### Experiment B

```text
LR = 0.0001
Batch = 32
Dropout = 0.2
Weight decay = 0.0001
Optimizer = Adam
```

---

# 🚨 DON'T CHANGE EVERYTHING AT ONCE

Bad experiment:

```text
Experiment A

LR = 0.001
Batch = 32
Dropout = 0.2
Adam
128 neurons
```

Then:

```text
Experiment B

LR = 0.01
Batch = 128
Dropout = 0.5
AdamW
512 neurons
```

Now performance changed.

But **why?**

You don't know.

Was it:

- Learning rate?
- Batch size?
- Dropout?
- Optimizer?
- Network size?

You cannot isolate the cause.

---

# ✅ Better Experimentation

Change one important variable at a time.

```text
Baseline
   ↓
Change LR
   ↓
Compare
   ↓
Change Batch Size
   ↓
Compare
   ↓
Change Dropout
   ↓
Compare
```

This gives you a much better understanding of what actually affects your model.

---

# 🧪 Hyperparameter Tuning Workflow

A practical workflow is:

```text
1. Choose baseline
       ↓
2. Train model
       ↓
3. Measure validation performance
       ↓
4. Change one important hyperparameter
       ↓
5. Train again
       ↓
6. Compare
       ↓
7. Keep the better configuration
       ↓
8. Repeat
```

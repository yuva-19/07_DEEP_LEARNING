## 1. Why Manual Backpropagation?

Manual backpropagation connects everything from Blocks **6.1–6.4**:

```text
Forward Pass
     ↓
Prediction
     ↓
Loss
     ↓
Chain Rule
     ↓
Gradients
     ↓
Parameter Update
```

The purpose is **not** to manually train large neural networks in practice.

The purpose is to understand exactly what frameworks such as PyTorch are doing when you call:

```python
loss.backward()
```

---

# 2. Simple Single-Neuron Network 🔥🔥🔥

Consider a single linear neuron:

$$
z=wx+b
$$

Since there is no activation:

$$
\hat y=z
$$

Use squared loss:

$$
L=\frac12(\hat y-y)^2
$$

The computational graph is:

```text
x ──→ ×w ──→ z ──→ ŷ ──→ Loss
                 ↑
                 b
```

The backward path is:

```text
Loss
  ↓
ŷ
  ↓
z
 ↙ ↘
w   b
```

---

# 3. Numerical Example

Given:

$$
x=2
$$

$$
w=3
$$

$$
b=1
$$

$$
y=10
$$

---

## Forward Pass

Calculate:

$$
z=wx+b
$$

$$
z=(3)(2)+1=7
$$

Therefore:

$$
\hat y=7
$$

---

## Loss

$$
L=\frac12(\hat y-y)^2
$$

$$
L=\frac12(7-10)^2
$$

$$
\boxed{L=4.5}
$$

---

# 4. Calculate the Gradients

## Gradient with respect to prediction

$$
\frac{\partial L}{\partial\hat y}
=
\hat y-y
$$

Therefore:

$$
\frac{\partial L}{\partial\hat y}
=
7-10=-3
$$

---

## Gradient with respect to weight

The dependency path is:

$$
w\rightarrow z\rightarrow\hat y\rightarrow L
$$

Using the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial\hat y}
\frac{\partial\hat y}{\partial z}
\frac{\partial z}{\partial w}
$$

Since:

$$
\frac{\partial\hat y}{\partial z}=1
$$

and:

$$
\frac{\partial z}{\partial w}=x
$$

we get:

$$
\frac{\partial L}{\partial w}
=
(-3)(1)(2)
$$

$$
\boxed{
\frac{\partial L}{\partial w}=-6
}
$$

---

## Gradient with respect to bias

$$
\frac{\partial z}{\partial b}=1
$$

Therefore:

$$
\frac{\partial L}{\partial b}
=
(-3)(1)(1)
$$

$$
\boxed{
\frac{\partial L}{\partial b}=-3
}
$$

---

# 5. Update the Parameters

Choose:

$$
\eta=0.1
$$

### Weight

$$
w_{\text{new}}
=
w-\eta\frac{\partial L}{\partial w}
$$

$$
=3-(0.1)(-6)
$$

$$
\boxed{w_{\text{new}}=3.6}
$$

### Bias

$$
b_{\text{new}}
=
b-\eta\frac{\partial L}{\partial b}
$$

$$
=1-(0.1)(-3)
$$

$$
\boxed{b_{\text{new}}=1.3}
$$

---

# 6. Check Whether the Update Helped

New prediction:

$$
\hat y=(3.6)(2)+1.3
$$

$$
\hat y=8.5
$$

New loss:

$$
L=\frac12(8.5-10)^2
$$

$$
\boxed{L=1.125}
$$

Before:

$$
L=4.5
$$

After:

$$
L=1.125
$$

So the loss decreased:

$$
\boxed{4.5\rightarrow1.125}
$$

This demonstrates the complete learning process with one manual update. 🚀

# 7. Multi-Layer Manual Backpropagation 🔥🔥🔥

Now consider:

```text
Input
  ↓
Hidden Layer
  ↓
Output Layer
  ↓
Loss
```

We'll use:

* ReLU in the hidden layer
* Linear output
* Squared loss

### Forward equations

Hidden layer:

$$
z^{[1]}=w_1x+b_1
$$

$$
a^{[1]}=\operatorname{ReLU}(z^{[1]})
$$

Output:

$$
z^{[2]}=w_2a^{[1]}+b_2
$$

$$
\hat y=z^{[2]}
$$

Loss:

$$
L=\frac12(\hat y-y)^2
$$

---

# 8. Numerical Values

Use:

$$
x=2
$$

$$
w_1=2,\quad b_1=1
$$

$$
w_2=3,\quad b_2=1
$$

$$
y=20
$$

---

# 9. Forward Pass

### Hidden layer

$$
z^{[1]}=w_1x+b_1
$$

$$
=(2)(2)+1=5
$$

ReLU:

$$
a^{[1]}=\max(0,5)
$$

$$
\boxed{a^{[1]}=5}
$$

Since:

$$
z^{[1]}=5>0
$$

we have:

$$
\operatorname{ReLU}'(5)=1
$$

---

### Output layer

$$
z^{[2]}=w_2a^{[1]}+b_2
$$

$$
=(3)(5)+1=16
$$

Therefore:

$$
\hat y=16
$$

---

### Loss

$$
L=\frac12(16-20)^2
$$

$$
\boxed{L=8}
$$

---

# 10. Backpropagation Through Output Layer

Start from the loss:

$$
\frac{\partial L}{\partial\hat y}
=
\hat y-y
$$

$$
=16-20=-4
$$

Because:

$$
\hat y=z^{[2]}
$$

we get:

$$
\boxed{
\delta^{[2]}
=
\frac{\partial L}{\partial z^{[2]}}
=-4
}
$$

---

# 11. Output-Layer Weight Gradient

The output weight is $w_2$.

$$
z^{[2]}=w_2a^{[1]}+b_2
$$

Therefore:

$$
\frac{\partial z^{[2]}}{\partial w_2}=a^{[1]}
$$

So:

$$
\frac{\partial L}{\partial w_2}
=
\delta^{[2]}a^{[1]}
$$

$$
=(-4)(5)
$$

$$
\boxed{
\frac{\partial L}{\partial w_2}=-20
}
$$

---

# 12. Output-Layer Bias Gradient

Because:

$$
\frac{\partial z^{[2]}}{\partial b_2}=1
$$

we get:

$$
\boxed{
\frac{\partial L}{\partial b_2}=-4
}
$$

---

# 13. Propagate Gradient to Hidden Layer

Now we need:

$$
\frac{\partial L}{\partial a^{[1]}}
$$

Since:

$$
z^{[2]}=w_2a^{[1]}+b_2
$$

we have:

$$
\frac{\partial z^{[2]}}{\partial a^{[1]}}
=
w_2
$$

Therefore:

$$
\frac{\partial L}{\partial a^{[1]}}
=
\frac{\partial L}{\partial z^{[2]}}
\frac{\partial z^{[2]}}{\partial a^{[1]}}
$$

$$
=(-4)(3)
$$

$$
\boxed{
\frac{\partial L}{\partial a^{[1]}}=-12
}
$$

---

# 14. Pass Through ReLU

We know:

$$
a^{[1]}=\operatorname{ReLU}(z^{[1]})
$$

Since:

$$
z^{[1]}=5
$$

and:

$$
\operatorname{ReLU}'(5)=1
$$

then:

$$
\frac{\partial L}{\partial z^{[1]}}
=
\frac{\partial L}{\partial a^{[1]}}
\frac{\partial a^{[1]}}{\partial z^{[1]}}
$$

$$
=(-12)(1)
$$

Therefore:

$$
\boxed{
\delta^{[1]}=-12
}
$$

---

# 15. Hidden-Layer Weight Gradient

Since:

$$
z^{[1]}=w_1x+b_1
$$

we have:

$$
\frac{\partial z^{[1]}}{\partial w_1}=x
$$

Therefore:

$$
\frac{\partial L}{\partial w_1}
=
\delta^{[1]}x
$$

$$
=(-12)(2)
$$

$$
\boxed{
\frac{\partial L}{\partial w_1}=-24
}
$$

---

# 16. Hidden-Layer Bias Gradient

Since:

$$
\frac{\partial z^{[1]}}{\partial b_1}=1
$$

we get:

$$
\boxed{
\frac{\partial L}{\partial b_1}=-12
}
$$

---

# 17. Complete Gradient Table 🔥

| Parameter | Gradient |
| --------- | -------: |
| $w_1$     |    $-24$ |
| $b_1$     |    $-12$ |
| $w_2$     |    $-20$ |
| $b_2$     |     $-4$ |

So the entire network has now been differentiated manually.

# 20. Numerical Gradient Checking 🔥🔥

Manual derivatives can contain mistakes.

We can independently verify a gradient numerically.

For a parameter $\theta$, use the **central difference approximation**:

$$
\boxed{
\frac{\partial L}{\partial\theta}
\approx
\frac{
L(\theta+\epsilon)-L(\theta-\epsilon)
}{
2\epsilon
}
}
$$

where $\epsilon$ is a small number.

---

# 21. Why Numerical Gradient Checking Works

Suppose the loss is a function of $\theta$:

```text
Loss
 │
 │       /
 │      /
 │     /
 │____/________ θ
      θ
```

Evaluate the loss slightly to the right:

$$
L(\theta+\epsilon)
$$

and slightly to the left:

$$
L(\theta-\epsilon)
$$

The difference estimates the slope at $\theta$.

That slope is approximately:

$$
\frac{\partial L}{\partial\theta}
$$

---

# 22. Numerical Check on Our Single Neuron

Recall:

$$
L=\frac12(wx+b-y)^2
$$

with:

$$
x=2,\quad w=3,\quad b=1,\quad y=10
$$

Our analytical result was:

$$
\boxed{
\frac{\partial L}{\partial w}=-6
}
$$

Choose:

$$
\epsilon=0.0001
$$

Calculate:

$$
L(w+\epsilon)
$$

and:

$$
L(w-\epsilon)
$$

Then:

$$
\frac{\partial L}{\partial w}
\approx
\frac{
L(3.0001)-L(2.9999)
}{
0.0002
}
$$

which gives approximately:

$$
\boxed{-6}
$$

So:

$$
g_{\text{analytical}}
\approx
g_{\text{numerical}}
$$

Our derivative is therefore consistent.

---

# 23. Analytical vs Numerical Gradient

```text
                Gradient
                   │
          ┌────────┴────────┐
          ↓                 ↓
     Analytical         Numerical
       Gradient           Gradient
          │                 │
      Chain Rule       Finite Difference
          │                 │
          └────────┬────────┘
                   ↓
                Compare
                   ↓
             Should be close
```

### Analytical gradient

Derived using calculus:

$$
\frac{\partial L}{\partial\theta}
$$

### Numerical gradient

Approximated using:

$$
\frac{
L(\theta+\epsilon)-L(\theta-\epsilon)
}{
2\epsilon
}
$$

---

# 24. Why Gradient Checking Is Useful

Gradient checking is mainly a **debugging and verification technique**.

Useful when:

* implementing backpropagation manually
* implementing a custom layer
* implementing a custom loss
* writing custom differentiable operations
* debugging gradient calculations

It is **not normally used during every training iteration** because it requires additional forward evaluations and is computationally expensive.

---

# 25. Choosing $\epsilon$

A common starting point is something around:

$$
\epsilon=10^{-5}
$$

but there isn't one universally correct value.

### Too large

The approximation may no longer accurately represent the local derivative.

### Too small

Floating-point numerical errors can become significant.

So gradient checking uses a **small but numerically meaningful** $\epsilon$.

---

# 26. Relative Gradient Difference

For larger gradient vectors, compare them using a relative difference such as:

$$
\text{Difference}
=
\frac{
\|g_{\text{analytic}}-g_{\text{numeric}}\|
}{
\|g_{\text{analytic}}\|
+
\|g_{\text{numeric}}\|
}
$$

A very small value indicates that the two gradients agree closely.

---

# 27. Manual Backpropagation vs PyTorch

### Manual

You explicitly calculate:

$$
\frac{\partial L}{\partial w}
$$

using the chain rule.

### PyTorch

You write:

```python
loss.backward()
```

and automatic differentiation calculates the gradients.

Then:

```python
optimizer.step()
```

updates the parameters.

```text
MANUAL

Forward
   ↓
Loss
   ↓
Chain Rule
   ↓
Gradients
   ↓
Update


PYTORCH

Forward
   ↓
Loss
   ↓
loss.backward()
   ↓
Gradients
   ↓
optimizer.step()
```

The mathematics is the same. PyTorch is simply automating the differentiation.

# 28. Common Mistakes ❌

### 1. Forgetting the chain rule

A parameter usually affects the loss through several intermediate operations.

Always trace:

$$
\theta\rightarrow\cdots\rightarrow L
$$

and multiply the derivatives along the path.

---

### 2. Forgetting the activation derivative

For a hidden layer:

$$
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
$$

The activation derivative is essential.

---

### 3. Confusing $z$ and $a$

Remember:

$$
z=Wa+b
$$

then:

$$
a=f(z)
$$

So:

* $z$ = pre-activation
* $a$ = activation

---

### 4. Thinking gradients update parameters

Gradients tell us **how parameters should change**.

The update rule actually changes them:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta L
$$

---

### 5. Expecting numerical and analytical gradients to be exactly identical

Due to floating-point precision:

$$
g_{\text{analytic}}
\neq
g_{\text{numeric}}
$$

may occur by a tiny amount.

What matters is whether they are **sufficiently close**.

---

# 29. ⭐ Master Formula Sheet

### Forward

$$
z^{[l]}
=
W^{[l]}a^{[l-1]}+b^{[l]}
$$

$$
a^{[l]}=f(z^{[l]})
$$

### Output gradient

$$
\delta^{[L]}
=
\frac{\partial L}{\partial z^{[L]}}
$$

### Hidden gradient

$$
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
$$

### Weight gradient

$$
\frac{\partial L}{\partial W^{[l]}}
=
\delta^{[l]}(a^{[l-1]})^T
$$

### Bias gradient

$$
\frac{\partial L}{\partial b^{[l]}}
=
\delta^{[l]}
$$

### Parameter update

$$
W
\leftarrow
W-\eta\frac{\partial L}{\partial W}
$$

$$
b
\leftarrow
b-\eta\frac{\partial L}{\partial b}
$$

### Numerical gradient

$$
\boxed{
\frac{\partial L}{\partial\theta}
\approx
\frac{
L(\theta+\epsilon)-L(\theta-\epsilon)
}{
2\epsilon
}
}
$$

# 🧠 Final Mental Model

```text
             FORWARD
                ↓
        Calculate prediction
                ↓
              LOSS
                ↓
            BACKWARD
                ↓
          Chain Rule
                ↓
        Calculate gradients
                ↓
       Update W and b
                ↓
          Repeat training
```

And for debugging:

```text
Analytical Gradient
        ↕
Numerical Gradient
        ↓
   Compare them
```

### 🥇 One sentence to remember

> **Manual backpropagation means calculating gradients by applying the chain rule backward through every operation, updating the parameters using those gradients, and optionally using numerical gradient checking to verify that the derivatives are correct.**

## 1. Big Picture 🔥🔥🔥

Parameter updates are the **final stage of one training iteration**.

```text
Input
  ↓
Forward Propagation
  ↓
Prediction
  ↓
Loss
  ↓
Backpropagation
  ↓
Gradients
  ↓
Parameter Update
  ↓
Updated Weights & Biases
  ↓
Repeat
```

The central idea is:

$$
\boxed{
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
}
$$

where:

* $\theta$ = model parameters
* $\eta$ = learning rate
* $\nabla_\theta L$ = gradient of loss with respect to parameters

---

# 2. Gradient Calculation 🔥🔥🔥

Backpropagation calculates how the loss changes with respect to each parameter.

For weights:

$$
\frac{\partial L}{\partial W}
$$

For biases:

$$
\frac{\partial L}{\partial b}
$$

Together:

$$
\nabla_\theta L
$$

### What does the gradient tell us?

The gradient points toward the direction of **steepest local increase in loss**.

Therefore, to reduce loss, gradient descent moves in the opposite direction:

$$
-\nabla_\theta L
$$

### 🧠 Memory

> **Gradient = direction information**

> **Learning rate = step-size information**

---

# 3. Learning Rate 🔥🔥🔥

The learning rate is represented by:

$$
\eta
$$

It determines **how large the parameter update is**.

The basic rule is:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
$$

### Learning rate too small

```text
• → • → • → • → • → minimum
```

Very small steps.

**Result:** training can be very slow.

### Appropriate learning rate

```text
• →   • →    • →     • → minimum
```

Reasonable progress toward a lower-loss region.

### Learning rate too large

```text
      •
   ↙     ↘
 •         •
   ↖     ↙
      •
```

The optimizer may jump across the minimum.

**Possible result:**

* Overshooting
* Oscillation
* Instability
* Divergence

# 4. Weight Update 🔥🔥🔥

For one weight $w$:

$$
\boxed{
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial L}{\partial w}
}
$$

### Example

Suppose:

$$
w_{\text{old}}=5
$$

$$
\frac{\partial L}{\partial w}=2
$$

$$
\eta=0.1
$$

Then:

$$
w_{\text{new}}
=
5-(0.1)(2)
$$

$$
\boxed{w_{\text{new}}=4.8}
$$

So the weight moves from:

$$
5\rightarrow4.8
$$

because the gradient was positive.

---

# 5. What If the Gradient Is Negative?

Suppose:

$$
w_{\text{old}}=5
$$

$$
\frac{\partial L}{\partial w}=-2
$$

and:

$$
\eta=0.1
$$

Then:

$$
w_{\text{new}}
=
5-(0.1)(-2)
$$

$$
\boxed{w_{\text{new}}=5.2}
$$

The weight **increases**.

### Why?

Because:

$$
-\eta(-2)=+\eta(2)
$$

So:

> **The sign of the gradient determines the direction of the parameter update.**

---

# 6. Bias Update 🔥🔥

Biases are trainable parameters too.

For a bias:

$$
\boxed{
b_{\text{new}}
=
b_{\text{old}}
-
\eta
\frac{\partial L}{\partial b}
}
$$

### Example

Suppose:

$$
b_{\text{old}}=2
$$

$$
\frac{\partial L}{\partial b}=0.5
$$

$$
\eta=0.1
$$

Then:

$$
b_{\text{new}}
=
2-(0.1)(0.5)
$$

$$
\boxed{b_{\text{new}}=1.95}
$$

---

# 7. Weight and Bias Updates Together

For a neuron:

$$
z=wx+b
$$

Backpropagation gives:

$$
\frac{\partial L}{\partial w}
$$

and:

$$
\frac{\partial L}{\partial b}
$$

Then:

### Weight

$$
w
\leftarrow
w-\eta\frac{\partial L}{\partial w}
$$

### Bias

$$
b
\leftarrow
b-\eta\frac{\partial L}{\partial b}
$$

So:

```text
Backpropagation
      ↓
┌─────────────────┐
│ ∂L/∂w           │
│ ∂L/∂b           │
└─────────────────┘
      ↓
Parameter update
      ↓
┌─────────────────┐
│ New weight      │
│ New bias        │
└─────────────────┘
```

---

# 8. Matrix Parameter Updates 🔥🔥🔥

Real neural networks contain many parameters.

For layer $l$:

$$
z^{[l]}
=
W^{[l]}a^{[l-1]}+b^{[l]}
$$

The weight matrix is updated as:

$$
\boxed{
W^{[l]}
\leftarrow
W^{[l]}
-
\eta
\frac{\partial L}{\partial W^{[l]}}
}
$$

Bias:

$$
\boxed{
b^{[l]}
\leftarrow
b^{[l]}
-
\eta
\frac{\partial L}{\partial b^{[l]}}
}
$$

All weights and biases in the layer can therefore be updated together using vectorized operations.

---

# 9. Connection to Block 6.3 🔥🔥🔥

This is extremely important.

In **6.3**, we calculated:

$$
\frac{\partial L}{\partial W^{[l]}}
$$

and:

$$
\frac{\partial L}{\partial b^{[l]}}
$$

In **6.4**, we use them.

```text
6.3 Backpropagation
        ↓
Calculate gradients
        ↓
∂L/∂W     ∂L/∂b
        ↓
6.4 Parameter Update
        ↓
W ← W - η(∂L/∂W)
b ← b - η(∂L/∂b)
```

### 🧠 The distinction

**Backpropagation:**

> "How should each parameter change?"

**Parameter update:**

> "Actually change the parameters."

---

# 10. Complete Training Step 🔥🔥🔥

A single training iteration can be represented as:

### Step 1 — Forward

$$
z=Wx+b
$$

$$
a=f(z)
$$

### Step 2 — Loss

$$
L=L(y,\hat y)
$$

### Step 3 — Backpropagation

Calculate:

$$
\frac{\partial L}{\partial W}
$$

$$
\frac{\partial L}{\partial b}
$$

### Step 4 — Update

$$
W
\leftarrow
W-\eta\frac{\partial L}{\partial W}
$$

$$
b
\leftarrow
b-\eta\frac{\partial L}{\partial b}
$$

### Step 5 — Repeat

Use the new parameters for the next training iteration.

---

# 11. Complete Numerical Example

Suppose:

$$
w=4
$$

$$
b=2
$$

Backpropagation produces:

$$
\frac{\partial L}{\partial w}=3
$$

$$
\frac{\partial L}{\partial b}=-2
$$

Learning rate:

$$
\eta=0.1
$$

### Weight update

$$
w_{\text{new}}
=
4-(0.1)(3)
$$

$$
\boxed{w_{\text{new}}=3.7}
$$

### Bias update

$$
b_{\text{new}}
=
2-(0.1)(-2)
$$

$$
\boxed{b_{\text{new}}=2.2}
$$

Therefore:

```text
BEFORE

w = 4
b = 2

       ↓
   Backpropagation

∂L/∂w = +3
∂L/∂b = -2

       ↓
   Parameter Update

w = 3.7
b = 2.2
```

---

# 12. Why the Update Uses the Negative Gradient

Suppose:

$$
\frac{\partial L}{\partial w}>0
$$

A positive gradient means that locally, increasing $w$ tends to increase the loss.

So we decrease $w$:

$$
w_{\text{new}}
=
w-\eta\frac{\partial L}{\partial w}
$$

If:

$$
\frac{\partial L}{\partial w}<0
$$

then increasing $w$ tends to reduce the loss locally, so the update increases $w$.

Thus:

$$
\boxed{
\text{Gradient descent moves opposite to the gradient}
}
$$

# 14. PyTorch Connection 💻

The mathematics maps directly to the standard PyTorch training loop:

```python
optimizer.zero_grad()

output = model(x)

loss = loss_function(output, y)

loss.backward()

optimizer.step()
```

### What happens?

| Code                    | Mathematical operation   |
| ----------------------- | ------------------------ |
| `model(x)`              | Forward propagation      |
| `loss_function(...)`    | Calculate $L$            |
| `loss.backward()`       | Calculate gradients      |
| `optimizer.step()`      | Update parameters        |
| `optimizer.zero_grad()` | Clear previous gradients |

Conceptually:

```text
loss.backward()
       ↓
∂L/∂W
∂L/∂b
       ↓
optimizer.step()
       ↓
Wnew
bnew
```

# 15. Parameters vs Gradients vs Learning Rate

| Term          | Meaning                        |
| ------------- | ------------------------------ |
| Parameter     | Value the model learns         |
| Weight        | Connection strength            |
| Bias          | Adjustable offset              |
| Gradient      | Direction/sensitivity of loss  |
| Learning rate | Size of parameter update       |
| Optimizer     | Algorithm that applies updates |

### 🧠 Easy memory trick

$$
\boxed{
\text{Parameter + Gradient + Learning Rate}
\rightarrow
\text{New Parameter}
}
$$

# 16. Common Mistakes ❌

### 1. "Learning rate determines direction"

❌ No.

Gradient determines the direction.

Learning rate determines the magnitude.

---

### 2. "Higher learning rate always means faster learning"

❌ No.

Too large can cause instability or divergence.

---

### 3. "Only weights are updated"

❌ No.

Both:

$$
W
$$

and:

$$
b
$$

are trainable parameters.

---

### 4. "Backpropagation updates the weights"

❌ Not directly.

Backpropagation calculates:

$$
\nabla_\theta L
$$

The optimizer applies the update.

---

### 5. "Gradient and learning rate are the same"

❌ No.

Gradient:

> **Which way?**

Learning rate:

> **How far?**

---

# 17. 🔥 Interview / GATE Points

### What is the parameter update equation?

$$
\boxed{
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
}
$$

### What does the learning rate control?

**Step size of the parameter update.**

### What happens when learning rate is too small?

Training becomes slow.

### What happens when learning rate is too large?

The optimization may overshoot, oscillate, become unstable, or diverge.

### Does backpropagation update parameters?

No. It computes gradients.

### What updates the parameters?

The optimizer.

### General parameter update

$$

$$

$$

$$

$$

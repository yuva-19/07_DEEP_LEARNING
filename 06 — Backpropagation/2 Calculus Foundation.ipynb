# 📚 Block 6.2 — Calculus Foundation for Backpropagation

## 1. Core Idea

Backpropagation is built mainly on **three mathematical ideas**:

1. **Partial derivatives** → tell us how sensitive the loss is to each variable.
2. **Chain rule** → connects derivatives through multiple operations.
3. **Computational graphs** → organize the operations so gradients can be calculated efficiently.

### The big picture

$$
\text{Forward Pass}
\rightarrow
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradients}
\rightarrow
\text{Parameter Update}
$$

And mathematically:

$$
\boxed{\text{Backpropagation} \approx \text{Chain Rule applied through a Computational Graph}}
$$

---

# 2. Partial Derivatives 🔥🔥🔥

A **partial derivative** measures how a function changes with respect to **one variable**, while treating the other variables as constant.

Consider:

$$
L(x,y)=x^2+3y
$$

Partial derivative with respect to $x$:

$$
\frac{\partial L}{\partial x}=2x
$$

Partial derivative with respect to $y$:

$$
\frac{\partial L}{\partial y}=3
$$

So:

$$
\nabla L=
\begin{bmatrix}
\frac{\partial L}{\partial x}\\
\frac{\partial L}{\partial y}
\end{bmatrix}
=
\begin{bmatrix}
2x\\
3
\end{bmatrix}
$$

### 🧠 Neural-network meaning

A neural network has thousands or millions of parameters:

$$
\theta=[w_1,w_2,w_3,\ldots,b_1,b_2,\ldots]
$$

We need to know:

$$
\frac{\partial L}{\partial w_1},
\frac{\partial L}{\partial w_2},
\frac{\partial L}{\partial w_3},\ldots
$$

These tell us:

> **"If I slightly change this parameter, how does the loss change?"**

That is exactly the information the optimizer needs.

---

# 3. Gradient

All the partial derivatives are collected into a **gradient vector**.

$$
\nabla_\theta L=
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots\\
\frac{\partial L}{\partial w_n}
\end{bmatrix}
$$

### Important distinction

**Partial derivative**

→ sensitivity with respect to **one variable**.

**Gradient**

→ collection of sensitivities with respect to **all parameters**.

The gradient tells us the direction of **steepest increase** of the loss.

Therefore, gradient descent moves in the opposite direction:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
$$

---

# 4. Chain Rule 🔥🔥🔥

This is the **most important mathematical idea for backpropagation**.

Suppose:

$$
x\rightarrow z\rightarrow y
$$

where:

$$
z=g(x)
$$

and

$$
y=f(z)
$$

Then:

$$
\frac{dy}{dx}
=
\frac{dy}{dz}
\frac{dz}{dx}
$$

The key idea:

> When one variable depends on another through intermediate variables, multiply the derivatives along the path.

### Simple example

Suppose:

$$
z=2x
$$

and:

$$
y=z^2
$$

We know:

$$
\frac{dy}{dz}=2z
$$

and:

$$
\frac{dz}{dx}=2
$$

Therefore:

$$
\frac{dy}{dx}
=
\frac{dy}{dz}
\frac{dz}{dx}
$$

$$
=2z\times2
$$

Since $z=2x$:

$$
\frac{dy}{dx}=8x
$$

That's the chain rule.

---

# 5. Why Chain Rule Matters in Neural Networks 🔥🔥🔥

A neural network isn't just:

$$
x\rightarrow L
$$

There are many intermediate computations.

For one neuron:

$$
x\rightarrow z\rightarrow a\rightarrow L
$$

where:

$$
z=wx+b
$$

$$
a=f(z)
$$

$$
L=L(a,y)
$$

We want:

$$
\frac{\partial L}{\partial w}
$$

But $w$ doesn't directly affect $L$.

It affects:

$$
w\rightarrow z\rightarrow a\rightarrow L
$$

So we apply the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
\frac{\partial z}{\partial w}
$$

Since:

$$
z=wx+b
$$

we have:

$$
\frac{\partial z}{\partial w}=x
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
x
}
$$

Similarly:

$$
\frac{\partial z}{\partial b}=1
$$

so:

$$
\boxed{
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
}
$$

This is the mathematical foundation of backpropagation.

---

# 6. Computational Graph 🔥🔥🔥

A **computational graph** represents mathematical operations as connected nodes.

For a neuron:

```text
       x ─────┐
              ↓
             × w
              ↓
              z
              ↓
        Activation f()
              ↓
              a
              ↓
            Loss
              ↓
              L
```

More explicitly:

```text
x ────────┐
          ↓
        z = wx + b
          ↑
w ────────┤
          ↑
b ────────┘
          │
          ↓
       a = f(z)
          │
          ↓
       L(a, y)
          │
          ↓
          L
```

The graph tells us how one variable depends on another.

---

# 7. Forward Pass vs Backward Pass

### Forward pass

Information moves:

$$
x\rightarrow z\rightarrow a\rightarrow L
$$

We calculate the prediction and loss.

### Backward pass

Gradient information moves:

$$
L\rightarrow a\rightarrow z\rightarrow w,b
$$

We calculate how the loss changes with respect to each parameter.

# 8. Full Computational Example 🔥🔥🔥

Consider:

$$
z=wx+b
$$

$$
a=f(z)
$$

$$
L=(a-y)^2
$$

The computational graph is:

```text
x, w, b
   ↓
z = wx + b
   ↓
a = f(z)
   ↓
L = (a-y)²
```

We want:

$$
\frac{\partial L}{\partial w}
$$

The path is:

$$
w\rightarrow z\rightarrow a\rightarrow L
$$

Therefore:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
\frac{\partial z}{\partial w}
$$

We already know:

$$
\frac{\partial z}{\partial w}=x
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
x
}
$$

Likewise:

$$
\boxed{
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
}
$$

This demonstrates exactly how the gradient travels backward through the network.

---

# 9. Computational Graph + Chain Rule = Backpropagation

This is the most important connection in this block:

```text
Partial derivatives
       ↓
Chain rule
       ↓
Computational graph
       ↓
Backpropagation
       ↓
Gradients
       ↓
Optimizer
       ↓
Updated parameters
```

So don't memorize backpropagation as some mysterious algorithm.

Understand it as:

> **Backpropagation efficiently applies the chain rule through the computational graph to calculate gradients of the loss with respect to model parameters.**

🔥 This sentence is worth remembering.

---

# 10. Connection to PyTorch

This is where the mathematics becomes practical.

When you write:

```python
loss.backward()
```

PyTorch performs **automatic differentiation** through the computation graph and calculates gradients.

Conceptually:

```text
Forward
   ↓
Prediction
   ↓
Loss
   ↓
loss.backward()
   ↓
Gradients
   ↓
optimizer.step()
   ↓
Updated parameters
```

For example:

```python
loss.backward()
optimizer.step()
```

means:

* `loss.backward()` → calculate gradients
* `optimizer.step()` → update parameters using those gradients

And:

```python
optimizer.zero_grad()
```

clears previously accumulated gradients.

### Important distinction

**Backpropagation does NOT update weights.**

It calculates gradients.

**The optimizer updates weights.**

# 11. The Complete Training Pipeline 🔥🔥🔥

Put everything from the previous blocks together:

```text
Input Data
    ↓
Forward Propagation
    ↓
Prediction
    ↓
Loss Function
    ↓
Computational Graph
    ↓
Backpropagation
    ↓
Gradients
    ↓
Optimizer
    ↓
Updated Parameters
    ↓
Repeat
```

Mathematically:

$$
\hat y=f(x;\theta)
$$

Calculate:

$$
L(y,\hat y)
$$

Then:

$$
\nabla_\theta L
$$

Then update:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta L
$$

Repeat for many batches and epochs.

# 12. Partial Derivative vs Gradient vs Chain Rule

| Concept             | Meaning                                             |
| ------------------- | --------------------------------------------------- |
| Partial derivative  | Change in output with respect to one variable       |
| Gradient            | Collection of all relevant partial derivatives      |
| Chain rule          | Connects derivatives through intermediate variables |
| Computational graph | Represents mathematical operations and dependencies |
| Backpropagation     | Efficiently applies chain rule through the graph    |
| Optimizer           | Uses gradients to update parameters                 |

### 🧠 One-line memory

> **Partial derivatives give the pieces → chain rule connects the pieces → computational graph organizes them → backpropagation calculates them efficiently → optimizer uses them.**

# 13. Common Mistakes ❌

### Mistake 1: Backpropagation updates weights

❌ Wrong.

Backpropagation calculates:

$$
\frac{\partial L}{\partial\theta}
$$

The optimizer performs:

$$
\theta\leftarrow\theta-\eta\nabla_\theta L
$$

---

### Mistake 2: Gradient points toward the minimum

❌ Not exactly.

The gradient points toward the direction of **steepest increase**.

Gradient descent uses:

$$
-\nabla L
$$

to move downhill.

---

### Mistake 3: Partial derivative and gradient are identical

Not exactly.

A partial derivative is generally one component:

$$
\frac{\partial L}{\partial w_1}
$$

The gradient contains many:

$$
\nabla_\theta L=
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots
\end{bmatrix}
$$

---

### Mistake 4: Chain rule is only a calculus topic

❌ For deep learning, chain rule is fundamental.

It allows gradients to travel through:

$$
L\rightarrow a^{[L]}\rightarrow z^{[L]}
\rightarrow a^{[L-1]}\rightarrow\cdots
\rightarrow W^{[1]}
$$

# 1. What Is Backpropagation Through Layers? 🔥🔥🔥

A neural network performs calculations layer by layer during the **forward pass**:

$$
x
\rightarrow
z^{[1]}
\rightarrow
a^{[1]}
\rightarrow
z^{[2]}
\rightarrow
a^{[2]}
\rightarrow
\cdots
\rightarrow
L
$$

After calculating the loss, we need to determine:

> **How much did each layer's weights and biases contribute to the loss?**

Backpropagation answers this by moving **gradient information backward** through the network.

$$
L
\rightarrow
\delta^{[L]}
\rightarrow
\delta^{[L-1]}
\rightarrow
\delta^{[L-2]}
\rightarrow
\cdots
$$

The standard layer-wise equations follow directly from the chain rule. ([Bartosz Naskręcki's Home Page][1])

---

# 2. Forward Pass Refresher

For layer $l$:

$$
z^{[l]}=W^{[l]}a^{[l-1]}+b^{[l]}
$$

Then:

$$
a^{[l]}=f^{[l]}(z^{[l]})
$$

where:

* $W^{[l]}$ = weights
* $b^{[l]}$ = biases
* $z^{[l]}$ = weighted sum / pre-activation
* $a^{[l]}$ = activation
* $a^{[l-1]}$ = previous layer's output

For the input:

$$
a^{[0]}=x
$$

---

# 3. The Complete Forward → Backward Picture

```text
                 FORWARD
                   ↓

Input
  ↓
Layer 1
  ↓
Layer 2
  ↓
Layer 3
  ↓
Prediction
  ↓
Loss

                   ↑
                   │
              BACKPROP
                   │
                   ↑

Layer 1 ← Layer 2 ← Layer 3 ← Loss
```

### Forward pass

Calculates:

$$
\hat y
$$

and then:

$$
L(y,\hat y)
$$

### Backward pass

Calculates:

$$
\frac{\partial L}{\partial W^{[l]}}
$$

and:

$$
\frac{\partial L}{\partial b^{[l]}}
$$

for every layer.

---

# 4. Output-Layer Gradient 🔥🔥🔥

Start from the output because the loss is directly connected to the output.

For the final layer:

$$
z^{[L]}=W^{[L]}a^{[L-1]}+b^{[L]}
$$

$$
a^{[L]}=f(z^{[L]})
$$

and:

$$
L=L(a^{[L]},y)
$$

We first calculate:

$$
\frac{\partial L}{\partial z^{[L]}}
$$

Using the chain rule:

$$
\frac{\partial L}{\partial z^{[L]}}
=
\frac{\partial L}{\partial a^{[L]}}
\frac{\partial a^{[L]}}{\partial z^{[L]}}
$$

Define:

$$
\boxed{
\delta^{[L]}
=
\frac{\partial L}{\partial z^{[L]}}
}
$$

This $\delta$ is commonly called the **error signal** or **backpropagated error at the pre-activation**.

---

# 5. Example: Binary Classification Output

For binary classification:

$$
a^{[L]}=\sigma(z^{[L]})
$$

Using binary cross-entropy:

$$
L=
-\left[
y\log(a^{[L]})
+
(1-y)\log(1-a^{[L]})
\right]
$$

A very useful simplification is:

$$
\boxed{
\delta^{[L]}=a^{[L]}-y
}
$$

So if:

$$
a^{[L]}=0.8
$$

and:

$$
y=1
$$

then:

$$
\delta^{[L]}=0.8-1
$$

$$
\delta^{[L]}=-0.2
$$

This is the gradient signal that starts propagating backward.

---

# 6. Weight Gradients 🔥🔥🔥

For a layer:

$$
z^{[l]}
=
W^{[l]}a^{[l-1]}+b^{[l]}
$$

We want:

$$
\frac{\partial L}{\partial W^{[l]}}
$$

The result for a single example is:

$$
\boxed{
\frac{\partial L}{\partial W^{[l]}}
=
\delta^{[l]}(a^{[l-1]})^T
}
$$

### 🧠 Intuition

Think:

$$
\boxed{
\text{Weight gradient}
=
\text{current error signal}
\times
\text{incoming activation}
}
$$

Why?

A weight determines how strongly the previous activation contributes to the current neuron.

So the gradient depends on:

1. **How much error is present at this neuron**
2. **How much input came through that connection**

---

# 7. Single-Neuron Version

For a single neuron:

$$
z=wx+b
$$

Suppose:

$$
\delta=\frac{\partial L}{\partial z}
$$

Then:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial z}
\frac{\partial z}{\partial w}
$$

Since:

$$
\frac{\partial z}{\partial w}=x
$$

we get:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\delta x
}
$$

The matrix equation:

$$
\frac{\partial L}{\partial W}
=
\delta a^T
$$

is simply the vectorized version of this idea.

---

# 8. Bias Gradients 🔥🔥🔥

For:

$$
z=Wa+b
$$

the derivative of $z$ with respect to $b$ is:

$$
\frac{\partial z}{\partial b}=1
$$

Therefore:

$$
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial z}
$$

Since:

$$
\delta=\frac{\partial L}{\partial z}
$$

we get:

$$
\boxed{
\frac{\partial L}{\partial b}
=
\delta
}
$$

### 🧠 Memory trick

> **Weight gradient = error × input**

> **Bias gradient = error**

For a batch, the bias gradients are accumulated/averaged across examples according to the loss reduction being used.

---

# 9. Hidden-Layer Gradients 🔥🔥🔥

This is the most important part of **backpropagation through layers**.

A hidden layer isn't directly connected to the loss.

So how does it receive a gradient?

Through the **next layer**.

Consider:

```text
Hidden Layer
     ↓
Output Layer
     ↓
   Loss
```

The hidden layer affects the output layer, which affects the loss.

Therefore:

$$
L
\rightarrow
z^{[L]}
\rightarrow
a^{[L-1]}
\rightarrow
z^{[L-1]}
$$

The chain rule connects these dependencies.

---

# 10. Hidden-Layer Error Formula

For a hidden layer $l$:

$$
\boxed{
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
}
$$

This is one of the **most important equations in backpropagation**.

### Meaning of each part

$$
(W^{[l+1]})^T
$$

→ sends gradient information backward through the next layer's weights.

$$
\delta^{[l+1]}
$$

→ error signal from the next layer.

$$
f'(z^{[l]})
$$

→ derivative of the current layer's activation.

$$
\odot
$$

→ element-wise multiplication.

---

# 11. Why Is the Weight Matrix Transposed?

Forward propagation:

$$
z^{[l+1]}
=
W^{[l+1]}a^{[l]}+b^{[l+1]}
$$

Backward propagation:

$$
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot f'(z^{[l]})
$$

Conceptually:

```text
FORWARD

a[l] ──── W[l+1] ────→ z[l+1]


BACKWARD

δ[l] ←── W[l+1]ᵀ ───── δ[l+1]
```

The transpose allows the gradient from the next layer to be mapped back onto the dimensions of the previous layer.

This is also why **shape checking** is extremely important when implementing backpropagation.

---

# 12. Why Does the Activation Derivative Appear?

Remember:

$$
a^{[l]}=f(z^{[l]})
$$

Therefore:

$$
\frac{\partial a^{[l]}}
{\partial z^{[l]}}
=
f'(z^{[l]})
$$

So while moving backward through a layer, the gradient must pass through the activation function.

That's why:

$$
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
$$

---

# 13. Connection to Activation Functions

This connects directly to what we studied in **Block 3**.

### ReLU

$$
f(z)=\max(0,z)
$$

Derivative:

$$
f'(z)=
\begin{cases}
0 & z<0\\
1 & z>0
\end{cases}
$$

If:

$$
z<0
$$

then:

$$
f'(z)=0
$$

Therefore:

$$
\delta^{[l]}\approx0
$$

for that path.

This is why activation derivatives are so important for gradient flow.

---

# 14. Hidden-Layer Weight Gradient

Once we calculate:

$$
\delta^{[l]}
$$

the weight gradient becomes:

$$
\boxed{
\frac{\partial L}{\partial W^{[l]}}
=
\delta^{[l]}(a^{[l-1]})^T
}
$$

And bias:

$$
\boxed{
\frac{\partial L}{\partial b^{[l]}}
=
\delta^{[l]}
}
$$

Notice something beautiful:

> **The same weight-gradient and bias-gradient formulas work for every layer.**

Only the way we calculate $\delta$ changes.

---

# 15. The Complete Layer-by-Layer Algorithm 🔥🔥🔥

For the final layer:

$$
\boxed{
\delta^{[L]}
=
\frac{\partial L}{\partial a^{[L]}}
\odot
f'(z^{[L]})
}
$$

Then for hidden layers:

$$
\boxed{
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
}
$$

Then for every layer:

$$
\boxed{
\frac{\partial L}{\partial W^{[l]}}
=
\delta^{[l]}(a^{[l-1]})^T
}
$$

$$
\boxed{
\frac{\partial L}{\partial b^{[l]}}
=
\delta^{[l]}
}
$$

This is the core mathematical algorithm.

The general $L$-layer formulation is consistent with standard backpropagation derivations from university deep-learning materials. ([Bartosz Naskręcki's Home Page][1])

# 17. Complete Example: Input → Hidden → Output

Consider:

```text
Input
  ↓
Hidden Layer
  ↓
Output Layer
  ↓
Loss
```

Forward:

$$
z^{[1]}=W^{[1]}x+b^{[1]}
$$

$$
a^{[1]}=f(z^{[1]})
$$

$$
z^{[2]}=W^{[2]}a^{[1]}+b^{[2]}
$$

$$
a^{[2]}=f(z^{[2]})
$$

$$
L=L(a^{[2]},y)
$$

### Step 1 — Output gradient

$$
\delta^{[2]}
=
\frac{\partial L}{\partial z^{[2]}}
$$

### Step 2 — Output weight gradient

$$
\frac{\partial L}{\partial W^{[2]}}
=
\delta^{[2]}(a^{[1]})^T
$$

### Step 3 — Output bias gradient

$$
\frac{\partial L}{\partial b^{[2]}}
=
\delta^{[2]}
$$

### Step 4 — Hidden gradient

$$
\delta^{[1]}
=
(W^{[2]})^T
\delta^{[2]}
\odot
f'(z^{[1]})
$$

### Step 5 — Hidden weight gradient

$$
\frac{\partial L}{\partial W^{[1]}}
=
\delta^{[1]}x^T
$$

### Step 6 — Hidden bias gradient

$$
\frac{\partial L}{\partial b^{[1]}}
=
\delta^{[1]}
$$

Now every trainable parameter has a gradient.

---

# 18. Numerical Mini-Example

Suppose the hidden layer has one neuron.

Input:

$$
x=2
$$

Hidden error signal:

$$
\delta^{[1]}=0.1
$$

Then:

$$
\frac{\partial L}{\partial w^{[1]}}
=
\delta^{[1]}x
$$

$$
=0.1\times2
$$

$$
=0.2
$$

Bias gradient:

$$
\frac{\partial L}{\partial b^{[1]}}
=
0.1
$$

So:

```text
Hidden error = 0.1

Input = 2
   ↓
Weight gradient = 0.1 × 2 = 0.2

Bias gradient = 0.1
```

This simple relationship is the foundation of the matrix formula.

# 19. Why Hidden Neurons Receive Multiple Gradient Contributions

Here's an important detail when the network becomes wider.

Suppose one hidden neuron connects to **three neurons in the next layer**:

```text
              → Output 1
Hidden neuron
              → Output 2
              → Output 3
```

That hidden neuron affects the loss through **three different paths**.

Therefore, its gradient receives contributions from all those paths.

Conceptually:

$$
\frac{\partial L}{\partial a^{[l]}}
=
\sum_j
\frac{\partial L}{\partial z_j^{[l+1]}}
\frac{\partial z_j^{[l+1]}}
{\partial a^{[l]}}
$$

The matrix multiplication:

$$
(W^{[l+1]})^T\delta^{[l+1]}
$$

performs this aggregation efficiently.

This is one reason matrix/vectorized backpropagation is so powerful. ([InDepth][3])

# 20. Backpropagation Flow — The Mental Model 🧠

Remember this picture:

```text
                    LOSS
                     │
                     ↓
              Output error δ[L]
                 ↙         ↘
                ↓           ↓
          ∂L/∂W[L]      ∂L/∂b[L]
                │
                ↓
       propagate through W[L]ᵀ
                │
                ↓
          Hidden error δ[L-1]
                 ↙       ↘
                ↓         ↓
        ∂L/∂W[L-1]   ∂L/∂b[L-1]
                │
                ↓
              repeat
                │
                ↓
           Earlier layers
```

### The golden rule 🥇

> **Calculate the error signal for a layer → use it to calculate that layer's weight and bias gradients → move one layer backward.**

---

# 21. Backpropagation vs Gradient Descent

These two are often confused.

### Backpropagation

Calculates:

$$
\frac{\partial L}{\partial W}
$$

and:

$$
\frac{\partial L}{\partial b}
$$

### Gradient descent / optimizer

Uses those gradients:

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\frac{\partial L}{\partial W}
$$

$$
b_{\text{new}}
=
b_{\text{old}}
-
\eta
\frac{\partial L}{\partial b}
$$

So:

```text
Backpropagation
      ↓
Calculates gradients
      ↓
Optimizer
      ↓
Updates parameters
```

**Backpropagation does not itself update the weights.**

---

# 22. PyTorch Connection 💻

In PyTorch, you normally don't manually calculate these derivatives.

```python
loss.backward()
```

performs automatic differentiation through the computation graph.

Then:

```python
optimizer.step()
```

updates the parameters.

And:

```python
optimizer.zero_grad()
```

clears the previous gradients.

So:

```python
optimizer.zero_grad()

output = model(x)
loss = loss_function(output, y)

loss.backward()
optimizer.step()
```

maps conceptually to:

```text
Forward
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradients
   ↓
Optimizer
   ↓
Updated weights
```

# 23. Common Mistakes ❌

### ❌ 1. "Backpropagation updates weights"

No.

**Backpropagation → gradients**

**Optimizer → parameter updates**

---

### ❌ 2. Forgetting the activation derivative

For hidden layers:

$$
\delta^{[l]}
=
(W^{[l+1]})^T
\delta^{[l+1]}
\odot
f'(z^{[l]})
$$

The $f'(z)$ term is essential.

---

### ❌ 3. Forgetting the transpose

The hidden-layer equation uses:

$$
(W^{[l+1]})^T
$$

not simply:

$$
W^{[l+1]}
$$

when using this column-vector convention.

---

### ❌ 4. Thinking bias gradient is zero

Bias has a gradient:

$$
\frac{\partial L}{\partial b}=\delta
$$

Biases are trainable parameters.

---

### ❌ 5. Confusing $a$ and $z$

Remember:

$$
z=Wa+b
$$

then:

$$
a=f(z)
$$

So:

* $z$ = before activation
* $a$ = after activation

The error signal $\delta$ is conventionally:

$$
\delta=\frac{\partial L}{\partial z}
$$

```text
FORWARD
─────────────────────────────
z[l] = W[l] a[l-1] + b[l]
a[l] = f(z[l])


BACKWARD
─────────────────────────────
Output:
δ[L] = ∂L/∂z[L]

Hidden:
δ[l] = W[l+1]ᵀ δ[l+1] ⊙ f'(z[l])

Weight:
∂L/∂W[l] = δ[l] (a[l-1])ᵀ

Bias:
∂L/∂b[l] = δ[l]


UPDATE
─────────────────────────────
W ← W - η ∂L/∂W
b ← b - η ∂L/∂b
```

### 🥇 One mental model

> **Loss produces the first gradient → gradient travels backward through each layer → activation derivative controls the flow → every layer gets its own error signal → error × incoming activation gives weight gradients → error gives bias gradients → optimizer updates the parameters.**

A university backpropagation derivation presents essentially this same layer-by-layer sequence: compute the output error, propagate hidden-layer errors backward, then compute weight and bias gradients. ([Bartosz Naskręcki's Home Page][1])


**Core takeaway:**

$$
\boxed{
\text{Backpropagation through layers}
=
\text{propagate error backward}
+
\text{calculate } \frac{\partial L}{\partial W}
+
\text{calculate } \frac{\partial L}{\partial b}
}
$$

# 📒 Module 41 — Regression Evaluation

Regression models predict **continuous numerical values** such as:

- 🏠 House price → ₹65,00,000
- 🌡️ Temperature → 31.5°C
- 🚗 EV range → 142.7 km
- ⚡ Motor temperature → 78.4°C

Unlike classification, where we ask **“Was the prediction correct?”**, regression asks:

> **“How far was the prediction from the actual value?”**

The main regression evaluation metrics in this module are:

1. **MAE — Mean Absolute Error**
2. **MSE — Mean Squared Error**
3. **RMSE — Root Mean Squared Error**
4. **R² — R-squared / Coefficient of Determination**

---

# 1. 🔹 First Understand: Prediction Error

Suppose our model predicts house prices.

| Actual | Predicted | Error |
|---:|---:|---:|
| ₹50L | ₹48L | ₹2L |
| ₹60L | ₹63L | ₹3L |
| ₹70L | ₹68L | ₹2L |

The basic error/residual is:

$$
e_i = y_i-\hat{y}_i
$$

where:

- \(y_i\) = actual value
- \(\hat{y}_i\) = predicted value
- \(e_i\) = prediction error/residual

### 📌 Residual

The **residual** is simply the distance between the actual value and predicted value.

![Regression residual error](https://encord.com/wp-content/uploads/2023/05/Mean-Square-Error-MSE-1.png)

The smaller the prediction errors, generally, the better the regression model.

---

# 2. 🔹 MAE — Mean Absolute Error

## What is MAE?

**Mean Absolute Error** calculates the average magnitude of the prediction errors.

It asks:

> **“On average, how much is my model wrong?”**

The absolute value is used so that positive and negative errors don't cancel each other.

### Formula

$$
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

Where:

- \(n\) = number of observations
- \(y_i\) = actual value
- \(\hat{y}_i\) = predicted value
- \(|y_i-\hat{y}_i|\) = absolute error

---

## 🧠 Example

Actual:

$$
[100,\ 200,\ 300]
$$

Predicted:

$$
[90,\ 220,\ 280]
$$

Errors:

$$
[10,\ -20,\ 20]
$$

Absolute errors:

$$
[10,\ 20,\ 20]
$$

Therefore:

$$
MAE=\frac{10+20+20}{3}
$$

$$
\boxed{MAE=16.67}
$$

### Meaning

An MAE of **16.67** means:

> The model is wrong by about **16.67 units on average**.

If you're predicting temperature in °C:

> MAE = 2°C → predictions are off by about 2°C on average.

If you're predicting ₹:

> MAE = ₹2 lakh → predictions are off by about ₹2 lakh on average.

---

## ⭐ Important Property of MAE

MAE is in the **same unit as the target variable**.

If predicting:

- ₹ → MAE is in ₹
- km → MAE is in km
- °C → MAE is in °C

This makes MAE very easy to interpret.

---

# 3. 🔹 MSE — Mean Squared Error

MSE stands for:

> **Mean Squared Error**

Instead of taking the absolute value of errors, we **square them**.

### Formula

$$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

The process is:

**Error → Square → Average**

![MSE residual visualization](https://encord.com/wp-content/uploads/2023/05/Mean-Square-Error-MSE-1.png)

---

## 🧠 Example

Actual:

$$
[100,\ 200,\ 300]
$$

Predicted:

$$
[90,\ 220,\ 280]
$$

Errors:

$$
[10,\ -20,\ 20]
$$

Square the errors:

$$
[100,\ 400,\ 400]
$$

Therefore:

$$
MSE=\frac{100+400+400}{3}
$$

$$
\boxed{MSE=300}
$$

---

# 4. 🔥 Why Does MSE Penalize Large Errors?

This is one of the **most important concepts**.

Consider two errors:

### Error 1

$$
5^2=25
$$

### Error 2

$$
20^2=400
$$

The error became **4× larger**, but the squared penalty became:

$$
\frac{400}{25}=16\times
$$

So MSE strongly penalizes large errors.

### Example

Suppose two models have errors:

**Model A**

$$
[2,2,2,2]
$$

**Model B**

$$
[1,1,1,7]
$$

Both may have comparable average absolute error, but Model B has one large mistake.

MSE will punish that large mistake much more heavily.

---

# 5. 🔹 MAE vs MSE

This distinction is extremely important for ML interviews.

| Property | MAE | MSE |
|---|---|---|
| Error treatment | Absolute | Squared |
| Large errors | Less sensitive | **Highly sensitive** |
| Outliers | More robust | **Strongly affected** |
| Units | Same as target | Squared units |
| Interpretation | Easy | Less intuitive |
| Optimization | Less smooth at 0 | Smooth |
| Penalty for large errors | Linear | **Quadratic** |

### 🧠 Remember

> **MAE treats errors proportionally.**

> **MSE makes large errors much more expensive.**

---

# 6. 🔹 RMSE — Root Mean Squared Error

RMSE stands for:

> **Root Mean Squared Error**

It is simply the square root of MSE.

### Formula

$$
RMSE=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
$$

So the pipeline is:

**Error → Square → Mean → Square Root**

![RMSE visualization](https://media.licdn.com/dms/image/v2/D4E22AQGupY76Ur1aiQ/feedshare-shrink_800/feedshare-shrink_800/0/1715642487563?e=2147483647&t=Lu9Vbqfa8alvcxU4svs7dwczXjF90BMKVD3seQIjoIs&v=beta)

---

## 🧠 Why Take the Square Root?

MSE has squared units.

For example, if predicting temperature:

$$
MSE=9\ ^\circ C^2
$$

That isn't very intuitive.

Taking the square root:

$$
RMSE=\sqrt{9}=3^\circ C
$$

Now we're back to the original unit.

Therefore:

> **RMSE is easier to interpret than MSE because it has the same unit as the target.**

---

# 7. 🔥 RMSE vs MAE

Both are useful, but they behave differently toward large errors.

Suppose:

### Model A

Errors:

$$
[2,2,2,2]
$$

### Model B

Errors:

$$
[1,1,1,5]
$$

RMSE will penalize Model B more because of the large error.

A useful mental model:

> **MAE = “How wrong am I on average?”**

> **RMSE = “How wrong am I, while strongly caring about big mistakes?”**

---

# 8. 🔹 Relationship Between MAE, MSE and RMSE

Think of them as a pipeline:

$$
\boxed{\text{Errors}}
$$

↓

### MAE

$$
\boxed{\text{Absolute} \rightarrow \text{Mean}}
$$

↓

### MSE

$$
\boxed{\text{Square} \rightarrow \text{Mean}}
$$

↓

### RMSE

$$
\boxed{\text{Square} \rightarrow \text{Mean} \rightarrow \text{Root}}
$$

---

# 9. ⭐ Important Relationship

For the same dataset:

$$
\boxed{RMSE\geq MAE}
$$

Generally, RMSE will be greater than or equal to MAE.

The bigger the difference between RMSE and MAE, the more strongly the model's error distribution contains some larger errors.


---

# 10. 🔥 R² — R-Squared

Now we move to a **different type of regression metric**.

MAE, MSE and RMSE ask:

> **“How large are my prediction errors?”**

R² asks something different:

> **“How much of the variation in the target variable can my model explain?”**

It is called:

- **R²**
- **R-squared**
- **Coefficient of Determination**

---

# 11. 🧠 The Basic Idea Behind R²

Suppose you're predicting house prices.

House prices naturally vary:

- ₹40L
- ₹50L
- ₹70L
- ₹90L
- ₹1.2Cr

Even before building a model, there is already variation in the data.

A regression model tries to explain this variation using features such as:

- Area
- Location
- Number of bedrooms
- Age
- Parking
- etc.

R² measures how much of that variation is explained by the model.

---

# 12. 🔹 R² Formula

The standard formula is:

$$
R^2=1-\frac{SS_{res}}{SS_{tot}}
$$

where:

### \(SS_{res}\)

Residual Sum of Squares:

$$
SS_{res}=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

This represents the squared prediction errors.

### \(SS_{tot}\)

Total Sum of Squares:

$$
SS_{tot}=\sum_{i=1}^{n}(y_i-\bar{y})^2
$$

This represents the total variation around the mean.

![R² visual intuition](https://media.licdn.com/dms/image/v2/D5612AQEQpfFGGCWpOA/article-inline_image-shrink_1000_1488/B56ZsqZVm1J4AQ-/0/1765942847673?e=2147483647&t=mKi1SkjoDEV6-aMcudjE739pPdwV8XaMp8xjFNpTRAI&v=beta)

---

# 13. 🧠 R² Intuition

Imagine:

### Model 1

Explains almost none of the variation.

$$
R^2\approx0
$$

### Model 2

Explains a large amount of variation.

$$
R^2\approx0.8
$$

### Model 3

Explains almost all the variation.

$$
R^2\approx1
$$

So:

$$
\boxed{R^2=0}
$$

means the model provides no improvement over simply predicting the mean, in the usual R² interpretation.

And:

$$
\boxed{R^2=1}
$$

means perfect prediction on the evaluated data.

---

# 14. 🔥 Example of R²

Suppose:

$$
R^2=0.85
$$

A common interpretation is:

> The model explains approximately **85% of the variance** in the target values relative to the mean baseline.

The remaining approximately:

$$
100-85=15\%
$$

is unexplained by the model under this interpretation.

⚠️ **Important:** Don't interpret \(R^2=0.85\) as “the model is 85% accurate.”

That's wrong.

**R² is NOT accuracy.**

---

# 15. 🔹 Can R² Be Negative?

Yes! 🚨

Unlike many beginners expect, R² can be **less than 0**.

For example:

$$
R^2=-0.5
$$

This means the model performs worse than the baseline that simply predicts the mean, on the evaluated dataset.

Because:

$$
R^2=1-\frac{SS_{res}}{SS_{tot}}
$$

If:

$$
SS_{res}>SS_{tot}
$$

then:

$$
R^2<0
$$

---

# 16. 🔥 MAE vs MSE vs RMSE vs R²

| Metric | What does it tell us? | Lower is better? | Same target units? |
|---|---|---|---|
| **MAE** | Average absolute error | ✅ Yes | ✅ Yes |
| **MSE** | Average squared error | ✅ Yes | ❌ No |
| **RMSE** | Error with stronger large-error penalty | ✅ Yes | ✅ Yes |
| **R²** | Variance explained relative to mean baseline | ❌ Higher is better | N/A |

For R²:

$$
\boxed{\text{Higher is generally better}}
$$

For MAE/MSE/RMSE:

$$
\boxed{\text{Lower is better}}
$$

---

# 17. 🎯 When Should You Use Which Metric?

## MAE

Use MAE when:

- You want an easy-to-understand error measure.
- You want the metric in the same units as the target.
- You don't want large errors to dominate as strongly.

### Example

Predicting house prices where occasional unusual houses shouldn't completely dominate evaluation.

---

## MSE

Use MSE when:

- Large errors are particularly undesirable.
- You want to heavily penalize large mistakes.
- You are working with optimization/loss functions where squared error is appropriate.

---

## RMSE

Use RMSE when:

- Large errors matter.
- You want stronger punishment for large errors.
- You still want the final metric in the original target units.

### Example

Predicting EV battery temperature:

A prediction error of **30°C** could be much more concerning than several small errors.

RMSE would emphasize that large error.

---

## R²

Use R² when:

- You want to understand how much variation the model explains.
- You want a scale-independent goodness-of-fit measure.
- You're comparing model explanatory performance.

But don't use R² alone.

A model can have a good R² while still having errors that are unacceptable for your actual application.

---

# 18. 🧠 Very Important: Don't Evaluate Using Only One Metric

Suppose:

### Model A

$$
MAE=5
$$

$$
RMSE=6
$$

$$
R^2=0.92
$$

### Model B

$$
MAE=4
$$

$$
RMSE=15
$$

$$
R^2=0.91
$$

Which model is better?

There isn't automatically one universal answer.

Model B has a lower MAE but much higher RMSE.

That suggests Model B may have **some very large errors**.

So we should investigate the **error distribution**, not blindly choose one metric.

---

# 19. 🚨 Outliers and Regression Metrics

Consider these errors:

$$
[2,3,2,4,50]
$$

MAE:

$$
MAE=\frac{2+3+2+4+50}{5}=12.2
$$

The 50-error affects the result.

But MSE squares it:

$$
50^2=2500
$$

So that one error has an **enormous influence** on MSE and RMSE.

Therefore:

### MAE

🟢 More resistant to extreme errors.

### MSE

🔴 Very sensitive to large errors.

### RMSE

🔴 Also strongly influenced by large errors.

---

# 20. 🧩 One Table to Remember Everything

| Metric | Formula | Main Idea |
|---|---|---|
| **MAE** | \( \frac{1}{n}\sum|y-\hat y| \) | Average absolute mistake |
| **MSE** | \( \frac{1}{n}\sum(y-\hat y)^2 \) | Penalize large errors |
| **RMSE** | \( \sqrt{MSE} \) | Large-error-sensitive metric in target units |
| **R²** | \(1-\frac{SS_{res}}{SS_{tot}}\) | Variance explained |

---

# 🧠🔥 Final Mental Model

Remember this sequence:

```text
                    REGRESSION EVALUATION
                           │
             ┌─────────────┴─────────────┐
             │                           │
       ERROR-BASED                    VARIANCE
             │                           │
      ┌──────┼──────┐                    │
      │      │      │                    │
     MAE    MSE    RMSE                  R²
      │      │      │                    │
   absolute square square               explained
   average   average + root             variance
      │      │      │                    │
   simple   big-error  big-error      goodness of
   error    penalty    penalty         fit
```

### 🔑 The four things to remember:

> **MAE → average absolute error**

> **MSE → squares errors, so large errors hurt much more**

> **RMSE → square root of MSE, bringing it back to target units**

> **R² → how much target variance the model explains**

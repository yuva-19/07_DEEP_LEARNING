# 📒 Module 41 — Deep Learning Evaluation — NOTES

> **Core goal:** A deep learning model is not considered good merely because it performs well on training data. We care about how well it performs on **unseen data**.

---

# 1. 🔹 Training vs Validation

A dataset is commonly divided into different subsets:

```text
                    COMPLETE DATASET
                           │
             ┌─────────────┼─────────────┐
             ↓             ↓             ↓
         TRAINING      VALIDATION       TEST
           DATA           DATA           DATA
             │             │              │
        Learn weights   Development    Final
                       & monitoring   evaluation
```

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Train-Test-Validation.png)

### 🟢 Training Set

The **training set** is used to learn the model's parameters.

During training:

```text
Input
  ↓
Neural Network
  ↓
Prediction
  ↓
Loss
  ↓
Backpropagation
  ↓
Update weights
```

The model repeatedly sees training examples and adjusts its weights to minimize the loss.

### 🟡 Validation Set

The **validation set** is used during model development.

It helps us determine:

- Which architecture is better?
- Which hyperparameters work better?
- Is the model overfitting?
- When should training stop?
- Is regularization helping?

The validation data should **not be used to update the model's weights**.

### 🔴 Test Set

The test set is reserved for the **final evaluation** after the model and development choices have been finalized.

### 🧠 Simple analogy

**Training = practice questions** 📚

**Validation = mock test** 📝

**Test = final exam** 🎓

---

# 2. 🔥 Overfitting

## Definition

**Overfitting occurs when a model learns the training data too specifically and performs poorly on unseen data.**

In simple words:

> **The model memorizes instead of learning patterns that generalize.**

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Overfitting.svg)

---

## Example

Suppose an image classifier gives:

```text
Training Accuracy      = 99%
Validation Accuracy    = 78%
```

The huge gap is a warning sign.

The model is excellent on data it has already seen, but much worse on new examples.

This indicates possible **overfitting**.

---

# 3. 📈 Detecting Overfitting Using Learning Curves

During training, we can plot:

- Training loss
- Validation loss

against:

- Epoch number

Typical behavior:

```text
Loss
 │
 │\
 │ \
 │  \ Training Loss
 │   \
 │    \____________
 │
 │      \____
 │           \____ Validation Loss
 │                /
 │               /
 └──────────────────────→ Epochs
                  ↑
             Overfitting
             begins
```

Initially:

$$
Training\ Loss\downarrow
$$

and

$$
Validation\ Loss\downarrow
$$

This is good.

But after some point:

$$
Training\ Loss\downarrow
$$

while:

$$
Validation\ Loss\uparrow
$$

🚨 This is a classic sign of overfitting.

![Image](https://www.geeksforgeeks.org/wp-content/uploads/2025/06/Training-and-Validation-Loss.png)

---

# 4. 🧠 Why Does Overfitting Happen?

Common causes:

### 1. Model is too complex

A model with very high capacity can memorize training examples.

### 2. Dataset is too small

A large neural network trained on limited data can easily overfit.

### 3. Training for too long

The model may first learn useful patterns and later start fitting noise or training-specific details.

### 4. Insufficient regularization

Without suitable constraints, the model can become too specialized.

### 5. Noisy data

The model may learn patterns caused by noise rather than meaningful relationships.

---

# 5. 🛠️ How to Reduce Overfitting

Common approaches:

### 🔹 More data

More diverse training examples can improve generalization.

### 🔹 Data augmentation

For image data:

- Rotation
- Cropping
- Flipping
- Scaling
- Color transformations

### 🔹 Regularization

Examples:

- L1 regularization
- L2 regularization
- Weight decay
- Dropout

### 🔹 Early stopping

Stop training when validation performance stops improving.

### 🔹 Reduce model complexity

Use a smaller or simpler architecture when appropriate.

### 🔹 Transfer learning

Start from a pretrained model that has already learned useful representations.

# 6. 🔹 Underfitting

## Definition

**Underfitting occurs when a model is too simple or insufficiently trained to capture the underlying patterns in the data.**

In simple words:

> **The model hasn't learned enough.**

---

## Example

```text
Training Accuracy      = 65%
Validation Accuracy    = 63%
```

Both performances are poor.

This suggests the model may be **underfitting**.

Compare this with:

```text
Training Accuracy      = 99%
Validation Accuracy    = 65%
```

That pattern suggests **overfitting**, not underfitting.

---

# 7. 🧠 Underfitting Intuition

Imagine the true relationship is complicated:

```text
        ╭──╮
    ╭──╯  ╰──╮
───╯        ╰───
```

But the model can only learn a very simple relationship:

```text
────────────────
```

The model doesn't have enough capacity to represent the underlying pattern.

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Underfitting_e_overfitting.png)

---

# 8. Causes of Underfitting

Common causes:

- Model is too simple
- Insufficient training
- Excessive regularization
- Poor feature/input representation
- Inappropriate learning rate
- Insufficient model capacity

---

# 9. 🛠️ How to Reduce Underfitting

Possible solutions:

- Increase model capacity
- Train for longer
- Improve input representation
- Reduce excessive regularization
- Tune learning rate
- Improve architecture
- Improve preprocessing/features

# 10. ⚔️ Underfitting vs Good Fit vs Overfitting

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Underfitting_e_overfitting.png)

| | Underfitting | Good Fit | Overfitting |
|---|---|---|---|
| Training performance | ❌ Poor | ✅ Good | 🟢 Excellent |
| Validation performance | ❌ Poor | ✅ Good | ❌ Poor |
| Model capacity | Often too low | Appropriate | Often too high |
| Learning | Insufficient | Appropriate | Too specific |
| Generalization | Poor | Good | Poor |

### 🧠 Memorize this:

```text
UNDERFITTING
Train ❌
Validation ❌

GOOD FIT
Train ✅
Validation ✅

OVERFITTING
Train 🟢
Validation 🔴
```

# 11. 🔥 Bias–Variance Intuition

Underfitting and overfitting are closely related to **bias and variance**.

### Underfitting

Usually associated with:

$$
\boxed{\text{High Bias}}
$$

The model makes overly simplistic assumptions.

### Overfitting

Usually associated with:

$$
\boxed{\text{High Variance}}
$$

The model becomes highly sensitive to the particular training dataset.

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Overfitting.svg)

Another useful representation:

![Image](https://commons.wikimedia.org/wiki/Special:Redirect/file/Underfitting_e_overfitting.png)

### Mental model

```text
Model Complexity
       →
       
UNDERFIT ─────── GOOD FIT ─────── OVERFIT
 High Bias                         High Variance
```

# 12. 🌍 Generalization

## Definition

**Generalization is the ability of a trained model to perform well on previously unseen data.**

This is ultimately what we want from machine learning.

A model with:

```text
Training Accuracy = 99.9%
```

is not necessarily good.

We want:

```text
Training Performance      → Good
Validation Performance    → Good
Test Performance          → Good
```

with a reasonably small gap.

---

# 13. 🧠 Generalization Example

Suppose we train a dog/cat classifier using:

> 100,000 training images.

After training, the model receives completely new images.

If it correctly classifies those unseen images, the model has **generalized**.

It has learned useful characteristics of dogs and cats rather than simply memorizing the training images.

---

# 14. 🔹 Generalization Gap

The difference between training and unseen-data performance can be thought of as the **generalization gap**.

For loss:

$$
\boxed{
\text{Generalization Gap}
=
L_{validation}-L_{training}
}
$$

For example:

$$
Training\ Loss=0.05
$$

$$
Validation\ Loss=0.42
$$

Then:

$$
\text{Gap}=0.42-0.05=0.37
$$

A large gap can indicate that the model fits the training data much better than unseen data.

---

# 15. 🎯 What Does Good Generalization Look Like?

```text
             TRAINING
                 │
                 ↓
              Learn
                 │
                 ↓
         ┌───────────────┐
         │    MODEL      │
         └───────┬───────┘
                 │
          ┌──────┴──────┐
          ↓             ↓
      Validation       Test
          ↓             ↓
        Good           Good
```

The ideal situation is:

$$
\boxed{
Training\ Performance
\approx
Validation\ Performance
\approx
Test\ Performance
}
$$

while all are sufficiently good for the application.

# 16. 🔍 Error Analysis

## Definition

**Error analysis is the systematic examination of incorrect predictions to understand why the model fails.**

Instead of simply saying:

> "My model has 10% errors."

we ask:

> **"What types of errors make up that 10%?"**

---

# 17. 🧠 Example — Image Classification

Suppose:

$$
Accuracy=90\%
$$

We examine the incorrect predictions.

| Error type | Number |
|---|---:|
| Blurry images | 25 |
| Poor lighting | 20 |
| Occlusion | 15 |
| Similar classes | 30 |
| Labeling errors | 10 |

Now we know **why** the model is failing.

The model isn't simply "bad."

We've identified specific **failure modes**.

---

# 18. 🔥 Error Analysis Workflow

```text
             MODEL PREDICTIONS
                    │
                    ↓
             Identify Errors
                    │
                    ↓
          Categorize Error Types
                    │
                    ↓
        Find Major Failure Modes
                    │
                    ↓
            Investigate Cause
                    │
                    ↓
        Apply Targeted Solution
                    │
                    ↓
            Retrain / Evaluate
```

The key idea:

> **Don't just count errors — understand them.**

---

# 19. 🔎 What Errors Should We Investigate?

### Classification

Look for:

- False positives
- False negatives
- Confusion between similar classes
- Rare classes
- Class imbalance
- Low-quality samples

### Regression

Look for:

- Large residuals
- Outliers
- Systematic errors
- Regions where predictions are consistently too high/low

### Computer Vision

Look for:

- Blur
- Occlusion
- Poor lighting
- Background problems
- Scale differences
- Object position

### NLP

Look for:

- Long inputs
- Ambiguous language
- Rare words
- Domain-specific terminology
- Negation
- Context-dependent meaning

---

# 20. 🧩 Example — Medical Image Model

Imagine a model detecting a disease from medical images.

You inspect incorrect predictions and discover:

> Most false negatives occur in low-quality images.

Instead of immediately making the neural network larger, investigate:

- Image resolution
- Noise
- Lighting/contrast
- Preprocessing
- Data collection
- Label quality

This gives you a **targeted improvement strategy**.

---

# 21. 🧠 Error Analysis vs Evaluation Metrics

This distinction is extremely important.

### Evaluation metrics tell us:

> **WHAT is happening?**

Example:

$$
Accuracy=90\%
$$

### Error analysis tells us:

> **WHY is it happening?**

Example:

> 60% of the errors occur because two visually similar classes are being confused.

Therefore:

$$
\boxed{
\text{Metrics} \rightarrow WHAT
}
$$

$$
\boxed{
\text{Error Analysis} \rightarrow WHY
}
$$

# 22. 🔥 Complete Deep Learning Evaluation Pipeline

```text
                    TRAIN MODEL
                         │
                         ↓
                Evaluate Training
                         │
                         ↓
               Evaluate Validation
                         │
                         ↓
                 Compare Results
                         │
             ┌───────────┴───────────┐
             ↓                       ↓
         Similar                 Large Gap
             │                       │
             ↓                       ↓
      Good Generalization       Overfitting
             │
             ↓
       Final Test Evaluation
             │
             ↓
        Error Analysis
             │
             ↓
       Identify Failure Modes
             │
             ↓
        Improve the Model
```

# 23. 🚨 Three Patterns You MUST Remember

## 🟥 Underfitting

```text
Training      ❌ Poor
Validation    ❌ Poor
```

**Meaning:** Model hasn't learned enough.

---

## 🟢 Good Generalization

```text
Training      ✅ Good
Validation    ✅ Good
Test          ✅ Good
```

**Meaning:** Model learned useful patterns that transfer to unseen data.

---

## 🟥 Overfitting

```text
Training      🟢 Excellent
Validation    🔴 Poor
```

**Meaning:** Model learned training-specific patterns too closely.

# 24. 🏆 Final Cheat Sheet

| Concept | Meaning | Main Signal |
|---|---|---|
| **Training** | Data used to learn model parameters | Model learns from it |
| **Validation** | Data used during development | Monitor/generalize |
| **Test** | Final unseen evaluation | Final performance |
| **Overfitting** | Learns training data too specifically | Train ≫ Validation |
| **Underfitting** | Model fails to learn enough | Train poor + Validation poor |
| **Generalization** | Performs well on unseen data | Train ≈ Validation/Test |
| **Error Analysis** | Study why predictions are wrong | Identify failure modes |

---

# 🧠🔥 One-Minute Revision

> **Training data** → teaches the model.

> **Validation data** → helps us monitor and improve the model during development.

> **Test data** → gives the final unbiased performance estimate.

> **Overfitting** → model learns training data too specifically.

> **Underfitting** → model hasn't learned enough.

> **Generalization** → model performs well on unseen data.

> **Error analysis** → investigates **why** the model makes mistakes.

### The ultimate goal:

$$
\boxed{
\text{Learn useful patterns}
\rightarrow
\text{Generalize}
\rightarrow
\text{Perform well on unseen data}
}
$$

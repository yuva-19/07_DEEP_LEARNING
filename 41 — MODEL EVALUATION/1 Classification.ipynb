# 📒 MODULE 41 — CLASSIFICATION EVALUATION METRICS

> **Scope:** Confusion Matrix → Accuracy → Precision → Recall → F1 → ROC-AUC → PR-AUC

Classification metrics tell us **how well a classification model is performing and what kinds of mistakes it is making**.

---

# 1. 🧩 CONFUSION MATRIX

A **confusion matrix** summarizes the predictions made by a classification model compared with the actual labels.

It contains four fundamental outcomes:

- **TP — True Positive**
- **TN — True Negative**
- **FP — False Positive**
- **FN — False Negative**

![Image](https://miro.medium.com/1%2Amuwa5CfbiTtG4W6dIIGuIw.png)

### Structure

| | Actual Positive | Actual Negative |
|---|---:|---:|
| **Predicted Positive** | **TP** | **FP** |
| **Predicted Negative** | **FN** | **TN** |

### Meaning

**True Positive (TP)**  
Model predicts positive → actual is positive ✅

**True Negative (TN)**  
Model predicts negative → actual is negative ✅

**False Positive (FP)**  
Model predicts positive → actual is negative ❌

> Also called a **false alarm**.

**False Negative (FN)**  
Model predicts negative → actual is positive ❌

> Also called a **miss**.

### 🧠 Memory

```text
TRUE  → Prediction is correct
FALSE → Prediction is wrong

POSITIVE → Model predicted positive
NEGATIVE → Model predicted negative
```

---

# 2. 🎯 ACCURACY

### Definition

**Accuracy** measures the percentage of all predictions that were correct.

$$
\boxed{
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

### 🧠 Simple meaning

> **Accuracy = Correct predictions / Total predictions**

---

### 🔢 Example

Suppose:

$$
TP=80,\quad TN=90,\quad FP=20,\quad FN=10
$$

Then:

$$
Accuracy=
\frac{80+90}
{80+90+20+10}
$$

$$
=\frac{170}{200}
$$

$$
\boxed{Accuracy=85\%}
$$

---

## ⚠️ Accuracy and Class Imbalance

Accuracy can be misleading when one class is much more common than the other.

Example:

```text
1000 samples

990 → Negative
10  → Positive
```

A model predicts:

```text
Everything → Negative
```

It gets:

$$
\frac{990}{1000}=99\%
$$

accuracy.

But it detects:

$$
\boxed{0/10}
$$

positive samples.

### 🔥 Key point

> **High accuracy does NOT necessarily mean a good classifier.**

This is why we need other metrics.

---

# 3. 🎯 PRECISION

### Definition

Precision tells us:

> **Out of all samples predicted as positive, how many were actually positive?**

$$
\boxed{
Precision =
\frac{TP}
{TP+FP}
}
$$

### 🧠 Simple meaning

> **Precision = "Can I trust my positive predictions?"**

Precision is mainly concerned with **False Positives (FP)**.

---

### 🔢 Example

Suppose the model predicts 100 people as positive.

```text
80 → Actually Positive
20 → Actually Negative
```

Therefore:

$$
TP=80
$$

$$
FP=20
$$

So:

$$
Precision=
\frac{80}{80+20}
$$

$$
\boxed{Precision=80\%}
$$

---

### High Precision

```text
High Precision
      ↓
Few False Positives
      ↓
Few False Alarms
```

### Example

In **fraud detection**, low precision means many legitimate transactions are incorrectly flagged as fraudulent.

---

# 4. 🎯 RECALL

### Definition

Recall tells us:

> **Out of all actual positive samples, how many did the model successfully identify?**

$$
\boxed{
Recall =
\frac{TP}
{TP+FN}
}
$$

Recall is mainly concerned with **False Negatives (FN)**.

---

### 🔢 Example

Suppose there are 100 actual positive cases.

```text
90 → Detected correctly
10 → Missed
```

Therefore:

$$
TP=90
$$

$$
FN=10
$$

So:

$$
Recall=
\frac{90}{90+10}
$$

$$
\boxed{Recall=90\%}
$$

---

### 🧠 Simple meaning

> **Recall = "Did I find the actual positives?"**

High recall means:

```text
High Recall
     ↓
Few False Negatives
     ↓
Few Missed Positives
```

Recall is also called:

- **Sensitivity**
- **True Positive Rate (TPR)**

$$
\boxed{
TPR=Recall=
\frac{TP}{TP+FN}
}
$$

---

# 5. ⚔️ PRECISION vs RECALL

| | **Precision** | **Recall** |
|---|---|---|
| Main question | Can I trust positive predictions? | Did I find the actual positives? |
| Formula | \(TP/(TP+FP)\) | \(TP/(TP+FN)\) |
| Main concern | **FP** | **FN** |
| High value means | Few false alarms | Few missed positives |

### 🧠 Easy memory trick

```text
PRECISION
↓
Predicted Positives
↓
"How many were actually positive?"

RECALL
↓
Actual Positives
↓
"How many did I successfully find?"
```

---

# 6. ⚖️ F1 SCORE

Precision and recall can sometimes give different views of model performance.

**F1 Score** combines them into one metric.

$$
\boxed{
F1 =
2\frac{Precision\times Recall}
{Precision+Recall}
}
$$

F1 is the **harmonic mean** of precision and recall.

---

### 🔢 Example

Suppose:

$$
Precision=0.8
$$

$$
Recall=0.6
$$

Then:

$$
F1=
2\frac{(0.8)(0.6)}
{0.8+0.6}
$$

$$
=\frac{0.96}{1.4}
$$

$$
\boxed{F1\approx0.686}
$$

Therefore:

$$
\boxed{F1\approx68.6\%}
$$

---

## 🧠 Why harmonic mean?

F1 penalizes situations where one of precision or recall is very high while the other is very low.

For example:

$$
Precision=100\%
$$

$$
Recall=0\%
$$

Then:

$$
\boxed{F1=0}
$$

The model cannot claim good overall positive-class performance if it completely misses the positive samples.

---

Suppose:

| | Actual Positive | Actual Negative |
|---|---:|---:|
| **Predicted Positive** | **80** | **20** |
| **Predicted Negative** | **10** | **90** |

Therefore:

$$
TP=80
$$

$$
FP=20
$$

$$
FN=10
$$

$$
TN=90
$$

Total:

$$
200
$$

### Accuracy

$$
Accuracy=
\frac{80+90}{200}
=
\boxed{85\%}
$$

### Precision

$$
Precision=
\frac{80}{80+20}
=
\boxed{80\%}
$$

### Recall

$$
Recall=
\frac{80}{80+10}
=
\boxed{88.9\%}
$$

### F1

$$
F1=
2\frac{(0.8)(0.889)}
{0.8+0.889}
$$

$$
\boxed{F1\approx84.4\%}
$$

### Summary

| Metric | Value |
|---|---:|
| Accuracy | **85%** |
| Precision | **80%** |
| Recall | **88.9%** |
| F1 | **84.4%** |

# 8. 🎚️ CLASSIFICATION THRESHOLD

A neural network often outputs a **probability/score** rather than directly producing a class.

Example:

```text
Model output = 0.82
```

If:

```text
Threshold = 0.5
```

then:

```text
0.82 > 0.5
      ↓
Positive
```

But if we change the threshold:

```text
0.3
0.5
0.7
0.9
```

the predicted classes can change.

Therefore:

> **Changing the classification threshold changes TP, TN, FP and FN.**

This is the foundation for understanding **ROC-AUC and PR-AUC**.

---

# 9. 📈 ROC CURVE

## ROC = Receiver Operating Characteristic

The ROC curve evaluates model behaviour across different classification thresholds.

It plots:

### Y-axis — True Positive Rate

$$
\boxed{
TPR=
\frac{TP}{TP+FN}
}
$$

TPR is the same as **Recall**.

### X-axis — False Positive Rate

$$
\boxed{
FPR=
\frac{FP}{FP+TN}
}
$$

Therefore:

$$
\boxed{
ROC = TPR\;vs\;FPR
}
$$

![Image](https://images.prismic.io/turing/65a53cb47a5e8b1120d5886a_Formula_of_AUC_curve_986065ab5e.webp?auto=format%2Ccompress)

### 🧠 Intuition

As the classification threshold changes:

```text
Threshold changes
       ↓
Predictions change
       ↓
TPR changes
       ↓
FPR changes
       ↓
ROC curve
```

---

# 10. 📐 ROC-AUC

### AUC

**AUC = Area Under the Curve**

Therefore:

> **ROC-AUC = Area Under the ROC Curve**

![Image](https://images.prismic.io/turing/65a53cb37a5e8b1120d5886a_Formula_of_AUC_curve_986065ab5e.webp?auto=format%2Ccompress)

### 🧠 Core intuition

ROC-AUC tells us how well the model can **distinguish/rank positive examples above negative examples across thresholds**.

### General interpretation

| ROC-AUC | General interpretation |
|---:|---|
| **1.0** | Perfect separation |
| **0.9** | Excellent |
| **0.8** | Good |
| **0.7** | Moderate |
| **0.5** | Random-like |
| **< 0.5** | Worse-than-random ranking |

⚠️ These are only rough interpretation guidelines. The acceptable value depends on the application.

---

# 11. 🧠 ROC-AUC INTUITION

Suppose the model produces:

```text
Positive samples:

A → 0.95
B → 0.87
C → 0.81


Negative samples:

D → 0.20
E → 0.12
F → 0.05
```

The model ranks:

```text
Positive
Positive
Positive
      ↓
Negative
Negative
Negative
```

This represents strong class separation.

### Remember:

> **ROC-AUC → "How well can my model separate the two classes across thresholds?"**

---

# 12. 📊 PRECISION–RECALL CURVE

A Precision-Recall curve evaluates the relationship between **precision and recall** as the classification threshold changes.

### X-axis

$$
\boxed{Recall}
$$

### Y-axis

$$
\boxed{Precision}
$$

Therefore:

$$
\boxed{
PR\ Curve =
Precision\;vs\;Recall
}
$$

![Image](https://sinyi-chou.github.io/images/prauc/PR_auc_area_plot.png)

---

# 13. 🔥 PR-AUC

### PR-AUC

**PR-AUC = Area Under the Precision-Recall Curve**

It summarizes the relationship between precision and recall across different thresholds.

$$
\boxed{PR\text{-}AUC}
$$

### 🧠 Core intuition

> **PR-AUC tells us how well precision and recall are maintained across thresholds.**

---

# 14. 🚨 WHY PR-AUC IS IMPORTANT FOR IMBALANCED DATA

Consider fraud detection:

```text
1,000,000 transactions

999,000 → Legitimate
1,000   → Fraud
```

Fraud is extremely rare.

In such situations, **precision and recall of the positive class become particularly important**.

Precision asks:

> "Of everything I predicted as fraud, how much was actually fraud?"

Recall asks:

> "Of all actual fraud cases, how many did I find?"

Therefore:

$$
\boxed{
PR\text{-}AUC
}
$$

can be particularly useful when the positive class is rare.

---

# 15. ⚔️ ROC-AUC vs PR-AUC

| | **ROC-AUC** | **PR-AUC** |
|---|---|---|
| Curve | TPR vs FPR | Precision vs Recall |
| Y-axis | TPR / Recall | Precision |
| X-axis | FPR | Recall |
| Main focus | Class discrimination | Positive-class performance |
| Thresholds | Across thresholds | Across thresholds |
| Imbalanced data | Can sometimes look optimistic | Often more informative |
| Rare positive class | Less directly focused | Particularly useful |

### 🧠 Memory trick

> **ROC → "Can I separate the classes?"**

> **PR → "How well can I find positives while keeping my positive predictions reliable?"**

# 17. 🎯 WHEN TO USE WHICH METRIC?

### 🩺 Disease Detection

If missing a positive case is very costly:

$$
\boxed{\text{Prioritize Recall}}
$$

because:

```text
False Negative
     ↓
Actual disease
     ↓
Model says negative
```

---

### 💳 Fraud Detection

Fraud is usually a rare positive class.

You care about:

- catching fraud
- avoiding excessive false alarms

Therefore:

$$
\boxed{\text{Precision + Recall + PR-AUC}}
$$

can be especially useful.

---

### 📧 Spam Detection

False positives can cause legitimate emails to be classified as spam.

Therefore:

$$
\boxed{\text{Precision is important}}
$$

---

### ⚖️ Need a Balance

When both precision and recall matter:

$$
\boxed{F1}
$$

---

### 📊 Class Separation

When evaluating ranking/discrimination across thresholds:

$$
\boxed{ROC-AUC}
$$

# 18. 🏆 MASTER TABLE

| Metric | Formula | Main Question | Main Concern |
|---|---|---|---|
| **Accuracy** | \(\frac{TP+TN}{TP+TN+FP+FN}\) | How many predictions were correct? | Overall correctness |
| **Precision** | \(\frac{TP}{TP+FP}\) | Can I trust positive predictions? | **FP** |
| **Recall** | \(\frac{TP}{TP+FN}\) | Did I find the positives? | **FN** |
| **F1** | \(2\frac{PR}{P+R}\) | How well are precision & recall balanced? | FP + FN |
| **Confusion Matrix** | TP/TN/FP/FN | What exactly did the model get right/wrong? | Error types |
| **ROC-AUC** | Area under ROC | How well are classes separated across thresholds? | TPR vs FPR |
| **PR-AUC** | Area under PR curve | How well are precision & recall maintained across thresholds? | Positive-class performance |

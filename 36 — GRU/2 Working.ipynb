## 1. What enters the GRU?

At time \(t\), the GRU receives two things:

* Current input → \(x_t\)
* Previous hidden state → \(h_{t-1}\)

```text
          Previous memory
              hₜ₋₁
                │
                ▼
              ┌─────┐
Current xₜ ──►│ GRU │──► hₜ
              └─────┘
```

The GRU then decides:

1. How much previous information should be kept?
2. How much previous information should influence the new candidate?
3. What should the new hidden state be?

These decisions are made using the **update gate** and **reset gate**.

---

# 2. Step 1 — Calculate the Update Gate 🔄

The update gate is:

$$
\boxed{
z_t=\sigma(W_z[h_{t-1},x_t]+b_z)
}
$$

The sigmoid produces values between 0 and 1:

$$
0\leq z_t\leq1
$$

The update gate determines how much **old information is retained** versus how much **new information is incorporated**.

### Intuition

```text
zₜ ≈ 1
   ↓
Keep more previous information

zₜ ≈ 0
   ↓
Use more new information
```

Think:

> **"How much should I update my memory?"**

---

# 3. Step 2 — Calculate the Reset Gate 🧹

The reset gate is:

$$
\boxed{
r_t=\sigma(W_r[h_{t-1},x_t]+b_r)
}
$$

Again:

$$
0\leq r_t\leq1
$$

The reset gate controls how much of the **previous hidden state** should influence the new candidate information.

```text
rₜ ≈ 1
   ↓
Use previous information strongly

rₜ ≈ 0
   ↓
Ignore much of previous information
```

Think:

> **"How much of the old information should I consider when creating the new information?"**

---

# 4. Step 3 — Create Candidate Hidden State 🧠

Now the reset gate is used to control the previous hidden state.

A common formulation is:

$$
\boxed{
\tilde h_t=
\tanh
\left(
W_h[r_t\odot h_{t-1},x_t]+b_h
\right)
}
$$

Here:

$$
r_t\odot h_{t-1}
$$

means that the reset gate selectively filters the previous hidden state.

### Example

Suppose:

```text
hₜ₋₁ = [0.8, 0.6, 0.9]

rₜ   = [1.0, 0.0, 0.2]
```

Then:

```text
rₜ ⊙ hₜ₋₁

= [1.0×0.8, 0.0×0.6, 0.2×0.9]

= [0.8, 0, 0.18]
```

So some previous information is retained while other information is suppressed.

Then the GRU combines this with \(x_t\) and applies `tanh` to produce:

$$
\tilde h_t
$$

the **candidate hidden state**.

---

# 5. Step 4 — Calculate the New Hidden State

Now the update gate determines how the old hidden state and candidate state are combined.

One common convention is:

$$
\boxed{
h_t=(1-z_t)\odot h_{t-1}
+
z_t\odot\tilde h_t
}
$$

This is the final hidden state.

⚠️ **Important:** Some textbooks/framework descriptions define the update gate with the opposite convention, resulting in an algebraically different-looking equation. The underlying idea is the same: the gate controls the balance between **previous state** and **candidate state**.

---

# 6. What Does This Equation Mean?

Look at:

$$
h_t=(1-z_t)\odot h_{t-1}
+
z_t\odot\tilde h_t
$$

There are two components:

### Old information

$$
(1-z_t)\odot h_{t-1}
$$

### New information

$$
z_t\odot\tilde h_t
$$

So the update gate acts like a **soft switch** between old and new information.

---

## 🔥 Case 1 — Update gate is large

Suppose:

$$
z_t\approx1
$$

Then:

$$
h_t\approx\tilde h_t
$$

The GRU uses mostly the **new candidate information**.

---

## 🔥 Case 2 — Update gate is small

Suppose:

$$
z_t\approx0
$$

Then:

$$
h_t\approx h_{t-1}
$$

The GRU keeps mostly the **previous hidden state**.

---

# 7. Complete GRU Working

The complete process is:

```text
                 xₜ
                  │
                  │
          ┌───────┴───────┐
          │               │
          ▼               ▼
     Update Gate      Reset Gate
         zₜ               rₜ
          │               │
          │               ▼
          │        rₜ ⊙ hₜ₋₁
          │               │
          │               ▼
          │         Candidate
          │           h̃ₜ
          │               │
          └───────┬───────┘
                  │
                  ▼
          Combine old + new
                  │
                  ▼
                 hₜ
```

Mathematically:

### ① Update gate

$$
z_t=\sigma(W_z[h_{t-1},x_t]+b_z)
$$

### ② Reset gate

$$
r_t=\sigma(W_r[h_{t-1},x_t]+b_r)
$$

### ③ Candidate state

$$
\tilde h_t=
\tanh(W_h[r_t\odot h_{t-1},x_t]+b_h)
$$

### ④ New hidden state

$$
\boxed{
h_t=(1-z_t)\odot h_{t-1}
+z_t\odot\tilde h_t
}
$$

---

# 8. Simple Real-Life Analogy 🧠

Imagine you're reading a long article.

At each new sentence:

### Reset gate 🧹

> "How much of the previous context do I need to consider for understanding this sentence?"

### Candidate state 📝

> "Based on the current sentence and relevant previous context, what new information should I remember?"

### Update gate 🔄

> "Should I mostly keep what I already know, or update my memory with this new information?"

### Hidden state 🧠

> "This is my updated understanding after processing the current sentence."

---

# 9. GRU Across Multiple Time Steps

```text
             h₀
              │
              ▼
x₁ ─────────► GRU ─────► h₁
                         │
                         ▼
x₂ ─────────► GRU ─────► h₂
                         │
                         ▼
x₃ ─────────► GRU ─────► h₃
                         │
                         ▼
x₄ ─────────► GRU ─────► h₄
```

At every time step:

$$
(x_t,h_{t-1})\rightarrow h_t
$$

The **same GRU parameters** are reused at every time step.

Therefore, the hidden state carries useful information from earlier parts of the sequence to later parts.

---

# 🧠 LSTM vs GRU Working

The easiest way to distinguish them:

```text
LSTM

xₜ + hₜ₋₁ + Cₜ₋₁
       ↓
   3 gates
       ↓
Cₜ + hₜ


GRU

xₜ + hₜ₋₁
       ↓
   2 gates
       ↓
      hₜ
```

### LSTM

Has a **separate cell state** for long-term memory.

### GRU

Uses its **hidden state as the main memory** and controls it using the update and reset gates.

## 🔥 One-Minute Revision

Remember this flow:

$$
\boxed{x_t+h_{t-1}}
$$

↓

### 🔄 Update gate

$$
z_t=\sigma(...)
$$

↓

### 🧹 Reset gate

$$
r_t=\sigma(...)
$$

↓

### 🧠 Candidate

$$
\tilde h_t=\tanh(...)
$$

↓

### 🔀 Combine

$$
h_t=(1-z_t)h_{t-1}+z_t\tilde h_t
$$

↓

$$
\boxed{h_t}
$$

**In one sentence:**

> **GRU uses the reset gate to control how much past information contributes to new information, and the update gate to control how the previous hidden state and new candidate are combined into the next hidden state.** 🚀

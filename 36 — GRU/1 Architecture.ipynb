# 🚀 MODULE 36 — GRU

## Architecture

GRU stands for **Gated Recurrent Unit**. It is a gated recurrent architecture designed to handle sequence dependencies while being **simpler than LSTM**.

A GRU has **two gates**:

1. 🔄 **Update gate**
2. 🧹 **Reset gate**

and one main state:

3. 🧠 **Hidden state**

Unlike LSTM, GRU does **not** maintain a separate cell state.

At each time step, a GRU receives:

* Current input \(x_t\)
* Previous hidden state \(h_{t-1}\)

and produces:

* New hidden state \(h_t\)

```text
             Previous Hidden State
                    hₜ₋₁
                      │
                      ▼
                ┌───────────┐
                │    GRU    │
                │           │
Current Input ─►│  Gates +  │──► hₜ
     xₜ         │  Candidate│
                └───────────┘
                      │
                      ▼
                New Hidden State
```

The key difference from LSTM:

```text
LSTM:
xₜ + hₜ₋₁ + Cₜ₋₁
       ↓
   multiple gates
       ↓
hₜ + Cₜ

GRU:
xₜ + hₜ₋₁
       ↓
   2 gates
       ↓
      hₜ
```

So:

> **GRU combines memory and hidden representation into a single hidden state.**

---

# 2. Update Gate 🔄

The **update gate** determines how much of the **previous hidden state** should be retained versus how much **new information** should be incorporated.

It is usually represented as:

$$
\boxed{
z_t=\sigma(W_z[h_{t-1},x_t]+b_z)
}
$$

where:

* \(z_t\) = update gate
* \(h_{t-1}\) = previous hidden state
* \(x_t\) = current input
* \(\sigma\) = sigmoid
* \(W_z,b_z\) = learnable parameters

Because sigmoid produces values between 0 and 1:

$$
0\leq z_t\leq1
$$

### Intuition

```text
zₜ ≈ 0
   ↓
Use more NEW information

zₜ ≈ 1
   ↓
Keep more OLD information
```

Think of the update gate as asking:

> 🧠 **"How much of the old information should I keep?"**

This gate is one of the main mechanisms that allows GRU to preserve useful information across time.

---

# 3. Reset Gate 🧹

The **reset gate** determines how much of the **previous hidden state** should be used when creating the new candidate information.

Equation:

$$
\boxed{
r_t=\sigma(W_r[h_{t-1},x_t]+b_r)
}
$$

where:

* \(r_t\) = reset gate
* \(h_{t-1}\) = previous hidden state
* \(x_t\) = current input

Again:

$$
0\leq r_t\leq1
$$

### Intuition

```text
rₜ ≈ 0
   ↓
Ignore much of the previous state

rₜ ≈ 1
   ↓
Use much of the previous state
```

The reset gate essentially asks:

> 🧹 **"How much of the previous information should I use when forming the new information?"**

---

# 4. Hidden State 🧠

Unlike LSTM, GRU doesn't have a separate:

$$
C_t
$$

cell state.

Instead, the **hidden state itself acts as the main memory**.

```text
Previous hidden state
        hₜ₋₁
          │
          ▼
      ┌────────┐
xₜ ──►│  GRU   │
      └────────┘
          │
          ▼
       hₜ
          │
          ▼
      Next time step
```

Therefore:

$$
\boxed{\text{GRU memory} \approx \text{hidden state}}
$$

This makes GRU structurally simpler than LSTM.

---

# 5. How the Two Gates Work Together

The two gates have different jobs:

| Gate                   | Main purpose                                      | Intuition                                    |
| ---------------------- | ------------------------------------------------- | -------------------------------------------- |
| 🔄 Update gate \(z_t\) | Controls old vs new information                   | **"How much should I keep?"**                |
| 🧹 Reset gate \(r_t\)  | Controls previous-state contribution to candidate | **"How much old information should I use?"** |

Think about processing a sentence:

> **"The students went to the laboratory because they had an experiment."**

The GRU needs to decide which previous information remains useful as new words arrive.

The **reset gate** can control how much previous context is considered when forming new information.

The **update gate** controls how much of the existing hidden representation is retained when creating the next hidden state.

---

# 6. GRU Architecture Flow

The basic conceptual flow is:

```text
             hₜ₋₁
               │
          ┌────┴────┐
          │         │
          ▼         ▼
     Update Gate  Reset Gate
        zₜ           rₜ
          │         │
          │         ▼
          │    Candidate
          │    information
          │         │
          └────┬────┘
               ▼
              hₜ
```

And across time:

```text
x₁ ──► GRU ──► h₁
               │
x₂ ──► GRU ◄───┘
         │
         ▼
        h₂
         │
x₃ ──► GRU ◄───┘
         │
         ▼
        h₃
```

The same GRU parameters are reused at every time step, just like an RNN and LSTM.

---

# 🔥 Key Difference From LSTM

This is the most important architectural point to remember.

### LSTM

```text
                ┌─────────────┐
xₜ,hₜ₋₁ ───────►│    LSTM     │
                └──────┬──────┘
                       │
                ┌──────┴──────┐
                ▼             ▼
               Cₜ            hₜ
        Cell State       Hidden State
```

### GRU

```text
                ┌─────────────┐
xₜ,hₜ₋₁ ───────►│     GRU     │
                └──────┬──────┘
                       │
                       ▼
                      hₜ
                 Hidden State
                 = main memory
```

So:

> **LSTM → separate cell state + hidden state**

> **GRU → only hidden state**

And:

> **LSTM → 3 main gates**

> **GRU → 2 main gates**

---

## 🧠 One-Line Memory Trick

**GRU = Update + Reset + Hidden state**

```text
UPDATE → How much OLD information survives?
RESET  → How much OLD information influences NEW information?
HIDDEN → Stores the resulting representation/memory
```

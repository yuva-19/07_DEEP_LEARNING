# 1. Why Activation Functions Matter

A neuron computes:

$$
z=Wx+b
$$

Then applies an activation:

$$
a=f(z)
$$

So:

$$
\boxed{a=f(Wx+b)}
$$

The activation introduces **non-linearity**.

Without activation:

$$
Linear \rightarrow Linear \rightarrow Linear
$$

can collapse into:

$$
\boxed{Linear}
$$

With nonlinear activation:

$$
Linear\rightarrow Nonlinear\rightarrow Linear\rightarrow Nonlinear
$$

the network can learn complex relationships.

# 2. Activation Function Analysis — Master Table

| Activation     | Formula                          | Gradient behavior    | Main advantage           | Main disadvantage   | Typical use          |
| -------------- | -------------------------------- | -------------------- | ------------------------ | ------------------- | -------------------- |
| **Step**       | $0/1$ threshold                  | 0 almost everywhere  | Simple                   | No useful gradient  | Classical perceptron |
| **Sigmoid**    | $\frac{1}{1+e^{-z}}$             | Can vanish           | Output 0–1               | Vanishing gradients | Binary output        |
| **Tanh**       | $\frac{e^z-e^{-z}}{e^z+e^{-z}}$  | Can vanish           | Zero-centered            | Vanishing gradients | RNN/selected layers  |
| **ReLU**       | $\max(0,z)$                      | 0 or 1               | Simple + efficient       | Dying ReLU          | Common hidden layers |
| **Leaky ReLU** | $\max(z,\alpha z)$               | $\alpha$ or 1        | Keeps negative gradient  | Requires $\alpha$   | ReLU alternative     |
| **ELU**        | $z$ / $\alpha(e^z-1)$            | Smooth negative side | Smooth negative region   | More expensive      | ReLU alternative     |
| **GELU**       | $x\Phi(x)$                       | Smooth               | Smooth gating            | More complex        | Transformers         |
| **Softmax**    | $\frac{e^{z_i}}{\sum_j e^{z_j}}$ | Coupled Jacobian     | Probability distribution | Not multi-label     | Multiclass output    |

$$

$$

almost everywhere.

At $z=0$, it is not differentiable in the ordinary sense.

### Advantages

* Historical neural networks
* Conceptual understanding

S-shaped curve.

## Derivative

$$
\boxed{\sigma'(z)=\sigma(z)(1-\sigma(z))}
$$

Maximum derivative:

$$
0.25
$$

at:

$$
z=0
$$

For large positive/negative $z$:

$$
\sigma'(z)\rightarrow0
$$

### Advantages

* Smooth
* Differentiable
* Output between 0 and 1
* Natural probability-like output
* Useful for binary classification

### Disadvantages

#### 1. Vanishing gradient

When saturated:

$$
\sigma'(z)\approx0
$$

#### 2. Not zero-centered

Output is:

$$
(0,1)
$$

#### 3. More expensive than ReLU

Requires exponential calculation.

#### 4. Usually not preferred for hidden layers

### Where Used?

🔥 **Binary classification output**

Also:

* Multi-label classification
* Some gating mechanisms

### 🎯 Remember

> **Sigmoid → binary/multi-label output.**

## Derivative

$$
\boxed{\tanh'(z)=1-\tanh^2(z)}
$$

At zero:

$$
\tanh'(0)=1
$$

At extreme values:

$$
\tanh'(z)\rightarrow0
$$

### Advantages

* Differentiable
* Zero-centered
* Output includes positive and negative values
* Stronger gradient around zero than sigmoid

### Disadvantages

* Vanishing gradients
* Saturates at $-1$ and $+1$
* More computationally expensive than ReLU
* Not usually the default hidden activation in modern networks

### Where Used?

* Historical hidden layers
* Recurrent neural networks
* Selected architectures

### 🎯 Remember

> **Tanh = zero-centered sigmoid-like function, but still suffers from saturation.**

## Derivative

$$
ReLU'(z)=
\begin{cases}
0 & z<0\\
1 & z>0
\end{cases}
$$

At $z=0$, the derivative is technically undefined; implementations use a chosen subgradient convention.

### Advantages

🔥 Very important:

* Simple
* Fast
* Computationally efficient
* Positive region has gradient 1
* Creates sparse activations
* Widely used

### Disadvantages

## ⚠️ Dying ReLU

For:

$$
z<0
$$

we have:

$$
ReLU(z)=0
$$

and:

$$
ReLU'(z)=0
$$

A neuron can therefore stop receiving useful gradient updates.

Also:

* Not zero-centered
* Negative information is completely removed

### Where Used?

* MLPs
* CNNs
* Computer vision
* Many traditional deep networks
* Hidden layers

### 🎯 Remember

> **ReLU = common default hidden activation for many basic neural networks.**

# 7. Leaky ReLU

## Formula

$$
f(z)=
\begin{cases}
z & z>0\\
\alpha z & z\leq0
\end{cases}
$$

$$

Unlike ReLU, the negative side has a small slope.

## Derivative

$$
f'(z)=
\begin{cases}
1 & z>0\\
\alpha & z<0
\end{cases}
$$

Therefore, if:

$$
\alpha=0.01
$$

negative inputs still have a gradient:

$$
f'(z)=0.01
$$

### Advantages

* Reduces dying ReLU problem
* Keeps a small negative gradient
* Computationally simple
* Preserves some negative information

### Disadvantages

* Requires selecting $\alpha$
* Not zero-centered
* Doesn't always outperform ReLU

### Where Used?

* Alternative to ReLU
* Some CNNs
* Some MLPs
* Networks experiencing dying-ReLU issues

### 🎯 Remember

> **Leaky ReLU = ReLU with a small negative slope.**

# 9. GELU 🔥

## Formula

Exact:

$$
GELU(x)=x\Phi(x)
$$

where $\Phi(x)$ is the standard normal CDF.

Common approximation:

$$
GELU(x)\approx
\frac{x}{2}
\left[
1+
\tanh
\left(
\sqrt{\frac{2}{\pi}}
(x+0.044715x^3)
\right)
\right]
$$

### 🚨 Important

You **do not need to memorize the approximation**.

## Derivative

Exact derivative:

$$

$$

where:

* $\Phi(x)$ = standard normal CDF
* $\phi(x)$ = standard normal PDF

Again:

> **Understand the concept, don't memorize the derivative.**

### Advantages

* Smooth
* Smooth gating behavior
* Works well in modern architectures
* Important in Transformer-style models

### Disadvantages

* More computationally expensive than ReLU
* More mathematically complex
* Not necessary for every simple neural network

### Where Used?

🔥 Important:

* Transformers
* BERT-style architectures
* Vision Transformers
* Modern neural architectures

### 🎯 Remember

> **GELU = smooth activation strongly associated with modern Transformer architectures.**

# 10. Softmax

Softmax is different because it works across a **vector of logits**.

## Formula

$$
\boxed{
P_i=
\frac{e^{z_i}}
{\sum_{j=1}^{K}e^{z_j}}
}
$$

## Example

Suppose:

$$
z=[2,1,0]
$$

Softmax approximately gives:

$$
[0.665,0.245,0.090]
$$

and:

$$
0.665+0.245+0.090=1
$$

## Derivative

Softmax uses a Jacobian.

$$
\frac{\partial P_i}{\partial z_j}
=
P_i(\delta_{ij}-P_j)
$$

For $i=j$:

$$
\frac{\partial P_i}{\partial z_i}
=
P_i(1-P_i)
$$

For $i\neq j$:

$$
\frac{\partial P_i}{\partial z_j}
=
-P_iP_j
$$

### For your level

You only need to understand:

> **Softmax outputs are coupled. Changing one logit affects the other class probabilities.**

You don't need to memorize the Jacobian.

### Advantages

* Converts logits into probability distribution
* Outputs lie between 0 and 1
* Probabilities sum to 1
* Appropriate for mutually exclusive classes

### Disadvantages

* Not suitable for multi-label classification
* Naive implementation can have numerical overflow issues
* More complex than sigmoid

### Where Used?

🔥 **Multiclass classification**

Example:

```text
Input image
     ↓
Neural network
     ↓
[Cat, Dog, Horse]
     ↓
Softmax
     ↓
[0.70, 0.20, 0.10]
```

# 🔥 11. Vanishing Gradient vs Dying ReLU

This distinction is **very important for interviews**.

## Vanishing Gradient

Commonly associated with saturated:

* Sigmoid
* Tanh

The derivative becomes extremely small:

$$
\frac{\partial L}{\partial w}\approx0
$$

Therefore, earlier layers may learn extremely slowly.

```text
Sigmoid/Tanh
      ↓
Saturation
      ↓
Small derivative
      ↓
Small gradients
      ↓
Slow learning
```

---

## Dying ReLU

For ReLU:

$$
z<0
$$

gives:

$$
ReLU(z)=0
$$

and:

$$
ReLU'(z)=0
$$

A neuron can therefore become inactive and stop learning.

```text
ReLU
 ↓
Negative region
 ↓
Output = 0
 ↓
Gradient = 0
 ↓
Neuron may stop learning
```

### Memory trick 🧠

> **Sigmoid/Tanh → vanishing gradients**
> **ReLU → dying neurons**

# 🔥 13. ReLU vs Leaky ReLU

| Property          | ReLU                | Leaky ReLU       |
| ----------------- | ------------------- | ---------------- |
| Positive region   | $z$                 | $z$              |
| Negative region   | $0$                 | $\alpha z$       |
| Negative gradient | 0                   | $\alpha$         |
| Dying ReLU        | Possible            | Reduced          |
| Complexity        | Very low            | Very low         |
| Main purpose      | Standard activation | ReLU alternative |

### Mental picture

```text
ReLU:

negative ───────┐
                └──────── positive


Leaky ReLU:

negative ──────/
              /
             └──────── positive
```

# 🔥 14. Multiclass vs Multi-label

This is one of the most important practical distinctions.

## Multiclass

Only **one** class is correct.

Example:

```text
Cat
Dog
Horse
```

Use:

$$
\boxed{\text{Softmax}}
$$

because probabilities compete and sum to 1.

---

## Multi-label

Multiple classes can be true simultaneously.

Example:

```text
☑ Dog
☑ Car
☑ Person
```

Use:

$$
\boxed{\text{Independent Sigmoids}}
$$

Each output is independent.

### Memory trick

> **One class → Softmax**
> **Many classes → Sigmoid per class**

# 🔥 15. Output Activation Cheat Sheet

| Problem                    |    Output | Activation                       | Common PyTorch loss |
| -------------------------- | --------: | -------------------------------- | ------------------- |
| Regression                 | 1 or more | Linear / none                    | MSE / MAE / Huber   |
| Binary classification      |         1 | Sigmoid conceptually             | `BCEWithLogitsLoss` |
| Multiclass classification  |       $K$ | Softmax conceptually             | `CrossEntropyLoss`  |
| Multi-label classification |       $K$ | Independent sigmoid conceptually | `BCEWithLogitsLoss` |

### 🚨 PyTorch rule

During training:

```python
nn.BCEWithLogitsLoss()
```

expects **raw logits**.

And:

```python
nn.CrossEntropyLoss()
```

also expects **raw logits**.

So normally don't manually apply sigmoid/softmax before these losses.

# 🧠 17. Interview Questions You Should Be Able to Answer

### Q1. Why do neural networks need activation functions?

Because without nonlinear activation functions, stacking linear layers still produces an overall linear transformation.

---

### Q2. Why is ReLU preferred over sigmoid in many hidden layers?

Because ReLU is computationally simple and maintains a derivative of 1 in its positive region, helping gradient flow, while sigmoid can saturate and produce very small gradients.

---

### Q3. What is the dying ReLU problem?

When a ReLU neuron remains in the negative region:

$$
ReLU(z)=0
$$

and:

$$
ReLU'(z)=0
$$

so it may stop receiving useful gradient updates.

---

### Q4. How does Leaky ReLU help?

It gives the negative region a small slope:

$$
f(z)=\alpha z
$$

so the gradient isn't zero there.

---

### Q5. Why is sigmoid used for binary classification?

Because it maps a logit to a value between 0 and 1, which can represent the model's predicted probability for the positive class.

---

### Q6. Why is softmax used for multiclass classification?

Because it converts multiple logits into a probability distribution whose values sum to 1, representing competing mutually exclusive classes.

---

### Q7. Why can't we use softmax for multi-label classification?

Because softmax forces the outputs to compete and sum to 1, while multiple labels can independently be true.

---

### Q8. What activation is associated with Transformers?

**GELU** is an important activation commonly encountered in Transformer-style architectures.

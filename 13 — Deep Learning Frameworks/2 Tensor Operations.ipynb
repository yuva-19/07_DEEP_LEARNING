# 3. 🔥 Reshaping

Reshaping changes the tensor's shape while preserving its number of elements.

```python
x = torch.arange(12)

x = x.reshape(3, 4)
```

Before:

$$
(12)
$$

After:

$$
(3,4)
$$

because:

$$
12=3\times4
$$

### Fundamental rule

$$
\boxed{
\text{elements before}
=
\text{elements after}
}
$$

So:

```text
(2,3,4) → (4,6)
```

is valid because:

$$
2\times3\times4=4\times6=24
$$

but:

```text
(2,3,4) → (5,5)
```

is invalid because:

$$
24\neq25
$$

---

## `-1`

PyTorch can infer one dimension:

```python
x.reshape(3, -1)
```

For 12 elements:

$$
3\times?=12
$$

therefore:

$$
?=4
$$

Result:

```text
(3,4)
```

### Memory trick

> **Reshape changes organization, not the number of elements.**

## 🔥 Broadcasting rules

Compare dimensions **from right to left**.

Two dimensions are compatible when:

1. They are equal, or
2. One is `1`, or
3. One dimension is missing.

Example:

```text
(2,3)
(3,)
```

Align:

```text
(2,3)
   (3)
```

Compatible → ✅

Another:

```text
(4,3,2)
(1,2)
```

Align:

```text
(4,3,2)
    (1,2)
```

Compatible → ✅

But:

```text
(2,3)
(2,)
```

aligns as:

```text
(2,3)
   (2)
```

`3` and `2` don't match → ❌

# 5. 🔥 Broadcasting in Neural Networks

This is particularly important for:

$$
Z=WX+b
$$

Suppose:

```text
Z.shape = (32,10)
b.shape = (10,)
```

PyTorch can broadcast the bias across all 32 samples:

$$
(32,10)+(10)
\rightarrow
(32,10)
$$

So one bias vector can be applied to every sample.

> **Broadcasting is one reason tensor operations can closely match the mathematical notation used in neural networks.**

# 6. 🔥 Matrix Multiplication

For:

$$
A\in\mathbb{R}^{m\times n}
$$

and:

$$
B\in\mathbb{R}^{n\times p}
$$

the result is:

$$
AB\in\mathbb{R}^{m\times p}
$$

Shape rule:

```text
(m,n) × (n,p)
      ↓
    (m,p)
```

The **inner dimensions must match**.

---

## PyTorch

```python
C = A @ B
```

or:

```python
C = torch.matmul(A, B)
```

Example:

```python
A = torch.tensor([
    [1, 2],
    [3, 4]
])

B = torch.tensor([
    [5, 6],
    [7, 8]
])

C = A @ B
```

Result:

```text
[[19, 22],
 [43, 50]]
```

For example:

$$
1(5)+2(7)=19
$$

# 7. 🚨 `*` vs `@`

Very important:

### `*` → element-wise multiplication

```python
A * B
```

Each corresponding element is multiplied.

### `@` → matrix multiplication

```python
A @ B
```

Uses matrix multiplication rules.

```text
*  → element by element

@  → linear algebra matrix multiplication
```

---

The core operation is:

So matrix multiplication appears throughout Deep Learning.

# 13. 🔥 Stacking — `torch.stack()`

Stacking combines tensors by creating a **new dimension**.

Suppose:

```text
A = (2,3)
B = (2,3)
```

### `cat`

```python
torch.cat([A, B], dim=0)
```

→

```text
(4,3)
```

### `stack`

```python
torch.stack([A, B], dim=0)
```

→

```text
(2,2,3)
```

So:

$$
\boxed{\text{cat → existing dimension}}
$$

$$
\boxed{\text{stack → new dimension}}
$$

---

# 14. 🔥 `cat()` vs `stack()`

|                           | `torch.cat()` | `torch.stack()` |
| ------------------------- | ------------- | --------------- |
| Creates new dimension?    | ❌             | ✅               |
| Joins existing dimension? | ✅             | ❌               |
| Example input             | `(2,3)+(2,3)` | `(2,3)+(2,3)`   |
| Example output            | `(4,3)`       | `(2,2,3)`       |

### 🧠 Memory trick

> **CAT = concatenate into an existing axis.**

> **STACK = stack into a new axis.**

# 16. 🔥 Operation → Shape Mental Model

This is the most important revision table:

| Operation    | What happens to data?   | Shape effect                 |
| ------------ | ----------------------- | ---------------------------- |
| Indexing     | Select values           | Usually reduces/selects axes |
| Slicing      | Select range            | Selects part of tensor       |
| Reshape      | Rearrange shape         | Same number of elements      |
| Broadcasting | Align compatible shapes | Implicit expansion           |
| `@`          | Matrix multiplication   | `(m,n)×(n,p) → (m,p)`        |
| `sum()`      | Combine values          | Reduces dimensions           |
| `mean()`     | Average values          | Reduces dimensions           |
| `cat()`      | Join tensors            | Existing dimension grows     |
| `stack()`    | Group tensors           | New dimension appears        |


❌ because:

$$
12\neq25
$$

---

### ❌ Confusing `*` and `@`

```python
A * B
```

≠

```python
A @ B
```

---

### ❌ Forgetting broadcasting rules

```text
(2,3) + (2,)
```

❌ because the aligned dimensions are `3` and `2`.

---

### ❌ Confusing `cat` and `stack`

```text
cat   → existing dimension
stack → new dimension
```

---

### ❌ Forgetting `dim`

```python
x.sum()
```

and:

```python
x.sum(dim=0)
```

do **different reductions**.

# 🎯 Interview / GATE-Style Points

### Q1. What must remain constant during reshape?

The total number of elements.

$$
\boxed{\prod \text{old dimensions}
=
\prod \text{new dimensions}}
$$

### Q2. What is broadcasting?

Automatic alignment/expansion of compatible tensor dimensions during operations.

### Q3. What are the matrix multiplication shape requirements?

For:

$$
(m,n)(n,p)
$$

the inner dimensions must match.

### Q4. Difference between `*` and `@`?

```text
* → element-wise multiplication
@ → matrix multiplication
```

### Q5. Difference between `cat` and `stack`?

```text
cat   → existing dimension
stack → new dimension
```

### Q6. What does `dim=0` mean?

The first axis/dimension.

### Q7. What does `keepdim=True` do?

It retains the reduced dimension with size `1`.

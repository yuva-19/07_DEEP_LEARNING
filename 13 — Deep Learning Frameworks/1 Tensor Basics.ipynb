# 1. 🔥 What Is a Tensor?

A **tensor** is a multidimensional numerical data structure.

The easiest way to understand it is as a progression:

```text
Scalar
   ↓
Vector
   ↓
Matrix
   ↓
Higher-dimensional tensor
```

| Type      | Dimensions | Example shape    |
| --------- | ---------: | ---------------- |
| Scalar    |         0D | `()`             |
| Vector    |         1D | `(3,)`           |
| Matrix    |         2D | `(2,3)`          |
| 3D Tensor |         3D | `(2,3,4)`        |
| 4D Tensor |         4D | `(32,3,224,224)` |

### 🧠 Important

In PyTorch, a tensor is more than just its values. We care about:

```text
Tensor
 ├── Values
 ├── Shape
 ├── Dimension / rank
 ├── Dtype
 └── Device
```

# 3. 🔥 Vector

A vector is a **1D collection of values**.

$$
x=
\begin{bmatrix}
10\\
20\\
30
\end{bmatrix}
$$

PyTorch:

```python
x = torch.tensor([10, 20, 30])
```

Shape:

```python
x.shape
```

```text
torch.Size([3])
```

Therefore:

```text
Elements   = 3
Dimensions = 1
Shape      = (3,)
```

### 🧠 Don't confuse:

> **3 elements ≠ 3 dimensions**

A vector containing 3 numbers has **1 dimension**.

# 4. 🔥 Matrix

A matrix is a **2D tensor**.

Example:

$$
X=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
$$

PyTorch:

```python
X = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
```

Shape:

```python
X.shape
```

```text
torch.Size([2, 3])
```

So:

```text
Rows       = 2
Columns    = 3
Dimensions = 2
Elements   = 2 × 3 = 6
```

# 5. 🔥 Higher-Dimensional Tensors

A tensor can have more than two dimensions.

Example:

```python
x = torch.randn(2, 3, 4)
```

Shape:

$$
(2,3,4)
$$

Number of dimensions:

$$
3
$$

Number of elements:

$$
2\times3\times4=24
$$

Think of the shape as describing the size along each axis:

```text
(2, 3, 4)

 ↑  ↑  ↑
 │  │  └── Dimension 3 size
 │  └───── Dimension 2 size
 └──────── Dimension 1 size
```

You don't need to physically visualize a 10D tensor. 😄

Just understand:

> **Each additional axis adds another dimension to the data structure.**

# 6. 🔥 Shape

**Shape tells us the size of the tensor along every dimension.**

Example:

```python
x = torch.randn(32, 128, 768)
```

Then:

$$
\boxed{\text{shape}=(32,128,768)}
$$

This means:

```text
Dimension 1 → 32
Dimension 2 → 128
Dimension 3 → 768
```

---

# 7. 🔥 Shape vs Dimension vs Elements

This is one of the most important distinctions in this topic.

Suppose:

```python
x = torch.randn(32, 128, 768)
```

### Shape

$$
(32,128,768)
$$

### Number of dimensions

$$
3
$$

### Number of elements

$$
32\times128\times768
=
3,145,728
$$

So:

| Concept              | Meaning              | Example        |
| -------------------- | -------------------- | -------------- |
| **Shape**            | Size along each axis | `(32,128,768)` |
| **Dimension / rank** | Number of axes       | `3`            |
| **Elements**         | Total values         | `3,145,728`    |

### 🧠 Memory trick

> **Shape = sizes**

> **Dimension = number of sizes**

---

# 8. 🔥 Why Shape Matters in Neural Networks

Neural networks perform mathematical operations between tensors.

From earlier modules:

$$
Z=WX+b
$$

Suppose:

$$
W=(3,4)
$$

and:

$$
X=(4,10)
$$

Then:

$$
(3,4)(4,10)
$$

The inner dimensions match:

$$
\boxed{4=4}
$$

so:

$$
Z=(3,10)
$$

But:

$$
(3,4)(5,10)
$$

❌ cannot be multiplied because:

$$
4\neq5
$$

This is why **tensor shape errors are among the most common errors you'll encounter when implementing neural networks.**

## Important PyTorch dtypes

| dtype                        | Typical purpose                    |
| ---------------------------- | ---------------------------------- |
| `torch.float32`              | General neural-network computation |
| `torch.float64`              | Higher numerical precision         |
| `torch.float16`              | Lower-precision computation        |
| `torch.bfloat16`             | Modern deep-learning computation   |
| `torch.int64` / `torch.long` | Integer labels / indexing          |
| `torch.bool`                 | Boolean masks                      |

For your current learning stage:

$$
\boxed{\texttt{float32}}
$$

is the most important default to understand.

# 14. 🔥 Moving a Tensor to a Device

You can move a tensor using `.to()`:

```python

```

You can also create it directly:

```python

)
```

For a model:

```python
model = model.to(device)
```

---

# 15. ⚠️ Device Mismatch

This becomes VERY important when you start coding models.

Suppose:

```text
Model → GPU
Input → CPU
```

and you try:

```python
output = model(x)
```

You can get a device mismatch error.

Usually you want:

```text
Model → GPU
Input → GPU
```

So:

```python
model = model.to(device)
x = x.to(device)

output = model(x)
```

### 🧠 Remember

> **Tensors participating in the same computation generally need compatible devices.**

# 17. 🔥 `shape` vs `ndim`

In PyTorch:

```python
x.shape
```

gives the shape.

```python
x.ndim
```

gives the number of dimensions.

Example:

```python
x = torch.randn(32, 128, 768)

print(x.shape)
print(x.ndim)
```

Output:

```text
torch.Size([32, 128, 768])
3
```

So:

$$
\boxed{\texttt{x.ndim}=\text{number of dimensions}}
$$

# 22. ⚠️ Common Mistakes

### ❌ Mistake 1 — "A vector of 3 numbers is 3D"

No.

```text
[1, 2, 3]
```

has:

```text
Elements   = 3
Dimensions = 1
Shape      = (3,)
```

---

### ❌ Mistake 2 — "Shape is the number of dimensions"

No.

```text
shape = (32,128,768)
```

but:

```text
dimensions = 3
```

---

### ❌ Mistake 3 — Ignoring dtype

A tensor containing:

```text
1.5, 2.7, 3.1
```

and one containing:

```text
1, 2, 3
```

can have different dtypes.

Dtype matters for operations, memory, performance and some PyTorch APIs.

---

### ❌ Mistake 4 — Ignoring device

```text
Model → GPU
Input → CPU
```

can cause an error.

---

### ❌ Mistake 5 — Thinking higher-dimensional means "more complicated mathematics"

Not necessarily.

A tensor is simply data organized across more axes.

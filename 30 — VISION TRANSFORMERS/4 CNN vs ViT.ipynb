# 1. 🧠 Core Idea

The most important thing to understand is:

> **CNNs are designed around image-specific spatial assumptions, while ViTs represent an image as tokens and use self-attention to learn relationships between those tokens.**

A simplified view:

```text
CNN
IMAGE
  ↓
CONVOLUTION
  ↓
LOCAL FEATURES
  ↓
HIERARCHICAL FEATURES
  ↓
PREDICTION
```

```text
ViT
IMAGE
  ↓
PATCHES
  ↓
PATCH TOKENS
  ↓
SELF-ATTENTION
  ↓
CONTEXTUALIZED TOKENS
  ↓
PREDICTION
```

The original ViT demonstrated that a Transformer could operate directly on sequences of image patches rather than requiring a convolutional backbone. ([arXiv][1])

# 2. 🔲 CNN — Convolution

A CNN processes an image using **convolutional filters**.

A small kernel moves across the image:

```text
Image

┌───┬───┬───┬───┬───┐
│   │   │   │   │   │
├───┼───┼───┼───┼───┤
│   │███│███│███│   │
├───┼───┼───┼───┼───┤
│   │███│███│███│   │
├───┼───┼───┼───┼───┤
│   │███│███│███│   │
├───┼───┼───┼───┼───┤
│   │   │   │   │   │
└───┴───┴───┴───┴───┘
       ↑
   local region
```

The same learned kernel is applied at different spatial positions.

### Therefore CNNs naturally have:

* **Local receptive fields**
* **Weight sharing**
* **Spatially structured processing**

# 3. 👀 ViT — Self-Attention

ViT first converts the image into patches.

For example:

$$
224\times224
$$

with:

$$
16\times16
$$

patches gives:

$$
14\times14=196
$$

patch tokens.

Then self-attention operates on the token sequence.

```text
P1  P2  P3  P4 ... P196
│   │   │   │       │
└───┴───┼───┴───────┘
        │
        ▼
   Self-Attention
```

Unlike a convolution kernel, self-attention computes relationships between tokens based on their representations.

# 4. 📍 Local Processing

## CNN

CNNs begin with local processing.

Suppose a kernel has a \(3\times3\) receptive field:

```text
┌───┬───┬───┐
│ x │ x │ x │
├───┼───┼───┤
│ x │ x │ x │
├───┼───┼───┤
│ x │ x │ x │
└───┴───┴───┘
```

The output at that position depends directly on this local neighborhood.

This is why CNNs have a strong **locality inductive bias**.

# 5. 🌍 But CNNs Can Still Understand Global Context

This is an important correction to a common oversimplification.

It is **wrong** to say:

> “CNNs can only understand local information.”

As CNN layers are stacked, the effective receptive field can grow.

```text
Layer 1
  ↓
Small local region

Layer 2
  ↓
Larger region

Layer 3
  ↓
Even larger region

Deep network
  ↓
Large/global contextual information
```

# 6. 🌐 ViT — Global Relationships

Self-attention allows token representations to interact across the sequence.

Conceptually:

```text
Patch A ───────────────────► Patch Z
   │                            │
   └──────── attention ─────────┘
```

A patch doesn't have to be spatially adjacent to another patch to exchange information through self-attention.

For example:

```text
┌──────────────────────────────┐
│                              │
│  🐶 head              tail   │
│                              │
└──────────────────────────────┘

     Head patch
         │
         │ self-attention
         ▼
     Tail patch
```

This is what your syllabus means by:

> **Global relationships**

The original ViT formulation uses self-attention over the sequence of image patches. ([arXiv][1])

# 7. 🧠 CNN's Strong Image Inductive Bias

This is one of the most important terms in the entire comparison.

## What is inductive bias?

An **inductive bias** is an assumption built into a learning system that influences what kinds of patterns it can learn efficiently.

For images, CNNs build in useful assumptions such as:

### 1. Locality

Nearby pixels/features are related.

### 2. Weight sharing

The same kernel can detect a feature at different locations.

### 3. Spatial structure

The arrangement of features on the image grid matters.

### 4. Translation-related structure

A visual feature can occur at different spatial locations.

These are not merely things the CNN learns from data — the architecture itself incorporates them.

Research comparing convolutional inductive biases with ViTs describes convolutional architectures as having stronger built-in image priors, particularly around locality and translation-related structure. ([arXiv][2])

---

# 8. 🧩 ViT — More Flexible Global Interactions

ViT does not impose the same hard convolutional locality structure.

Instead:

```text
Patch 1
   ↕
Patch 2
   ↕
Patch 3
   ↕
...
Patch N
```

Self-attention learns which token interactions matter.

Therefore:

> **ViT provides a more flexible mechanism for modeling relationships among visual tokens.**

This is what your syllabus means by:

> **More flexible global interactions**

⚠️ **Do not interpret “more flexible” as “always better.”**

It describes the architectural mechanism, not a universal performance ranking.

---

# 9. ⚖️ Inductive Bias — The Tradeoff

This is an important deeper concept.

### CNN

```text
Strong image assumptions
          ↓
Locality
Weight sharing
Spatial structure
          ↓
Less has to be learned
```

### ViT

```text
More flexible token interactions
          ↓
Fewer CNN-specific assumptions
          ↓
More has to be learned from data
```

The original ViT paper's results helped establish the importance of large-scale pretraining for the architecture. Later research also explicitly studied how adding convolutional/locality biases can improve data efficiency in some settings. ([arXiv][1])

### 🧠 Memory trick

> **CNN gives the model more knowledge about images beforehand.**

> **ViT gives the model a more flexible interaction mechanism and lets training learn more of the visual structure.**

# 12. 🔥 Hierarchy vs Tokens

This is the deepest conceptual contrast in 30.4.

### CNN

```text
IMAGE
  ↓
LOCAL FEATURES
  ↓
COMBINED FEATURES
  ↓
LARGER FEATURES
  ↓
HIGH-LEVEL FEATURES
```

The representation is built through a **spatial hierarchy**.

### ViT

```text
IMAGE
  ↓
PATCHES
  ↓
PATCH TOKENS
  ↓
SELF-ATTENTION
  ↓
CONTEXTUALIZED TOKENS
```

The representation is built through **token interactions and repeated Transformer blocks**.

---

# 13. 🐶 Example: Object With Distant Parts

Imagine an image of a dog:

```text
┌─────────────────────────────┐
│                             │
│  🐶                         │
│  HEAD                  TAIL │
│                             │
└─────────────────────────────┘
```

The dog's head and tail may be far apart.

### CNN

Early layers:

```text
Head → local features
Tail → local features
```

Then deeper layers progressively combine broader information.

```text
Head
 ↓
larger receptive field
 ↓
body
 ↓
tail
```

### ViT

The patch containing the head and the patch containing the tail can directly participate in self-attention:

```text
Head patch
     │
     │ attention
     ▼
Tail patch
```

That illustrates the architectural distinction.

---

# 14. 🧠 A Crucial Nuance About "Global"

Don't say in an interview:

> ❌ “CNN is local and ViT is global.”

Say:

> ✅ **CNNs have a strong locality inductive bias and build larger receptive fields progressively through layers, whereas standard ViT self-attention allows direct interactions between tokens across the image.**

That's a much stronger technical answer.

# 15. 📊 Complete CNN vs ViT Comparison

| Feature                    | CNN                                  | ViT                                                     |
| -------------------------- | ------------------------------------ | ------------------------------------------------------- |
| Core operation             | Convolution                          | Self-attention                                          |
| Basic input representation | Spatial grid                         | Patch/token sequence                                    |
| Primary interaction        | Local neighborhood                   | Token-to-token                                          |
| Locality                   | Strong built-in bias                 | Not imposed in the same way                             |
| Global relationships       | Built progressively                  | Direct token interactions through attention             |
| Image inductive bias       | Strong                               | Weaker in vanilla ViT                                   |
| Weight sharing             | Convolution kernels shared spatially | Transformer parameters shared across sequence positions |
| Spatial structure          | Built into convolutional processing  | Positional information explicitly represented           |
| Feature representation     | Spatial feature maps                 | Token embeddings                                        |
| Feature hierarchy          | Naturally hierarchical               | Repeated contextualized token representations           |
| Early processing           | Local feature extraction             | Patch embedding                                         |
| Global representation      | Deep feature hierarchy / pooling     | CLS or another aggregation mechanism                    |
| Main mental model          | **Build a hierarchy**                | **Model token relationships**                           |

# 16. 🧮 Important Terminology

### **Local receptive field**

The region of the previous representation that directly influences a particular output.

---

### **Inductive bias**

Architectural assumptions that influence what patterns the model learns efficiently.

---

### **Global relationship**

A relationship between representations that may correspond to distant spatial regions of the image.

---

### **Hierarchical feature**

A representation formed progressively from lower-level visual patterns to increasingly higher-level structures.

---

### **Patch token**

A vector representing an image patch after embedding.

# 17. 🎤 Interview Questions

### Q1. What is the fundamental difference between CNN and ViT?

**Answer:**

CNNs use convolution to process local spatial neighborhoods, while ViTs represent images as patch tokens and use self-attention to model relationships among those tokens.

---

### Q2. Why are CNNs said to have strong inductive bias?

Because their architecture explicitly incorporates image-related assumptions such as locality, spatial structure, and shared convolutional filters.

---

### Q3. Does a CNN only understand local information?

**No.**

Deep CNNs can obtain large receptive fields and represent global context. Their key distinction is that locality is built into the architecture and broader context is developed progressively.

---

### Q4. Why can ViT model long-range relationships?

Because self-attention allows tokens to directly interact with other tokens across the sequence.

---

### Q5. What does "hierarchical spatial features" mean?

CNNs progressively transform low-level patterns such as edges into increasingly complex representations such as textures, shapes, object parts, and objects.

---

### Q6. What is the equivalent basic representation in ViT?

Image patches are converted into **patch embeddings**, which act as visual tokens.

---

### Q7. Does ViT have no inductive bias?

Not literally.

A vanilla ViT has **less image-specific inductive bias than a CNN**, but it still contains architectural structure and assumptions, including the tokenization process and positional representation.

---

### Q8. Why might convolutional inductive bias be useful?

It can make learning image structure more efficient because some useful assumptions about visual data are already encoded in the architecture. Research such as ConViT explicitly investigates combining these convolutional biases with the flexibility of attention. ([arXiv][2])

---

# 18. ⚠️ Common Mistakes

### ❌ "CNN cannot model global relationships."

Wrong.

CNNs can build global representations through depth and increasing receptive fields.

---

### ❌ "ViT has zero inductive bias."

Too strong.

The better statement is:

> **Vanilla ViT has substantially less image-specific inductive bias than CNNs.**

---

### ❌ "Global attention means every image pixel attends to every other pixel."

Not necessarily.

Standard ViT operates on **patch tokens**, not individual pixels.

---

### ❌ "CNN directly detects objects in the first layer."

Usually not.

Early layers generally learn lower-level patterns, with deeper layers building increasingly complex representations.

---

### ❌ "ViT has no hierarchy."

Be careful.

Vanilla ViT does not have the same **explicit hierarchical spatial feature structure** as conventional CNNs. But its representations still evolve through stacked Transformer layers.

---

### ❌ "ViT replaced CNN because CNNs became obsolete."

No.

The architectural families continue to coexist, and many modern vision architectures combine convolution and attention ideas.

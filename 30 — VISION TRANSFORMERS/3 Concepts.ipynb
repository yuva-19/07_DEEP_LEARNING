# 1. 🧠 Complete ViT Pipeline

The core Vision Transformer pipeline is:

```text
                         IMAGE
                           │
                           ▼
                    Split into patches
                           │
                           ▼
                   Patch embeddings
                           │
                           ▼
                Add positional embeddings
                           │
                           ▼
                      Add [CLS]
                           │
                           ▼
              Multi-Head Self-Attention
                           │
                           ▼
                  Transformer Encoder
                           │
                           ▼
                  Final [CLS] vector
                           │
                           ▼
                       MLP Head
                           │
                           ▼
                    Class Prediction
```

# 2. 🖼️ Image Patches

A ViT does not directly feed the entire image into the Transformer.

Instead, it divides the image into fixed-size patches.

For example:

```text
Image = 224 × 224 × 3
Patch = 16 × 16 × 3
```

Number of patches:

$$
N = \frac{H}{P}\times\frac{W}{P}
$$

Therefore:

$$
N = \frac{224}{16}\times\frac{224}{16}
$$

$$
N = 14\times14
$$

$$
\boxed{N=196}
$$

So:

```text
224 × 224 image

┌──┬──┬──┬──┬──┬──┐
│  │  │  │  │  │  │
├──┼──┼──┼──┼──┼──┤
│  │  │  │  │  │  │
├──┼──┼──┼──┼──┼──┤
│  │  │  │  │  │  │
├──┼──┼──┼──┼──┼──┤
│  │  │  │  │  │  │
└──┴──┴──┴──┴──┴──┘

14 × 14 = 196 patches
```

# 3. 🔢 Patch Embeddings

The raw patch is not directly used by the Transformer.

It is converted into a fixed-dimensional vector called a **patch embedding**.

### Concept

```text
Raw image patch
      │
      ▼
Flatten
      │
      ▼
Linear projection
      │
      ▼
Patch embedding
```

For example:

```text
16 × 16 × 3
      ↓
768 values
      ↓
Linear projection
      ↓
768-dimensional embedding
```

Conceptually:

$$
z = xE+b
$$

where:

* \(x\) = flattened image patch
* \(E\) = learnable projection matrix
* \(b\) = bias
* \(z\) = patch embedding

### 🧠 Important intuition

Think of:

> **Patch embedding = converting raw pixels into a meaningful visual token representation.**

This is similar to word embeddings in NLP.

```text
NLP:

"cat"
 ↓
word embedding
 ↓
vector


ViT:

image patch
 ↓
patch embedding
 ↓
vector
```

**Patch ≠ Patch embedding**

# 4. 📍 Positional Embeddings

There is a major problem.

Suppose we have:

```text
Patch A
Patch B
Patch C
```

If the Transformer only sees their feature vectors, it does not inherently know:

```text
A is at the top-left
B is in the center
C is at the bottom-right
```

Transformers need positional information because the image patches have a spatial arrangement.

Therefore ViT adds **positional embeddings**.

---

## Patch + Position

Conceptually:

$$
Z = E_{patch}+E_{position}
$$

So each token contains:

```text
WHAT is this patch?
        +
WHERE is this patch?
```

### 🧠 Best memory trick

> **Patch embedding = WHAT**
> **Positional embedding = WHERE**

---

### Example

```text
Patch embedding
     │
     │  "visual content"
     ▼
   [768]


Position embedding
     │
     │  "location"
     ▼
   [768]

       +
       ↓

Final token
   [768]
```

# 5. 🎯 CLS Token

ViT introduces a special learnable token:

$$
\boxed{[CLS]}
$$

This token is added to the beginning of the patch sequence.

Without CLS:

```text
P1 P2 P3 ... P196
```

With CLS:

```text
[CLS] P1 P2 P3 ... P196
```

Therefore:

$$
196+1=197
$$

tokens.

---

## Why CLS?

The CLS token is used as a global representation of the image.

During self-attention, the CLS token can interact with all patch tokens.

Conceptually:

```text
              ┌──── P1
              │
              ├──── P2
              │
[CLS] ────────┼──── P3
              │
              ├──── P4
              │
              └──── ...
```

After the Transformer:

```text
[CLS]
  ↓
Image representation
  ↓
MLP head
  ↓
Class prediction
```

# 6. 👀 Self-Attention

This is one of the most important concepts in ViT.

Self-attention allows every token to determine how strongly it should interact with other tokens.

For a token, we construct:

* Query \(Q\)
* Key \(K\)
* Value \(V\)

The standard scaled dot-product attention is:

$$
\boxed{
Attention(Q,K,V)
=
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
}
$$

## 6.1 Query, Key and Value

A useful intuition:

### Query

> **What information am I looking for?**

### Key

> **What information do I contain that others can match against?**

### Value

> **What information should I actually provide?**

---

### Example

Suppose the image contains:

```text
        🐶
     Dog's face
```

One patch around the dog's eye may need information from another patch containing the dog's ear.

Self-attention allows:

```text
Eye patch
   │
   ├──────► Ear patch
   │
   ├──────► Nose patch
   │
   └──────► Face patch
```

The model learns which relationships are useful.

---

# 7. 🔢 Self-Attention Step-by-Step

Suppose the token matrix is:

$$
X
$$

We create:

$$
Q=XW_Q
$$

$$
K=XW_K
$$

$$
V=XW_V
$$

where \(W_Q,W_K,W_V\) are learnable matrices.

Then:

### Step 1 — Similarity

$$
QK^T
$$

This measures how strongly tokens relate.

### Step 2 — Scaling

$$
\frac{QK^T}{\sqrt{d_k}}
$$

### Step 3 — Softmax

$$
A=
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)
$$

This produces attention weights.

### Step 4 — Weighted aggregation

$$
Output=AV
$$

So the output of each token contains information gathered from other tokens.

---

# 8. 🌐 Why Self-Attention Matters for Images

CNNs naturally focus on local neighborhoods.

Self-attention allows a token to interact directly with other tokens.

For example:

```text
Patch A ───────────────────► Patch Z
       direct interaction
```

rather than requiring many convolutional layers to gradually expand the receptive field.

This gives the Transformer a mechanism for modeling **long-range relationships between image patches**.

---

# 9. 🧠 Multi-Head Attention

A single attention mechanism can learn one type of relationship.

Instead, Transformers use multiple attention heads.

This is called:

$$
\boxed{\text{Multi-Head Self-Attention}}
$$

Conceptually:

```text
                 Input Tokens
                      │
          ┌───────────┼───────────┐
          ▼           ▼           ▼
       Head 1       Head 2       Head 3
          │           │           │
          ▼           ▼           ▼
      Attention    Attention    Attention
          │           │           │
          └───────────┼───────────┘
                      ▼
                  Concatenate
                      │
                      ▼
                 Linear layer
```

## Why multiple heads?

Different heads can learn different relationships.

For example, conceptually:

```text
Head 1 → local visual relationships
Head 2 → object-part relationships
Head 3 → long-range relationships
Head 4 → shape relationships
```

⚠️ These are **illustrative interpretations**, not a guarantee that each head specializes exactly this way.

The important idea is:

> **Different attention heads operate in different learned representation subspaces.**

# 10. 🔢 Multi-Head Attention Dimensions

Suppose:

$$
D=768
$$

and:

$$
h=12
$$

attention heads.

Each head receives:

$$
d_{head}=\frac{768}{12}
$$

$$
\boxed{d_{head}=64}
$$

So:

```text
Embedding dimension = 768

       ↓ split across 12 heads

Head 1 → 64
Head 2 → 64
Head 3 → 64
...
Head 12 → 64
```

Then the outputs are concatenated back to:

$$
12\times64=768
$$

The PyTorch `MultiheadAttention` documentation similarly defines multi-head attention as multiple attention operations performed in parallel over projected representations.

# 12. 🔄 Residual Connections

Transformer blocks use residual/skip connections.

Conceptually:

```text
          ┌──────────────────────┐
          │                      │
Input ────┼──► Self-Attention ──► (+)
          │                      │
          └──────────────────────┘
```

The residual connection helps information flow through deep networks.

Similarly:

```text
          ┌──────────────────────┐
          │                      │
Input ────┼──► MLP ─────────────► (+)
          │                      │
          └──────────────────────┘
```

---

# 13. 🧮 MLP / Feed-Forward Network

The Transformer encoder contains an MLP/FFN.

Its role is different from self-attention.

### Self-attention

> Mixes information **between tokens**.

### MLP/FFN

> Transforms the representation **within each token**.

This distinction is extremely important.

```text
Self-Attention
       ↓
Token ↔ Token
       ↓
Information exchange


MLP / FFN
       ↓
Individual token
       ↓
Feature transformation
```

---

## Typical FFN

Conceptually:

$$
FFN(x)=W_2\sigma(W_1x+b_1)+b_2
$$

Usually:

```text
Embedding
   ↓
Linear
   ↓
Activation
   ↓
Linear
   ↓
Output
```

---

# 14. 🏢 Stacking Transformer Encoder Layers

A ViT does not use just one encoder layer.

Multiple encoder layers are stacked:

```text
Patch + Position + CLS
          │
          ▼
     Encoder Layer 1
          │
          ▼
     Encoder Layer 2
          │
          ▼
     Encoder Layer 3
          │
          ▼
          ⋮
          │
          ▼
     Encoder Layer L
```

The PyTorch Transformer encoder is likewise defined as a stack of encoder layers.

Each layer progressively transforms the token representations.

# 15. 🎯 Final CLS Representation

After passing through the Transformer encoder:

```text
Input sequence:

[CLS] P1 P2 P3 ... P196

            ↓

Transformer Encoder

            ↓

[CLS]' P1' P2' P3' ... P196'
```

We take the transformed CLS token:

$$
\boxed{CLS'\in\mathbb{R}^{D}}
$$

For example:

$$
CLS'\in\mathbb{R}^{768}
$$

This vector serves as the image representation for classification.

---

# 16. 🧠 MLP Classification Head

The final MLP head converts the CLS representation into class logits.

Example:

```text
CLS representation
      │
      ▼
   Linear
      │
      ▼
   Activation
      │
      ▼
   Linear
      │
      ▼
Class logits
```

For 10 classes:

$$
[768]\rightarrow[10]
$$

So the final output is:

$$
\boxed{[B,10]}
$$

for batch size \(B\).

---

# 17. ⚠️ Don't Confuse Two MLPs

This is a common interview trap.

There are two conceptually different MLP roles:

### MLP inside Transformer encoder

```text
Self-Attention
      ↓
MLP / FFN
```

Its purpose:

> Transform features within each token.

### Final MLP classification head

```text
CLS
 ↓
MLP Head
 ↓
Class logits
```

Its purpose:

> Convert the final image representation into the task output.

# 20. 🔑 The Most Important Conceptual Map

Memorize this table:

| Component                | Main question it answers                                     |
| ------------------------ | ------------------------------------------------------------ |
| **Image patch**          | What region of the image are we looking at?                  |
| **Patch embedding**      | What visual information does this patch contain?             |

> **Patch = region → Embedding = WHAT → Position = WHERE → Attention = RELATIONSHIPS → Encoder = CONTEXT + TRANSFORMATION → CLS = SUMMARY → MLP Head = PREDICTION**

# 23. ⚠️ Common Mistakes

### ❌ Mistake 1

> "ViT divides the image into individual pixels as tokens."

Not in the standard original ViT setup.

It divides the image into **patches**.

---

### ❌ Mistake 2

> "Patch embedding is positional embedding."

No.

```text
Patch embedding → visual content
Positional embedding → spatial information
```

---

### ❌ Mistake 3

> "CLS is one of the image patches."

No.

CLS is a **special learnable token**.

---

### ❌ Mistake 4

> "Self-attention and MLP do the same thing."

No.

```text
Attention → interaction between tokens
MLP       → feature transformation within tokens
```

---

### ❌ Mistake 5

> "The Transformer encoder is only self-attention."

No.

A Transformer encoder layer includes self-attention plus a feed-forward/MLP block and surrounding residual/normalization structure.

---

### ❌ Mistake 6

> "The final MLP and encoder MLP are the same."

They have different roles.

---

### ❌ Mistake 7

> "CLS is required for every modern ViT."

Not universally.

The **original ViT** uses the CLS-token approach, while other vision-transformer architectures can use different aggregation mechanisms.

For **30.3**, remember the original ViT flow.

# 24. 🎤 Interview Questions

### Q1. Why does ViT divide an image into patches?

To convert the 2D image into a sequence of visual tokens that can be processed by a Transformer.

---

### Q2. Why not use every pixel as a token?

Because the resulting sequence would be extremely long, making self-attention computationally expensive.

---

### Q3. What is a patch embedding?

A learned vector representation obtained by projecting an image patch into the Transformer embedding space.

---

### Q4. Why are positional embeddings required?

Because the Transformer needs information about the spatial position/order of image patches.

---

### Q5. What is the CLS token?

A learnable special token whose final contextualized representation can be used as the global image representation for classification.

---

### Q6. What does self-attention do?

It allows each token to weight and aggregate information from other tokens.

---

### Q7. Why use multiple attention heads?

To allow attention to operate through multiple learned representation subspaces and capture different relationships.

---

### Q8. What is the difference between attention and the MLP?

**Attention:** exchanges information between tokens.

**MLP:** transforms features within each token.

---

### Q9. What happens to the CLS token during the Transformer?

It participates in self-attention like the other tokens and becomes a contextualized representation containing information gathered from the sequence.

---

### Q10. What is the role of the final MLP head?

It maps the final image representation, commonly the CLS representation, to task-specific output logits.

---

# 25. 🧪 PyTorch Mental Model

A simplified conceptual implementation looks like:

```python
# image
x = image

# convert image into patches
patches = patchify(x)

# convert patches into embeddings
tokens = patch_embedding(patches)

# add CLS token
tokens = add_cls_token(tokens)

# add positional information
tokens = tokens + positional_embedding

# Transformer encoder
tokens = transformer_encoder(tokens)

# extract CLS
cls = tokens[:, 0]

# classification head
logits = mlp_head(cls)
```

A common implementation technique is to perform patch projection with a `Conv2d` whose `kernel_size` and `stride` equal the patch size. Conceptually, this performs the patch extraction/projection efficiently; it does **not** turn ViT into a conventional CNN architecture.

# 1. 🧠 What is a Vision Transformer?

**Vision Transformer (ViT)** applies the Transformer architecture to visual data.

The fundamental idea is:

```text
IMAGE
  ↓
Split into patches
  ↓
Convert patches → tokens
  ↓
Add positional information
  ↓
Transformer
  ↓
Self-Attention
  ↓
Visual representation
  ↓
Classification / downstream task
```

The original ViT work showed that a **pure Transformer operating directly on sequences of image patches** can perform image classification effectively, particularly with large-scale pretraining. ([arXiv][1])

> ⚠️ **ViT is NOT a CNN.**

Its fundamental feature-processing mechanism is **Transformer self-attention**, not convolution.

---

# 2. 🧩 Why Was ViT Introduced?

Before ViT, CNNs dominated computer vision.

CNNs are extremely effective because they contain strong **image-specific inductive biases**.

Researchers asked a different question:

> **Can we process an image using a general Transformer architecture instead of relying heavily on convolutional inductive biases?**

ViT demonstrated that this was possible.

The conceptual transition is:

```text
CNN-based vision
       ↓
Strong image-specific assumptions
       ↓
Convolution + locality
       ↓
       ?
       ↓
Transformer-based vision
       ↓
Image → Tokens
       ↓
Self-Attention
       ↓
Learn relationships between tokens
```

The original ViT paper explicitly explored removing the dependence on CNNs and applying a Transformer directly to image-patch sequences. ([arXiv][1])

---

# 3. 🧠 CNN Inductive Bias

An **inductive bias** is an assumption built into a model that helps it learn particular types of patterns.

CNNs have strong image-related inductive biases, particularly:

* **locality**
* **local receptive fields**
* **shared filters**
* **spatial structure**
* **hierarchical feature extraction**

A CNN therefore does not have to discover all of these properties entirely from data.

### Example

A convolution kernel observes a small neighborhood:

```text
┌───┬───┬───┬───┬───┐
│   │   │   │   │   │
├───┼───┼───┼───┼───┤
│   │ 🔍│ 🔍│   │   │
├───┼───┼───┼───┼───┤
│   │ 🔍│ 🔍│   │   │
├───┼───┼───┼───┼───┤
│   │   │   │   │   │
└───┴───┴───┴───┴───┘
```

The CNN effectively assumes:

> **Nearby pixels are likely to have meaningful relationships.**

That is an extremely useful assumption for images.

---

# 4. 🖼️ Local Receptive Fields

A CNN initially processes **local regions** of an image.

```text
IMAGE
┌───────────────────────────┐
│                           │
│      ┌────────┐           │
│      │ Kernel │           │
│      └────────┘           │
│                           │
└───────────────────────────┘
             ↓
       Local features
```

As the network becomes deeper:

```text
Small regions
     ↓
Edges / textures
     ↓
Shapes / parts
     ↓
Object structures
     ↓
Global representation
```

Thus CNNs naturally build hierarchical visual representations.

---

# 5. 🌐 Locality vs Global Relationships

A major conceptual question is:

> **How does a model understand relationships between distant parts of an image?**

Consider:

```text
┌──────────────────────────┐
│ 🐶                       │
│                          │
│                          │
│                       🐾 │
└──────────────────────────┘
```

The dog's head and paw may be far apart.

A CNN processes local neighborhoods and gradually propagates information through its layers.

Conceptually:

```text
Region A
   ↓
Local processing
   ↓
Intermediate representation
   ↓
Intermediate representation
   ↓
Region B
   ↓
Global representation
```

This does **not** mean CNNs cannot learn global relationships. Deep CNNs absolutely can.

The distinction is that **locality is built directly into convolution**.

---

# 6. 🔗 Self-Attention and Global Relationships

Transformers provide another mechanism.

With self-attention, tokens can interact with other tokens:

```text
Token 1 ──────────────→ Token 4
   │                       ↑
   └────→ Token 2 ─────────┘
             │
             ↓
          Token 3
```

The important idea:

> A token can directly use information from other tokens, including tokens representing distant image regions.

This gives the Transformer a mechanism for modeling **global relationships**.

⚠️ Detailed self-attention mathematics belongs to later sections and is **not part of 30.1**.

---

# 7. 🔄 Image → Patches → Tokens

A Transformer expects a sequence.

An image is not naturally represented as a sequence of words.

ViT therefore converts the image into a sequence of **image patches**.

```text
Original Image

┌────┬────┬────┬────┐
│ P1 │ P2 │ P3 │ P4 │
├────┼────┼────┼────┤
│ P5 │ P6 │ P7 │ P8 │
├────┼────┼────┼────┤
│ P9 │P10 │P11 │P12 │
└────┴────┴────┴────┘
          ↓
     Patch sequence

[P1] [P2] [P3] ... [P12]
          ↓
       Transformer
```

The original ViT formulation divides an image into fixed-size patches and processes the resulting patch sequence with a Transformer. ([arXiv][1])

# 8. 📍 Why Positional Information?

Once the image becomes a sequence:

```text
[P1] [P2] [P3] [P4]
```

the model also needs to know **where each patch came from**.

For example:

```text
P1 P2
P3 P4
```

contains different spatial information from another arrangement of the same patches.

Therefore, ViT incorporates **positional information** into the patch representations.

Conceptually:

```text
Patch content
      +
Position
      ↓
Patch representation
```

### Key point

> **Patch embedding tells the model what a patch contains; positional information tells it where the patch belongs.**

The exact implementation of positional embeddings belongs to later architecture sections.

# 9. ⚖️ CNN vs ViT

| Feature                    | CNN                               | ViT                                               |
| -------------------------- | --------------------------------- | ------------------------------------------------- |
| Fundamental operation      | Convolution                       | Self-attention                                    |
| Input representation       | Image grid                        | Patch-token sequence                              |
| Locality                   | Strong built-in bias              | Not imposed in the same way                       |
| Receptive field            | Builds progressively              | Attention can connect tokens globally             |
| Spatial hierarchy          | Naturally supported by CNN design | Learned through Transformer representations       |
| Image-specific assumptions | Strong                            | Weaker                                            |
| Long-range relationships   | Develop through layers            | Direct token interactions through attention       |
| Data requirements          | Strong image priors can help      | Large-scale pretraining is particularly important |

⚠️ **Do not interpret this table as "ViT is better than CNN."**

The important difference is **architectural assumptions and mechanisms**.

# 10. 🧠 CNN Inductive Bias: Advantage and Limitation

CNN inductive bias is **not a flaw**.

It is useful because the model already knows certain properties of images.

### Advantage

```text
Image
 ↓
Built-in locality
 ↓
Efficient learning of local patterns
```

### Potential limitation

The architecture imposes assumptions about how visual information should be processed.

ViT explores a more general approach:

```text
Image
 ↓
Patch tokens
 ↓
Transformer
 ↓
Learn relationships from data
```

So the fundamental research question became:

> **How much visual structure needs to be built into the architecture, and how much can be learned from data?**

# 12. 🎯 Why ViT Was Significant

The significance of ViT was not simply:

> "Transformers can classify images."

The broader significance was:

> **A Transformer can operate directly on visual tokens without requiring a conventional CNN backbone.**

The original ViT study showed strong transfer performance after large-scale pretraining across several image-recognition benchmarks. ([arXiv][1])

This helped establish Transformers as an important architecture family for computer vision.

---

# 13. ⚠️ Important Nuance: ViT Did Not Make CNNs Obsolete

This is an important interview point.

Do **not** say:

> ❌ "ViT replaced CNNs because CNNs cannot model global information."

That's wrong.

Instead:

> ✅ **CNNs and ViTs use different architectural mechanisms and inductive biases. CNNs explicitly encode locality through convolution, while ViTs represent images as tokens and use attention to learn relationships between them.**

There have also been many later architectures that combine convolutional and attention mechanisms.

---

# 14. 🧮 Key Formula: Number of Image Patches

For an image of size:

$$
H \times W
$$

with square patches of size:

$$
P \times P
$$

the number of patches is:

$$
\boxed{N=\frac{HW}{P^2}}
$$

### Example

For a:

$$
224\times224
$$

image with:

$$
16\times16
$$

patches:

$$
N=\frac{224\times224}{16^2}
$$

$$
=\frac{50176}{256}
$$

$$
\boxed{N=196}
$$

So the image becomes **196 patch tokens** before accounting for any additional special token used by a particular ViT configuration.

This formula becomes extremely important when you study ViT architecture and computational complexity.

---

# 15. 💡 Why Patch Size Matters

Suppose the image is:

$$
224\times224
$$

### With 16×16 patches:

$$
N=196
$$

### With 32×32 patches:

$$
N=49
$$

Therefore:

```text
Smaller patch
     ↓
More tokens
     ↓
More detailed spatial representation
     ↓
More attention computation
```

while:

```text
Larger patch
     ↓
Fewer tokens
     ↓
Less computation
     ↓
Less fine-grained spatial representation
```

⚠️ The detailed computational consequences belong to later ViT architecture sections, but this relationship is important to remember.

# 16. 📌 Core Terminology

### **Inductive Bias**

Assumptions built into a model that influence what it learns efficiently.

### **Local Receptive Field**

The local region of an input that a neuron/unit initially receives information from.

### **Global Relationship**

A relationship between potentially distant regions/tokens in an image.

### **Patch**

A fixed-size region extracted from an image.

### **Patch Token**

A vector representation of an image patch used as a Transformer input token.

### **Self-Attention**

A mechanism allowing tokens to compute relationships with other tokens.

### **Positional Information**

Information representing the spatial/sequence position of a token.

# 🎤 Interview Questions

### Q1. Is ViT a CNN?

**No.** ViT is a Transformer-based vision architecture.

### Q2. Why are images divided into patches?

To convert the 2D image into a sequence of manageable visual tokens that can be processed by a Transformer.

### Q3. What is the major CNN inductive bias?

Strong spatial locality and local connectivity through convolution.

### Q4. Why is global attention useful?

It provides a mechanism for tokens to directly interact with potentially distant parts of the image.

### Q5. Does CNN completely lack global understanding?

**No.** Deep CNNs can learn global relationships through successive layers. The distinction is how those relationships are formed.

### Q6. Why does ViT require positional information?

Because after converting an image into a sequence of patches, the model needs information about each patch's spatial location.

### Q7. Why was large-scale pretraining important for the original ViT?

The original ViT results relied heavily on large-scale pretraining before transfer to downstream vision tasks. ([arXiv][1])

# 🚨 Common Mistakes

### ❌ Mistake 1

> "ViT is a type of CNN."

**Correct:** ViT is Transformer-based.

### ❌ Mistake 2

> "CNNs only understand local information."

**Correct:** CNNs progressively build larger receptive fields and can capture global information.

### ❌ Mistake 3

> "Attention automatically makes ViT better."

**Correct:** Attention is a different mechanism; model performance depends on architecture, training, data, compute, and task.

### ❌ Mistake 4

> "Patch = token."

More precisely:

```text
Image patch
    ↓
Embedding/projection
    ↓
Patch token representation
```

### ❌ Mistake 5

> "ViT has no spatial information."

Incorrect. Positional information is incorporated so the Transformer can use patch positions.

> **CNNs encode strong assumptions about visual locality into the architecture, whereas ViT represents an image as a sequence of patches and uses Transformer attention to learn relationships among those visual tokens.**

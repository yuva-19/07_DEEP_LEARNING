## 1. What is Batch Normalization?

Batch Normalization (BatchNorm) normalizes activations within a **mini-batch** and then applies a learnable scaling and shifting operation.

The complete equation is:

$$
y_i =
\gamma
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
+\beta
$$

Think of it as:

```text
Layer activations
       ↓
Calculate mean
       ↓
Calculate variance
       ↓
Normalize
       ↓
Scale using γ
       ↓
Shift using β
       ↓
Output
```

---

# 2. Mean

For a mini-batch containing $m$ values:

$$
\mu_B=\frac{1}{m}\sum_{i=1}^{m}x_i
$$

### Example

$$
x=[2,4,6,8]
$$

Then:

$$
\mu_B=\frac{2+4+6+8}{4}=5
$$

### Intuition

**Mean tells us where the activation distribution is centered.**

---

# 3. Variance

Variance tells us how spread out the activations are.

$$
\sigma_B^2=
\frac{1}{m}
\sum_{i=1}^{m}(x_i-\mu_B)^2
$$

For:

$$
x=[2,4,6,8]
$$

with $\mu_B=5$:

$$
\sigma_B^2=
\frac{(2-5)^2+(4-5)^2+(6-5)^2+(8-5)^2}{4}
$$

$$
\sigma_B^2=5
$$

### Easy memory

> **Mean = center**

> **Variance = spread**

---

# 4. Normalization 🔥

BatchNorm first standardizes the activations:

$$
\hat{x}_i=
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
$$

This does two things:

### Centering

$$
x_i-\mu_B
$$

moves the distribution toward zero.

### Scaling

$$
\frac{1}{\sqrt{\sigma_B^2+\epsilon}}
$$

controls its spread.

The resulting values have approximately:

$$
E[\hat{x}]\approx0
$$

and

$$
Var(\hat{x})\approx1
$$

# 5. Why $\epsilon$?

The equation contains:

$$
\sqrt{\sigma_B^2+\epsilon}
$$

where $\epsilon$ is a very small positive number.

For example:

$$
\epsilon=10^{-5}
$$

Its purpose is **numerical stability**.

If variance becomes extremely small, $\epsilon$ prevents division by zero or an unstable denominator.

---

# 6. Scaling — $\gamma$

After normalization, BatchNorm applies a learnable scale parameter:

$$
y=\gamma\hat{x}
$$

$\gamma$ controls the spread of the output.

```text
γ > 1  → stretch distribution
γ < 1  → compress distribution
γ = 1  → unchanged scale
```

### Memory trick 🧠

> **$\gamma$ = Scale**

---

# 7. Shifting — $\beta$

BatchNorm also applies a learnable shift:

$$
y=\gamma\hat{x}+\beta
$$

$\beta$ controls where the distribution is centered.

```text
β > 0  → shift upward
β < 0  → shift downward
β = 0  → no shift
```

### Memory trick 🧠

> **$\beta$ = Bias/Shift**

---

# 8. Why Do We Need $\gamma$ and $\beta$?

A common question:

> If we already normalized the values, why change them again?

Because forcing every activation to permanently have:

$$
\text{mean}=0,\qquad\text{variance}=1
$$

could unnecessarily restrict the network.

Instead, BatchNorm lets the network **learn the appropriate scale and location**.

The final transformation is:

$$
\boxed{y=\gamma\hat{x}+\beta}
$$

where:

* $\gamma$ → learnable scale
* $\beta$ → learnable shift

So BatchNorm is not simply:

> "Force everything to mean 0 and variance 1."

It is:

> **"Normalize the activations, then let the network learn the useful scale and shift."**

---

# 9. Complete BatchNorm Formula 🔥🔥🔥

All steps combined:

$$
\boxed{
y_i=
\gamma
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
+\beta
}
$$

Break it into:

| Step | Operation                              | Purpose         |
| ---- | -------------------------------------- | --------------- |
| 1    | $\mu_B$                                | Find center     |
| 2    | $\sigma_B^2$                           | Find spread     |
| 3    | $x-\mu_B$                              | Center          |
| 4    | Divide by $\sqrt{\sigma_B^2+\epsilon}$ | Normalize scale |
| 5    | $\gamma$                               | Learn scale     |
| 6    | $\beta$                                | Learn shift     |

---

# 10. Training vs Inference 🚨

This is one of the **most important BatchNorm concepts**.

## During Training

BatchNorm uses the **current mini-batch** to calculate:

$$
\mu_B
$$

and

$$
\sigma_B^2
$$

```text
Training batch
      ↓
Batch mean
      ↓
Batch variance
      ↓
Normalize
      ↓
γ and β
      ↓
Output
```

Every batch can have slightly different statistics.

---

# 11. Running Mean and Running Variance

If BatchNorm only remembered the current batch, inference would be problematic.

So during training it maintains estimates called:

* **Running mean**
* **Running variance**

Conceptually:

```text
Batch 1 ──┐
Batch 2 ──┤
Batch 3 ──┼──→ Running statistics
Batch 4 ──┤
Batch 5 ──┘
```

These statistics represent the activation distribution learned during training.

---

# 12. During Inference

At inference time, we generally don't want to calculate statistics from the incoming batch.

Instead, BatchNorm uses its **stored running mean and running variance**.

```text
                 TRAINING
                    ↓
             Batch statistics
                    ↓
            Running statistics
                    ↓
                 stored
                    ↓
                INFERENCE
                    ↓
          Running statistics
                    ↓
              Normalization
                    ↓
                 γ, β
                    ↓
                 Output
```

# 13. Training vs Inference Table 🔥

|          | **Training**                | **Inference**       |
| -------- | --------------------------- | ------------------- |
| Mean     | Current mini-batch mean     | Running mean        |
| Variance | Current mini-batch variance | Running variance    |
| $\gamma$ | Learned                     | Fixed learned value |
| $\beta$  | Learned                     | Fixed learned value |
| Purpose  | Learn model                 | Make predictions    |

### One-line memory:

> **Training → batch statistics**

> **Inference → running statistics**

---

# 14. PyTorch Connection

In PyTorch:

```python
model.train()
```

puts the model into training mode.

BatchNorm uses **batch statistics**.

During evaluation:

```python
model.eval()
```

BatchNorm uses its **running statistics**.

For inference, you will commonly also use:

```python
with torch.no_grad():
    output = model(x)
```

⚠️ These are different:

| Command           | Purpose                       |
| ----------------- | ----------------------------- |
| `model.train()`   | Training behavior             |
| `model.eval()`    | Evaluation/inference behavior |
| `torch.no_grad()` | Disable gradient tracking     |

# 15. Numerical Example 🧮

Suppose:

$$
x=[2,4,6,8]
$$

Mean:

$$
\mu_B=5
$$

Variance:

$$
\sigma_B^2=5
$$

Ignoring $\epsilon$ for simplicity:

$$
\sqrt{5}\approx2.236
$$

Normalized values:

$$
\hat{x}
=
[-1.342,-0.447,0.447,1.342]
$$

Now suppose BatchNorm has learned:

$$
\gamma=2
$$

and:

$$
\beta=1
$$

Then:

$$
y=2\hat{x}+1
$$

giving approximately:

$$
y=
[-1.684,0.106,1.894,3.684]
$$

So:

```text
Original
[2, 4, 6, 8]
       ↓
Mean + Variance
       ↓
Normalize
[-1.342, -0.447, 0.447, 1.342]
       ↓
γ = 2, β = 1
       ↓
[-1.684, 0.106, 1.894, 3.684]
```

# 16. The Complete Mental Model 🧠

Think of BatchNorm as:

### **Center → Normalize → Learn Scale → Learn Shift**

```text
x
│
├── Calculate μ
│
├── Calculate σ²
│
↓
(x - μ) / √(σ² + ε)
│
↓
Normalized activation
│
├── × γ
│
├── + β
│
↓
BatchNorm output
```

# 17. 🔥 Interview / Implementation Points

You should be able to answer these confidently:

### Q1. What does BatchNorm calculate?

**Mean and variance of the current mini-batch during training.**

### Q2. What is the normalization equation?

$$
\hat{x}=
\frac{x-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
$$

### Q3. Why $\epsilon$?

**Numerical stability.**

### Q4. What are $\gamma$ and $\beta$?

$$
y=\gamma\hat{x}+\beta
$$

* $\gamma$ → learnable scale
* $\beta$ → learnable shift

### Q5. What statistics are used during inference?

**Running mean and running variance.**

### Q6. Why not use the current test batch's statistics?

Because predictions should use the statistics learned from training rather than depend on the particular inference batch.

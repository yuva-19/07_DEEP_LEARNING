## 1. Why do we need normalization?

During neural-network training, the activations produced by earlier layers are continuously changing.

```text
Input
 ↓
Layer 1
 ↓
Activations
 ↓
Layer 2
 ↓
Activations
 ↓
Layer 3
 ↓
...
```

When the scale/distribution of activations becomes difficult to optimize, training can become less stable or inefficient.

Normalization helps **control activation statistics and make optimization easier**.

A basic normalization operation is:

$$
\boxed{
\hat{x}=
\frac{x-\mu}
{\sqrt{\sigma^2+\epsilon}}
}
$$

where:

* $\mu$ = mean
* $\sigma^2$ = variance
* $\epsilon$ = small value for numerical stability

---

## 2. Normalization does NOT mean "make everything Gaussian"

A common misconception is:

> Normalization forces activations to become normally distributed.

Not necessarily.

The basic objective is to control **center and scale**.

After standardization, we generally obtain approximately:

$$
E[\hat{x}]\approx0
$$

and

$$
Var(\hat{x})\approx1
$$

---

## 3. Learnable scale and shift

If we only normalized, we might unnecessarily restrict the network.

Therefore normalization layers commonly use:

$$
\boxed{y=\gamma\hat{x}+\beta}
$$

where:

* $\gamma$ → learnable scale
* $\beta$ → learnable shift

Think:

```text
Normalize
   ↓
"Put activations into a controlled range"
   ↓
γ → stretch/compress
   ↓
β → shift
   ↓
Final activation
```

---

## 4. Important nuance: Internal Covariate Shift

You may see the term **Internal Covariate Shift** associated with BatchNorm.

Historically, BatchNorm was motivated partly as a way of addressing changes in activation distributions during training.

However, modern understanding is more nuanced: the benefits of normalization are not completely explained by internal covariate shift alone.

For your level:

> **Normalization primarily matters because it can improve optimization, training stability, and conditioning.**

# 11.2 Batch Normalization 🔥🔥🔥

## 1. What is BatchNorm?

Batch Normalization normalizes activations using statistics calculated from a **mini-batch**.

For a feature:

$$
\mu_B=
\frac{1}{m}\sum_{i=1}^{m}x_i
$$

Variance:

$$
\sigma_B^2=
\frac{1}{m}
\sum_{i=1}^{m}(x_i-\mu_B)^2
$$

Normalize:

$$
\hat{x}_i=
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
$$

Then scale and shift:

$$
\boxed{
y_i=\gamma\hat{x}_i+\beta
}
$$

---

## 2. Mean

Example:

$$
x=[2,4,6,8]
$$

Mean:

$$
\mu_B=\frac{2+4+6+8}{4}=5
$$

**Mean = center of the batch's activation values.**

---

## 3. Variance

$$
\sigma_B^2=
\frac{1}{m}
\sum_{i=1}^{m}(x_i-\mu_B)^2
$$

For the example:

$$
\sigma_B^2=5
$$

**Variance = spread of the activation values.**

---

## 4. Normalization

$$
\hat{x}_i=
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
$$

This:

1. Centers the values.
2. Controls their scale.

Approximately:

$$
E[\hat{x}]\approx0
$$

$$
Var(\hat{x})\approx1
$$

---

## 5. Why $\epsilon$?

The denominator is:

$$
\sqrt{\sigma_B^2+\epsilon}
$$

$\epsilon$ is a tiny positive value.

Its purpose:

> **Numerical stability and avoiding division by zero.**

---

## 6. Scaling — $\gamma$

After normalization:

$$
y=\gamma\hat{x}
$$

$\gamma$ controls the scale.

```text
γ > 1 → increase spread
γ < 1 → decrease spread
γ = 1 → unchanged
```

---

## 7. Shifting — $\beta$

Then:

$$
y=\gamma\hat{x}+\beta
$$

$\beta$ controls the shift.

```text
β > 0 → shift upward
β < 0 → shift downward
β = 0 → no shift
```

### Easy memory:

> **$\gamma$ = Scale**

> **$\beta$ = Shift**

---

## 8. Complete BatchNorm formula

$$
\boxed{
y_i=
\gamma
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
+\beta
}
$$

# 11.2.1 Training vs Inference 🚨

This is **extremely important**.

## During training

BatchNorm calculates:

$$
\mu_B,\sigma_B^2
$$

from the **current mini-batch**.

```text
Mini-batch
    ↓
Mean + variance
    ↓
Normalize
    ↓
γ + β
    ↓
Output
```

---

## Running statistics

Because inference shouldn't depend on whatever batch happens to arrive, BatchNorm maintains:

* Running mean
* Running variance

Conceptually:

```text
Batch 1 ──┐
Batch 2 ──┤
Batch 3 ──┼──→ Running statistics
Batch 4 ──┤
Batch 5 ──┘
```

---

## During inference

BatchNorm uses the stored running statistics.

```text
New input
   ↓
Running mean
Running variance
   ↓
Normalize
   ↓
γ + β
   ↓
Prediction
```

### Memorize:

$$
\boxed{\text{Training → batch statistics}}
$$

$$
\boxed{\text{Inference → running statistics}}
$$

# 11.3 Layer Normalization 🔥🔥🔥

## 1. What is LayerNorm?

Layer Normalization normalizes **the features of each individual sample**.

Suppose:

$$
X=
\begin{bmatrix}
2&4&6&8\\
1&3&5&7\\
4&6&8&10
\end{bmatrix}
$$

For LayerNorm:

```text
Sample 1 → [2,4,6,8]   ← normalize here
Sample 2 → [1,3,5,7]   ← normalize here
Sample 3 → [4,6,8,10]  ← normalize here
```

Each sample gets its **own statistics**.

---

## 2. LayerNorm mean

For one sample with $d$ features:

$$
\mu=
\frac{1}{d}
\sum_{i=1}^{d}x_i
$$

Variance:

$$
\sigma^2=
\frac{1}{d}
\sum_{i=1}^{d}(x_i-\mu)^2
$$

Normalize:

$$
\hat{x}_i=
\frac{x_i-\mu}
{\sqrt{\sigma^2+\epsilon}}
$$

Then:

$$
\boxed{
y_i=\gamma_i\hat{x}_i+\beta_i
}
$$

---

### BatchNorm

Looks **down the batch**:

```text
Feature 1

Sample 1  ●
Sample 2  ●
Sample 3  ●
          ↑
       statistics
```

### LayerNorm

Looks **across the features**:

```text
Sample 1

● → ● → ● → ●
└────────────┘
  statistics
```

### 🔥 Core memory

$$
\boxed{\text{BatchNorm → across batch}}
$$

$$
\boxed{\text{LayerNorm → across features}}
$$

---

# 11.3.1 Why LayerNorm Doesn't Need Running Statistics

BatchNorm:

```text
Training
   ↓
Current batch statistics
   ↓
Running statistics
   ↓
Inference
```

LayerNorm:

```text
Training:
Sample → its own statistics → normalize

Inference:
Sample → its own statistics → normalize
```

It doesn't need the batch to calculate its statistics.

Therefore:

> **LayerNorm doesn't need BatchNorm-style running mean and variance.**

---

# 11.4 BatchNorm vs LayerNorm 🔥🔥🔥

## Side-by-side

| Property                            | BatchNorm               | LayerNorm                      |
| ----------------------------------- | ----------------------- | ------------------------------ |
| Statistics                          | Across batch            | Across features                |
| Other samples affect normalization? | Yes                     | No                             |
| Batch-dependent?                    | Yes                     | No                             |
| Running statistics                  | Yes                     | No                             |
| Inference behavior                  | Uses running statistics | Uses current sample statistics |
| Learnable $\gamma,\beta$            | Yes                     | Yes                            |
| Common use                          | CNNs                    | Sequence models / Transformers |
| Small-batch sensitivity             | More relevant           | Much less relevant             |

# 11.4.2 Sequence Data

Sequence models commonly deal with:

$$
X\in\mathbb{R}^{B\times L\times D}
$$

where:

* $B$ = batch
* $L$ = sequence length
* $D$ = feature dimension

For example:

$$
32\times512\times768
$$

means:

```text
32 sequences
   ×
512 tokens
   ×
768 features
```

Each token has a feature representation:

$$
x=[x_1,x_2,\ldots,x_{768}]
$$

LayerNorm can normalize that feature vector independently.

```text
Token
  ↓
[768 features]
  ↓
LayerNorm
  ↓
normalized representation
```

# 11.4.4 Why LayerNorm Fits Transformers

### Reason 1 — No batch dependency

Transformer representations don't need normalization to depend on the other examples in the batch.

LayerNorm:

$$
\text{sample → its own feature statistics}
$$

---

### Reason 2 — Works naturally with variable sequence structures

Transformer input has:

$$
B\times L\times D
$$

LayerNorm can operate on the feature dimension:

$$
D
$$

for each token.

---

### Reason 3 — Works naturally at inference

A language model may perform inference with:

```text
Batch = 1
```

or another batch size.

LayerNorm doesn't need a meaningful batch distribution.

---

### Reason 4 — Same basic normalization mechanism during training and evaluation

LayerNorm calculates statistics from the input itself.

So:

```text
Training → input statistics
Evaluation → input statistics
```

Unlike BatchNorm, it doesn't switch to stored running mean/variance.

---

# 11.4.5 Attention Connection

Self-attention uses:

$$
Q=XW_Q
$$

$$
K=XW_K
$$

$$
V=XW_V
$$

and:

$$
Attention(Q,K,V)
=
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

The Transformer repeatedly transforms representations through attention and feed-forward layers.

Normalization helps keep the representations and optimization behavior controlled as they pass through these deep blocks.

⚠️ Don't reduce the explanation to:

> "LayerNorm prevents attention from exploding."

Its role is broader: it helps maintain manageable representation scales and supports optimization.

---

# 11.4.6 Residual Connections + LayerNorm

Transformers also rely heavily on residual connections:

$$
y=x+F(x)
$$

A simplified block:

```text
              ┌───────────────┐
              │               ↓
x ─────────→ LayerNorm → Attention
│                           │
│                           ↓
└──────────── Add ←─────────┘
                ↓
             LayerNorm
                ↓
          Feed Forward
                ↓
              Add
```

This combination of:

* LayerNorm
* Residual connections
* Self-attention
* Feed-forward networks

forms the foundation of Transformer blocks.

# 11.4.7 The Deep Concept: Dependency

This is the **best mental model** for the entire chapter.

### BatchNorm

```text
Sample A ──┐
Sample B ──┤
Sample C ──┼──→ Statistics
Sample D ──┤
Sample E ──┘
```

Your normalization depends on the **batch**.

### LayerNorm

```text
Sample A
   ↓
Its features
   ↓
Statistics

Sample B
   ↓
Its features
   ↓
Statistics
```

Your normalization depends on the **sample's own features**.

---

# 11.4.8 Why CNN → BatchNorm and Transformer → LayerNorm?

Don't memorize it as an absolute rule.

Instead understand the architecture:

```text
CNN
 │
 ├── Image/spatial data
 ├── Mini-batch training
 └── BatchNorm is commonly useful
          ↓
       BatchNorm


Transformer
 │
 ├── Sequence data
 ├── B × L × D representation
 ├── Per-token feature vectors
 └── Avoid batch dependency
          ↓
       LayerNorm
```

So:

$$
\boxed{\text{CNNs → BatchNorm is common}}
$$

$$
\boxed{\text{Transformers → LayerNorm is common}}
$$

But neither statement is an absolute rule.

# 🔥 Interview Questions

### Q1. What is the fundamental difference between BatchNorm and LayerNorm?

**BatchNorm computes statistics across the batch for each feature, while LayerNorm computes statistics across features within each individual sample.**

---

### Q2. Why doesn't LayerNorm need running statistics?

Because its statistics are calculated from the current sample rather than from a mini-batch.

---

### Q3. Why is LayerNorm common in Transformers?

Because Transformer representations are naturally organized as:

$$
B\times L\times D
$$

and LayerNorm can normalize the feature dimension $D$ independently for each token/sample without depending on other samples in the batch.

---

### Q4. What are $\gamma$ and $\beta$?

$$
y=\gamma\hat{x}+\beta
$$

* $\gamma$ → learnable scale
* $\beta$ → learnable shift

---

### Q5. What happens during BatchNorm inference?

BatchNorm generally uses **running mean and running variance** accumulated during training.

---

### Q6. What happens during LayerNorm inference?

LayerNorm calculates statistics from the input itself; it doesn't switch to BatchNorm-style running statistics.

# ⚠️ Common Mistakes

### ❌ "BatchNorm normalizes the whole batch into one distribution."

Too vague.

Better:

> It calculates feature-wise statistics using the batch dimension.

### ❌ "LayerNorm normalizes the whole neural-network layer."

Too vague.

Better:

> It normalizes the specified feature dimensions for each individual sample.

### ❌ "LayerNorm is always better."

No.

The choice depends on architecture and training setup.

### ❌ "BatchNorm only works for CNNs."

No.

CNNs are simply a major/common use case.

### ❌ "Transformers use LayerNorm because BatchNorm cannot work."

Too strong.

LayerNorm is **naturally suited** to Transformer sequence representations and avoids batch-statistics dependence.

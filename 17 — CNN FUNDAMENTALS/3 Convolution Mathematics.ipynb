## 16.3 Convolution Mathematics

### 🎯 What you must master

This block has **3 🔥🔥🔥 MUST-KNOW concepts**:

1. **Output dimensions**
2. **Number of parameters**
3. **Receptive field**

The big picture:

```text
                 CONVOLUTION
                      │
        ┌─────────────┼─────────────┐
        ↓             ↓             ↓
  Output Size     Parameters    Receptive Field
        │             │             │
        ↓             ↓             ↓
  H, W, K, S, P   K, Cin, Cout   Kernel + Layers
```

---

# 1. 🔥🔥🔥 Output Dimensions

A convolution changes the spatial dimensions of an image depending on:

* Input size
* Kernel size
* Stride
* Padding

The fundamental formula is:

$$
\boxed{
O=
\left\lfloor
\frac{N-F+2P}{S}
\right\rfloor+1
}
$$

Where:

| Symbol | Meaning             |
| ------ | ------------------- |
| \(N\)  | Input spatial size  |
| \(F\)  | Kernel/filter size  |
| \(P\)  | Padding             |
| \(S\)  | Stride              |
| \(O\)  | Output spatial size |

For height and width separately:

$$
\boxed{
H_{out}
=
\left\lfloor
\frac{H_{in}-K_H+2P_H}{S_H}
\right\rfloor+1
}
$$

$$
\boxed{
W_{out}
=
\left\lfloor
\frac{W_{in}-K_W+2P_W}{S_W}
\right\rfloor+1
}
$$

## 🧮 Example 1 — No Padding

Given:

```text
Input      = 32 × 32
Kernel     = 3 × 3
Stride     = 1
Padding    = 0
```

$$
O=
\left\lfloor
\frac{32-3+0}{1}
\right\rfloor+1
$$

$$
O=30
$$

Therefore:

$$
\boxed{30\times30}
$$

```text
32 × 32
   ↓
3 × 3 kernel
   ↓
30 × 30
```

---

# 🧮 Example 2 — Padding = 1

Given:

```text
Input   = 32 × 32
Kernel  = 3 × 3
Stride  = 1
Padding = 1
```

$$
O=
\frac{32-3+2(1)}{1}+1
$$

$$
\boxed{O=32}
$$

Therefore:

```text
32 × 32
   ↓
3×3 kernel
padding = 1
stride = 1
   ↓
32 × 32
```

### 🧠 Important pattern

For an odd-sized `3×3` kernel:

```text
Kernel  = 3
Stride  = 1
Padding = 1
```

the spatial dimensions remain unchanged.

---

# 🧮 Example 3 — Stride = 2

Given:

```text
Input   = 32 × 32
Kernel  = 3 × 3
Stride  = 2
Padding = 1
```

$$
O=
\left\lfloor
\frac{32-3+2}{2}
\right\rfloor+1
$$

$$
O=
\left\lfloor
\frac{31}{2}
\right\rfloor+1
$$

$$
O=16
$$

Therefore:

$$
\boxed{16\times16}
$$

So:

```text
32 × 32
   ↓
Conv
stride = 2
   ↓
16 × 16
```

### 🧠 Intuition

```text
Stride ↑
   ↓
Kernel moves farther
   ↓
Fewer positions
   ↓
Smaller output
```

# 2. 🔥🔥🔥 Input Channels vs Output Channels

This is extremely important before calculating parameters.

Suppose:

```text
Input:
3 × 64 × 64
```

The `3` means:

```text
3 input channels
```

because it's an RGB image.

Now suppose:

```text
Conv2D
out_channels = 32
```

That means the layer has:

$$
\boxed{32\text{ filters}}
$$

and therefore produces:

$$
\boxed{32\text{ output channels}}
$$

So:

```text
Input
3 × 64 × 64
      ↓
   Conv2D
   32 filters
      ↓
Output
32 × Hout × Wout
```

### 🔥 Key relationship

$$
\boxed{
\text{Number of filters}
=
\text{Number of output channels}
}
$$

# 3. 🔥🔥🔥 Number of Parameters

Now we ask:

> **How many learnable weights does a convolutional layer have?**

For a standard convolution:

$$
\boxed{
\text{Parameters}
=
K_H\times K_W\times C_{in}\times C_{out}
+
C_{out}
}
$$

The first part is the **weights**.

The second part is the **biases**.

---

## 🧠 Why?

Suppose:

```text
Input channels = 3
Kernel = 3 × 3
```

One filter needs:

$$
3\times3\times3
$$

weights.

Why?

Because the filter spans all 3 input channels:

```text
Red   → 3 × 3
Green → 3 × 3
Blue  → 3 × 3
```

Therefore:

$$
3\times3\times3=27
$$

weights **per filter**.

If we have 32 filters:

$$
27\times32=864
$$

weights.

Each filter also has one bias:

$$
32
$$

biases.

Therefore:

$$
864+32=\boxed{896}
$$

parameters.

---

# 🧮 Parameter Example

Given:

```text
Cin  = 3
Cout = 64
Kernel = 3 × 3
Bias = enabled
```

Weights:

$$
3\times3\times3\times64
$$

$$
=1728
$$

Bias:

$$
64
$$

Total:

$$
1728+64
$$

$$
\boxed{1792}
$$

---

# 🔥🔥🔥 The General Formula

### Bias enabled

$$
\boxed{
P=K_HK_WC_{in}C_{out}+C_{out}
}
$$

### Bias disabled

$$
\boxed{
P=K_HK_WC_{in}C_{out}
}
$$

---

# 4. 🧠 Why Isn't Image Height/Width in the Formula?

This is one of the **most important CNN ideas**.

Look at:

$$
P=K_HK_WC_{in}C_{out}+C_{out}
$$

Notice:

```text
H → not present
W → not present
```

Why?

Because CNN filters use **parameter sharing**.

The same filter is reused at every spatial position.

# 5. 🔥🔥🔥 Receptive Field

Now the third major concept.

### Definition

The **receptive field** of a neuron/output location is the region of the original input that can influence that output.

Simple definition:

> 🧠 **Receptive field = how much of the original image a particular feature can see.**

# 6. 🔥 Why Receptive Field Matters

CNNs learn features hierarchically.

### Early layers

Small receptive fields:

```text
Pixels
  ↓
Edges
  ↓
Simple textures
```

### Middle layers

Larger receptive fields:

```text
Edges + textures
       ↓
Object parts
```

### Deep layers

Much larger receptive fields:

```text
Object parts
      ↓
Larger structures
      ↓
Object-level information
```

# 7. 🔥🔥 Receptive Field Grows With Depth

Suppose we stack two:

```text
3 × 3
```

convolution layers.

Assume:

```text
Stride = 1
```

### Layer 1

A neuron sees:

$$
3\times3
$$

of the original image.

### Layer 2

A Layer-2 neuron sees a `3×3` region of Layer 1.

But each Layer-1 neuron already sees a `3×3` region of the original image.

Therefore the Layer-2 neuron sees:

$$
5\times5
$$

of the original image.

```text
Input
 ↓
3×3 Conv
 ↓
RF = 3×3
 ↓
3×3 Conv
 ↓
RF = 5×5
```

🔥 This is why stacking convolutional layers increases the effective receptive field.

---

# 8. 📐 Receptive Field Mathematics

For a sequence of convolution/pooling layers:

$$
\boxed{
r_l=r_{l-1}+(k_l-1)j_{l-1}
}
$$

where:

* \(r_l\) = receptive field at layer \(l\)
* \(k_l\) = kernel size
* \(j_{l-1}\) = jump between neighboring features measured in original-image pixels

The jump is:

$$
\boxed{
j_l=j_{l-1}s_l
}
$$

where \(s_l\) is the stride.

Initially:

$$
r_0=1
$$

$$
j_0=1
$$

---

# 🧮 Example

Two layers:

```text
Layer 1:
kernel = 3
stride = 1

Layer 2:
kernel = 3
stride = 1
```

Start:

$$
r_0=1,\qquad j_0=1
$$

### Layer 1

$$
r_1=1+(3-1)(1)
$$

$$
r_1=3
$$

Jump:

$$
j_1=1\times1=1
$$

### Layer 2

$$
r_2=3+(3-1)(1)
$$

$$
r_2=5
$$

Therefore:

$$
\boxed{RF=5\times5}
$$

---

# 9. 🔥 Stride and Receptive Field

Stride also affects receptive-field growth.

Example:

```text
Layer 1:
kernel = 3
stride = 2
```

Start:

$$
r_0=1,\qquad j_0=1
$$

Then:

$$
r_1=1+(3-1)(1)=3
$$

and:

$$
j_1=1\times2=2
$$

Now Layer 2:

```text
kernel = 3
stride = 1
```

$$
r_2=3+(3-1)(2)
$$

$$
r_2=7
$$

Therefore:

$$
\boxed{RF=7\times7}
$$

# ⚠️ Common Mistakes

### ❌ Mistake 1 — Image size determines parameter count

No.

Parameters depend on:

$$
K_H,K_W,C_{in},C_{out}
$$

not `H × W`.

---

### ❌ Mistake 2 — RGB 3×3 filter has only 9 parameters

For RGB:

$$
3\times3\times3=27
$$

weights **per filter**.

---

### ❌ Mistake 3 — `in_channels` = number of filters

No.

```text
in_channels
→ channels coming INTO the layer

out_channels
→ number of filters / channels produced
```

---

### ❌ Mistake 4 — One 3×3 layer always means 3×3 receptive field

Only for that layer's direct input.

After stacking layers, the receptive field grows.

```text
3×3
 ↓
3×3
 ↓
5×5 effective RF
```

# 🎯 Interview Questions

### Q1. What is the output-size formula?

$$
\boxed{
\left\lfloor\frac{N-F+2P}{S}\right\rfloor+1
}
$$

### Q2. What determines the number of convolution parameters?

$$
\boxed{
K_HK_WC_{in}C_{out}+C_{out}
}
$$

when bias is enabled.

### Q3. Why doesn't image height/width affect parameter count?

Because convolution uses **parameter sharing**.

### Q4. What is the relationship between filters and output channels?

$$
\boxed{
\text{Number of filters}=\text{Number of output channels}
}
$$

### Q5. What is receptive field?

The region of the original input that can influence a particular output feature.

### Q6. Why does receptive field grow deeper into the network?

Because each deeper feature depends on multiple features from the previous layer, which themselves correspond to regions of the original image.

\frac{N-F+2P}{S}
\right\rfloor+1
}
$$

## 👁️ Receptive Field

```text
One 3×3 convolution
        ↓
     RF = 3×3

Two 3×3 convolutions
        ↓
     RF = 5×5
```

General:

$$
\boxed{
r_l=r_{l-1}+(k_l-1)j_{l-1}
}
$$

$$
\boxed{
j_l=j_{l-1}s_l
}
$$

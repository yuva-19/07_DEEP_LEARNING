## 16.2 Convolution — Complete Revision Notes

> **Core idea:** A CNN uses small learnable filters to scan an image, detect local patterns, and produce **feature maps**.

The complete flow is:

```text
Input Image
     ↓
Kernel / Filter
     ↓
Sliding Window
     ↓
Element-wise Multiply
     ↓
Sum (+ bias)
     ↓
One Output Value
     ↓
Repeat Across Image
     ↓
Feature Map
```

# 1. 🔥🔥🔥 Kernel / Filter

A **kernel** (also called a **filter**) is a small matrix of learnable weights that moves across the input image.

Example:

```text
┌────┬────┬────┐
│  1 │  0 │ -1 │
├────┼────┼────┤
│  1 │  0 │ -1 │
├────┼────┼────┤
│  1 │  0 │ -1 │
└────┴────┴────┘
```

A filter examines a **local region** of the image and produces a response indicating how strongly that region matches the pattern represented by the filter.

CNN filters are **learned during training**, rather than manually programmed for a particular feature.

### Mental model

```text
Kernel
  ↓
Pattern detector 🔍
  ↓
Scans image
  ↓
Detects learned feature
```

---

# 2. 🔥🔥🔥 Sliding Window

The kernel doesn't examine the whole image simultaneously.

It moves across the image one local region at a time.

Suppose:

```text
Input  = 5 × 5
Kernel = 3 × 3
```

The first region is:

```text
┌───┬───┬───┐
│ 1 │ 2 │ 3 │
├───┼───┼───┤
│ 6 │ 7 │ 8 │
├───┼───┼───┤
│11 │12 │13 │
└───┴───┴───┘
```

Then the kernel moves:

```text
        →
```

and examines:

```text
┌───┬───┬───┐
│ 2 │ 3 │ 4 │
├───┼───┼───┤
│ 7 │ 8 │ 9 │
├───┼───┼───┤
│12 │13 │14 │
└───┴───┴───┘
```

It continues until the entire valid region has been processed.

# 3. 🔥🔥🔥 Convolution Operation

At every position, the kernel and the corresponding image region are multiplied **element by element**, and the results are summed.

Suppose:

### Input region

```text
1  2  3
4  5  6
7  8  9
```

### Kernel

```text
1  0  1
0  1  0
1  0  1
```

Multiply corresponding elements:

```text
1×1   2×0   3×1
4×0   5×1   6×0
7×1   8×0   9×1
```

Then add:

$$
1+0+3+0+5+0+7+0+9
$$

$$
\boxed{25}
$$

So that particular image region produces **one output value: 25**.

### 🧠 Memorize

$$
\boxed{\text{Multiply corresponding elements → Sum}}
$$

In an actual convolution layer, a bias is generally added as well:

$$
z=\sum X_{ij}K_{ij}+b
$$

# 4. 🔥🔥🔥 Feature Map

The kernel repeats this calculation across the entire input.

Each position produces one number.

Those numbers are arranged into a new matrix called a:

$$
\boxed{\text{Feature Map}}
$$

```text
Input
  ↓
Filter
  ↓
┌────┬────┬────┐
│ 25 │ 30 │ 18 │
├────┼────┼────┤
│ 12 │ 41 │ 20 │
├────┼────┼────┤
│ 17 │ 22 │ 35 │
└────┴────┴────┘
     Feature Map
```

### What does the feature map mean?

It tells us **where the pattern detected by that particular filter appears strongly or weakly**.

For example, if a filter learns a vertical-edge pattern:

```text
High activation → strong evidence of vertical edge
Low activation  → weak evidence
```

So:

> **Feature map = spatial map of the filter's responses.**

# 5. 🔥🔥🔥 One Filter → One Feature Map

A convolutional layer normally contains multiple filters.

For example:

```text
                 INPUT
                   │
        ┌──────────┼──────────┐
        ↓          ↓          ↓
     Filter 1   Filter 2   Filter 3
        ↓          ↓          ↓
     Map 1      Map 2      Map 3
```

Each filter can learn to respond to different patterns.

Therefore:

$$
\boxed{\text{1 filter} \rightarrow \text{1 feature map}}
$$

and:

$$
\boxed{\text{multiple filters} \rightarrow \text{multiple feature maps}}
$$

# 6. 🌈 RGB Images and Convolution

An RGB image has:

```text
3 channels
```

For example:

```text
Input:

3 × 64 × 64
```

A convolution filter must operate across **all three input channels**.

For a `3 × 3` spatial kernel:

$$
\boxed{3\times3\times3}
$$

It can be visualized as:

```text
       FILTER
 ┌───────────────┐
 │ Red   → 3×3   │
 │ Green → 3×3   │
 │ Blue  → 3×3   │
 └───────────────┘
        ↓
   One output value
```

The responses from the three channels are combined to produce one output value at each spatial location.

# 7. 🔥🔥 Stride

**Stride** tells us how far the filter moves after each operation.

### Stride = 1

```text
Kernel
  ↓
move 1 pixel
  ↓
move 1 pixel
  ↓
move 1 pixel
```

### Stride = 2

```text
Kernel
  ↓
move 2 pixels
  ↓
move 2 pixels
  ↓
```

So:

$$
\boxed{\text{Stride = movement size of the filter}}
$$

# 9. 🔥 Why Padding Is Used

### Reason 1 — Preserve spatial dimensions

Without padding:

```text
5 × 5 input
3 × 3 kernel
stride = 1

→ 3 × 3 output
```

With padding = 1:

```text
5 × 5 input
3 × 3 kernel
stride = 1
padding = 1

→ 5 × 5 output
```

### Reason 2 — Process border pixels

Without padding, pixels near the edges participate in fewer convolution operations.

Padding allows the filter to operate around the boundaries.

---

# 10. 🔥 Valid vs Same

### Valid

```text
Padding = 0
```

For example:

```text
5 × 5
 ↓
3 × 3 kernel
 ↓
3 × 3 output
```

### Same

Padding is selected so that, for stride 1, the output keeps the same spatial dimensions.

For a `3 × 3` kernel:

```text
Kernel = 3
Padding = 1
Stride = 1
```

gives:

```text
Input H × W
      ↓
Output H × W
```

---

# 11. 📐🔥🔥🔥 Output Size Formula

For one spatial dimension:

$$
\boxed{
O=
\left\lfloor
\frac{N-F+2P}{S}
\right\rfloor+1
}
$$

Where:

| Symbol | Meaning            |
| ------ | ------------------ |
| `N`    | Input size         |
| `F`    | Filter/kernel size |
| `P`    | Padding            |
| `S`    | Stride             |
| `O`    | Output size        |

For height:

$$
H_{out}
=
\left\lfloor
\frac{H_{in}-K_H+2P}{S}
\right\rfloor+1
$$

For width:

$$
W_{out}
=
\left\lfloor
\frac{W_{in}-K_W+2P}{S}
\right\rfloor+1
$$

# 14. 🧠 The Most Important Mental Model

Don't think:

> "CNN understands the entire image at once."

Think:

```text
Small region
     ↓
Filter checks it
     ↓
Produces one number
     ↓
Move filter
     ↓
Check next region
     ↓
Repeat
     ↓
Feature map
```

Then use several filters:

```text
                     INPUT
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       Filter 1     Filter 2     Filter 3
          ↓            ↓            ↓
       Map 1         Map 2        Map 3
          │            │            │
          └────────────┼────────────┘
                       ↓
               Multiple Features
```

# ⚠️ Common Mistakes

### ❌ Kernel = feature map

No.

```text
Kernel
 ↓
operates on input
 ↓
Feature map
```

### ❌ Stride = kernel size

No.

```text
Kernel size → size of the filter

Stride → how far the filter moves
```

### ❌ Padding = extra useful image information

Usually padding consists of artificial values, commonly zeros. It primarily controls spatial dimensions and boundary processing.

### ❌ One filter = entire CNN

No.

A convolutional layer can have many filters:

```text
32 filters
↓
32 output feature maps
```

### ❌ Larger stride means more output

Usually the opposite:

```text
Stride ↑
↓
Number of positions ↓
↓
Output spatial dimensions ↓
```

# 🎯 Interview Essentials

### What is a convolution kernel?

> A small set of learnable weights that slides over an input and computes local weighted sums to detect patterns.

### What is a feature map?

> The spatial output produced by applying one filter across the input.

### What is stride?

> The number of pixels by which the kernel moves between successive positions.

### Why is padding used?

> To control output dimensions and allow convolution to process regions near image boundaries.

### What happens when stride increases?

> The filter evaluates fewer spatial positions, generally reducing output spatial dimensions.

### What happens when padding increases?

> The effective input size increases, allowing more boundary positions to be processed and potentially preserving larger output dimensions.

# 💻 Implementation Knowledge

Eventually you'll encounter:

```python
nn.Conv2d(
    in_channels=3,
    out_channels=32,
    kernel_size=3,
    stride=1,
    padding=1
)
```

Understand it as:

```text
in_channels = 3
      ↓
RGB input

out_channels = 32
      ↓
32 filters
      ↓
32 output feature maps

kernel_size = 3
      ↓
3 × 3 spatial filter

stride = 1
      ↓
move 1 pixel

padding = 1
      ↓
one-pixel border
```

You **do not need to memorize the API syntax yet**. You need to understand what each parameter physically means.

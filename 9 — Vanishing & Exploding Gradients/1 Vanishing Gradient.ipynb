# 9.1 Vanishing Gradient

> **Core idea:** During backpropagation, gradients can become extremely small as they travel through many layers. Early layers then receive tiny parameter updates and learn very slowly.

---

## 1. What Is the Vanishing Gradient Problem? 🔥🔥🔥

During training:

$$
\text{Input}
\rightarrow
\text{Forward Pass}
\rightarrow
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradients}
\rightarrow
\text{Parameter Update}
$$

The gradient travels **backward**:

```text
Loss
  ↓
Layer L
  ↓
Layer L-1
  ↓
Layer L-2
  ↓
   .
   .
   .
  ↓
Layer 2
  ↓
Layer 1
```

If the gradient becomes smaller at every layer:

```text
1.0
 ↓
0.5
 ↓
0.25
 ↓
0.125
 ↓
0.0625
 ↓
...
 ↓
≈ 0
```

the gradient has effectively **vanished**.

### Definition

> **Vanishing gradient = a situation where gradients become extremely small during backpropagation, causing earlier layers to receive very small updates.**

# 3. Why Does It Happen? 🔥🔥🔥

The fundamental reason is the **chain rule**.

For an early-layer parameter:

$$
\frac{\partial L}{\partial W^{[1]}}
=
\frac{\partial L}{\partial z^{[L]}}
\frac{\partial z^{[L]}}{\partial z^{[L-1]}}
\cdots
\frac{\partial z^{[2]}}{\partial z^{[1]}}
\frac{\partial z^{[1]}}{\partial W^{[1]}}
$$

Notice the important part:

> **Many derivatives are multiplied together.**

If many of these derivatives are smaller than $1$:

$$
0.5\times0.5\times0.5\times0.5
$$

the result becomes smaller and smaller.

### Example

After 10 layers:

$$
0.5^{10}\approx0.00098
$$

After 20 layers:

$$
0.5^{20}\approx0.000001
$$

So the gradient can become almost zero.

---

# 4. Why Does a Tiny Gradient Cause a Problem?

Gradient descent updates parameters using:

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\frac{\partial L}{\partial W}
$$

Suppose:

$$
\eta=0.01
$$

and:

$$
\frac{\partial L}{\partial W}=0.000001
$$

Then:

$$
\Delta W
=
-0.01(0.000001)
$$

$$
\Delta W=-0.00000001
$$

That's an extremely tiny update.

Therefore:

```text
Vanishing gradient
       ↓
Tiny gradient
       ↓
Tiny parameter update
       ↓
Very slow learning
       ↓
Early layers learn poorly
```

---

# 5. Why Are Deep Networks More Vulnerable? 🔥🔥🔥

A shallow network has relatively few derivative multiplications.

A deep network has many:

```text
Loss
 ↓
× derivative
 ↓
Layer 5
 ↓
× derivative
 ↓
Layer 4
 ↓
× derivative
 ↓
Layer 3
 ↓
× derivative
 ↓
Layer 2
 ↓
× derivative
 ↓
Layer 1
```

Every additional layer can introduce another multiplication factor.

So if the factors are consistently smaller than $1$:

$$
a\times b\times c\times d\times e\rightarrow\text{very small}
$$

### Important nuance ⚠️

**Deep networks do not automatically suffer from vanishing gradients.**

The problem depends on:

* network depth
* activation functions
* weight initialization
* architecture
* normalization
* gradient scale

Modern architectures use several techniques to make gradient flow easier.

# 6. Sigmoid Connection 🔥🔥🔥

Sigmoid is:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Its derivative is:

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

The maximum derivative is:

$$
\sigma'(z)\leq0.25
$$

So every sigmoid activation can introduce a derivative whose maximum is only $0.25$.

Imagine repeatedly multiplying by $0.25$:

$$
0.25\times0.25\times0.25\times\cdots
$$

After 10 such factors:

$$
0.25^{10}\approx9.54\times10^{-7}
$$

That's extremely small.

---

# 7. Sigmoid Saturation 🔥🔥

The sigmoid curve looks approximately like this:

```text
1 ─────────────────────
                  ────
              ───
           ──
        ──
     ──
0 ─────────────────────
```

For very large positive values:

$$
z\rightarrow+\infty
$$

we get:

$$
\sigma(z)\approx1
$$

Therefore:

$$
\sigma'(z)
=
\sigma(z)(1-\sigma(z))
\approx
1(1-1)
=
0
$$

For very large negative values:

$$
z\rightarrow-\infty
$$

we get:

$$
\sigma(z)\approx0
$$

Therefore:

$$
\sigma'(z)\approx0(1-0)=0
$$

So sigmoid has **saturation regions** where the derivative becomes almost zero.

# 9. Tanh Connection 🔥🔥

Tanh:

$$
\tanh(z)
=
\frac{e^z-e^{-z}}
{e^z+e^{-z}}
$$

Derivative:

$$
\frac{d}{dz}\tanh(z)
=
1-\tanh^2(z)
$$

Its maximum derivative is:

$$
1
$$

So tanh is better than sigmoid regarding the **maximum derivative**.

However, tanh also saturates.

For large positive $z$:

$$
\tanh(z)\approx1
$$

Therefore:

$$
1-\tanh^2(z)\approx0
$$

For large negative $z$:

$$
\tanh(z)\approx-1
$$

Therefore:

$$
1-\tanh^2(z)\approx0
$$

So:

> **Tanh can also produce vanishing gradients when it saturates.**

# 10. Sigmoid vs Tanh

| Property                      | Sigmoid     | Tanh        |
| ----------------------------- | ----------- | ----------- |
| Range                         | $(0,1)$     | $(-1,1)$    |
| Maximum derivative            | $0.25$      | $1$         |
| Saturates                     | ✅           | ✅           |
| Can cause vanishing gradients | ✅           | ✅           |
| Zero-centered                 | ❌           | ✅           |
| Modern hidden-layer usage     | Less common | Less common |

### Important

Tanh is **not immune** to vanishing gradients.

It simply has some properties that make it more favorable than sigmoid in certain situations.

# 12. Why Does ReLU Help? 🔥🔥🔥

ReLU:

$$
f(z)=\max(0,z)
$$

Derivative:

$$
f'(z)=
\begin{cases}
0 & z<0\\
1 & z>0
\end{cases}
$$

For positive inputs:

$$
f'(z)=1
$$

Therefore, on the positive side, ReLU does **not continuously shrink the gradient** like saturated sigmoid/tanh.

This was one important reason ReLU became popular in deep networks. Stanford's CS231n notes also describe ReLU's non-saturating positive side as a major practical advantage over sigmoid/tanh. ([CS231n][1])

---

# 13. But ReLU Is Not Perfect ⚠️

For negative inputs:

$$
f'(z)=0
$$

So a neuron can potentially stop receiving useful gradient.

This is called:

> **Dying ReLU**

```text
ReLU

Positive side:
gradient = 1
     ↓
gradient can flow

Negative side:
gradient = 0
     ↓
gradient blocked
```

So:

$$
\text{ReLU reduces one gradient problem}
$$

but:

$$
\text{does not eliminate every gradient problem}
$$

# 14. Complete Gradient-Flow Picture 🧠

```text
                FORWARD
Input ──→ Layer 1 ──→ Layer 2 ──→ Layer 3 ──→ Output
                                                  ↓
                                                 Loss
                                                  │
                BACKWARD                          │
Input ←── Layer 1 ←── Layer 2 ←── Layer 3 ←──────┘
          ↑
          │
      Tiny gradient
```

The farther back we go, the gradient may become smaller.

# 15. The Core Mathematical Pattern

### Normal gradient flow

$$
1.0\rightarrow0.9\rightarrow0.8\rightarrow0.7
$$

Still useful.

### Vanishing gradient

$$
1.0\rightarrow0.5\rightarrow0.25\rightarrow0.125
\rightarrow\cdots\rightarrow0
$$

### Exploding gradient

$$
1.0\rightarrow1.5\rightarrow2.25\rightarrow3.375
\rightarrow\cdots
$$

This gives the fundamental distinction:

$$
\boxed{\text{Repeated factors}<1\Rightarrow\text{Vanishing}}
$$

$$
\boxed{\text{Repeated factors}>1\Rightarrow\text{Exploding}}
$$

# 16. Relationship With Initialization

Initialization controls the initial scale of weights, which affects the scale of activations and gradients.

### Xavier / Glorot

$$
Var(W)=\frac{2}{n_{in}+n_{out}}
$$

Commonly associated with:

$$
\text{Sigmoid / Tanh}
$$

### He / Kaiming

$$
Var(W)=\frac{2}{n_{in}}
$$

Commonly associated with:

$$
\text{ReLU}
$$

The objective is to maintain a reasonable scale for activations and gradients as they move through the network.

This connects directly to **Module 8 — Initialization**.

# 18. Why Does This Matter in Real Networks?

Consider an image-classification network:

```text
Image
  ↓
Early layers
  ↓
Edges
  ↓
Shapes
  ↓
Parts
  ↓
Objects
  ↓
Classification
```

Early layers learn useful lower-level representations.

If the gradient reaching those layers becomes tiny:

```text
Output layers
     ↓
learn normally

     ↓

Early layers
     ↓
tiny gradient
     ↓
barely update
```

The network may therefore struggle to learn good representations.

# 19. How Can Vanishing Gradients Be Reduced?

Common approaches include:

### 1. Better initialization

Examples:

* Xavier
* He

### 2. Better activation functions

Especially:

* ReLU
* Leaky ReLU
* GELU

depending on architecture.

### 3. Normalization

Examples:

* Batch Normalization
* Layer Normalization

### 4. Residual connections

Used heavily in modern deep architectures.

```text
x ────────────────┐
│                 │
↓                 │
Layer → Layer ────┤
                  ↓
                 +
                  ↓
                 output
```

The shortcut provides an easier path for information and gradients.

# 20. Important Distinction: Vanishing ≠ Zero Loss ❌

Vanishing gradient does **not** mean:

$$
L=0
$$

It means:

$$
\frac{\partial L}{\partial W}\approx0
$$

The loss can still be large.

The problem is that the optimizer has very little gradient information with which to update the parameter.

---

# 21. Common Mistakes ❌

### ❌ "Only sigmoid causes vanishing gradients."

No.

Tanh can also suffer from it, and gradient scale depends on the overall network.

### ❌ "Deep networks always have vanishing gradients."

No.

Modern architectures are specifically designed to improve gradient flow.

### ❌ "ReLU completely solves vanishing gradients."

No.

ReLU has the dying-ReLU problem and other architectural considerations.

### ❌ "Increasing learning rate always fixes it."

No.

A tiny gradient is fundamentally a gradient-flow problem. Increasing the learning rate blindly can cause instability elsewhere.

### ❌ "Vanishing gradient means weights become zero."

No.

The **gradient** becomes tiny; the weights don't necessarily become zero.

# 22. Interview / GATE Points 🎯

### Q1. What is vanishing gradient?

> A problem in which gradients become extremely small during backpropagation, causing early layers to receive tiny parameter updates.

### Q2. Why does it happen?

> Backpropagation uses the chain rule, which involves multiplying many derivatives. Repeated factors smaller than one can cause the gradient to shrink exponentially.

### Q3. Why are sigmoid and tanh associated with vanishing gradients?

> They saturate. In saturation regions, their derivatives approach zero.

### Q4. What is sigmoid's maximum derivative?

$$
0.25
$$

### Q5. What is tanh's maximum derivative?

$$
1
$$

### Q6. Why does ReLU help?

For positive inputs:

$$
ReLU'(x)=1
$$

so it avoids saturation on the positive side.

### Q7. Does ReLU completely eliminate gradient problems?

> No. It can suffer from dying ReLU for negative inputs.

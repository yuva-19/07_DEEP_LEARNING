## 🔥 1. Core Idea

The three types of Gradient Descent differ mainly in **how many training samples are used to calculate one parameter update**.

Suppose the dataset contains **10,000 samples**:

| Method                                | Samples used for one update |
| ------------------------------------- | --------------------------: |
| **Batch Gradient Descent**            |                  All 10,000 |
| **Stochastic Gradient Descent (SGD)** |                           1 |
| **Mini-Batch Gradient Descent**       |      A small group, e.g. 32 |

### 🧠 Easy memory trick

> **Batch = ALL**
> **SGD = ONE**
> **Mini-Batch = SOME**

---

# 2. 🔥 Batch Gradient Descent

Batch Gradient Descent uses the **entire training dataset** to calculate the gradient before updating the parameters.

If there are $N$ training examples:

$$
J(\theta)=\frac{1}{N}\sum_{i=1}^{N}L_i(\theta)
$$

The gradient is:

$$
\nabla_\theta J(\theta)
=
\frac{1}{N}
\sum_{i=1}^{N}
\nabla_\theta L_i(\theta)
$$

Then:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta J(\theta)
$$

### Example

Dataset = 1,000 samples.

```text
1000 samples
     ↓
Forward Pass
     ↓
Calculate Loss
     ↓
Backpropagation
     ↓
Calculate Gradient
     ↓
Update Parameters
```

So:

$$
1000\text{ samples}\rightarrow1\text{ update}
$$

### Advantages

* Stable gradient estimate
* Smooth optimization behavior
* Uses information from the complete dataset

### Disadvantages

* Can be computationally expensive
* Requires processing the entire dataset before an update
* Not practical for very large datasets

---

# 3. 🔥 Stochastic Gradient Descent

**Stochastic Gradient Descent (SGD)** uses **one training example** to calculate the gradient and update the parameters.

For example:

```text
Sample 1 → Gradient → Update
Sample 2 → Gradient → Update
Sample 3 → Gradient → Update
Sample 4 → Gradient → Update
      ...
```

Mathematically:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta L_i(\theta)
$$

Here, the gradient is calculated using only **one example**.

### Example

Dataset = 1,000 samples.

```text
1 sample → update
1 sample → update
1 sample → update
...
1 sample → update
```

Therefore, approximately:

$$
1000\text{ samples}\rightarrow1000\text{ updates}
$$

per epoch.

### Advantages

* Frequent updates
* Low memory requirement per update
* Can start learning immediately from individual samples

### Disadvantages

* Updates are noisy
* Loss may fluctuate
* Optimization can be less stable

# 4. 🔥🔥 Mini-Batch Gradient Descent

Mini-Batch Gradient Descent uses a **small group of training examples** for each update.

Example:

$$
N=10,000
$$

and:

$$
B=32
$$

Then:

```text
32 samples → update
32 samples → update
32 samples → update
32 samples → update
...
```

So instead of using:

* all 10,000 → Batch GD
* one → SGD

we use:

* **32 → Mini-Batch GD**

---

# 5. 🔥 Batch Size

**Batch size** = number of training examples used before one parameter update.

For example:

```python
batch_size = 32
```

means:

> Process 32 samples → calculate gradient → update parameters.

Then process the next batch.

### Example

Dataset:

$$
N=1000
$$

Batch size:

$$
B=100
$$

Then:

```text
Batch 1 → samples 1–100     → update
Batch 2 → samples 101–200   → update
Batch 3 → samples 201–300   → update
...
Batch 10 → samples 901–1000 → update
```

Therefore:

$$
\text{Updates per epoch}
=
\frac{N}{B}
$$

For this example:

$$
\frac{1000}{100}=10
$$

---

# 6. 🔥 Mini-Batch Mathematical Form

For a mini-batch $B$:

$$
\nabla_\theta J_B
=
\frac{1}{B}
\sum_{i\in B}
\nabla_\theta L_i(\theta)
$$

Then:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta J_B
$$

So mini-batch uses a **small sample of the dataset to estimate the gradient**.

---

# 7. 🔥 Why Mini-Batch is So Important

Mini-batch provides a practical middle ground:

```text
Batch GD
   ↓
Very stable
but expensive

      ↓

Mini-Batch GD
   ↓
Good balance
between efficiency
and stability

      ↓

SGD
   ↓
Very frequent
but noisy updates
```

# 8. 🔥 Complete Comparison

| Feature            | Batch GD    | SGD         | Mini-Batch GD     |
| ------------------ | ----------- | ----------- | ----------------- |
| Samples/update     | **All**     | **1**       | **Small subset**  |
| Gradient estimate  | Full        | Noisy       | Approximate       |
| Updates/epoch      | Fewest      | Most        | Middle            |
| Noise              | Low         | High        | Moderate          |
| Memory requirement | High        | Low         | Moderate          |
| Optimization       | Smooth      | Fluctuating | Relatively smooth |
| Modern DL usage    | Less common | Sometimes   | **Very common**   |

# 10. 🔥 Epoch vs Batch Size

These are **different concepts**.

### Batch size

Number of samples used for **one update**.

### Epoch

One complete pass through the **entire training dataset**.

Example:

```text
Dataset = 1000
Batch size = 100
```

One epoch contains:

```text
100 samples → update
100 samples → update
100 samples → update
...
100 samples → update
```

Total:

$$
10\text{ batches}
$$

Therefore:

$$
\boxed{
\text{Steps per epoch}\approx\frac{N}{B}
}
$$

# 11. 🔥 PyTorch Connection

This concept becomes very important when you start implementation.

You'll commonly see:

```python
DataLoader(
    dataset,
    batch_size=32,
    shuffle=True
)
```

The important part:

```python
batch_size=32
```

means the model receives approximately **32 training samples at a time**.

The training process becomes:

```text
Dataset
   ↓
DataLoader
   ↓
Batch of 32
   ↓
Forward Pass
   ↓
Loss
   ↓
Backward Pass
   ↓
Optimizer Step
   ↓
Next Batch
   ↓
Repeat
```

This is the practical form of **mini-batch training** that you'll encounter heavily in PyTorch.

# 12. ⚠️ Important Terminology

You may sometimes see people simply say **"SGD"** even when the actual implementation uses mini-batches.

Strictly for learning:

$$
\boxed{\text{Batch GD}=N}
$$

$$
\boxed{\text{SGD}=1}
$$

$$
\boxed{\text{Mini-Batch GD}=1<B<N}
$$

But in modern ML discussions, **SGD is also sometimes used as a broad term for stochastic/mini-batch gradient-based optimization**.

Don't get confused when you encounter that terminology.

# 13. 🎯 Practical Mental Model

Imagine you have **1,000 students** and want feedback.

### Batch GD 👥

Ask **all 1,000 students** before making one decision.

### SGD 👤

Ask **one student**, make a decision, then ask another.

### Mini-Batch 👥👥

Ask **32 students**, average their feedback, then make a decision.

```text
              1000 Training Samples
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       BATCH          SGD       MINI-BATCH
          ↓            ↓            ↓
         ALL          ONE         SOME
          ↓            ↓            ↓
       Stable        Noisy       Balanced
```

**Number of training examples processed before one parameter update.**

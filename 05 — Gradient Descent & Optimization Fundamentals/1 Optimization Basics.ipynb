# 1. 🔥🔥🔥 Parameters

## Definition

**Parameters are the values learned by the neural network during training.**

The main neural-network parameters are:

* **Weights**
* **Biases**

For a single neuron:

$$
z=wx+b
$$

Here:

$$
\boxed{w=\text{weight}}
$$

$$
\boxed{b=\text{bias}}
$$

Both are **learnable parameters**.

---

## Example

Initially:

$$
w=2,\qquad b=1
$$

After training:

$$
w=1.7,\qquad b=0.8
$$

The optimizer has changed the parameters based on the calculated gradients.

### 🧠 Remember

$$
\boxed{\text{Parameters = learned values}}
$$

# 2. 🔥🔥🔥 Objective Function

## Definition

The **objective function** defines what the optimization process is trying to achieve.

In most neural-network training:

$$
\boxed{\min_{\theta}J(\theta)}
$$

where:

* $\theta$ = model parameters
* $J(\theta)$ = objective function

Often:

$$
J(\theta)=\frac{1}{m}\sum_{i=1}^{m}L_i
$$

So the objective may be the average training loss.

It can also contain additional terms such as regularization:

$$
J(\theta)=L(\theta)+\lambda R(\theta)
$$

---

## Simple Example

Suppose:

$$
J(w)=(w-3)^2
$$

We want:

$$
\min_w J(w)
$$

At:

$$
w=3
$$

we get:

$$
J(3)=0
$$

So the minimum occurs at:

$$
\boxed{w=3}
$$

### 🧠 Think:

> **Objective function = What are we trying to minimize?**

---

# 3. 🔥🔥🔥 Partial Derivatives

## Why do we need them?

Suppose our objective depends on several parameters:

$$
J(w_1,w_2)
$$

We want to know:

> **How does the objective change if I change $w_1$?**

That's:

$$
\frac{\partial J}{\partial w_1}
$$

And:

> **How does the objective change if I change $w_2$?**

That's:

$$
\frac{\partial J}{\partial w_2}
$$

---

## Definition

A **partial derivative** measures how a function changes with respect to **one variable**, while treating the other variables as fixed.

## Example

Consider:

$$
J(w_1,w_2)=w_1^2+w_2^2
$$

Partial derivative with respect to $w_1$:

$$
\frac{\partial J}{\partial w_1}=2w_1
$$

Partial derivative with respect to $w_2$:

$$

$$

Each derivative tells us how the objective responds to one parameter.

# 4. 🔥🔥🔥 Gradient Vector

Now we combine **all the partial derivatives** into one vector.

For:

$$
J(w_1,w_2)
$$

the gradient is:

$$
\nabla J=
\begin{bmatrix}
\frac{\partial J}{\partial w_1}\\
\frac{\partial J}{\partial w_2}
\end{bmatrix}
$$

For $n$ parameters:

$$
\boxed{
\nabla_\theta J=
\begin{bmatrix}
\frac{\partial J}{\partial\theta_1}\\
\frac{\partial J}{\partial\theta_2}\\
\vdots\\
\frac{\partial J}{\partial\theta_n}
\end{bmatrix}
}
$$

### 🧠 Simple meaning

> **Gradient = all the partial derivatives collected together.**

# 5. 🔥 What Does the Gradient Tell Us?

The gradient points in the direction of **steepest local increase** of the objective.

Therefore:

$$
\boxed{\nabla J=\text{steepest local increase}}
$$

If we want to **minimize** the objective, we move in the opposite direction:

$$
\boxed{-\nabla J}
$$

This gives us the fundamental idea behind **Gradient Descent**.

---

# 6. 🔥🔥🔥 Gradient Descent

The basic update equation is:

$$
\boxed{
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta J
}
$$

where:

* $\theta$ = parameters
* $\eta$ = learning rate
* $\nabla_\theta J$ = gradient

---

## Example

Suppose:

$$
w=3
$$

and:

$$
\frac{\partial J}{\partial w}=6
$$

Learning rate:

$$
\eta=0.1
$$

Then:

$$
w_{\text{new}}
=
3-(0.1)(6)
$$

$$
w_{\text{new}}=2.4
$$

So:

$$
3\rightarrow2.4
$$

The parameter moved in the direction intended to reduce the objective.

# 7. 🔥 Complete Worked Example

Let's put all five concepts together.

Consider:

$$
J(w_1,w_2)=w_1^2+w_2^2
$$

Suppose:

$$
w_1=3,\qquad w_2=2
$$

### Step 1 — Parameters

$$
\theta=
\begin{bmatrix}
3\\
2
\end{bmatrix}
$$

---

### Step 2 — Objective

$$
J(3,2)=3^2+2^2
$$

$$
J=13
$$

---

### Step 3 — Partial derivatives

$$
\frac{\partial J}{\partial w_1}=2w_1
$$

$$
\frac{\partial J}{\partial w_2}=2w_2
$$

Therefore:

$$
\frac{\partial J}{\partial w_1}=6
$$

$$
\frac{\partial J}{\partial w_2}=4
$$

---

### Step 4 — Gradient

$$
\nabla J=
\begin{bmatrix}
6\\
4
\end{bmatrix}
$$

This tells us the direction of steepest increase.

---

### Step 5 — Move opposite the gradient

If:

$$
\eta=0.1
$$

then:

$$
\theta_{\text{new}}
=
\begin{bmatrix}
3\\
2
\end{bmatrix}
-
0.1
\begin{bmatrix}
6\\
4
\end{bmatrix}
$$

Therefore:

$$
\theta_{\text{new}}
=
\begin{bmatrix}
2.4\\
1.6
\end{bmatrix}
$$

The parameters moved toward the minimum at:

$$
(w_1,w_2)=(0,0)
$$

# 8. 🔥 Parameters vs Hyperparameters

Don't mix these up.

| Parameters              | Hyperparameters      |
| ----------------------- | -------------------- |
| Learned by the model    | Chosen/configured    |
| Weights                 | Learning rate        |
| Biases                  | Batch size           |
| Updated during training | Number of epochs     |
| Optimizer changes them  | Optimizer type       |
| Model learns them       | Architecture choices |

### Memory trick

$$
\boxed{\text{Parameters = Learned}}
$$

$$
\boxed{\text{Hyperparameters = Configured}}
$$

# 9. 🔥 How Everything Connects to Neural Networks

A real neural network may have millions of parameters.

For example:

$$
\theta=
[W_1,b_1,W_2,b_2,\ldots,W_L,b_L]
$$

The training process is:

```text id="q0u2pr"
                 Parameters
                W₁, b₁, W₂, b₂
                      ↓
                 Forward Pass
                      ↓
                  Prediction
                      ↓
                Loss / Objective
                      ↓
              Partial Derivatives
                      ↓
                Gradient Vector
                      ↓
               Gradient Descent
                      ↓
             Updated Parameters
                      ↓
                  Repeat
```

This is the mathematical foundation of neural-network training.

---

# 🎯 Interview / GATE Points

### What are parameters?

> Learnable values such as weights and biases.

### What is an objective function?

> A function that defines what optimization is trying to minimize or maximize.

### What is a partial derivative?

> The rate of change of a function with respect to one variable while holding the other variables constant.

### What is a gradient?

> A vector containing the partial derivatives of a scalar function with respect to its variables.

### Which direction does the gradient point?

> The direction of steepest local increase.

### Why does gradient descent use $-\nabla J$?

Because we want to move toward decreasing objective values.

$$
\theta\leftarrow\theta-\eta\nabla_\theta J
$$

---

# 🚨 Common Mistakes

### ❌ Gradient = one derivative

No.

For multiple parameters:

$$
\text{Gradient}=
\begin{bmatrix}
\text{partial derivative 1}\\
\text{partial derivative 2}\\
\vdots
\end{bmatrix}
$$

---

### ❌ Gradient points toward the minimum

Not exactly.

$$
\nabla J
$$

points toward **steepest increase**.

Gradient descent uses:

$$
-\nabla J
$$

---

### ❌ Parameters = hyperparameters

No.

```text
Weights / Biases → Parameters
Learning rate    → Hyperparameter
```

---

### ❌ Objective function always equals loss

Not necessarily.

A simple objective may be the loss itself:

$$
J(\theta)=L(\theta)
$$

But it can also include regularization:

$$
J(\theta)=L(\theta)+\lambda R(\theta)
$$

### Parameters

$$
\boxed{\theta=\text{learnable weights and biases}}
$$

### Objective Function

$$
\boxed{\min_\theta J(\theta)}
$$

### Partial Derivative

$$
\boxed{\frac{\partial J}{\partial\theta_i}}
$$

**Effect of one parameter on the objective.**

$$
\boxed{
\nabla_\theta J=
\begin{bmatrix}
\frac{\partial J}{\partial\theta_1}\\
\frac{\partial J}{\partial\theta_2}\\
\vdots\\
\frac{\partial J}{\partial\theta_n}
\end{bmatrix}
}
$$

**All partial derivatives together.**

### Gradient Descent

$$
\boxed{
\theta\leftarrow\theta-\eta\nabla_\theta J
}
$$

# 🧠 Ultimate Mental Model

> 🧠 **Parameters** → what the model learns
> 🎯 **Objective** → what we want to minimize
> 📐 **Partial derivative** → effect of one parameter
> 🧭 **Gradient** → effects of all parameters together
> 🚶 **Gradient descent** → move opposite the gradient

$$
\boxed{
\text{Parameters}
\rightarrow
\text{Objective}
\rightarrow
\text{Derivatives}
\rightarrow
\text{Gradient}
\rightarrow
\text{Update}
}
$$

# 🧠 MODULE 33 — SEQUENCE TYPES

## 1. Overview

Sequence types describe the relationship between the **number of inputs** and **number of outputs** in a neural-network architecture.

There are four fundamental patterns:

1. **One-to-One**
2. **One-to-Many**
3. **Many-to-One**
4. **Many-to-Many**

The easiest way to identify them is:

> **Count the inputs → Count the outputs.**

---

# 2. 🔵 One-to-One

### Definition

A single input produces a single output.

```text
ONE INPUT
    │
    ▼
  MODEL
    │
    ▼
ONE OUTPUT
```

### Mapping

```text
1 → 1
```

### Example — Image Classification

```text
🖼️ Image
   │
   ▼
Neural Network
   │
   ▼
🐱 Cat
```

**Input:** One image
**Output:** One class

### Other examples

* Image → Cat/Dog
* Image → Class label
* Feature vector → Prediction

### Important

One-to-one is generally **not a sequence-processing architecture**, because there is only one input and one output.

---

# 3. 🟢 One-to-Many

### Definition

One input produces a sequence of outputs.

```text
                 ┌──→ Output₁
                 ├──→ Output₂
ONE INPUT ──→ MODEL
                 ├──→ Output₃
                 └──→ Output₄
```

### Mapping

```text
1 → MANY
```

## Example — Image Captioning

```text
🖼️ Image
    │
    ▼
  Model
    │
    ├──→ "A"
    ├──→ "dog"
    ├──→ "is"
    └──→ "running"
```

So:

```text
Image → "A" → "dog" → "is" → "running"
```

The model receives **one input** and generates **multiple outputs sequentially**.

### Other examples

* Image → Caption
* Initial input → Music sequence
* Image → Description

---

# 4. 🟠 Many-to-One

### Definition

A sequence of inputs produces one output.

```text
Input₁ ─┐
Input₂ ─┤
Input₃ ─┤
Input₄ ─┘
    │
    ▼
  MODEL
    │
    ▼
ONE OUTPUT
```

### Mapping

```text
MANY → 1
```

## Example — Sentiment Analysis

Sentence:

```text
"I really enjoyed this movie"
```

Sequence:

```text
"I"
 ↓
"really"
 ↓
"enjoyed"
 ↓
"this"
 ↓
"movie"
 ↓
MODEL
 ↓
❤️ Positive
```

### Other examples

* Sentence → Sentiment
* Sensor sequence → Fault/Normal
* Time series → Class
* Video sequence → Activity

### Key idea

The model receives the **whole sequence** and produces **one final prediction**.

---

# 5. 🔴 Many-to-Many

### Definition

A sequence of inputs produces a sequence of outputs.

```text
Input₁ → Model → Output₁
Input₂ → Model → Output₂
Input₃ → Model → Output₃
Input₄ → Model → Output₄
```

### Mapping

```text
MANY → MANY
```

There are two important forms.

---

# 6. Many-to-Many — Same Length

Here, every input position has a corresponding output.

```text
Input₁ → Output₁
Input₂ → Output₂
Input₃ → Output₃
Input₄ → Output₄
```

Therefore:

```text
Input length = Output length
```

## Example — Named Entity Recognition

Sentence:

```text
"Elon Musk founded SpaceX"
```

Input:

```text
Elon → Musk → founded → SpaceX
```

Output:

```text
PERSON → PERSON → OTHER → ORGANIZATION
```

Therefore:

```text
4 input tokens → 4 output labels
```

This is called **sequence labeling**.

### Examples

* Named Entity Recognition
* Part-of-Speech tagging
* Token-level classification
* Sequence labeling

---

# 7. 🔥 Many-to-Many — Different Length

This is commonly associated with **Sequence-to-Sequence (Seq2Seq)** architectures.

```text
INPUT SEQUENCE
      │
      ▼
    MODEL
      │
      ▼
OUTPUT SEQUENCE
```

The input and output sequences can have different lengths.

```text
Input length ≠ Output length
```

## Example — Machine Translation

English:

```text
"I love AI"
```

French:

```text
"J'aime l'IA"
```

Conceptually:

```text
"I" → "love" → "AI"
             │
             ▼
          Encoder
             │
             ▼
          Decoder
             │
             ▼
"J'" → "aime" → "l'IA"
```

### Examples

* Machine translation
* Sequence transformation
* Speech-to-text
* Other Seq2Seq tasks

---

# 8. 📊 Complete Comparison

| Type                       | Input | Output | Example              |
| -------------------------- | ----: | -----: | -------------------- |
| **One-to-One**             |     1 |      1 | Image classification |
| **One-to-Many**            |     1 |   Many | Image captioning     |
| **Many-to-One**            |  Many |      1 | Sentiment analysis   |
| **Many-to-Many**           |  Many |   Many | NER                  |
| **Many-to-Many / Seq2Seq** |  Many |   Many | Machine translation  |

# 9. 🧠 Input-Output Matrix

```text
                       OUTPUT
                  ONE          MANY
              ┌──────────┬──────────┐
         ONE  │ 1 → 1    │ 1 → M    │
              │          │          │
INPUT         ├──────────┼──────────┤
        MANY  │ M → 1    │ M → M    │
              │          │          │
              └──────────┴──────────┘
```

### Examples

```text
1 → 1
Image → Class

1 → M
Image → Caption

M → 1
Sentence → Sentiment

M → M
Words → NER Labels
```

---

# 10. 🔥 Two Forms of Many-to-Many

## A. Same-Length Many-to-Many

```text
x₁ → y₁
x₂ → y₂
x₃ → y₃
x₄ → y₄
```

```text
Input length = Output length
```

Example:

```text
Words → NER labels
```

---

## B. Different-Length Many-to-Many

```text
x₁ → x₂ → x₃ → x₄
          │
          ▼
       Encoder
          │
          ▼
       Decoder
          │
          ▼
y₁ → y₂ → y₃
```

```text
Input length ≠ Output length
```

Example:

```text
English → French
```

This is the basic idea behind **Seq2Seq** models.

# 12. 🧠 How to Identify the Architecture

Whenever you see a problem, ask:

### Step 1 — How many inputs?

```text
ONE or MANY?
```

### Step 2 — How many outputs?

```text
ONE or MANY?
```

### Step 3 — Match them

```text
ONE + ONE
    ↓
ONE-TO-ONE

ONE + MANY
    ↓
ONE-TO-MANY

MANY + ONE
    ↓
MANY-TO-ONE

MANY + MANY
    ↓
MANY-TO-MANY
```

# 14. 🔥 Golden Memory Trick

```text
ONE → ONE
One thing → One thing

ONE → MANY
One thing → Sequence

MANY → ONE
Sequence → One thing

MANY → MANY
Sequence → Sequence
```

> **Don't memorize the application first. Count the inputs and outputs.**

> **Core idea:** The training loop repeatedly takes a batch of data, performs a forward pass, calculates the loss, computes gradients with `backward()`, and updates the model parameters with the optimizer.

---

# 1. 🔥🔥🔥 Complete Training Loop

The conceptual flow is:

```text
Data
 ↓
Model
 ↓
Prediction
 ↓
Loss
 ↓
Backward()
 ↓
Optimizer
 ↓
Update
 ↺
```

The practical PyTorch version is:

```python
for X_batch, y_batch in train_loader:

    # Forward pass
    prediction = model(X_batch)

    # Calculate loss
    loss = loss_fn(prediction, y_batch)

    # Clear old gradients
    optimizer.zero_grad()

    # Backpropagation
    loss.backward()

    # Update parameters
    optimizer.step()
```

This is the **core PyTorch training loop**.

---

# 2. 🔥 Step 1 — Get a Batch

The `DataLoader` provides the current batch:

```python
for X_batch, y_batch in train_loader:
```

For example:

```text
X_batch → (32, 10)
y_batch → (32,)
```

Meaning:

```text
32 samples
10 features per sample
```

The overall flow is:

```text
Dataset
   ↓
DataLoader
   ↓
Batch
   ↓
Training Loop
```

---

# 3. 🔥🔥🔥 Step 2 — Forward Pass

```python
prediction = model(X_batch)
```

The batch passes through the neural network.

Example:

```text
X_batch
   ↓
Linear
   ↓
ReLU
   ↓
Linear
   ↓
Prediction
```

The output is:

$$
\hat{y}
$$

where $\hat{y}$ is the model's prediction.

---

# 4. 🔥🔥🔥 Step 3 — Calculate Loss

```python
loss = loss_fn(prediction, y_batch)
```

The loss function compares:

```text
Prediction
     ↓
    Loss
     ↑
  Target
```

For example, MSE:

$$
L=\frac{1}{n}\sum_i(y_i-\hat{y}_i)^2
$$

The loss is a scalar describing how far the predictions are from the targets according to the chosen loss function.

---

# 5. 🔥🔥🔥 Step 4 — `optimizer.zero_grad()`

```python
optimizer.zero_grad()
```

This clears the gradients from the previous iteration.

### Why?

PyTorch **accumulates gradients by default**.

Without clearing them:

```text
Batch 1
 ↓
gradient = g₁

Batch 2
 ↓
gradient = g₁ + g₂
```

Normally, for standard mini-batch training, we want the gradient for the **current batch**, not an accumulation of previous batches.

Therefore:

```python
optimizer.zero_grad()
```

clears the existing gradients before the new backward pass.

---

# 6. 🔥🔥🔥 Step 5 — Backpropagation

```python
loss.backward()
```

This activates PyTorch's **Autograd** system.

It traverses the computational graph backward and calculates gradients such as:

$$
\frac{\partial L}{\partial W}
$$

and:

$$
\frac{\partial L}{\partial b}
$$

Conceptually:

```text
Loss
 ↓
Backward
 ↓
Gradients
 ↓
parameter.grad
```

This connects directly to Module 14.1:

```text
Computational Graph
        ↓
Autograd
        ↓
backward()
        ↓
Gradients
```

---

# 7. 🔥🔥🔥 Step 6 — Optimizer Update

```python
optimizer.step()
```

The optimizer uses the calculated gradients to update the model's parameters.

For basic SGD:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
$$

Conceptually:

```text
Parameter
    +
Gradient
    +
Learning Rate
    ↓
Optimizer
    ↓
Updated Parameter
```

For Adam/AdamW, the mathematical update is more sophisticated, but the role remains the same:

> **Use gradients to update the learnable parameters.**

---

# 8. 🔥 Complete Flow

```text
                 DataLoader
                     │
                     ▼
                   Batch
                     │
                     ▼
                 model(X)
                     │
                     ▼
                Prediction
                     │
                     ▼
             loss_fn(pred, y)
                     │
                     ▼
                    Loss
                     │
                     ▼
           optimizer.zero_grad()
                     │
                     ▼
              loss.backward()
                     │
                     ▼
                 Gradients
                     │
                     ▼
             optimizer.step()
                     │
                     ▼
          Updated Parameters
                     │
                     └──────→ Next Batch
```

🔥 This is the fundamental PyTorch training cycle.

---

# 9. 🔥 Epochs

Usually, we train for multiple epochs.

```python
for epoch in range(num_epochs):

    for X_batch, y_batch in train_loader:

        prediction = model(X_batch)

        loss = loss_fn(prediction, y_batch)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()
```

Structure:

```text
Epoch 1
 ├── Batch 1
 ├── Batch 2
 ├── Batch 3
 └── ...

Epoch 2
 ├── Batch 1
 ├── Batch 2
 ├── Batch 3
 └── ...

Epoch 3
 ├── Batch 1
 ├── Batch 2
 ├── Batch 3
 └── ...
```

An **epoch** means one complete pass through the training dataset.

---

# 10. 🔥 Parameter Update Example

Suppose:

$$
w=2
$$

and:

$$
\frac{\partial L}{\partial w}=4
$$

Learning rate:

$$
\eta=0.1
$$

SGD update:

$$
w_{\text{new}}
=
w-\eta\frac{\partial L}{\partial w}
$$

Therefore:

$$
w_{\text{new}}
=
2-(0.1)(4)
$$

$$
w_{\text{new}}=1.6
$$

PyTorch's conceptual flow:

```text
loss.backward()
      ↓
w.grad = 4
      ↓
optimizer.step()
      ↓
w = 1.6
```

# 11. 🔥 Training vs Inference

### Training

```text
Input
 ↓
Forward
 ↓
Loss
 ↓
Backward
 ↓
Gradients
 ↓
Optimizer
 ↓
Update
```

### Inference

```text
Input
 ↓
Model
 ↓
Prediction
```

There is **no parameter update during inference**.

Typical inference:

```python
model.eval()

with torch.no_grad():
    prediction = model(X)
```

---

# 12. 🔥 `model.train()` vs `model.eval()`

### Training

```python
model.train()
```

Used to put the model into training mode.

Important for layers whose behavior differs between training and evaluation, especially:

* Dropout
* BatchNorm

### Evaluation

```python
model.eval()
```

Used for validation/testing/inference behavior.

---

# 13. 🔥 `torch.no_grad()`

Don't confuse:

```python
model.eval()
```

with:

```python
torch.no_grad()
```

They do different things.

### `model.eval()`

Changes the behavior of certain layers, such as:

* Dropout
* BatchNorm

### `torch.no_grad()`

Disables gradient tracking for the enclosed operations.

Typical evaluation:

```python
model.eval()

with torch.no_grad():
    prediction = model(X)
```

# 14. 🔥 Full Training Example

```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1)
)

loss_fn = nn.BCEWithLogitsLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001
)

for epoch in range(num_epochs):

    model.train()

    for X_batch, y_batch in train_loader:

        # Forward
        prediction = model(X_batch)

        # Loss
        loss = loss_fn(prediction, y_batch)

        # Clear old gradients
        optimizer.zero_grad()

        # Backward
        loss.backward()

        # Update
        optimizer.step()
```

This combines everything from:

```text
14.1 → Autograd
14.2 → Model API
14.3 → Dataset/DataLoader
14.4 → Training Loop
```

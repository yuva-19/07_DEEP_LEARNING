# 1. 🔥🔥🔥 `nn.Module`

`nn.Module` is the **base class for PyTorch neural-network models and layers**.

PyTorch's official documentation describes a neural network as a module that can contain other modules/layers. This nested structure lets PyTorch manage complex architectures and their parameters. ([PyTorch Docs][1])

```python
import torch
import torch.nn as nn

class MyNetwork(nn.Module):

    def __init__(self):
        super().__init__()

    def forward(self, x):
        pass
```

Think of it as:

```text
nn.Module
    │
    ├── Layers
    ├── Parameters
    ├── Forward computation
    └── Model structure
```

---

# 2. 🔥 `__init__()` vs `forward()`

These two methods are fundamental.

### `__init__()`

Defines the **layers/components** of the model.

```python
def __init__(self):
    super().__init__()

    self.fc1 = nn.Linear(10, 64)
    self.fc2 = nn.Linear(64, 1)
```

### `forward()`

Defines **how the input flows through those layers**.

```python
def forward(self, x):
    x = self.fc1(x)
    x = self.fc2(x)
    return x
```

So remember:

```text
__init__()
    ↓
"What components does my model contain?"

forward()
    ↓
"How does data flow through them?"
```

The official PyTorch tutorial follows exactly this pattern: define layers in `__init__()` and implement the computation in `forward()`. ([PyTorch Docs][1])

# 3. 🔥🔥🔥 `nn.Linear`

`nn.Linear` represents a **fully connected / dense layer**.

Mathematically:

$$
z=Wx+b
$$

PyTorch:

```python
nn.Linear(in_features, out_features)
```

Example:

```python
layer = nn.Linear(10, 64)
```

means:

```text
10 input features
       ↓
   Linear layer
       ↓
64 output features
```

---

## Parameters inside `nn.Linear`

For:

```python
nn.Linear(10, 64)
```

PyTorch creates:

### Weight

$$
W\in\mathbb{R}^{64\times10}
$$

### Bias

$$
b\in\mathbb{R}^{64}
$$

Therefore:

$$
\text{output}=xW^T+b
$$

when using the common PyTorch batch representation `(batch, features)`.

---

## Parameter count

Weights:

$$
10\times64=640
$$

Biases:

$$
64
$$

Total:

$$
640+64=704
$$

So:

```python
nn.Linear(10, 64)
```

contains **704 trainable parameters**.

---

# 4. 🔥 `nn.Linear` with batches

Suppose:

```python
x.shape
```

is:

```text
(32, 10)
```

Meaning:

```text
32 samples
10 features per sample
```

Then:

```python
layer = nn.Linear(10, 64)

output = layer(x)
```

gives:

```text
(32, 64)
```

because:

```text
(32, 10)
    ↓
Linear(10 → 64)
    ↓
(32, 64)
```

This is a very important shape pattern when implementing networks.

# 5. 🔥🔥🔥 Activation Layers

A linear layer performs:

$$
z=Wx+b
$$

But stacking only linear layers still produces an overall linear transformation.

Therefore neural networks need **nonlinear activation functions**.

PyTorch provides activation layers such as:

```python
nn.ReLU()
nn.Sigmoid()
nn.Tanh()
nn.GELU()
nn.LeakyReLU()
```

---

## ReLU

The most important basic example:

$$
ReLU(x)=\max(0,x)
$$

PyTorch:

```python
relu = nn.ReLU()
```

Example:

```python
x = torch.tensor([-2.0, 0.0, 3.0])

relu = nn.ReLU()

print(relu(x))
```

Conceptually:

```text
[-2, 0, 3]
     ↓
   ReLU
     ↓
[ 0, 0, 3]
```

---

## Typical network

```python
nn.Linear(10, 64)
nn.ReLU()
nn.Linear(64, 32)
nn.ReLU()
nn.Linear(32, 1)
```

Architecture:

```text
Input
  ↓
Linear
  ↓
ReLU
  ↓
Linear
  ↓
ReLU
  ↓
Linear
  ↓
Output
```

# 6. 🔥🔥🔥 Loss Functions

A loss function measures how different the model's prediction is from the target.

You already learned the mathematical side of this in Module 4.

PyTorch provides common losses through `torch.nn`.

---

## Regression → `MSELoss`

```python
loss_fn = nn.MSELoss()
```

Mathematically:

$$
MSE=\frac{1}{n}\sum_i(y_i-\hat y_i)^2
$$

Usage:

```python
loss = loss_fn(prediction, target)
```

---

## Binary Classification → `BCEWithLogitsLoss`

```python
loss_fn = nn.BCEWithLogitsLoss()
```

Important:

**It expects raw logits.**

Typical setup:

```text
Model
  ↓
Raw logits
  ↓
BCEWithLogitsLoss
```

So normally you **do not add `nn.Sigmoid()` before this loss**.

---

## Multiclass Classification → `CrossEntropyLoss`

```python
loss_fn = nn.CrossEntropyLoss()
```

Typical setup:

```text
Model
  ↓
Raw logits
  ↓
CrossEntropyLoss
```

It expects raw logits and integer class labels.

So during the normal training setup:

❌ Don't manually add `Softmax()` before `CrossEntropyLoss`.

This distinction is important because PyTorch's loss implementations handle the necessary numerical operations internally. ([PyTorch Docs][1])

# 7. 🔥 Loss vs Optimizer

Don't confuse these two.

### Loss function

Answers:

> **"How wrong is my prediction?"**

### Optimizer

Answers:

> **"How should I change my parameters using the gradients?"**

Full flow:

```text
Input
 ↓
Model
 ↓
Prediction
 ↓
Loss Function
 ↓
Loss
 ↓
backward()
 ↓
Gradients
 ↓
Optimizer
 ↓
Updated Parameters
```

For example:

```python
loss_fn = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001
)
```

# 8. 🔥🔥 `nn.Sequential`

`nn.Sequential` is an **ordered container of modules**.

PyTorch passes the input through the modules in the order you specify. ([PyTorch Docs][1])

Instead of writing a custom `forward()`:

```python
class MyNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.fc1 = nn.Linear(10, 64)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(64, 1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x
```

we can write:

```python
model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1)
)
```

The data automatically flows:

```text
Input
  ↓
Linear(10 → 64)
  ↓
ReLU
  ↓
Linear(64 → 1)
  ↓
Output
```

([PyTorch Docs][1])

---

# 9. 🔥 `nn.Module` vs `nn.Sequential`

This is an important practical distinction.

### `nn.Sequential`

Best for straightforward architectures:

```text
A → B → C → D
```

Example:

```python
model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1)
)
```

### Custom `nn.Module`

Better when you need custom data flow:

```text
             ┌── Layer A ──┐
Input ───────┤              ├──→ Combine
             └── Layer B ──┘
```

Examples:

* residual/skip connections
* branching
* multiple inputs
* multiple outputs
* custom operations
* attention architectures

So:

> **`nn.Sequential` is a convenient container; `nn.Module` is the general foundation for PyTorch models.**

PyTorch's own examples use `nn.Sequential` as a submodule inside a custom `nn.Module`, which is a very common pattern. ([PyTorch Docs][1])

# 10. 🔥 Complete PyTorch Model

Let's put everything together.

Suppose we have:

* 10 input features
* 64 hidden neurons
* 32 hidden neurons
* 1 output

```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),

    nn.Linear(64, 32),
    nn.ReLU(),

    nn.Linear(32, 1)
)
```

Architecture:

```text
        10 features
             │
             ▼
      ┌─────────────┐
      │ Linear      │
      │ 10 → 64     │
      └─────────────┘
             │
             ▼
          ReLU
             │
             ▼
      ┌─────────────┐
      │ Linear      │
      │ 64 → 32     │
      └─────────────┘
             │
             ▼
          ReLU
             │
             ▼
      ┌─────────────┐
      │ Linear      │
      │ 32 → 1      │
      └─────────────┘
             │
             ▼
          Output
```

This is essentially the same model structure PyTorch demonstrates in its official beginner model-building tutorial. ([PyTorch Docs][1])

# 14.6 GPU

## 1. CPU vs GPU 🔥🔥🔥

### CPU — Central Processing Unit

A CPU is a **general-purpose processor** designed to handle many different kinds of tasks.

Typical characteristics:

* Fewer, powerful cores
* Good at sequential/branching operations
* Handles general Python/program execution
* Good for small or irregular workloads

### GPU — Graphics Processing Unit

A GPU is designed to perform **large numbers of similar operations in parallel**.

This makes it particularly suitable for Deep Learning because neural networks perform huge numbers of:

* Matrix multiplications
* Tensor operations
* Convolutions
* Element-wise operations

### Simple intuition

Imagine calculating 10,000 multiplication operations.

```text
CPU

Core 1
  ↓
  calculation
  ↓
  calculation
  ↓
  calculation
```

A GPU can divide the work across many processing units:

```text
GPU

Core 1 ──→ calculation
Core 2 ──→ calculation
Core 3 ──→ calculation
Core 4 ──→ calculation
Core 5 ──→ calculation
   ...
Many operations simultaneously
```

So:

> **CPU = powerful general-purpose workers**

> **GPU = huge number of parallel workers**

# 2. Why GPUs Are Useful for Deep Learning 🔥🔥🔥

Consider a neural-network layer:

$$
Z = XW+b
$$

For a large dataset and large network, this can involve **millions or billions of arithmetic operations**.

The forward pass performs these calculations.

Then backpropagation performs many more operations to calculate gradients.

```text
Input
  ↓
Forward Pass
  ↓
Loss
  ↓
Backward Pass
  ↓
Gradients
  ↓
Parameter Update
```

Therefore, GPU parallelism can dramatically accelerate large neural-network workloads.

### Important nuance ⚠️

GPU ≠ automatically faster for everything.

For very small models or tiny tensors, GPU overhead can outweigh the benefit.

# 4. Device in PyTorch 🔥🔥🔥

A PyTorch tensor exists on a particular **device**.

Common devices:

```text
cpu
cuda:0
cuda:1
cuda:2
```

For example:

```python
x.device
```

could return:

```text
device(type='cpu')
```

or:

```text
device(type='cuda', index=0)
```

### What does `cuda:0` mean?

```text
cuda:0
  ↑  ↑
  │  └── GPU number 0
  └───── CUDA device
```

If you have multiple GPUs:

```text
cuda:0 → first GPU
cuda:1 → second GPU
cuda:2 → third GPU
```

PyTorch's device abstraction identifies both the device type and, when applicable, the device index. ([PyTorch Docs][2])

# 6. Selecting the Device 🔥🔥🔥

Instead of hard-coding GPU:

```python
device = "cuda"
```

use:

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
```

This means:

```text
CUDA available?
       │
   ┌───┴───┐
  Yes      No
   ↓        ↓
 CUDA      CPU
```

This is called **device-agnostic code**.

PyTorch itself recommends this style for code that can switch between CPU and CUDA depending on available hardware. ([PyTorch Docs][1])

# 9. Model and Input Must Be on Compatible Devices ⭐⭐⭐

This is one of the most common beginner GPU mistakes.

### ❌ Wrong

```text
Model → GPU

Input → CPU
```

Then:

```python
prediction = model(X)
```

can produce a device mismatch error.

Conceptually:

```text
CPU                         GPU

X ────────────────────────→ Model
       ❌ mismatch
```

### ✅ Correct

```text
CPU                         GPU

X ── .to(device) ─────────→ Model
                              ↑
                         Model parameters
```

Therefore:

```python
model = model.to(device)
X = X.to(device)
```

Now:

```python
prediction = model(X)
```

works on the same device.

# 10. Complete GPU Example 🔥🔥🔥

```python
import torch
import torch.nn as nn

# Select device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Create model
model = nn.Sequential(
    nn.Linear(10, 64),
    nn.ReLU(),
    nn.Linear(64, 1)
)

# Move model
model = model.to(device)

# Create input
X = torch.randn(32, 10)

# Move input
X = X.to(device)

# Forward pass
output = model(X)

print("Device:", device)
print("Input:", X.device)
print("Output:", output.device)
```

If CUDA is available:

```text
Device: cuda
Input: cuda:0
Output: cuda:0
```

### GPU training

### Complete mental flow

```text
Dataset
   ↓
DataLoader
   ↓
CPU batch
   ↓
.to(device)
   ↓
GPU
   ↓
Model
   ↓
Forward
   ↓
Loss
   ↓
Backward
   ↓
Gradients
   ↓
Optimizer
   ↓
Updated GPU parameters
```

# 12. `.to(device)` vs `.cuda()`

You may see:

```python
x.cuda()
```

This directly moves a tensor to CUDA.

But for your projects, prefer:

```python
x.to(device)
```

because it works for both:

```text
CPU
CUDA
```

Example:

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

x = x.to(device)
model = model.to(device)
```

This lets the same code work on machines with or without a CUDA GPU.

# 14. Don't Transfer Data Unnecessarily ⚠️

Moving data between CPU and GPU has overhead.

Avoid doing this repeatedly:

```text
CPU
 ↓
GPU
 ↓
CPU
 ↓
GPU
 ↓
CPU
```

Instead, keep the data where the computation happens as much as practical:

```text
CPU
 ↓
GPU
 ↓
GPU computation
 ↓
GPU computation
 ↓
GPU computation
```

This becomes increasingly important with large models and datasets.

# 15. GPU Memory / VRAM 🔥🔥

A GPU has its own memory, commonly called **VRAM**.

During training, GPU memory can contain:

```text
GPU Memory
│
├── Model parameters
├── Gradients
├── Optimizer states
├── Activations
└── Input batch
```

So even if your computer has plenty of system RAM, the GPU can still run out of memory.

You may see:

```text
CUDA out of memory
```

### Common solutions

If this happens:

* Reduce `batch_size`
* Reduce model size
* Reduce image resolution
* Reduce sequence length
* Use mixed precision
* Use gradient accumulation

We will properly encounter these techniques later.

# 17. GPU vs `model.train()` vs `model.eval()` vs `no_grad()` 🔥🔥🔥

These are **four completely different concepts**.

| Code                | Purpose                   |
| ------------------- | ------------------------- |
| `model.to(device)`  | Move model                |
| `tensor.to(device)` | Move tensor               |
| `model.train()`     | Training behavior         |
| `model.eval()`      | Evaluation behavior       |
| `torch.no_grad()`   | Disable gradient tracking |

For example:

```python
model = model.to(device)
```

asks:

> **Where should computation happen?**

While:

```python
model.eval()
```

asks:

> **How should the model behave during evaluation?**

And:

```python
with torch.no_grad():
```

asks:

> **Should PyTorch track gradients?**

🔥 Don't confuse these.

# 18. GPU During Training vs Inference

### Training

```python
model.train()

for X, y in train_loader:

    X = X.to(device)
    y = y.to(device)

    prediction = model(X)

    loss = loss_fn(prediction, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

### Inference

```python
model.eval()

with torch.no_grad():

    X = X.to(device)

    prediction = model(X)
```

Notice:

```text
Training:
GPU + gradients + parameter updates

Inference:
GPU + forward computation
(no gradient tracking)
```

# 19. Industry-Style Device Pattern ⭐⭐⭐

This pattern is worth memorizing:

```python
import torch

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = model.to(device)

for X_batch, y_batch in train_loader:

    X_batch = X_batch.to(device)
    y_batch = y_batch.to(device)

    prediction = model(X_batch)

    loss = loss_fn(prediction, y_batch)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()
```

This is the basic foundation you'll use repeatedly when working with PyTorch.

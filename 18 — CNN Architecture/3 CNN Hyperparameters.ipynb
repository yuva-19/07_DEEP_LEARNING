## 🎯 What are CNN Hyperparameters?

**Hyperparameters** are design settings chosen **before training**.

They control how a CNN processes images and how its architecture behaves.

The five important hyperparameters in this block are:

| Hyperparameter        | Controls                                 |
| --------------------- | ---------------------------------------- |
| **Kernel size**       | How large a local region the filter sees |
| **Stride**            | How far the filter moves                 |
| **Padding**           | How borders are handled                  |
| **Number of filters** | How many different features are learned  |
| **Pooling size**      | Size of the region used for downsampling |

> 🧠 **Important:** These are generally **not learned directly through backpropagation**. We choose them when designing the network.

---

# 1️⃣ Kernel Size 🔥🔥🔥

### Definition

**Kernel size** determines the spatial size of the region examined by a convolution filter at each position.

Common choices:

```text
3 × 3
5 × 5
7 × 7
```

### Example

A `3×3` kernel looks at:

```text
[ x x x ]
[ x x x ]  ← 9 pixels
[ x x x ]
```

A `5×5` kernel looks at:

```text
25 pixels
```

So:

> **Kernel size = "How much area do I look at?"**

### Effect

Larger kernel:

* sees a larger local region
* has more parameters
* increases computation
* can capture broader spatial patterns

Example with:

```text
Cin = 3
Cout = 32
```

For `3×3`:

$$
P=(3\times3\times3\times32)+32
$$

$$
P=896
$$

For `5×5`:

$$
P=(5\times5\times3\times32)+32
$$

$$
P=2432
$$

So increasing kernel size significantly increases parameters.

### 🧠 Important insight

Instead of using one large kernel, CNNs can stack smaller kernels.

For example:

```text
3×3 Conv
   ↓
3×3 Conv
```

The effective receptive field becomes approximately:

```text
5 × 5
```

while using nonlinearities between layers.

That's one reason **3×3 kernels became very common**.

---

# 2️⃣ Stride 🔥🔥🔥

### Definition

**Stride** specifies how many pixels the kernel moves after each operation.

### Stride = 1

```text
→
→
→
```

The filter moves one pixel at a time.

### Stride = 2

```text
→→
→→
→→
```

The filter moves two pixels at a time.

So:

> **Stride = "How far do I move?"**

### Effect of increasing stride

Higher stride generally means:

```text
larger stride
      ↓
fewer kernel positions
      ↓
smaller output
      ↓
more spatial downsampling
```

### Example

Input:

```text
32 × 32
```

Kernel:

```text
3 × 3
```

Padding:

```text
0
```

#### Stride = 1

$$
O=\left\lfloor\frac{32-3}{1}\right\rfloor+1
$$

$$
O=30
$$

Output:

```text
30 × 30
```

#### Stride = 2

$$
O=\left\lfloor\frac{32-3}{2}\right\rfloor+1
$$

$$
O=15
$$

Output:

```text
15 × 15
```

---

# 3️⃣ Padding 🔥🔥🔥

### Definition

**Padding** adds extra values around the border of the input.

The most common type is **zero padding**.

Example:

```text
Original

1 2 3
4 5 6
7 8 9
```

With one-pixel zero padding:

```text
0 0 0 0 0
0 1 2 3 0
0 4 5 6 0
0 7 8 9 0
0 0 0 0 0
```

### Why use padding?

Without padding, convolution gradually reduces spatial dimensions.

Padding helps:

* preserve spatial dimensions
* allow border pixels to participate more fully
* control output size

---

## `valid` vs `same`

### Valid padding

```text
padding = 0
```

No extra border.

Example:

```text
32 × 32
↓
3×3 Conv
↓
30 × 30
```

### Same padding

Padding is selected so that, for stride 1, the spatial dimensions are preserved.

Example:

```text
32 × 32
↓
3×3 Conv
padding = 1
↓
32 × 32
```

Therefore:

> **Padding = "What do I do at the borders?"**

---

# 4️⃣ Number of Filters 🔥🔥🔥

This is one of the **most important CNN concepts**.

### Definition

The **number of filters** determines how many different feature detectors the convolution layer learns.

In PyTorch:

```python
nn.Conv2d(
    in_channels=3,
    out_channels=32,
    kernel_size=3
)
```

Here:

```text
in_channels  = 3
out_channels = 32
```

Therefore:

> **32 filters are learned.**

Each filter produces one output feature map.

```text
32 filters
    ↓
32 feature maps
```

So if the input is:

```text
3 × 224 × 224
```

and we use:

```text
32 filters
```

with spatial dimensions preserved:

```text
3 × 224 × 224
       ↓
   Conv2D
       ↓
32 × 224 × 224
```

### What do those filters learn?

Different filters can learn different patterns:

```text
Filter 1 → edges
Filter 2 → corners
Filter 3 → textures
Filter 4 → curves
...
Filter 32 → different learned patterns
```

The actual patterns are learned during training.

---

## Parameter relationship

For a convolution layer:

$$
P=K_HK_WC_{in}C_{out}+C_{out}
$$

when bias is enabled.

Example:

```text
Kernel = 3×3
Cin = 3
Cout = 32
```

$$
P=(3)(3)(3)(32)+32
$$

$$
P=896
$$

If we increase filters:

```text
Cout = 64
```

then:

$$
P=(3)(3)(3)(64)+64
$$

$$
P=1792
$$

### Important

Increasing the number of filters:

```text
More filters
     ↓
More output channels
     ↓
More feature representations
     ↓
More parameters + computation
```

It does **not directly increase H × W**.

### Definition

**Pooling size** specifies the spatial region examined by the pooling operation.

Common example:

```text
2 × 2
```

For max pooling:

```text
[1  5]
[3  2]
```

Output:

```text
5
```

because:

$$
\max(1,5,3,2)=5
$$

### Example

Input:

```text
28 × 28
```

MaxPool:

```text
kernel = 2×2
stride = 2
```

Output:

```text
14 × 14
```

So pooling reduces spatial dimensions.

### Important distinction

Don't confuse:

**Pooling size**

with

**Pooling stride**

For example:

```python
nn.MaxPool2d(
    kernel_size=2,
    stride=2
)
```

means:

```text
Pooling window = 2×2
Movement = 2 pixels
```

# 🧮 6️⃣ Output Size Formula

This formula is **🔥🔥🔥 MUST KNOW**.

For convolution and standard pooling:

$$
\boxed{
O=
\left\lfloor
\frac{N-F+2P}{S}
\right\rfloor+1
}
$$

Where:

| Symbol | Meaning             |
| ------ | ------------------- |
| \(N\)  | Input spatial size  |
| \(F\)  | Filter/kernel size  |
| \(P\)  | Padding             |
| \(S\)  | Stride              |
| \(O\)  | Output spatial size |

For height and width, apply the formula separately.

# 🧠 7️⃣ Complete CNN Example

Consider:

```text
Input
3 × 32 × 32
```

### Layer 1 — Convolution

```text
Kernel = 3×3
Stride = 1
Padding = 1
Filters = 16
```

Output:

```text
16 × 32 × 32
```

Why?

$$
O=\frac{32-3+2(1)}{1}+1
$$

$$
O=32
$$

And 16 filters produce 16 channels.

---

### Layer 2 — Pooling

```text
Pool size = 2×2
Stride = 2
```

Output:

```text
16 × 16 × 16
```

---

### Layer 3 — Convolution

```text
Kernel = 3×3
Stride = 1
Padding = 1
Filters = 32
```

Output:

```text
32 × 16 × 16
```

---

### Layer 4 — Pooling

```text
2×2
stride = 2
```

Output:

```text
32 × 8 × 8
```

### Complete shape flow

```text
Input
3 × 32 × 32
     ↓
Conv
16 × 32 × 32
     ↓
Pool
16 × 16 × 16
     ↓
Conv
32 × 16 × 16
     ↓
Pool
32 × 8 × 8
```

🔥 This shape-tracking ability is **very important for CNN implementation and interviews**.

# 🧩 8️⃣ All Hyperparameters Together

Consider:

```python
nn.Conv2d(
    in_channels=3,
    out_channels=32,
    kernel_size=3,
    stride=1,
    padding=1
)
```

Read it like this:

```text
Input channels
      ↓
     3
      ↓
RGB image
      ↓
Learn 32 different filters
      ↓
Each filter is 3×3
      ↓
Move 1 pixel at a time
      ↓
Add 1-pixel padding
```

In simple English:

> **"Look at 3×3 regions, move one pixel at a time, handle the border with one-pixel padding, and learn 32 different feature detectors."**

That's the mental model you want. 🧠🔥

# ⚠️ 🔟 Common Mistakes

### ❌ Mistake 1: Filters = input channels

No.

```text
in_channels → number of input channels

out_channels → number of filters/output feature maps
```

---

### ❌ Mistake 2: Kernel size determines number of filters

No.

You can have:

```text
3×3 kernel
32 filters
```

or:

```text
3×3 kernel
64 filters
```

Kernel size and filter count are separate choices.

---

### ❌ Mistake 3: Increasing filters increases image height/width

No.

Increasing:

```text
16 filters → 32 filters
```

changes:

```text
channels
```

not directly:

```text
H × W
```

---

### ❌ Mistake 4: Pooling size and stride are the same thing

Not necessarily.

```python
MaxPool2d(kernel_size=3, stride=2)
```

means:

```text
Pooling window = 3×3
Movement = 2
```

---

### ❌ Mistake 5: Stride only applies to pooling

No.

Stride can be used in:

* convolution
* pooling

# 🎯 11️⃣ Interview Essentials

Be able to answer these immediately:

### Q1. What is kernel size?

> The spatial size of the region examined by a convolution filter.

### Q2. What does stride control?

> How far the kernel moves between consecutive positions.

### Q3. Why is padding used?

> To control spatial dimensions and allow better processing of border pixels.

### Q4. What does `out_channels` represent in Conv2d?

> The number of filters and therefore the number of output feature maps.

### Q5. What happens when the number of filters increases?

> The number of output channels and learnable parameters increases.

### Q6. What happens when stride increases?

> The spatial output generally becomes smaller.

### Q7. What is a common CNN kernel size?

> **3×3**.

### Q8. What is a common pooling configuration?

> **2×2 pooling with stride 2**.

## 1. Basic CNN Architecture 🔥🔥🔥

A basic CNN follows:

```text
Input
  ↓
Convolution
  ↓
Activation
  ↓
Pooling
  ↓
Convolution
  ↓
Activation
  ↓
Pooling
  ↓
Flatten
  ↓
Fully Connected
  ↓
Output
```

### Two major stages

```text
FEATURE LEARNING
Conv → Activation → Pool
Conv → Activation → Pool
        ↓
CLASSIFICATION
Flatten → Fully Connected → Output
```

### 🧠 Core idea

> **CNN = Feature Extraction + Classification**

# 2. Input

Suppose the input is an RGB image:

```text
224 × 224
```

In PyTorch:

$$
\boxed{3\times224\times224}
$$

where:

```text
3   → Channels (RGB)
224 → Height
224 → Width
```

The image initially consists only of pixel values.

# 3. Convolution 🔥🔥🔥

Example:

```python
nn.Conv2d(
    in_channels=3,
    out_channels=16,
    kernel_size=3,
    padding=1
)
```

Input:

$$
3\times224\times224
$$

Output:

$$
16\times224\times224
$$

Why 16?

Because the layer contains **16 filters**, producing 16 output feature maps.

```text
3 × 224 × 224
       ↓
    Conv2D
       ↓
16 × 224 × 224
```

### What does convolution learn?

Early CNN filters can learn responses to patterns such as:

* Edges
* Lines
* Corners
* Simple textures

These patterns are learned from training data.

# 4. Activation 🔥🔥

A common activation is **ReLU**:

$$
\boxed{ReLU(x)=\max(0,x)}
$$

Example:

```text
Before ReLU:

-2   3
 5  -1

After ReLU:

0   3
5   0
```

### Why?

Activation introduces **non-linearity**, allowing the network to learn complex relationships.

Typical block:

```text
Convolution
     ↓
ReLU
```

---

# 5. Pooling 🔥🔥

Pooling reduces spatial dimensions.

Example:

```text
16 × 224 × 224
        ↓
     MaxPool
        ↓
16 × 112 × 112
```

Notice:

```text
Channels: 16 → 16
Height: 224 → 112
Width: 224 → 112
```

### Key rule

> **Pooling normally reduces H and W while keeping the number of channels unchanged.**

---

# 6. Second Convolution

Now another convolution operates on the learned feature maps.

Example:

```python
nn.Conv2d(
    in_channels=16,
    out_channels=32,
    kernel_size=3,
    padding=1
)
```

Input:

$$
16\times112\times112
$$

Output:

$$
32\times112\times112
$$

So:

```text
16 × 112 × 112
        ↓
      Conv
        ↓
32 × 112 × 112
```

Now the network can learn more complex combinations of the earlier features.

---

# 7. Second Activation + Pooling

After ReLU:

```text
32 × 112 × 112
```

Then:

```text
2×2 MaxPool, stride 2
```

gives:

$$
\boxed{32\times56\times56}
$$

Therefore:

```text
32 × 112 × 112
        ↓
      Pool
        ↓
32 × 56 × 56
```

---

# 8. The Important CNN Pattern 🔥🔥🔥

As we move deeper:

### Spatial dimensions generally decrease

```text
224 × 224
     ↓
112 × 112
     ↓
56 × 56
```

### Number of feature channels often increases

```text
3
↓
16
↓
32
```

So remember:

```text
        Deeper CNN
             ↓
Spatial size   ↓
Feature depth ↑
```

This is a **common architectural pattern**, not an absolute rule for every CNN.

# 10. Flatten 🔥🔥

Suppose the final convolutional feature maps are:

$$
32\times56\times56
$$

Flatten converts them into a vector.

$$
32\times56\times56=100352
$$

Therefore:

```text
32 × 56 × 56
       ↓
    Flatten
       ↓
   100352
```

### Important

Flatten **does not learn anything**.

It only changes the representation:

```text
3D tensor
   ↓
1D vector
```

---

# 11. Fully Connected Layer 🔥🔥🔥

Now the flattened vector enters a dense layer.

Example:

```python
nn.Linear(100352, 128)
```

Flow:

```text
100352 features
      ↓
Fully Connected
      ↓
128 neurons
```

The fully connected layer combines the learned features for the classification task.

---

# 12. Output Layer

Suppose there are 3 classes:

```text
Cat
Dog
Horse
```

Then:

```python
nn.Linear(128, 3)
```

Output:

$$
[2.4,\;0.7,\;-1.2]
$$

These are **logits**.

The classification system then uses the appropriate output/loss setup to determine the predicted class.

# 14. Feature Extraction vs Classification 🔥🔥🔥

## Feature Extraction

```text
Input
 ↓
Convolution
 ↓
Activation
 ↓
Pooling
 ↓
Convolution
 ↓
Activation
 ↓
Pooling
```

Purpose:

> **Learn useful representations from the image.**

---

## Classification

```text
Flatten
   ↓
Fully Connected
   ↓
Output
```

Purpose:

> **Use the learned representation to make the prediction.**

### Mental model

```text
IMAGE
  ↓
┌─────────────────────────┐
│   FEATURE EXTRACTION    │
│ Conv → ReLU → Pool      │
│ Conv → ReLU → Pool      │
└─────────────────────────┘
  ↓
Learned representation
  ↓
┌─────────────────────────┐
│      CLASSIFICATION     │
│ Flatten → FC → Output   │
└─────────────────────────┘
  ↓
Prediction
```

# 15. Basic CNN PyTorch Skeleton

```python
model = nn.Sequential(
    nn.Conv2d(3, 16, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),

    nn.Conv2d(16, 32, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),

    nn.Flatten(),

    nn.Linear(32 * 56 * 56, 128),
    nn.ReLU(),

    nn.Linear(128, 3)
)
```

For input:

```text
3 × 224 × 224
```

the shape flow is:

```text
3×224×224
    ↓
16×224×224
    ↓
16×112×112
    ↓
32×112×112
    ↓
32×56×56
    ↓
100352
    ↓
128
    ↓
3
```

🔥 At this stage, **understand the data flow rather than memorizing the code**.

# 16. Modern CNN Note ⚠️

The basic syllabus architecture uses:

```text
Pooling
 ↓
Flatten
 ↓
Fully Connected
```

But modern CNNs may instead use:

```text
Feature maps
     ↓
Global Average Pooling
     ↓
Feature vector
     ↓
Linear
     ↓
Output
```

You already learned GAP in **Module 16.4**.

So don't assume:

> Every CNN must use Flatten + a large fully connected layer.

The **basic architecture** is what we're learning here.

# 17. Layer-by-Layer Cheat Sheet

| Layer           | Main purpose                       |              Learnable? | Typical effect        |
| --------------- | ---------------------------------- | ----------------------: | --------------------- |
| Input           | Image data                         |                       ❌ | —                     |
| Convolution     | Feature extraction                 |                       ✅ | Channels can change   |
| Activation      | Non-linearity                      |               Usually ❌ | Same shape            |
| Pooling         | Downsampling                       |                       ❌ | H/W decrease          |
| Flatten         | Reshape                            |                       ❌ | Tensor → vector       |
| Fully Connected | Feature combination/classification |                       ✅ | Vector transformation |
| Output          | Class scores                       | ✅ in final linear layer | Number of classes     |

# 18. Common Mistakes ⚠️

### ❌ Convolution = classification

No.

Convolution primarily performs **feature extraction**.

### ❌ Pooling learns features

No.

Standard pooling has no learnable parameters.

### ❌ Flatten learns something

No.

Flatten only reshapes the data.

### ❌ Pooling changes channels

Normally no.

```text
32 × 28 × 28
      ↓ Pool
32 × 14 × 14
```

### ❌ Every CNN must have Flatten

No.

Global Average Pooling can replace a large flatten + dense classification head.

# 🎯 Interview Quick Questions

**What is a CNN?**

> A neural network architecture designed to learn spatial features from structured data such as images using convolutional operations.

**Why do we use convolution?**

> To learn local spatial patterns and construct feature maps.

**Why activation after convolution?**

> To introduce non-linearity.

**Why pooling?**

> To reduce spatial dimensions and computation while retaining useful information.

**What does Flatten do?**

> Converts a multidimensional feature tensor into a one-dimensional vector.

**Does Flatten have parameters?**

> No.

**Why do channels often increase as CNN depth increases?**

> Deeper layers learn increasingly diverse and complex feature representations.

**What is the basic CNN pipeline?**

$$
\boxed{
Input\rightarrow Conv\rightarrow Activation\rightarrow Pool
\rightarrow Conv\rightarrow Activation\rightarrow Pool
\rightarrow Flatten\rightarrow FC\rightarrow Output
}
$$

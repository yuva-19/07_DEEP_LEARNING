# 1. Feature Extraction 🔥🔥🔥

**Feature extraction** is the process of transforming raw image pixels into useful learned representations.

```text
Raw Image
   ↓
Convolution
   ↓
Feature Maps
   ↓
Useful Visual Features
```

CNN filters learn to detect patterns such as:

* Edges
* Lines
* Corners
* Textures
* Shapes
* Object parts

### Example

```text
Pixels
  ↓
Edges
  ↓
Textures / shapes
  ↓
Object parts
  ↓
Object representation
```

### Key idea

> **A convolution filter acts like a learned feature detector.**

A filter scans different regions of an image and produces a **feature map** indicating where that learned pattern is strongly present.

---

# 2. Feature Map

A **feature map** is the output produced by applying a convolution filter to an input.

```text
Input Image
     ↓
 Filter 1
     ↓
 Feature Map 1

Input Image
     ↓
 Filter 2
     ↓
 Feature Map 2
```

If:

```text
out_channels = 32
```

then the convolution produces:

$$
\boxed{32\text{ feature maps}}
$$

### Remember

```text
Filter
   ↓
detects a pattern

Feature Map
   ↓
shows where that pattern occurs
```

The network combines simpler patterns:

```text
Edges + corners + textures
             ↓
        Shapes / patterns
```

# 4. Parameter Sharing 🔥🔥🔥

One of the fundamental advantages of CNNs is **parameter sharing**.

A convolution filter is learned once and then reused across different spatial positions.

```text
       Same Filter
           ↓
 ┌─────────┼─────────┐
 ↓         ↓         ↓
Region 1  Region 2  Region 3
```

The filter does **not** learn separate weights for every location.

# 5. Why Parameter Sharing Matters

Suppose:

```text
Input channels = 3
Kernel = 3×3
Output channels = 32
```

Number of parameters:

$$
3\times3\times3\times32+32
$$

$$
=864+32
$$

$$
\boxed{896}
$$

Notice that **image height and width do not appear in this formula**.

Whether the spatial input is:

```text
32 × 32
```

or:

```text
224 × 224
```

the convolution still has the same number of parameters if the kernel/channel configuration is unchanged.

### Why?

Because the same weights are reused across the image.

# 6. Translation Equivariance vs Translation Invariance 🔥🔥🔥

This distinction is very important.

## Translation Equivariance

If an object moves in the input, its feature response can move correspondingly.

Conceptually:

```text
Input:
Object on left
      ↓
Feature:
Response on left

Input:
Object on right
      ↓
Feature:
Response on right
```

Mathematically:

$$
f(T(x))\approx T(f(x))
$$

where \(T\) represents a translation.

### Simple meaning

> **Shift the input → the feature response shifts.**

This is a property associated with convolution.

---

# 7. Translation Invariance

For image classification, we often want the final prediction to remain similar even if an object moves within the image.

Example:

```text
Cat on left
    ↓
  Cat

Cat on right
    ↓
  Cat
```

Conceptually:

$$
f(T(x))\approx f(x)
$$

### Important ⚠️

Don't say:

> "CNNs are perfectly translation invariant."

That's an oversimplification.

Standard CNNs have translation-related properties, but exact invariance is not guaranteed for arbitrary translations.

---

# 8. Role of Pooling

Pooling/downsampling can make representations less sensitive to small spatial changes.

Example:

```text
Feature map
224 × 224
     ↓
Pooling
     ↓
112 × 112
```

Small positional differences can become less important after downsampling.

So the rough picture is:

```text
Convolution
     ↓
Translation equivariance
     ↓
Pooling / downsampling
     ↓
Greater tolerance to small shifts
     ↓
Classification
     ↓
More position-insensitive prediction
```

### 🔑 Remember

> **Convolution → equivariance; pooling/downsampling can contribute to translation tolerance/invariance.**

---

# 9. Receptive Field 🔥🔥🔥

The **receptive field** is:

> **The region of the original input that can influence a particular neuron's activation.**

Imagine one neuron in a feature map.

It may only depend on a small region of the original image.

---

# 10. Receptive Field of One 3×3 Convolution

Consider:

```text
Input
  ↓
3×3 Conv
```

Each output neuron receives information from:

```text
3 × 3
```

of the input.

Therefore:

$$
\boxed{RF=3\times3}
$$

---

# 11. Receptive Field of Two 3×3 Convolutions 🔥

Consider:

```text
Input
 ↓
3×3 Conv
 ↓
3×3 Conv
```

The second-layer neuron sees a 3×3 region of the first feature map.

But each first-layer neuron already sees a 3×3 region of the original image.

Therefore the effective receptive field becomes:

$$
\boxed{5\times5}
$$

```text
3×3 Conv
    ↓
3×3 Conv
    ↓
5×5 effective RF
```

# 12. Receptive Field of Three 3×3 Convolutions

```text
3×3
 ↓
3×3
 ↓
3×3
```

Effective receptive field:

$$
\boxed{7\times7}
$$

So:

| Number of 3×3 layers | Effective RF |
| -------------------: | -----------: |
|                    1 |          3×3 |
|                    2 |          5×5 |
|                    3 |          7×7 |

### 🔥 Key insight

> **Stacking small convolution kernels increases the effective receptive field.**

---

# 13. Receptive Field Formula

For layer \(l\):

$$
\boxed{
r_l=r_{l-1}+(k_l-1)j_{l-1}
}
$$

where:

* \(r_l\) = receptive field at layer \(l\)
* \(k_l\) = kernel size
* \(j_{l-1}\) = jump between neighboring receptive-field centers

The jump evolves as:

$$
\boxed{
j_l=j_{l-1}s_l
}
$$

where:

* \(s_l\) = stride

Initial values:

$$
r_0=1
$$

$$
j_0=1
$$

# 14. Pooling and Receptive Field

Pooling also affects how much of the original input can influence deeper neurons.

Conceptually:

```text
Small local region
       ↓
Larger effective region
       ↓
Even larger region
       ↓
Object-level context
```

Therefore, as a CNN becomes deeper, neurons can incorporate information from increasingly larger portions of the original image.

# 15. Connecting All Five Concepts 🔥🔥🔥

Suppose the input is a **car image**.

### Step 1 — Feature Extraction

```text
Pixels
 ↓
Convolution
 ↓
Edges
```

### Step 2 — Hierarchical Features

```text
Edges
 ↓
Shapes
 ↓
Wheel-like structures
 ↓
Car parts
```

### Step 3 — Parameter Sharing

The same learned edge detector can find an edge:

```text
Left
Center
Right
```

### Step 4 — Translation Tolerance

If the feature moves spatially, convolution can still detect it, while pooling/downsampling can make the representation less sensitive to small shifts.

### Step 5 — Receptive Field

Deeper neurons can combine information from larger image regions:

```text
Small region
     ↓
Larger region
     ↓
Object part
     ↓
Larger object context
```

# ⚠️ COMMON MISTAKES

### ❌ Filter = Feature Map

Incorrect.

```text
Filter
 ↓
applied to input
 ↓
Feature Map
```

### ❌ Parameter sharing means all CNN layers share weights

Incorrect.

Parameter sharing occurs across **spatial positions within a convolutional layer**.

### ❌ CNN = perfectly translation invariant

Incorrect.

CNNs have translation-equivariant operations, while pooling and later processing can provide some translation tolerance.

### ❌ Receptive field = kernel size always

Incorrect.

A single 3×3 layer has a 3×3 receptive field, but stacked layers can produce much larger effective receptive fields.

### ❌ More channels automatically means larger receptive field

Not necessarily.

**Channels** represent feature depth; **receptive field** describes the spatial region influencing an activation.

```text
FEATURE EXTRACTION
→ Learn useful visual patterns

HIERARCHICAL FEATURES
→ Simple → complex representations

PARAMETER SHARING
→ Same filter reused across spatial positions

TRANSLATION EQUIVARIANCE
→ Shift input → feature response shifts correspondingly

TRANSLATION INVARIANCE
→ Final representation/prediction can be less sensitive
  to object position

RECEPTIVE FIELD
→ Input region influencing a neuron
```

> **CNNs learn features, build them hierarchically, reuse the same filters everywhere, tolerate some spatial shifts, and progressively increase receptive fields.** 🔥

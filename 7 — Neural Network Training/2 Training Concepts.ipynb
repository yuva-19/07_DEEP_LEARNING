# 1. 🔥🔥🔥 Dataset Splitting

A dataset is commonly divided into:

```text
Dataset
   │
   ├── Training Set
   │
   ├── Validation Set
   │
   └── Test Set
```

Each has a different purpose.

| Dataset        | Purpose                               | Parameter updates? |
| -------------- | ------------------------------------- | ------------------ |
| **Training**   | Learn model parameters                | ✅ Yes              |
| **Validation** | Monitor/tune model during development | ❌ No               |
| **Test**       | Final evaluation                      | ❌ No               |

### 🧠 Easy memory

> **Training → Learn**
> **Validation → Monitor / Tune**
> **Test → Final evaluation**

---

# 2. 🔥🔥🔥 Training Set

The **training set** contains examples used to learn the neural network's parameters.

The model performs:

$$
X_{train}\rightarrow\hat{Y}_{train}
$$

Then calculates training loss:

$$
L_{train}=L(Y_{train},\hat{Y}_{train})
$$

Then:

$$
L_{train}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradients}
\rightarrow
\text{Parameter Update}
$$

Therefore:

> **Training data directly participates in learning the weights and biases.**

---

# 3. 🔥🔥🔥 Validation Set

The **validation set** is held-out data used during model development.

It helps us:

* monitor generalization
* compare model configurations
* tune hyperparameters
* detect overfitting
* decide when training should stop

Conceptually:

```text
Training data
     ↓
Learn parameters
     ↓
Model

Validation data
     ↓
Evaluate model
     ↓
Validation loss / metrics
```

The validation set does **not normally cause parameter updates**.

---

# 4. 🔥🔥🔥 Test Set

The **test set** is held out for final evaluation.

```text
Training → Learn
Validation → Develop / Tune
Test → Final evaluation
```

The important idea is that the test set should remain separate from the development process as much as practical.

If we repeatedly inspect test performance and make decisions based on it, it can stop being a truly independent final evaluation.

---

# 5. 🔥🔥🔥 Epoch

An **epoch** is:

> **One complete pass through the entire training dataset.**

Suppose:

$$
N=10,000
$$

training samples.

If:

$$
B=100
$$

then approximately:

$$
\frac{10,000}{100}=100
$$

iterations occur in one epoch.

```text
Epoch 1
 ├── Batch 1
 ├── Batch 2
 ├── Batch 3
 ├── ...
 └── Batch 100
```

After Batch 100:

**Epoch 1 is complete.**

---

# 6. 🔥🔥🔥 Batch Size

**Batch size** is the number of training samples processed before one parameter update.

Example:

$$
B=100
$$

means the model processes 100 samples in one batch.

```text
10,000 samples

Batch 1 → 100 samples
Batch 2 → 100 samples
Batch 3 → 100 samples
...
Batch 100 → 100 samples
```

### Why batch size matters

It affects:

* memory requirements
* computational efficiency
* number of updates per epoch
* gradient noise
* hardware utilization
* training behavior

---

# 7. 🔥🔥🔥 Iteration

An **iteration** or **training step** generally means:

> Processing one batch and performing one parameter update.

So:

```text
1 batch
   ↓
1 forward pass
   ↓
1 loss calculation
   ↓
1 backward pass
   ↓
1 optimizer update
   ↓
1 iteration
```

For:

$$
N=10,000
$$

and:

$$
B=100
$$

we get:

$$
\text{Iterations per epoch}
=
\left\lceil\frac{N}{B}\right\rceil
$$

Therefore:

$$
\text{Iterations per epoch}=100
$$

---

# 8. 🔥🔥🔥 Epoch vs Batch vs Iteration

These are commonly confused.

| Term                 | Meaning                                |
| -------------------- | -------------------------------------- |
| **Batch**            | Group of samples                       |
| **Iteration / Step** | One batch processed + parameter update |
| **Epoch**            | Complete pass through training data    |

### Example

Suppose:

$$
N=10,000
$$

$$
B=100
$$

$$
E=20
$$

Then:

$$
\text{Iterations per epoch}
=
\frac{10,000}{100}
=
100
$$

Total iterations:

$$
100\times20=2,000
$$

So:

```text
1 batch       = 100 samples
1 iteration   = 1 batch + 1 update
100 iterations = 1 epoch
20 epochs     = 2,000 updates
```

---

# 9. 🔥🔥 Training Loss

Training loss measures how well the model performs on the **training data**.

For a dataset containing $N$ samples:

$$
L_{train}
=
\frac{1}{N}
\sum_{i=1}^{N}
L(y_i,\hat{y}_i)
$$

The exact loss function depends on the task.

Examples:

* Regression → MSE, MAE, Huber
* Binary classification → BCE
* Multiclass classification → Cross-Entropy

### Typical behavior

```text
Epoch       Training Loss

1              2.50
2              1.80
3              1.20
4              0.80
5              0.50
```

The model is learning from the training data, so training loss often decreases.

---

# 10. 🔥🔥 Validation Loss

Validation loss measures the model's loss on the **validation set**.

Conceptually:

$$
L_{val}
=
\frac{1}{N_{val}}
\sum_{i=1}^{N_{val}}
L(y_i,\hat{y}_i)
$$

Unlike training loss, it is **not used to directly update the parameters**.

It helps answer:

> "Is the model learning patterns that generalize beyond the training data?"

---

# 11. 🔥🔥🔥 Training Loss vs Validation Loss

This is one of the most important ideas in this block.

Suppose:

| Epoch | Training Loss | Validation Loss |
| ----: | ------------: | --------------: |
|     1 |          2.50 |            2.60 |
|     2 |          1.80 |            1.90 |
|     3 |          1.20 |            1.30 |
|     4 |          0.80 |            0.95 |
|     5 |          0.50 |            0.90 |
|     6 |          0.30 |            1.10 |
|     7 |          0.20 |            1.40 |

Notice:

```text
Training loss
2.5 → 1.8 → 1.2 → 0.8 → 0.5 → 0.3 → 0.2
                                  ↓
                              keeps falling

Validation loss
2.6 → 1.9 → 1.3 → 0.95 → 0.90 → 1.10 → 1.40
                                  ↓
                              starts rising
```

This pattern is a common indication of **overfitting**.

The model is continuing to fit the training data while validation performance gets worse.

# 13. 🔥🔥 Complete Training Timeline

Suppose:

* Training samples = 10,000
* Batch size = 100
* Epochs = 5

Then:

$$
\text{Iterations per epoch}
=
\left\lceil\frac{10,000}{100}\right\rceil
=
100
$$

Total training iterations:

$$
100\times5=500
$$

The process is:

```text
Epoch 1
 ├─ Batch 1 → Update
 ├─ Batch 2 → Update
 ├─ ...
 └─ Batch 100 → Update
        ↓
    Validation
        ↓
Epoch 2
 ├─ Batch 1 → Update
 ├─ ...
 └─ Batch 100 → Update
        ↓
    Validation
        ↓
       ...
```

---

# 14. 🔥🔥 One Iteration in Detail

One iteration looks like:

```text
Current Batch
     ↓
Forward Pass
     ↓
Prediction
     ↓
Calculate Loss
     ↓
Backpropagation
     ↓
Calculate Gradients
     ↓
Optimizer
     ↓
Parameter Update
```

Then the next batch begins.

---

# 15. 🔥🔥 PyTorch Mapping

A typical training loop:

```python
for epoch in range(num_epochs):

    for X_batch, y_batch in dataloader:

        optimizer.zero_grad()

        output = model(X_batch)

        loss = loss_function(output, y_batch)

        loss.backward()

        optimizer.step()
```

### Mapping

| Code               | Concept                     |
| ------------------ | --------------------------- |
| `num_epochs`       | Number of epochs            |
| `X_batch`          | Current batch               |
| `y_batch`          | Target batch                |
| `model(X_batch)`   | Forward pass                |
| `loss`             | Training loss for the batch |
| `loss.backward()`  | Backpropagation             |
| `optimizer.step()` | Parameter update            |
| Inner loop         | Iterations                  |

---

# 16. 🔥🔥 Training vs Validation During Code

Training:

```python
model.train()

for X_batch, y_batch in train_loader:
    optimizer.zero_grad()

    output = model(X_batch)
    loss = loss_function(output, y_batch)

    loss.backward()
    optimizer.step()
```

Validation conceptually:

```python
model.eval()

with torch.no_grad():
    for X_batch, y_batch in val_loader:
        output = model(X_batch)
        loss = loss_function(output, y_batch)
```

Notice that validation has:

```python
torch.no_grad()
```

and does not contain:

```python
loss.backward()
optimizer.step()
```

The exact evaluation code can vary, but the conceptual distinction is important:

> **Training updates parameters; validation evaluates them.**

# 17. 🔥🔥 Important Formula Collection

### Iterations per epoch

$$
\text{Iterations per epoch}
=
\left\lceil\frac{N}{B}\right\rceil
$$

where:

* $N$ = number of training samples
* $B$ = batch size

### Total iterations

$$
\text{Total iterations}
=
\text{Epochs}
\times
\text{Iterations per epoch}
$$

### Training loss

$$
L_{train}
=
\frac{1}{N}
\sum_{i=1}^{N}
L(y_i,\hat y_i)
$$

### Validation loss

$$
L_{val}
=
\frac{1}{N_{val}}
\sum_{i=1}^{N_{val}}
L(y_i,\hat y_i)
$$

# 18. ⚠️ Common Mistakes

### ❌ Batch = Epoch

No.

A batch is only a subset.

An epoch covers the complete training dataset.

---

### ❌ One epoch = one update

No.

One epoch can contain many updates.

$$
\text{Updates per epoch}\approx\frac{N}{B}
$$

---

### ❌ Validation data trains the model

No.

Validation data is used for evaluation/development.

---

### ❌ Test data should guide every training decision

No.

The test set should be reserved for final evaluation as much as possible.

---

### ❌ Training loss alone tells you whether the model is good

No.

You should also examine validation performance and task-appropriate evaluation metrics.

---

### ❌ Lower training loss always means better generalization

No.

Training loss can continue decreasing while validation loss increases.

That is a classic overfitting pattern.

# 🧠 Final Cheat Sheet

```text
TRAINING SET
→ Model learns weights & biases

VALIDATION SET
→ Monitor / tune the model

TEST SET
→ Final evaluation

BATCH
→ Group of samples

ITERATION
→ One batch processed + parameter update

EPOCH
→ Entire training dataset processed once

TRAINING LOSS
→ Error on training data

VALIDATION LOSS
→ Error on validation data
```

### ⭐ The most important relationship

$$
\boxed{
\text{Epoch}
=
\text{Many Iterations}
=
\text{Many Batches}
}
$$

More precisely:

$$
\text{Iterations per epoch}
=
\left\lceil\frac{N}{B}\right\rceil
$$

And the most important practical signal:

```text
Training loss ↓
Validation loss ↓
        → learning/generalization improving

Training loss ↓
Validation loss ↑
        → possible overfitting
```

## 🚀 One-minute revision

> **Training set** teaches the model.
> **Validation set** helps us monitor and tune it.
> **Test set** gives the final evaluation.
> **Batch size** tells how many samples are processed per update.
> **Iteration** is one update step.
> **Epoch** is one complete pass through training data.
> **Training loss** measures training performance.
> **Validation loss** measures performance on held-out validation data.

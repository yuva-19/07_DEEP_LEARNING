> **Once I have a pretrained model, how much of it should I allow to learn on my new dataset?**

There are two major strategies:

```text
                         PRETRAINED MODEL
                                │
                 ┌──────────────┴──────────────┐
                 ▼                             ▼
        FEATURE EXTRACTION              FINE-TUNING
                 │                             │
          Freeze backbone              Unfreeze some/all
          Train new classifier          Train with smaller LR
```

# 2. 🔒 What Does Freezing Mean?

Suppose a pretrained parameter is:

$$
W
$$

Normally:

$$
W_{new}
=
W_{old}
-
\eta\frac{\partial L}{\partial W}
$$

For a frozen parameter:

$$
W_{new}=W_{old}
$$

So:

```text
requires_grad = False
        ↓
parameter isn't updated
```

### ⚠️ Frozen does NOT mean removed.

The backbone still performs:

```text
Image
 ↓
Frozen CNN
 ↓
Features
```

It simply doesn't learn new parameter values.

# 3. 🧮 Feature Extraction — Numerical Example

Suppose:

```text
Pretrained backbone = 20,000,000 parameters
New classifier       =     10,000 parameters
```

### Total model parameters

$$
20,000,000+10,000
=
20,010,000
$$

But during feature extraction:

```text
Backbone
20,000,000 → 🔒

Classifier
10,000 → 🔓
```

Therefore:

$$
\boxed{\text{Trainable parameters}=10,000}
$$

### Important:

$$
\boxed{\text{Total parameters} \neq \text{Trainable parameters}}
$$

This distinction is very important in practical deep learning.

# 4. 🎯 Why Feature Extraction?

Suppose your target dataset contains only:

```text
1,000 images
```

A pretrained CNN may already know useful visual representations:

```text
Edges
 ↓
Textures
 ↓
Shapes
 ↓
Object parts
```

Instead of relearning these:

```text
Pretrained features 🔒
        ↓
New classifier 🔓
```

This gives a relatively inexpensive baseline.

### Typical advantages

* Lower number of trainable parameters
* Lower training cost
* Faster experiments
* Useful when target data is limited
* Less adaptation capacity than full fine-tuning

# 5. 🔥 Fine-Tuning

### Definition

> **Start with pretrained weights and continue training some or all of the pretrained layers on the target dataset.**

Example:

```text
Input
 ↓
Early layers       🔒
 ↓
Middle layers      🔒
 ↓
Later layers       🔓
 ↓
New classifier     🔓
 ↓
Output
```

This is **partial fine-tuning**.

You can also fine-tune the whole model:

```text
Input
 ↓
Early layers       🔓
 ↓
Middle layers      🔓
 ↓
Later layers       🔓
 ↓
Classifier         🔓
 ↓
Output
```

---

# 6. 🧠 Why Fine-Tune?

A pretrained model contains useful knowledge, but the target task may require different representations.

Fine-tuning allows:

```text
Pretrained representation
          ↓
      adaptation
          ↓
Target-specific representation
```

Mathematically, we start from:

$$
W_0=W_{\text{pretrained}}
$$

rather than:

$$
W_0=W_{\text{random}}
$$

Then gradient descent adjusts the selected parameters.

---

# 7. 🔥 Smaller Learning Rate

This is one of the most important points in this section.

During fine-tuning, pretrained weights already contain useful information.

Therefore, we generally don't want to make huge changes to them.

```text
Pretrained weight
       ●
       │
       └──→ small adjustment
```

rather than:

```text
Pretrained weight
       ●
       │
       └────────────────→ huge adjustment
```

So fine-tuning commonly uses a **smaller learning rate** for pretrained layers.

For example:

$$
\eta=10^{-4}
$$

instead of:

$$
\eta=10^{-3}
$$

The exact values are **not fixed rules**; they are hyperparameters chosen experimentally.

---

# 8. 🧠 Why Usually Adapt Later Layers?

CNNs learn hierarchical representations.

```text
Early layers
    ↓
Edges
Corners
Simple textures

Middle layers
    ↓
Patterns
Shapes
Complex textures

Later layers
    ↓
Object parts
Higher-level representations
```

Early features can be relatively generic.

Later features tend to be more specialized to the original task.

Therefore, a common strategy is:

```text
Early layers → 🔒
Later layers → 🔓
Classifier   → 🔓
```

# 9. 🌍 Source–Target Similarity

How much should you fine-tune?

One important factor is how similar the source and target problems are.

### Similar domains

```text
Source:
Natural photographs

Target:
Natural photographs
```

There may be substantial reusable representation.

### More different domains

```text
Source:
Natural photographs

Target:
Specialized scientific imagery
```

The target may require more adaptation.

Think about:

```text
Target dataset size
+
Source-target similarity
+
Available compute
+
Validation behavior
```

# 10. 🔥 Feature Extraction vs Fine-Tuning

|                      | Feature Extraction | Fine-Tuning     |
| -------------------- | ------------------ | --------------- |
| Pretrained model     | ✅                  | ✅               |
| Backbone             | 🔒 Frozen          | Some/all 🔓     |
| New classifier       | 🔓                 | 🔓              |
| Backbone LR          | Effectively 0      | Usually smaller |
| Trainable parameters | Fewer              | More            |
| Training cost        | Lower              | Higher          |
| Adaptation           | Limited            | Greater         |
| Flexibility          | Lower              | Greater         |

### Memory trick 🧠

> **Feature Extraction = REUSE**

> **Fine-Tuning = REUSE + ADAPT**

# 12. 💻 PyTorch — Feature Extraction

```python
import torch.nn as nn
from torchvision.models import resnet50, ResNet50_Weights

model = resnet50(
    weights=ResNet50_Weights.DEFAULT
)

# Freeze pretrained parameters
for param in model.parameters():
    param.requires_grad = False

# Replace original classifier
model.fc = nn.Linear(
    model.fc.in_features,
    num_classes
)
```

Result:

```text
ResNet backbone → 🔒
New FC          → 🔓
```

---

# 13. 💻 PyTorch — Partial Fine-Tuning

```python
for param in model.parameters():
    param.requires_grad = False

for param in model.layer4.parameters():
    param.requires_grad = True

model.fc = nn.Linear(
    model.fc.in_features,
    num_classes
)
```

Now:

```text
layer1 → 🔒
layer2 → 🔒
layer3 → 🔒
layer4 → 🔓
fc     → 🔓
```

Optimizer:

```python
optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4
)
```

The important relationship is:

```text
requires_grad
      ↓
Which parameters can learn?
      ↓
Optimizer
      ↓
Parameter updates
```

# 14. 🔬 Gradient Flow

### Feature extraction

```text
Image
 ↓
Frozen backbone
 ↓
Features
 ↓
Classifier
 ↓
Loss
 ↓
Backward
 ↓
Classifier updates
```

### Fine-tuning

```text
Image
 ↓
Backbone
 ↓
Classifier
 ↓
Loss
 ↓
Backward
 ↓
Trainable backbone layers + classifier update
```

So the fundamental question is:

> **Which parameters are allowed to change?**

# 16. 🚨 Common Mistakes

### ❌ "Fine-tuning means training from scratch."

No.

Fine-tuning starts from:

$$
W_{\text{pretrained}}
$$

---

### ❌ "Fine-tuning always means unfreezing every layer."

No.

It can be:

```text
some layers → 🔓
```

or:

```text
all layers → 🔓
```

---

### ❌ "Frozen layers aren't used."

Wrong.

```text
Frozen ≠ removed
```

They still perform the forward pass.

---

### ❌ "More trainable layers automatically means better."

Not necessarily.

More trainable layers mean:

```text
More adaptation
+
More computation
+
More optimization complexity
+
Potentially more overfitting
```

You evaluate experimentally.

---

### ❌ "Fine-tuning uses exactly 0.0001 learning rate."

No.

The important principle is:

> **Use an appropriately smaller learning rate when adapting pretrained weights.**

The actual value depends on the model, dataset, optimizer, batch size, and experiment.

# 17. 🎯 Interview Questions

### Q1. What are the two major transfer-learning strategies?

**Feature extraction and fine-tuning.**

### Q2. What happens during feature extraction?

The pretrained backbone is frozen and a new task-specific classifier is trained.

### Q3. What happens during fine-tuning?

Selected or all pretrained layers are allowed to update on the target dataset.

### Q4. Why use a smaller learning rate during fine-tuning?

Because the pretrained weights already contain useful representations, so smaller updates can adapt them without unnecessarily disrupting the existing representation.

### Q5. Can only the last few layers be fine-tuned?

**Yes.** Partial fine-tuning is possible.

### Q6. What determines how many layers to fine-tune?

Common considerations:

* target dataset size
* source-target similarity
* compute budget
* validation behavior

```text
Your dataset
   ↓
3,000 images
   ↓
5 target classes
```

# 2. 🧱 ResNet

## Core identity

> **ResNet → Residual learning**

Its defining mechanism is the skip/residual connection:

$$
\boxed{y=F(x)+x}
$$

Instead of forcing the stacked layers to directly learn the complete mapping, the block learns \(F(x)\), while \(x\) can bypass through the shortcut.

# 3. 📱 MobileNet

## Core identity

> **MobileNet → Efficient CNN design for resource-constrained environments**

A major MobileNet idea is **depthwise separable convolution**.

Instead of:

```text
Standard convolution
      ↓
Spatial + channel mixing together
```

we factorize it into:

```text
Depthwise convolution
      ↓
Spatial filtering
      ↓
1×1 pointwise convolution
      ↓
Channel mixing
```

It reduces the computational cost compared with a standard convolution.

* embedded systems
* edge inference
* memory constraints
* compute constraints

This does **not** mean MobileNet is automatically fastest on every hardware platform. Actual latency depends on hardware and implementation.

# 5. 🧮 EfficientNet Compound Scaling

The basic scaling relationships are:

$$
d=\alpha^\phi
$$

$$
w=\beta^\phi
$$

$$
r=\gamma^\phi
$$

where:

* \(d\) = depth
* \(w\) = width
* \(r\) = input resolution
* \(\phi\) = overall scaling coefficient
* \(\alpha\) = depth scaling factor
* \(\beta\) = width scaling factor
* \(\gamma\) = resolution scaling factor

The original compound-scaling formulation uses the approximate constraint:

$$
\boxed{\alpha\beta^2\gamma^2\approx2}
$$

### Mental picture

```text
                EfficientNet
                     │
        ┌────────────┼────────────┐
        ↓            ↓            ↓
      Depth        Width      Resolution
        ↑            ↑            ↑
        └────────────┼────────────┘
                     ↓
            Coordinated scaling
```

# 13. 🔥 The Common Transfer-Learning Workflow

Despite their architectural differences, the practical workflow is very similar:

```text
                 PRETRAINED MODEL
                        ↓
                  Load weights
                        ↓
             Use correct preprocessing
                        ↓
             Replace classification head
                        ↓
               Freeze backbone
                        ↓
                Train new head
                        ↓
                   Evaluate
                        ↓
             Unfreeze selected layers
                        ↓
                   Fine-tune
                        ↓
                   Evaluate
```

The model-specific differences are mainly:

```text
Which weights?
Which preprocessing?
Where is the classifier?
Which layers should be unfrozen?
What are the deployment constraints?
```

# 14. 🚨 CRITICAL: Weight-Specific Preprocessing

This is one of the most important practical points.

You should **not assume every pretrained model expects exactly the same preprocessing**.

TorchVision explicitly states that preprocessing can vary by:

* model family
* model variant
* weight version

and that using the correct preprocessing is important. ([PyTorch Docs][1])

The recommended pattern is:

```python
weights = ResNet50_Weights.DEFAULT

preprocess = weights.transforms()
```

Then:

```python
img = preprocess(img)
```

TorchVision's weight objects expose these transforms directly. ([PyTorch Docs][1])

# 15. 🧮 Why Does Preprocessing Matter?

Imagine the pretrained model learned from:

```text
Input distribution A
       ↓
Pretraining
       ↓
Learned weights
```

Then you give it:

```text
Input distribution B
       ↓
Pretrained model
```

The model may receive inputs that don't match the distribution it was trained with.

Therefore:

$$
\boxed{\text{Pretrained weights + expected preprocessing}}
$$

should be thought of as a practical pair.

### Example

Current TorchVision's ResNet18 ImageNet weights use:

```text
Resize → 256
Center Crop → 224
Normalize:
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]
```

while EfficientNet-B0's documented weights use bicubic resizing before the 224 crop. ([PyTorch Docs][2])

So:

> **Don't blindly copy preprocessing from one model to another.**

# 16. 💻 Generic PyTorch Pattern

A basic transfer-learning template:

```python
import torch.nn as nn
from torchvision.models import (
    resnet50,
    ResNet50_Weights
)

# Load pretrained weights
weights = ResNet50_Weights.DEFAULT

model = resnet50(
    weights=weights
)

# Freeze backbone
for param in model.parameters():
    param.requires_grad = False

# Replace classifier
model.fc = nn.Linear(
    model.fc.in_features,
    num_classes
)

# Weight-specific preprocessing
preprocess = weights.transforms()
```

Then:

```text
preprocess
    ↓
Dataset
    ↓
DataLoader
    ↓
Pretrained backbone 🔒
    ↓
New classifier 🔓
    ↓
Loss
    ↓
Optimizer
```

# 19. 🧠 Parameters vs FLOPs vs Latency

Remember from Module 26:

$$
\boxed{\text{Parameters} \neq \text{FLOPs} \neq \text{Latency}}
$$

### Parameters

How many learned values the model contains.

### FLOPs

Approximate amount of arithmetic computation.

### Latency

How long the actual inference takes on a particular system.

So:

```text
5M parameters
```

doesn't automatically mean:

```text
fastest inference
```

# 22. 🎯 Interview Questions

### Q1. What is ResNet mainly known for?

> Residual/skip connections that enable residual learning.

### Q2. Why is MobileNet useful?

> It emphasizes computational efficiency and is particularly relevant to mobile, embedded, and resource-constrained applications.

### Q3. What is EfficientNet's main contribution?

> Compound scaling of network depth, width, and input resolution.

### Q4. What is ConvNeXt?

> A modern pure CNN architecture that modernizes CNN design while retaining convolution as its core spatial-mixing mechanism.

### Q5. Do all pretrained models use identical preprocessing?

> **No.** Preprocessing can differ between model families, variants, and weight versions. ([PyTorch Docs][1])

### Q6. Does fewer FLOPs guarantee lower latency?

> **No.** Hardware, kernels, memory behavior, batch size, precision, and implementation affect actual latency.

# 🎓 FINAL MASTER UNDERSTANDING

If an interviewer gives you:

> **"You have a small image dataset. How would you use a pretrained CNN?"**

Your answer should be:

> **I would start with a suitable pretrained backbone such as ResNet, MobileNet, EfficientNet, or ConvNeXt, load its pretrained weights, use the preprocessing associated with those weights, replace the original classification head with one matching my target classes, and initially freeze the backbone and train the new head. If the target task needs further adaptation, I would progressively unfreeze selected layers and fine-tune them with an appropriately smaller learning rate. I would compare the resulting models using both predictive performance and deployment metrics such as parameter count, memory, and latency.**

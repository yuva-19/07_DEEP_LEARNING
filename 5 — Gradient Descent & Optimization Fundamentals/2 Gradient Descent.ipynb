## 1. 🔥 What is Gradient Descent?

**Gradient Descent is an optimization algorithm used to minimize the objective/loss function by repeatedly updating the model parameters in the direction that reduces the loss.**

In simple words:

> **Calculate which direction increases the loss → move in the opposite direction.**

The main goal is:

$$
\min_{\theta} J(\theta)
$$

Where:

* $\theta$ = model parameters
* $J(\theta)$ = objective/cost function
* $\nabla J(\theta)$ = gradient of the objective

# 3. 🔥 Core Gradient Descent Equation

The most important equation:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla J(\theta_{\text{old}})
$$

Where:

* $\theta_{\text{old}}$ = current parameter
* $\theta_{\text{new}}$ = updated parameter
* $\eta$ = learning rate
* $\nabla J(\theta)$ = gradient

### 🧠 Memorize this:

$$
\boxed{\text{New Parameter}=\text{Old Parameter}-\text{Learning Rate}\times\text{Gradient}}
$$

The **minus sign** is extremely important.

---

# 4. 🔥 Why Negative Gradient?

Suppose:

$$
\frac{dJ}{d\theta}=+5
$$

The positive gradient means that increasing $\theta$ locally increases the objective.

So we move in the opposite direction:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}-\eta(5)
$$

Therefore, $\theta$ decreases.

If:

$$
\frac{dJ}{d\theta}=-5
$$

Then:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}-\eta(-5)
$$

which means $\theta$ increases.

### Mental model

> **Gradient points uphill. Gradient descent walks downhill.**

If $\eta$ is too small:

* Updates are tiny
* Training becomes slow
* May require many iterations

### Large learning rate

If $\eta$ is too large:

* Updates can overshoot
* Training may oscillate
* Loss may increase
* Training can become unstable

### Good learning rate

Moves toward the minimum efficiently without excessive oscillation.

# 6. 🔥 Parameter Update

Gradient descent calculates the change:

$$
\Delta\theta=-\eta\nabla J(\theta)
$$

Then:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}+\Delta\theta
$$

Combining them:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla J(\theta)
$$

---

# 7. 🧮 Simple Numerical Example

Suppose:

$$
\theta_{\text{old}}=5
$$

Gradient:

$$
\nabla J=2
$$

Learning rate:

$$
\eta=0.1
$$

Apply:

$$
\theta_{\text{new}}
=
5-(0.1)(2)
$$

$$
\theta_{\text{new}}=4.8
$$

So:

**5 → 4.8**

The parameter moved in the direction that should reduce the objective locally.

---

# 8. 🔥 Multiple Parameters

Neural networks don't have just one parameter.

Suppose:

$$
\theta=
\begin{bmatrix}
5\\
3
\end{bmatrix}
$$

and:

$$
\nabla J=
\begin{bmatrix}
4\\
2
\end{bmatrix}
$$

with:

$$
\eta=0.1
$$

Then:

$$
\theta_{\text{new}}
=
\begin{bmatrix}
5\\
3
\end{bmatrix}
-
0.1
\begin{bmatrix}
4\\
2
\end{bmatrix}
$$

Therefore:

$$
\theta_{\text{new}}
=
\begin{bmatrix}
4.6\\
2.8
\end{bmatrix}
$$

Both parameters are updated **simultaneously**.

# 9. 🧠 One-Dimensional Example

Consider:

$$
J(\theta)=\theta^2
$$

Derivative:

$$
\frac{dJ}{d\theta}=2\theta
$$

Gradient descent:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta(2\theta_{\text{old}})
$$

Suppose:

$$
\theta=5,\qquad \eta=0.1
$$

### Step 1

Gradient:

$$
2(5)=10
$$

Update:

$$
5-(0.1)(10)=4
$$

### Step 2

$$
4-(0.1)(8)=3.2
$$

### Step 3

$$
3.2-(0.1)(6.4)=2.56
$$

### Step 4

$$
2.56-(0.1)(5.12)=2.048
$$

The parameter gradually approaches:

$$
\theta=0
$$

And therefore:

$$
J(\theta)=\theta^2\rightarrow0
$$

# 10. 🔥 Gradient Descent in Neural Networks

This is where Gradient Descent becomes extremely important.

The complete training process is:

```text
Input
  ↓
Neural Network
  ↓
Prediction
  ↓
Loss
  ↓
Backpropagation
  ↓
Gradients
  ↓
Optimizer
  ↓
Updated Parameters
  ↓
Repeat
```

More specifically:

$$
X
\rightarrow
\text{Forward Pass}
\rightarrow
\hat y
\rightarrow
L
\rightarrow
\text{Backpropagation}
\rightarrow
\nabla_\theta L
\rightarrow
\text{Parameter Update}
$$

---

# 11. 🔥 Gradient Descent vs Gradient

These are **not the same thing**.

### Gradient

The gradient is a mathematical quantity:

$$
\nabla J(\theta)
$$

It tells us the direction of steepest increase of the objective.

### Gradient Descent

Gradient Descent is the **optimization algorithm** that uses the gradient:

$$
\theta
\leftarrow
\theta-\eta\nabla J(\theta)
$$

### Easy distinction

> **Gradient = information**

> **Gradient Descent = algorithm using that information**

---

# 12. 🔥 Gradient Descent vs Learning Rate

Don't confuse these either.

### Gradient

Tells:

> **Which direction and how strongly the objective changes.**

### Learning rate

Tells:

> **How big a step we take in that direction.**

So:

$$
\text{Update magnitude}
\approx
\text{Learning Rate}\times\text{Gradient magnitude}
$$

---

# 13. 🔥 Role in Deep Learning

A neural network can contain:

* thousands
* millions
* billions

of parameters.

We need to find parameter values that produce low loss.

Gradient descent provides the basic mechanism:

$$
\boxed{
\theta
\leftarrow
\theta-\eta\nabla_\theta J
}
$$

Repeated many times, the model can move toward parameter values that reduce the training objective.

# 14. 🔥 PyTorch Connection

Later, when you start implementation, you'll see something like:

```python
loss.backward()
optimizer.step()
```

Conceptually:

### `loss.backward()`

Calculates gradients:

$$
\nabla_\theta L
$$

### `optimizer.step()`

Updates parameters.

For basic SGD, conceptually:

$$
\theta
\leftarrow
\theta-\eta\nabla_\theta L
$$

So the theory you're learning now directly becomes your PyTorch training loop later. 🔥

# 15. ⚠️ Common Mistakes

### ❌ Mistake 1: Thinking gradient points toward the minimum

No.

The gradient points toward the **steepest local increase**.

Gradient descent uses:

$$
-\nabla J
$$

to move downhill.

---

### ❌ Mistake 2: Learning rate determines direction

No.

The **gradient determines direction**.

The **learning rate determines step size**.

---

### ❌ Mistake 3: Bigger learning rate is always faster

Not necessarily.

Too large can cause:

```text
minimum
  ↑
overshoot ← → overshoot
```

and potentially prevent convergence.

---

### ❌ Mistake 4: Gradient descent updates only weights

It updates **all trainable parameters**, including:

* weights
* biases
* other trainable parameters

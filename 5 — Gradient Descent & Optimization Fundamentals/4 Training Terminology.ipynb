## 🔥 1. Training Terminology at a Glance

These terms describe **how training data moves through a neural network and how often the parameters are updated**.

| Term              | Meaning                                               |
| ----------------- | ----------------------------------------------------- |
| **Epoch**         | One complete pass through the entire training dataset |
| **Batch**         | A group of training samples processed together        |
| **Iteration**     | One training cycle/update using a batch               |
| **Step**          | Usually one parameter update                          |
| **Learning Rate** | Controls how large each parameter update is           |

### 🧠 Easy memory trick

> **Epoch = Whole dataset** 📚
> **Batch = Chunk of data** 📦
> **Iteration/Step = One update** 🔄
> **Learning Rate = Update size** 📏

---

# 🔥 2. Epoch

An **epoch** is one complete pass through the entire training dataset.

Suppose:

$$
N=1000
$$

training samples.

Processing all 1000 samples once =

$$
\boxed{1\text{ epoch}}
$$

Processing them again =

$$
\boxed{2\text{ epochs}}
$$

### Example

```text
Dataset: 1000 samples

1000 samples
     ↓
Process entire dataset
     ↓
   Epoch 1
     ↓
Process entire dataset again
     ↓
   Epoch 2
```

### Important

**One epoch does NOT necessarily mean one parameter update.**

The number of updates depends on the batch size.

---

# 🔥 3. Batch

A **batch** is a group of training samples processed together.

Suppose:

$$
N=1000
$$

and:

$$
B=100
$$

Then:

```text
Batch 1 → 100 samples
Batch 2 → 100 samples
Batch 3 → 100 samples
...
Batch 10 → 100 samples
```

So:

$$
\boxed{\text{Batch size}=100}
$$

means **100 samples are processed together before an update**.

---

# 🔥 4. Iteration

An **iteration** generally refers to **one training cycle using one batch**.

Typical process:

```text
Batch
  ↓
Forward Pass
  ↓
Loss
  ↓
Backward Pass
  ↓
Parameter Update
```

That is commonly considered **one iteration**.

If:

$$
N=1000
$$

and:

$$
B=100
$$

then:

$$
\frac{1000}{100}=10
$$

So there are approximately:

$$
\boxed{10\text{ iterations per epoch}}
$$

---

# 🔥 5. Steps

In deep-learning frameworks, **step** commonly refers to one optimizer update.

For example:

```text
Batch 1 → Step 1
Batch 2 → Step 2
Batch 3 → Step 3
...
Batch 10 → Step 10
```

So:

$$
\boxed{\text{1 batch} \rightarrow \text{1 step/update}}
$$

In many contexts:

$$
\boxed{\text{Iteration}\approx\text{Step}}
$$

The exact terminology can vary between libraries and textbooks, but this distinction is sufficient for your current learning.

---

# 🔥 6. Learning Rate

The **learning rate** controls the size of the parameter update.

It is usually represented by:

$$
\eta
$$

Gradient descent:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla J(\theta)
$$

Where:

* $\theta$ = parameters
* $\eta$ = learning rate
* $\nabla J(\theta)$ = gradient

### Small learning rate

Small parameter updates:

```text
• → • → • → • → •
```

Training may be slow.

### Large learning rate

Large parameter updates:

```text
• --------→
       ←--------
```

The optimizer may overshoot the minimum or become unstable.

### Good learning rate

Moves toward a lower-loss region efficiently.

# 🔥 7. Putting Them Together

Suppose:

$$
N=1000
$$

training samples.

Batch size:

$$
B=100
$$

Then:

$$
\frac{N}{B}
=
\frac{1000}{100}
=
10
$$

Therefore, one epoch contains approximately **10 batches/steps**.

```text
Epoch 1
│
├── Batch 1 → Step 1
├── Batch 2 → Step 2
├── Batch 3 → Step 3
├── Batch 4 → Step 4
├── Batch 5 → Step 5
├── Batch 6 → Step 6
├── Batch 7 → Step 7
├── Batch 8 → Step 8
├── Batch 9 → Step 9
└── Batch 10 → Step 10
```

After Batch 10:

$$
\boxed{\text{1 epoch completed}}
$$

---

# 🔥 8. Steps per Epoch

The basic formula is:

$$
\boxed{
\text{Steps per epoch}
\approx
\frac{\text{Dataset size}}{\text{Batch size}}
}
$$

For example:

$$
N=5000
$$

$$
B=50
$$

Therefore:

$$
\frac{5000}{50}=100
$$

So:

$$
\boxed{100\text{ steps per epoch}}
$$

---

# 🔥 9. Total Training Steps

Suppose:

* Dataset = 5,000 samples
* Batch size = 50
* Epochs = 10

Steps per epoch:

$$
\frac{5000}{50}=100
$$

Total steps:

$$
100\times10=1000
$$

Therefore:

$$
\boxed{1000\text{ parameter updates}}
$$

assuming every batch results in one optimizer update.

---

# ⚠️ 10. Batch ≠ Epoch

This is one of the most common beginner mistakes.

### Batch

A **portion** of the dataset.

### Epoch

The **entire dataset once**.

```text
Dataset
├── Batch 1
├── Batch 2
├── Batch 3
└── Batch 4

All batches together
        ↓
    1 Epoch
```

---

# ⚠️ 11. Batch ≠ Step

A batch is **data**.

A step is the **update operation** performed using that batch.

Think:

> **Batch = what the model sees**

> **Step = what the optimizer does afterward**

---

# ⚠️ 12. Epoch ≠ Update

One epoch can contain many parameter updates.

For example:

$$
N=1000,\quad B=100
$$

Then:

$$
1\text{ epoch}=10\text{ steps}
$$

So one epoch results in **10 updates**, not 1.

---

# 🔥 13. PyTorch Connection

You'll soon encounter code like:

```python
for epoch in range(num_epochs):

    for X_batch, y_batch in dataloader:

        predictions = model(X_batch)

        loss = loss_fn(predictions, y_batch)

        loss.backward()

        optimizer.step()

        optimizer.zero_grad()
```

Now you can identify each concept:

```text
for epoch
     ↓
Epoch

X_batch, y_batch
     ↓
Batch

One execution of inner loop
     ↓
Iteration / Step

optimizer.step()
     ↓
Parameter update

learning_rate
     ↓
Controls update magnitude
```

The training structure is therefore:

```text
Dataset
   ↓
Batches
   ↓
Forward Pass
   ↓
Loss
   ↓
Backward Pass
   ↓
Optimizer Step
   ↓
Next Batch
   ↓
...
   ↓
End of Epoch
   ↓
Next Epoch
```

# 🧠 14. Complete Example

Suppose:

$$
N=10,000
$$

$$
B=32
$$

$$
E=5
$$

Steps per epoch:

$$
\frac{10000}{32}=312.5
$$

Because you cannot have half a batch, the actual number of batches depends on how the data loader handles the final incomplete batch. If it keeps the final partial batch, there are typically **313 steps per epoch**.

Total:

$$
313\times5=1565
$$

So approximately:

$$
\boxed{1565\text{ optimizer updates}}
$$

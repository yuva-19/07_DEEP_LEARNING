# 🧠 Module 34.2 — RNN Architecture — Notes

## 1. Basic RNN Architecture

An RNN has three main components at each time step:

* **`xₜ` → Input**
* **`hₜ` → Hidden state**
* **`yₜ` → Output**

Basic structure:

```text
xₜ → [ RNN ] → hₜ → yₜ
```

But the RNN also receives the **previous hidden state**:

```text
                 hₜ₋₁
                   │
                   ▼
xₜ ───────────► [ RNN ] ─────► hₜ ─────► yₜ
```

## 2. Input `xₜ` 📥

`xₜ` represents the **input at time step `t`**.

For text:

```text
x₁ = "I"
x₂ = "love"
x₃ = "this"
x₄ = "movie"
```

For sensor data:

```text
x₁ = temperature at t₁
x₂ = temperature at t₂
x₃ = temperature at t₃
```

So:

> **`xₜ` = current information entering the RNN.**

---

## 3. Hidden State `hₜ` 🧠

The hidden state is the RNN's **internal representation of the sequence information available up to the current time step**.

The hidden states flow forward:

```text
h₁ → h₂ → h₃ → h₄
```

At every step, the new hidden state depends on:

```text
Current input + Previous hidden state
```

For example:

```text
x₁ + h₀ → h₁
x₂ + h₁ → h₂
x₃ + h₂ → h₃
```

Think of `hₜ` as a **learned numerical representation of information carried through the sequence**.

⚠️ It is not human-like memory. It is a vector of learned numerical values.

---

# 4. Recurrent Connection 🔄

The defining feature of an RNN is the connection from the previous hidden state to the next time step.

```text
x₁ → h₁
      ↓
x₂ → h₂
      ↓
x₃ → h₃
```

The vertical connection represents:

```text
h₁ → h₂ → h₃
```

This allows information from previous time steps to influence later time steps.

---

## 5. Output `yₜ` 📤

`yₜ` represents the output produced at time step `t`.

For an RNN producing an output at every step:

```text
x₁ → h₁ → y₁
x₂ → h₂ → y₂
x₃ → h₃ → y₃
```

This is useful for tasks where we need a prediction for each element of the sequence.

Example:

```text
John   lives   in   India
 ↓      ↓      ↓     ↓
PERSON   O      O   LOCATION
```

# 6. Expanded RNN Architecture

Your original architecture:

```text
x₁ → h₁ → y₁
      ↓
x₂ → h₂ → y₂
      ↓
x₃ → h₃ → y₃
```

Can be understood more precisely as:

```text
              h₀
               │
               ▼
x₁ ─────────► [RNN] ─────► h₁ ─────► y₁
                              │
                              ▼
x₂ ─────────► [RNN] ─────► h₂ ─────► y₂
                              │
                              ▼
x₃ ─────────► [RNN] ─────► h₃ ─────► y₃
```

At each step:

```text
xₜ + hₜ₋₁
    ↓
  RNN
    ↓
hₜ + yₜ
```

# 7. RNN Mathematical Structure

Conceptually:

```text
hₜ = f(xₜ, hₜ₋₁)
```

Meaning:

> The current hidden state depends on the current input and previous hidden state.

The output is obtained from the hidden state:

```text
yₜ = g(hₜ)
```

So the complete flow is:

```text
xₜ + hₜ₋₁
      ↓
     RNN
      ↓
     hₜ
      ↓
     yₜ
```

---

# 8. Same RNN Cell Is Reused 🔁

An important architectural property:

We don't create a completely new RNN for every time step.

Instead, **the same RNN cell is repeatedly applied**.

```text
        SAME RNN CELL
             ↓
x₁ → [RNN] → h₁
x₂ → [RNN] → h₂
x₃ → [RNN] → h₃
```

This is called **parameter sharing**.

The same learned parameters are used across the sequence.

This becomes particularly important when we study the mathematics of RNNs.

---

# 9. Example — Text Sequence 📖

Consider:

> **"I love this movie"**

The RNN processes the sequence step by step:

```text
"I"       → h₁
"love"    → h₂
"this"    → h₃
"movie"   → h₄
```

The information flows:

```text
"I"
 ↓
h₁
 ↓
"love"
 ↓
h₂
 ↓
"this"
 ↓
h₃
 ↓
"movie"
 ↓
h₄
```

The later hidden states can therefore contain information influenced by earlier inputs.

---

# 10. Example — Sensor Data ⚙️

Suppose we're monitoring an EV motor:

```text
Temperature:

40°C → 42°C → 45°C → 49°C
```

The RNN processes:

```text
40°C → h₁
          ↓
42°C → h₂
          ↓
45°C → h₃
          ↓
49°C → h₄
```

Therefore, `h₄` is influenced not only by the current temperature but also by information carried through previous hidden states.

---

# 11. Output at Every Time Step vs Final Output

### Many-to-many

Output at every time step:

```text
x₁ → h₁ → y₁
x₂ → h₂ → y₂
x₃ → h₃ → y₃
```

Example:

**Sequence labeling / NER**

---

### Many-to-one

Only a final output:

```text
x₁ → h₁
      ↓
x₂ → h₂
      ↓
x₃ → h₃ → y
```

Example:

**Sentiment classification**

```text
"I love this movie"
          ↓
         RNN
          ↓
      Positive 😊
```

This connects directly to the **sequence types** from Module 33.

# 🧠 The Complete Mental Model

Remember this:

```text
             Previous information
                    hₜ₋₁
                       │
                       ▼
Current input xₜ →  [ RNN ]  → hₜ → yₜ
```

Or mathematically:

```text
       xₜ + hₜ₋₁
            ↓
           RNN
            ↓
         hₜ + yₜ
```

### 🔑 One-line definition

> **An RNN processes a sequence one time step at a time, combining the current input with the previous hidden state to produce a new hidden state and output.**

---

## 🔥 Must Remember

**`xₜ`** → current input
**`hₜ₋₁`** → previous hidden state
**`hₜ`** → current hidden state
**`yₜ`** → output

Core relationship:

```text
xₜ + hₜ₋₁ → RNN → hₜ → yₜ
```

And across the sequence:

```text
x₁ → h₁ → y₁
      ↓
x₂ → h₂ → y₂
      ↓
x₃ → h₃ → y₃
```

**The most important idea:**
👉 **The hidden state carries information forward through time.** 🧠🔄

## 1. Hidden-State Equation 🧠

The core equation of an RNN is:

$$
\boxed{h_t=f(W_{xh}x_t+W_{hh}h_{t-1}+b_h)}
$$

For a standard vanilla RNN, `f` is commonly `tanh`:

$$
\boxed{
h_t=\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

### Meaning of each term

* **`xₜ`** → current input
* **`hₜ₋₁`** → previous hidden state
* **`Wₓₕ`** → input-to-hidden weights
* **`Wₕₕ`** → hidden-to-hidden weights
* **`bₕ`** → hidden-layer bias
* **`tanh`** → activation function
* **`hₜ`** → current hidden state

The calculation is:

```text
Current input ──────► Wₓₕ ──┐
                            │
Previous hidden state ─► Wₕₕ ──► (+ bₕ) ─► tanh ─► hₜ
```

---

## 2. What Does the Hidden-State Equation Mean?

The new hidden state combines **two sources of information**:

```text
                 ┌── Current information
                 │
xₜ ──────────────┤
                 │
hₜ₋₁ ────────────┤──► New hidden state hₜ
                 │
                 └── Previous information
```

So the key idea is:

> **Current input + previous hidden state → current hidden state**

This is what allows the RNN to process sequential information.

# 3. Output Equation 📤

Once `hₜ` has been calculated, the RNN can generate an output.

The output equation is:

$$
\boxed{
y_t=g(W_{hy}h_t+b_y)
}
$$

Where:

* **`hₜ`** → current hidden state
* **`Wₕᵧ`** → hidden-to-output weights
* **`bᵧ`** → output bias
* **`g`** → output activation function
* **`yₜ`** → output

The flow is:

```text
hₜ
 │
 ▼
Wₕᵧ
 │
 ▼
+ bᵧ
 │
 ▼
g()
 │
 ▼
yₜ
```

## 4. Output Activation Depends on the Task

`g()` isn't always the same function.

For example:

### Binary classification

Often uses **sigmoid**:

$$
y_t=\sigma(W_{hy}h_t+b_y)
$$

### Multi-class classification

Often uses **softmax**:

$$
y_t=softmax(W_{hy}h_t+b_y)
$$

So remember:

> **The output layer and activation depend on the task.**

# 5. Complete RNN Mathematics 🔥

Put the two equations together.

### Hidden state

$$
\boxed{
h_t=
\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

### Output

$$
\boxed{
y_t=g(W_{hy}h_t+b_y)
}
$$

Complete flow:

```text
                 hₜ₋₁
                   │
                   ▼
xₜ ──► Wₓₕ ──►   (+)   ◄── Wₕₕ
                   │
                  bₕ
                   │
                 tanh
                   │
                   ▼
                  hₜ
                   │
                 Wₕᵧ
                   │
                  bᵧ
                   │
                  g()
                   │
                   ▼
                  yₜ
```

# 6. Parameter Sharing 🔄

This is one of the **most important ideas in RNNs**.

Suppose the RNN processes:

```text
x₁ → x₂ → x₃
```

We might draw:

```text
x₁ → [RNN] → h₁
x₂ → [RNN] → h₂
x₃ → [RNN] → h₃
```

It may look like there are three different RNNs.

**There aren't.**

The **same parameters are reused** at every time step.

```text
         SAME PARAMETERS
              ↓
x₁ ─────► [RNN] ─────► h₁
x₂ ─────► [RNN] ─────► h₂
x₃ ─────► [RNN] ─────► h₃
```

# 7. Which Parameters Are Shared?

The same parameters are used at every time step:

$$
W_{xh}
$$

$$
W_{hh}
$$

$$
W_{hy}
$$

$$
b_h
$$

$$
b_y
$$

For example:

```text
t₁ → Wₓₕ, Wₕₕ, Wₕᵧ, bₕ, bᵧ
t₂ → Wₓₕ, Wₕₕ, Wₕᵧ, bₕ, bᵧ
t₃ → Wₓₕ, Wₕₕ, Wₕᵧ, bₕ, bᵧ
```

They are **the same learned parameters**.

---

# 8. Why Parameter Sharing? 🤔

Without parameter sharing, we could have:

```text
t₁ → W₁
t₂ → W₂
t₃ → W₃
...
```

The number of parameters would grow with the sequence length.

Instead:

```text
t₁ ─┐
t₂ ─┤
t₃ ─┼──► SAME PARAMETERS
t₄ ─┤
... ─┘
```

This gives us:

### ✅ Fewer parameters

We don't need a completely new set of weights for every time step.

### ✅ Same learned transformation

The same rule is applied to every position in the sequence.

### ✅ Different sequence lengths

The same RNN cell can be repeatedly applied to sequences of different lengths.

---

# 9. Very Important: Parameters ≠ Hidden States ⚠️

Don't confuse these two.

### Parameters remain the same:

```text
Wₓₕ
Wₕₕ
Wₕᵧ
bₕ
bᵧ
```

### Hidden state changes:

```text
h₀ → h₁ → h₂ → h₃ → h₄
```

So:

```text
       SAME PARAMETERS
             ↓
h₀ ──► h₁ ──► h₂ ──► h₃
```

The **weights are shared**, while the **hidden state evolves over time**.

# 🧠 Complete Mental Model

At every time step:

```text
                 Previous hidden state
                       hₜ₋₁
                         │
                         ▼
Current input xₜ ──►  [ RNN ] ──► hₜ ──► yₜ
```

Mathematically:

$$
\boxed{
h_t=\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

then:

$$
\boxed{
y_t=g(W_{hy}h_t+b_y)
}
$$

And these parameters are reused across the entire sequence:

```text
        SAME PARAMETERS
             ↓
t₁ ──► RNN ──► h₁
t₂ ──► RNN ──► h₂
t₃ ──► RNN ──► h₃
...
tₜ ──► RNN ──► hₜ
```

# 🔥 Must Remember

### Hidden-state equation

$$
\boxed{
h_t=f(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

### Vanilla RNN

$$
\boxed{
h_t=\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

### Output equation

$$
\boxed{
y_t=g(W_{hy}h_t+b_y)
}
$$

### Parameter sharing

> **The same RNN parameters are reused at every time step.**

### ⭐ Core intuition

```text
Current input
     +
Previous hidden state
     ↓
New hidden state
     ↓
Output
```

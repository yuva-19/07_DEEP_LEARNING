## 1. Why Do RNNs Have Problems?

From the previous section:

$$
h_t=\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
$$

The hidden state is passed through time:

```text
h₁ → h₂ → h₃ → h₄ → h₅ → ... → hₜ
```

During training, **Backpropagation Through Time (BPTT)** sends gradients backward through these time steps:

```text
hₜ → hₜ₋₁ → hₜ₋₂ → ... → h₁
```

Because gradients are repeatedly multiplied through many time steps, two major problems can occur:

```text
        BPTT
          ↓
Repeated gradient multiplication
          ↓
   ┌──────┴──────┐
   ↓             ↓
Too small      Too large
   ↓             ↓
Vanishing      Exploding
gradient       gradient
```

# 2. Vanishing Gradients 📉

A **vanishing gradient** occurs when the gradient becomes extremely small as it propagates backward through many time steps.

For example:

```text
1.0
 ↓
0.5
 ↓
0.25
 ↓
0.125
 ↓
0.0625
 ↓
...
 ↓
≈ 0
```

Mathematically, repeated multiplication by values smaller than 1 can cause:

$$
a^n \rightarrow 0
$$

as `n` becomes large.

---

## Why is this a problem?

Earlier time steps receive an extremely small learning signal.

So the RNN struggles to learn relationships between events that are **far apart in the sequence**.

```text
Early information
      ↓
h₁ → h₂ → h₃ → ... → h₅₀
                           ↓
                       prediction
```

The information may exist in the sequence, but the gradient reaching the earlier time step can become almost zero.

### 🔑 Remember

> **Vanishing gradient = gradient becomes too small to effectively update earlier parts of the network.**

---

# 3. Exploding Gradients 📈

An **exploding gradient** occurs when the gradient becomes extremely large during backpropagation.

For example:

```text
1
↓
2
↓
4
↓
8
↓
16
↓
32
↓
...
```

Repeated multiplication by values greater than 1 can cause rapid growth:

$$
a^n \rightarrow \text{very large}
$$

## Why is this a problem?

Very large gradients can cause:

* Extremely large parameter updates
* Unstable training
* Numerical instability
* Loss to become unstable
* Training to diverge

Instead of:

```text
small → stable → learning
```

we can get:

```text
small → large → HUGE → unstable
```

### 🔑 Remember

> **Exploding gradient = gradient becomes excessively large during training.**

---

# 4. Gradient Clipping ✂️

A common technique for controlling **exploding gradients** is **gradient clipping**.

If the gradient exceeds a chosen limit, its magnitude is restricted.

Conceptually:

```text
Gradient = 100
        ↓
Gradient clipping
        ↓
Gradient = 10
```

So:

> **Gradient clipping limits excessively large gradients.**

⚠️ Important:

Gradient clipping primarily addresses **exploding gradients**. It does not fundamentally solve the vanishing-gradient problem.

# 5. Long-Term Dependencies 🧠⏳

A **long-term dependency** occurs when information from an earlier part of a sequence is needed much later.

For example:

```text
x₁ → x₂ → x₃ → ... → x₅₀
↑                         ↑
Important information     Prediction
```

The model needs to preserve information from `x₁` until `x₅₀`.

# 6. Language Example 📖

Consider:

> **"I grew up in France ... [many words] ... I speak fluent ____."**

The earlier information:

```text
France
```

needs to influence the later prediction:

```text
French
```

Conceptually:

```text
France
  ↓
h₁ → h₂ → h₃ → ... → h₅₀
                         ↓
                      French
```

A vanilla RNN can struggle to preserve this information over a long sequence.

---

# 7. Why Are Long-Term Dependencies Difficult?

The hidden state must carry information through many steps:

```text
h₁ → h₂ → h₃ → h₄ → ... → h₅₀
```

And during training, the gradient travels backward:

```text
h₅₀ → h₄₉ → ... → h₂ → h₁
```

If the gradient vanishes:

```text
h₅₀ → h₄₉ → ... → h₂ → h₁
                         ↓
                    tiny gradient
```

the earlier time steps receive very little learning signal.

Therefore:

$$
\boxed{\text{Vanishing gradients} \rightarrow
\text{difficulty learning long-term dependencies}}
$$

---

# 8. Relationship Between the Three Problems 🔗

These three concepts are strongly connected:

```text
                RNN
                 ↓
          Long sequence
                 ↓
                BPTT
                 ↓
       Repeated multiplication
                 ↓
        ┌────────┴────────┐
        ↓                 ↓
 Gradient too small   Gradient too large
        ↓                 ↓
   Vanishing          Exploding
   gradient           gradient
        ↓
 Difficulty learning
 long-term dependencies
```

### Vanishing gradient

$$
\boxed{\text{Gradient} \rightarrow 0}
$$

### Exploding gradient

$$
\boxed{\text{Gradient} \rightarrow \text{very large}}
$$

### Long-term dependency

$$
\boxed{\text{Earlier information is needed much later}}
$$

---

# 9. The Big Connection to LSTM & GRU 🚀

This is the **main reason** LSTM and GRU are important.

```text
Vanilla RNN
     ↓
Problems with long sequences
     ↓
Vanishing / exploding gradients
     ↓
Difficulty with long-term dependencies
     ↓
LSTM / GRU
```

LSTM introduces mechanisms such as **gates and a cell state** to improve the handling of information over long sequences.

---

# 🧠 Final Mental Model

Remember this complete chain:

```text
RNN processes sequence
        ↓
BPTT propagates gradients through time
        ↓
Repeated gradient multiplication
        ↓
 ┌───────────────┬────────────────┐
 ↓               ↓                ↓
Too small       Too large       Distant
gradient        gradient        information
 ↓               ↓                ↓
Vanishing       Exploding       Long-term
gradient        gradient        dependency
```

### 🔥 Must Remember

**Vanishing gradient:**
Gradient becomes extremely small during propagation through many time steps.

**Exploding gradient:**
Gradient becomes extremely large during propagation through many time steps.

**Long-term dependency:**
The model needs information from a distant earlier time step to make a later prediction.

**Gradient clipping:**
A technique used to limit excessively large gradients.

### ⭐ Most important connection

> **Vanishing gradients are a major reason vanilla RNNs struggle to learn long-term dependencies.**

And this naturally leads into the next major architecture:

**LSTM — Long Short-Term Memory.**

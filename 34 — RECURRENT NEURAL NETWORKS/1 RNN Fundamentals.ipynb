## 🔥 What is an RNN?

A **Recurrent Neural Network (RNN)** is a neural network designed to process **sequential data**.

The key difference from a normal feed-forward network is that an RNN can carry information from an earlier time step to a later time step.

Think:

**Current Input + Previous Information → RNN → Updated Information**

# 1️⃣ Hidden State

The **hidden state** is the central idea behind an RNN.

You can think of it as the RNN's **running memory or context**.

At time `t`, the RNN receives:

* Current input → `xₜ`
* Previous hidden state → `hₜ₋₁`

and produces:

* New hidden state → `hₜ`

Conceptually:

**Previous Hidden State + Current Input**

↓

**RNN**

↓

**New Hidden State**

```text
hₜ₋₁ ──────┐
           ↓
xₜ ─────→ RNN ─────→ hₜ
```

The new hidden state contains information derived from the current input and information carried from previous time steps.

---

## 🧠 Example

Consider:

> "I went to the bank to deposit money."

The RNN processes the sequence step-by-step:

**I → went → to → the → bank → to → deposit → money**

At every step, information is carried forward:

**I → h₁**

**went + h₁ → h₂**

**to + h₂ → h₃**

**the + h₃ → h₄**

**bank + h₄ → h₅**

So when the model reaches **"bank"**, its hidden state has been influenced by the preceding words.

### Important idea

> The hidden state is a learned representation of relevant information from the sequence history.

# 2️⃣ Recurrent Connection

Now let's understand why it's called a **Recurrent** Neural Network.

A recurrent connection allows information from the previous time step to be fed into the next time step.

Without recurrence:

```text
x₁ → Network → y₁

x₂ → Network → y₂

x₃ → Network → y₃
```

Each input is processed independently.

With recurrence:

```text
x₁ → RNN → h₁
            ↓
x₂ → RNN → h₂
            ↓
x₃ → RNN → h₃
```

The hidden state moves forward through the sequence.

# 3️⃣ Basic RNN Equation

The basic hidden-state equation is:

$$
h_t = f(W_xx_t + W_hh_{t-1} + b)
$$

Let's understand each part.

### `xₜ`

Current input.

### `hₜ₋₁`

Previous hidden state.

### `Wₓ`

Weights associated with the current input.

### `Wₕ`

Weights associated with the previous hidden state.

### `b`

Bias.

### `f`

Activation function.

### `hₜ`

Current/new hidden state.

So conceptually:

$$
\boxed{h_t=f(x_t,h_{t-1})}
$$

means:

> **The current hidden state depends on the current input and the previous hidden state.**

# 4️⃣ 🔄 Unrolling

An RNN is often represented as a loop.

### Compact representation

```text
       ┌─────────┐
xₜ ──→ │   RNN   │ ──→ hₜ
       └────┬────┘
            ↑
            │
        previous state
```

But this representation hides what happens at each time step.

So we **unroll** the RNN.

### Unrolled representation

```text
x₁ → RNN → h₁
           ↓
x₂ → RNN → h₂
           ↓
x₃ → RNN → h₃
           ↓
x₄ → RNN → h₄
```

# 5️⃣ Why Is Unrolling Important?

Unrolling lets us see the RNN **one time step at a time**.

### Time step 1

**x₁ + h₀**

↓

RNN

↓

**h₁**

### Time step 2

**x₂ + h₁**

↓

RNN

↓

**h₂**

### Time step 3

**x₃ + h₂**

↓

RNN

↓

**h₃**

### Time step 4

**x₄ + h₃**

↓

RNN

↓

**h₄**

The important thing is:

> **The same RNN cell is reused at every time step.**

So we're not creating a completely different neural network for every word or measurement.

The same set of learned parameters is applied repeatedly.

# 6️⃣ 🔢 Sequence Processing

An RNN processes a sequence **one time step at a time**.

Suppose:

**x₁ → x₂ → x₃ → x₄**

The processing looks like:

```text
x₁ + h₀ → RNN → h₁

x₂ + h₁ → RNN → h₂

x₃ + h₂ → RNN → h₃

x₄ + h₃ → RNN → h₄
```

Or visually:

```text
x₁ ──→ RNN ──→ h₁
                 ↓
x₂ ──→ RNN ──→ h₂
                 ↓
x₃ ──→ RNN ──→ h₃
                 ↓
x₄ ──→ RNN ──→ h₄
```

The information flows forward through the hidden states.

# 7️⃣ 📝 Example — Sentiment Analysis

Consider:

> **"The movie was extremely good."**

The sequence becomes:

**The → movie → was → extremely → good**

The RNN processes each token:

**The → h₁**

**movie + h₁ → h₂**

**was + h₂ → h₃**

**extremely + h₃ → h₄**

**good + h₄ → h₅**

The final representation can then be used to predict:

**❤️ Positive**

This is a **many-to-one** sequence-processing problem.

---

# 8️⃣ 🏭 Example — Motor Sensor Data

Suppose a motor produces temperature measurements:

**40 → 42 → 45 → 49°C**

The RNN processes them sequentially:

```text
40 → RNN → h₁
            ↓
42 → RNN → h₂
            ↓
45 → RNN → h₃
            ↓
49 → RNN → h₄
```

At each step, the model receives:

**Current measurement + previous hidden state**

Therefore, the RNN can maintain a representation of the sequence history while processing the current measurement.

**Processing each sequence element while carrying hidden-state information forward.**

# 🎯 Must Remember

### 🟦 Hidden State

> Carries information from previous time steps.

### 🟩 Recurrent Connection

> Passes the previous hidden state into the next time step.

### 🟨 Unrolling

> Represents the recurrent computation separately across time steps.

### 🟥 Sequence Processing

> Processes sequential inputs step-by-step while maintaining hidden-state information.

---

# 🧠 One Equation to Remember

$$
\boxed{h_t=f(W_xx_t+W_hh_{t-1}+b)}
$$

Or conceptually:

$$
\boxed{\text{Current State}=\text{Current Input}+\text{Previous State}}
$$

Not literally addition alone—the actual computation includes learned weights, bias, and an activation function.

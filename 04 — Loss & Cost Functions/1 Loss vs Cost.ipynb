# 📚 4.1 — Loss vs Cost — Revision Notes

## 🔥 1. Core Idea

A neural network needs a numerical way to measure:

> **“How wrong is my prediction?”**

That is the role of the **loss function**.

The basic flow is:

```text
Input
  ↓
Neural Network
  ↓
Prediction ŷ
  ↓
Compare with true value y
  ↓
Loss
  ↓
Aggregate losses
  ↓
Objective
  ↓
Gradient
  ↓
Parameter update
```

---

# 🔥 2. Loss Function

A **loss function** measures the error of a model's prediction, typically for an individual training example.

$$
L(y,\hat y)
$$

where:

* $y$ = true value
* $\hat y$ = predicted value
* $L$ = loss

Generally:

$$
\boxed{\text{Lower loss} \rightarrow \text{better prediction}}
$$

### Example

True value:

$$
y=10
$$

Prediction:

$$
\hat y=8
$$

Using squared error:

$$
L=(y-\hat y)^2
$$

$$
L=(10-8)^2=4
$$

So the prediction has a loss of **4**.

---

# 🔥 3. Different Tasks → Different Losses

There is no single loss function for every problem.

| Task                      | Common loss          |
| ------------------------- | -------------------- |
| Regression                | MSE, MAE, Huber      |
| Binary classification     | Binary Cross Entropy |
| Multiclass classification | Cross Entropy        |

The loss function should reflect **what kind of error the model needs to minimize**.

---

# 🔥 4. Cost Function

When we have many training examples, we need an overall measure of error.

Suppose:

$$
L_1=2
$$

$$
L_2=5
$$

$$
L_3=1
$$

$$
L_4=4
$$

A common aggregate is the mean:

$$
J=
\frac{1}{m}
\sum_{i=1}^{m}L_i
$$

For the example:

$$
J=
\frac{2+5+1+4}{4}
$$

$$
\boxed{J=3}
$$

So:

> **Loss → individual prediction error**

> **Cost → aggregate/average prediction error**

---

# 🧠 5. Loss vs Cost

|          | Loss               | Cost                    |
| -------- | ------------------ | ----------------------- |
| Scope    | Individual example | Collection of examples  |
| Measures | Prediction error   | Aggregate error         |
| Symbol   | $L_i$              | $J$                     |
| Example  | $L(y_i,\hat y_i)$  | $\frac{1}{m}\sum_i L_i$ |

### Memory trick 🧠

> **Loss = one**
> **Cost = collection**

### ⚠️ Terminology

In modern Deep Learning, **loss** and **cost** are sometimes used interchangeably.

For example, people may say:

> “Calculate the batch loss”

when they actually mean the average loss over a batch.

So understand the **conceptual distinction**, but don't get stuck arguing about terminology.

---

# 🔥 6. Objective Function

The **objective function** is the quantity that the optimization algorithm tries to optimize.

Usually, training is formulated as:

$$
\boxed{\min_{\theta}J(\theta)}
$$

where:

* $\theta$ = model parameters
* $J(\theta)$ = objective

In simple supervised learning:

$$
\text{Objective}\approx\text{aggregate prediction loss}
$$

---

# 🔥 7. Objective Can Include More Than Prediction Loss

The objective can contain additional terms.

For example:

$$
J(\theta)
=
L(\theta)+\lambda R(\theta)
$$

where:

* $L(\theta)$ = prediction loss
* $R(\theta)$ = regularization term
* $\lambda$ = regularization strength

So the optimizer minimizes the **entire objective**, not necessarily just the prediction error.

Conceptually:

```text
Prediction Loss
       +
Regularization
       ↓
   Objective
```

---

# 🔥 8. Loss → Cost → Objective

A useful conceptual hierarchy is:

```text
Individual example
       ↓
      LOSS
       ↓
Aggregate losses
       ↓
COST / MEAN LOSS
       ↓
+ Regularization, if used
       ↓
OBJECTIVE
       ↓
Optimization
```

However, remember that **terminology varies** across textbooks and frameworks.

---

# 🔥 9. Connection to Gradient Descent

Suppose the objective is:

$$
J(\theta)
$$

We calculate its gradient:

$$
\nabla_\theta J(\theta)
$$

The gradient tells us how the objective changes with respect to the model parameters.

Gradient descent updates the parameters:

$$
\boxed{
\theta
\leftarrow
\theta-\eta\nabla_\theta J(\theta)
}
$$

where:

* $\theta$ = parameters
* $\eta$ = learning rate
* $\nabla_\theta J$ = gradient

### Why subtract?

The gradient points toward the direction of **increasing** objective.

So we move in the opposite direction to reduce it.

---

# 🔥 10. Complete Training Loop

```text
                 INPUT
                   ↓
              Neural Network
                   ↓
               Prediction
                   ↓
             Loss Function
                   ↓
          Aggregate / Objective
                   ↓
             Backpropagation
                   ↓
                Gradients
                   ↓
               Optimizer
                   ↓
          Updated Parameters
                   ↓
             Neural Network
                   ↓
                 Repeat 🔄
```

This is the foundation for understanding:

* Backpropagation
* Gradient descent
* Optimizers
* Training loops

---

# 🧠 11. Example: Regression

Suppose we're predicting motor temperature.

True:

$$
y=80^\circ C
$$

Prediction:

$$
\hat y=75^\circ C
$$

Squared error:

$$
L=(80-75)^2
$$

$$
L=25
$$

For thousands of training examples:

$$
J=
\frac{1}{m}
\sum_{i=1}^{m}L_i
$$

Then the optimizer tries to minimize:

$$
J(\theta)
$$

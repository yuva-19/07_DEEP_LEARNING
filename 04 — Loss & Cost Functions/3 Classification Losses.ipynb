# 1. 🔥🔥🔥 Binary Cross Entropy — BCE

## What is it?

**Binary Cross Entropy** is used for **binary classification**, where there are two possible classes.

### Examples

* Spam / Not Spam
* Pass / Fail
* Disease / No Disease
* Cat / Not Cat

Typical architecture:

```text
Input
  ↓
Neural Network
  ↓
1 Output Logit
  ↓
Sigmoid
  ↓
Probability
  ↓
Binary Cross Entropy
```

---

## Formula

For one example:

$$
L=-[y\log(\hat y)+(1-y)\log(1-\hat y)]
$$

Where:

* $y$ = actual label, 0 or 1
* $\hat y$ = predicted probability

For $n$ examples:

$$
BCE=-\frac{1}{n}\sum_{i=1}^{n}
[y_i\log(\hat y_i)+(1-y_i)\log(1-\hat y_i)]
$$

---

## 🧠 Intuition

Suppose:

$$
y=1
$$

The model should give a **high probability** to class 1.

### Correct and confident

$$
\hat y=0.9
$$

Then:

$$
L=-\log(0.9)\approx0.105
$$

Small loss. ✅

### Wrong and confident

$$
\hat y=0.01
$$

Then:

$$
L=-\log(0.01)\approx4.605
$$

Large loss. 🚨

Therefore:

> **BCE strongly penalizes confident wrong predictions.**

---

## Relationship with Sigmoid

Binary classification commonly uses:

$$
\hat y=\sigma(z)
$$

where:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

So conceptually:

$$
\boxed{\text{Logit}\rightarrow\text{Sigmoid}\rightarrow\text{Probability}\rightarrow\text{BCE}}
$$

---

## 🔥 PyTorch

In PyTorch, you'll commonly use:

```python
loss_fn = nn.BCEWithLogitsLoss()
```

This combines:

```text
Sigmoid + Binary Cross Entropy
```

in a numerically stable implementation.

Therefore, give it **raw logits**:

```python
logits = model(x)
loss = loss_fn(logits, target)
```

### 🚨 Don't do this

```python
prob = torch.sigmoid(logits)
loss = loss_fn(prob, target)
```

`BCEWithLogitsLoss` already handles the sigmoid internally.

## Softmax

The network produces logits:

$$
z_1,z_2,\ldots,z_K
$$

Softmax converts them into probabilities:

$$
P_i=
\frac{e^{z_i}}
{\sum_{j=1}^{K}e^{z_j}}
$$

The probabilities satisfy:

$$
\sum_{i=1}^{K}P_i=1
$$

---

## CCE Formula

For one-hot encoded targets:

$$
L=-\sum_{i=1}^{K}y_i\log(\hat y_i)
$$

Where:

* $y_i$ = actual one-hot label
* $\hat y_i$ = predicted probability
* $K$ = number of classes

---

## Example

Suppose:

```text
0 = Cat
1 = Dog
2 = Horse
```

Actual class = **Dog**.

One-hot target:

$$
y=[0,1,0]
$$

Model prediction:

$$
\hat y=[0.1,0.8,0.1]
$$

Then:

$$
L=
-[0\log(0.1)+1\log(0.8)+0\log(0.1)]
$$

Therefore:

$$
L=-\log(0.8)
$$

$$
L\approx0.223
$$

Small loss because the model gave the correct class a high probability. ✅

# 3. 🔥🔥🔥 Sparse Categorical Cross Entropy — SCCE

This is where beginners often get confused.

SCCE is used for the **same type of problem as CCE**:

> **Multiclass classification with exactly one correct class.**

The difference is **how the target label is stored**.

---

## CCE → One-Hot Target

For:

```text
0 = Cat
1 = Dog
2 = Horse
```

Dog is:

$$
[0,1,0]
$$

---

## SCCE → Integer Class Index

Instead of storing:

$$
[0,1,0]
$$

we simply store:

$$
1
$$

So:

```text
Cat   → 0
Dog   → 1
Horse → 2
```

That's what **sparse** means here.

> We store only the class index instead of the complete one-hot vector.

---

# 🔥 CCE vs SCCE

|                    | CCE                    | SCCE                   |
| ------------------ | ---------------------- | ---------------------- |
| Problem            | Multiclass             | Multiclass             |
| One correct class? | ✅                      | ✅                      |
| Target             | One-hot vector         | Integer class index    |
| Dog target         | `[0,1,0]`              | `1`                    |
| Model output       | K logits/probabilities | K logits/probabilities |
| Conceptual purpose | Same                   | Same                   |

### 🧠 Key point

$$
\boxed{\text{CCE and SCCE mainly differ in target representation}}
$$

They do **not** represent two different classification problems.

---

# 🔥 PyTorch Connection

PyTorch commonly uses:

```python
nn.CrossEntropyLoss()
```

for multiclass classification.

It expects:

```text
Model output → raw logits
Target       → integer class indices
```

Example:

```python
logits = model(x)

criterion = nn.CrossEntropyLoss()

loss = criterion(logits, target)
```

If Dog is class `1`:

```python
target = 1
```

not:

```python
target = [0, 1, 0]
```

---

## 🚨 Don't manually apply Softmax

Don't normally do:

```python
probabilities = torch.softmax(logits, dim=1)
loss = criterion(probabilities, target)
```

Instead:

```python
loss = criterion(logits, target)
```

`CrossEntropyLoss` internally handles the required log-softmax and negative-log-likelihood computation in a numerically stable way.

---

# 🧠 The BIG Picture

```text
                 CLASSIFICATION
                       │
             ┌─────────┴─────────┐
             ↓                   ↓
          BINARY             MULTICLASS
             │                   │
        2 classes             K classes
             │                   │
        1 output              K outputs
             │                   │
          Sigmoid             Softmax
             │                   │
            BCE           Cross Entropy
                                 │
                         ┌───────┴───────┐
                         ↓               ↓
                        CCE             SCCE
                         │               │
                      One-hot       Class index
```

# ⚡ Complete Comparison

| Loss     | Task       | Outputs | Target format | Activation concept |
| -------- | ---------- | ------: | ------------- | ------------------ |
| **BCE**  | Binary     |       1 | `0` / `1`     | Sigmoid            |
| **CCE**  | Multiclass |       K | One-hot       | Softmax            |
| **SCCE** | Multiclass |       K | Integer index | Softmax            |

### PyTorch equivalent

| Task       | PyTorch loss        | Model should output |
| ---------- | ------------------- | ------------------- |
| Binary     | `BCEWithLogitsLoss` | Raw logits          |
| Multiclass | `CrossEntropyLoss`  | Raw logits          |

# 🎤 Interview / GATE Points

### 1. What is BCE?

A loss function commonly used for **binary classification** that measures the difference between the actual binary label and predicted probability.

### 2. Why does cross entropy penalize confident wrong predictions?

Because of the logarithm:

$$
-\log(p)
$$

As the probability assigned to the correct class approaches zero, the loss becomes very large.

### 3. CCE vs SCCE?

> **CCE → one-hot target**
> **SCCE → integer class index**

### 4. Binary vs multiclass?

```text
Binary:
1 output → sigmoid → BCE

Multiclass:
K outputs → softmax conceptually → cross entropy
```

### 5. What does PyTorch `CrossEntropyLoss` expect?

> **Raw logits + integer class labels.**

# 🚨 Common Mistakes

### ❌ BCE for ordinary multiclass classification

For mutually exclusive classes like Cat/Dog/Horse, use multiclass cross entropy rather than treating the classes as independent binary outputs.

### ❌ Softmax for multi-label classification

If multiple classes can be correct simultaneously, use **independent sigmoid outputs**, not softmax.

### ❌ Confusing CCE and SCCE

Remember:

```text
CCE  → [0, 1, 0]
SCCE → 1
```

Same underlying class information.

### ❌ Softmax before `CrossEntropyLoss`

```python
# ❌
probs = torch.softmax(logits, dim=1)
loss = criterion(probs, target)
```

Use:

```python
# ✅
loss = criterion(logits, target)
```

### ❌ Sigmoid before `BCEWithLogitsLoss`

```python
# ❌
prob = torch.sigmoid(logits)
loss = criterion(prob, target)
```

Use:

```python
# ✅
loss = criterion(logits, target)
```

# 📚 4.2 Regression Losses — Revision Notes

Regression losses measure **how far the model's predicted continuous value is from the actual value**.

The three important losses in this block are:

1. 🔥🔥🔥 **MSE — Mean Squared Error**
2. 🔥🔥🔥 **MAE — Mean Absolute Error**
3. 🔥🔥🔥 **Huber Loss**

---

## 1. 🔥 MSE — Mean Squared Error

### Core Concept

MSE calculates the **average of squared prediction errors**.

$$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Where:

* $y_i$ = actual value
* $\hat{y}_i$ = predicted value
* $n$ = number of samples

### Intuition

```text
Actual ───────┐
              ↓
           Difference
              ↓
           Square it
              ↓
          Average
              ↓
             MSE
```

### Example

Suppose the errors are:

$$
1,\;2,\;5
$$

Then:

$$
MSE=\frac{1^2+2^2+5^2}{3}
$$

$$
=\frac{1+4+25}{3}
$$

$$
=10
$$

### 🚨 Key Property: Large Errors Are Heavily Penalized

Because the error is **squared**:

$$
2^2=4
$$

but

$$
10^2=100
$$

Therefore:

> **MSE is highly sensitive to large errors and outliers.**

### Gradient

For a single example:

$$
L=(y-\hat y)^2
$$

Derivative with respect to prediction:

$$
\frac{\partial L}{\partial\hat y}
=
2(\hat y-y)
$$

So a larger error produces a larger gradient.

### Typical Uses

* House-price prediction
* Temperature prediction
* Energy-demand prediction
* Power-consumption prediction
* Motor-speed prediction

# 2. 🔥 MAE — Mean Absolute Error

### Core Concept

MAE calculates the **average absolute prediction error**.

$$
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat y_i|
$$

Instead of squaring the error, we simply take its magnitude.

### Example

Errors:

$$
1,\;2,\;5
$$

Then:

$$
MAE=\frac{|1|+|2|+|5|}{3}
$$

$$
=\frac{1+2+5}{3}
$$

$$
=2.67
$$

So the average absolute error is approximately **2.67 units**.

---

## 🚨 Key Property: More Robust to Outliers

Consider an error of 10.

### MSE

$$
10^2=100
$$

### MAE

$$
|10|=10
$$

MSE makes the error **100**, while MAE keeps it at **10**.

Therefore:

> **MAE is less sensitive to outliers than MSE.**

---

## Mathematical Behavior

For:

$$
L=|y-\hat y|
$$

the gradient away from zero is approximately:

$$
\frac{\partial L}{\partial\hat y}
=
\begin{cases}
-1 & \hat y<y\\
+1 & \hat y>y
\end{cases}
$$

At zero error, the absolute-value function has a sharp corner.

### Important idea

**MSE:**

> Larger error → larger gradient

**MAE:**

> Gradient magnitude remains constant away from zero

# 3. 🔥 Huber Loss

### Core Concept

Huber Loss combines the useful properties of **MSE and MAE**.

It behaves:

* like **MSE for small errors**
* like **MAE for large errors**

## Formula

Let:

$$
e=y-\hat y
$$

Then:

$$
L_\delta(e)=
\begin{cases}
\frac{1}{2}e^2 & |e|\leq\delta\\
\delta\left(|e|-\frac{1}{2}\delta\right) & |e|>\delta
\end{cases}
$$

where $\delta$ determines the point where the behavior changes.

---

## 🧠 Understand It Visually

```text
             Huber Loss

Loss
 ↑
 |                         /
 |                       /
 |                     /
 |                  __/
 |               __/
 |            __/
 |         __/
 |_______/________________→ Error
         ↑
         δ
         
     MSE       MAE-like
   behavior    behavior
```

### Small error

Huber behaves like:

$$
\frac12e^2
$$

So it is **quadratic**, like MSE.

### Large error

Huber behaves linearly:

$$
\delta\left(|e|-\frac12\delta\right)
$$

So it becomes **MAE-like**.

---

# 🔥 Why Huber Loss?

Imagine one training sample has a very large error.

### MSE

The error is squared:

$$
20^2=400
$$

That large error can have a huge influence on training.

### MAE

The error is:

$$
|20|=20
$$

Much less influence.

### Huber

Huber behaves like MSE when the error is small, but once the error becomes sufficiently large, it switches to linear behavior.

So:

> **Huber gives smooth behavior around small errors while limiting the influence of very large errors.**

| Property                  | MSE                   | MAE                   | Huber                       |   |           |
| ------------------------- | --------------------- | --------------------- | --------------------------- | - | --------- |
| Formula idea              | Error²                | $                     | Error                       | $ | MSE → MAE |
| Large errors              | 🔴 Strongly penalized | 🟢 Less influence     | 🟢 Controlled               |   |           |
| Outlier sensitivity       | High                  | Low                   | Lower than MSE              |   |           |
| Smooth around zero        | ✅                     | ❌ Sharp corner        | ✅                           |   |           |

| Robust to outliers        | ❌                     | ✅                     | ✅                           |   |           |
| Main idea                 | Punish large errors   | Treat errors linearly | Balance MSE + MAE           |   |           |

# 🧠 The Best Mental Model

### MSE

$$
\boxed{\text{Error}^2}
$$

**"Big mistakes are very expensive."**

---

### MAE

$$
\boxed{|Error|}
$$

**"Mistake size matters directly."**

---

### Huber

$$
\boxed{\text{MSE for small errors + MAE-like for large errors}}
$$

**"Be sensitive to normal errors, but don't let huge errors dominate."**

---

# 🎯 When to Use What?

### MSE

Choose when:

* Large errors should be strongly penalized.
* Outliers are meaningful.
* Smooth optimization is desirable.

---

### MAE

Choose when:

* Outliers shouldn't dominate training.
* Robustness is important.
* You care about average absolute deviation.

---

### Huber

Choose when:

* You want smooth behavior around small errors.
* Your dataset may contain outliers.
* You want a compromise between MSE and MAE.

---

# 🔗 Connection to Neural Network Training

For a regression neural network:

```text
Input
  ↓
Neural Network
  ↓
Prediction ŷ
  ↓
Compare with actual y
  ↓
Regression Loss
  ↓
Gradient
  ↓
Backpropagation
  ↓
Optimizer
  ↓
Update Weights
```

The loss function therefore determines **how prediction errors influence learning**.

$$
\boxed{\text{Prediction}\rightarrow\text{Loss}\rightarrow\text{Gradient}\rightarrow\text{Weight Update}}
$$

# 🎤 Interview / GATE Points

### Q1. Why is MSE sensitive to outliers?

Because errors are squared:

$$
e\rightarrow e^2
$$

Large errors therefore grow disproportionately.

---

### Q2. Which is more robust to outliers: MSE or MAE?

**MAE.**

Because it uses:

$$
|e|
$$

instead of:

$$
e^2
$$

---

### Q3. What is Huber Loss?

A loss that behaves like **MSE for small errors** and **MAE for large errors**.

---

### Q4. Why might Huber be preferred over MAE?

Huber retains a **smooth quadratic region around zero**, while MAE has a sharp corner at zero.

---

### Q5. What does $\delta$ do in Huber Loss?

It determines the **transition point** between the quadratic MSE-like region and the linear MAE-like region.

---

# 🚨 Common Mistakes

❌ **"MAE squares the error."**

No.

$$
MAE=|y-\hat y|
$$

---

❌ **"MSE is robust to outliers."**

No.

MSE is **sensitive** to outliers.

---

❌ **"Huber is exactly the average of MSE and MAE."**

No.

Huber is **piecewise defined**: quadratic for small errors and linear for large errors.

---

❌ **"MAE is always better than MSE."**

No universal winner. The appropriate loss depends on the problem and how you want errors/outliers to influence training.

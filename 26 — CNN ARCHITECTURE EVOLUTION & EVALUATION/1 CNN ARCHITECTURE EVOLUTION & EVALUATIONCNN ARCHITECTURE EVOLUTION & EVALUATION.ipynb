# 3. 🟦 LeNet — THE BEGINNING

## 1. What problem existed?

Early vision systems needed to recognize things such as **handwritten characters**.

A major challenge was handling the spatial structure of images efficiently rather than treating every pixel as an unrelated input.

LeCun and collaborators demonstrated CNN-based systems for document/handwriting recognition, including LeNet-5. ([IEEE Xplore][2])

---

## 2. What idea was introduced?

Use:

* convolution
* subsampling/pooling
* hierarchical feature extraction
* shared weights

Conceptually:

```text
Image
  ↓
Convolution
  ↓
Pooling
  ↓
Convolution
  ↓
Pooling
  ↓
Fully Connected
  ↓
Classification
```

---

## 3. How does it work?

Instead of learning a completely independent weight for every pixel relationship:

```text
pixels
 ↓
local filters
 ↓
feature maps
 ↓
higher-level features
```

The CNN exploits two important image properties:

### Locality

Nearby pixels are strongly related.

### Weight sharing

The same filter can detect the same pattern at different positions.

---

## 4. Advantage

Huge reduction in unnecessary connections compared with a fully connected network.

It introduced the fundamental CNN philosophy:

$$
\boxed{\text{local connectivity + shared weights}}
$$

---

## 5. Trade-off

LeNet was designed for relatively constrained recognition problems.

It wasn't a giant ImageNet-scale architecture.

The hardware, datasets and training ecosystem of the era also limited how far this approach could be scaled.

---

## 6. What came later?

**AlexNet** demonstrated that CNNs could be scaled dramatically with much larger datasets, GPUs and deeper networks.

## 1. Problem

CNNs worked, but the question was:

> **Can deep CNNs handle huge, realistic image datasets?**

ImageNet provided millions of images and 1000 classes.

---

## 2. Innovation

AlexNet combined several important ingredients:

* deeper CNN
* ReLU
* GPU-based training
* data augmentation
* dropout
* max pooling
* large-scale ImageNet training

The original paper reports roughly **60 million parameters** and five convolutional layers followed by fully connected layers. Its 2012 competition submission achieved a top-5 test error of 15.3%, substantially ahead of the second-place 26.2%. ([NeurIPS Papers][3])

---

## 3. How?

```text
Image
 ↓
Conv + ReLU
 ↓
Conv + ReLU
 ↓
...
 ↓
Fully Connected
 ↓
Softmax
```

The crucial difference wasn't one magical layer.

It was:

$$
\boxed{\text{CNN}+\text{big data}+\text{GPU}+\text{better training}}
$$

---

## 4. Advantage

AlexNet demonstrated that **deep CNNs could dominate large-scale visual recognition**.

This became a major turning point for modern deep learning. ([NeurIPS Papers][3])

---

## 5. Trade-off

The model was computationally and memory intensive.

Approximately:

$$
\boxed{60M\text{ parameters}}
$$

The fully connected layers were especially parameter-heavy.

---

## 6. What came later?

**VGG** asked:

> Can we make the architecture deeper while keeping the design simple?

## 1. Problem

AlexNet worked extremely well, but its architecture was somewhat heterogeneous.

Researchers wanted to understand:

> **What happens if we systematically increase CNN depth?**

---

## 2. Innovation

VGG used very simple repeated:

$$
3\times3
$$

convolutions.

Instead of complicated filters:

```text
3×3
3×3
3×3
3×3
...
```

The VGG paper systematically evaluated increasing depth and found strong improvements with 16–19 weight layers. ([arXiv][4])

---

## 3. How?

```text
Input
 ↓
3×3 Conv
 ↓
3×3 Conv
 ↓
Pooling
 ↓
3×3 Conv
 ↓
3×3 Conv
 ↓
Pooling
 ↓
...
```

---

## 4. Advantage

### Simple architecture

Easy to understand and modify.

### Greater depth

More nonlinear transformations:

```text
edges
 ↓
textures
 ↓
parts
 ↓
objects
```

The VGG work demonstrated the importance of depth for ImageNet recognition. ([arXiv][4])

---

## 5. Trade-off

VGG became **very parameter- and computation-heavy**, particularly its fully connected layers.

VGG-16 is roughly:

$$
\boxed{138M\text{ parameters}}
$$

So:

```text
Simple architecture
      ↓
More depth
      ↓
Better representation
      ↓
Huge computation/model size
```

---

## 6. What came later?

Two major directions:

```text
VGG
 ├──► Inception
 │     "Can we use computation more intelligently?"
 │
 └──► ResNet
       "Can we train much deeper networks?"
```

## 1. Problem

Simply increasing depth and width causes computation to explode.

The question:

> **How can we increase representational power without proportionally increasing computation?**

---

## 2. Innovation

The **Inception module** processes features through multiple branches:

```text
              Input
                │
       ┌────────┼────────┐
       ↓        ↓        ↓
      1×1      3×3      5×5
       │        │        │
       └────────┼────────┘
                ↓
           Concatenate
```

It also used 1×1 convolutions to reduce channel dimensionality before expensive operations.

The original Inception paper explicitly targeted better utilization of computational resources while increasing depth/width under a constrained computational budget. ([arXiv][5])

---

## 3. How?

Different branches capture different spatial scales:

```text
1×1 → local/channel transformation
3×3 → medium spatial pattern
5×5 → larger spatial pattern
pool → alternative representation
```

Then:

$$
\boxed{\text{Concatenate}}
$$

---

## 4. Advantage

Better computational utilization and multi-scale feature extraction.

GoogLeNet/Inception achieved strong ImageNet results while being much smaller in parameter count than architectures such as VGG.

---

## 5. Trade-off

The architecture becomes:

* more complicated
* harder to design manually
* harder to reason about
* dependent on carefully chosen branches and dimensions

---

## 6. What came later?

**ResNet** attacked a different problem:

> Even with good architecture design, **very deep networks were becoming difficult to optimize.**

## 1. Problem

This was a huge problem:

> **Simply making a network deeper did not guarantee better training.**

The ResNet paper showed that deeper plain networks could have **higher training error**, indicating an optimization problem rather than merely overfitting. ([DOI][6])

---

## 2. Innovation

Instead of directly learning:

$$
H(x)
$$

learn:

$$
\boxed{F(x)=H(x)-x}
$$

and output:

$$
\boxed{y=F(x)+x}
$$

This is the residual connection.

---

## 3. How?

```text
        ┌───────────────────┐
        │                   │
        │                   ▼
Input ──┴──► Conv → BN → ReLU → Conv
        │                   │
        └──────────────────►+
                            │
                            ▼
                          Output
```

The shortcut creates a direct path through the network.

---

## 4. Advantage

It made substantially deeper networks easier to optimize.

The original work evaluated networks up to **152 layers**, much deeper than VGG, while maintaining lower complexity for comparable configurations. ([arXiv][7])

---

## 5. Trade-off

Residual connections themselves aren't free:

* additional architectural complexity
* projection shortcuts when dimensions change
* deeper models can still require significant compute

But the bigger conceptual trade-off is that ResNet established a **new structural pattern that later architectures heavily reused**.

---

## 6. What came later?

ResNet created multiple branches of evolution:

```text
                 ResNet
              /    |     \
             ↓     ↓      ↓
         DenseNet ResNeXt MobileNet
```

Not all directly descended from ResNet, but they explored related questions around **connectivity, transformations and efficiency**.

---

# 8. 🟪 DenseNet — FEATURE REUSE

## 1. Problem

ResNet gives a shortcut from earlier features to later layers.

But:

> **Why only connect a layer to the immediately previous representation through a residual addition?**

Could we provide later layers with **all previous feature maps**?

---

## 2. Innovation

Dense connectivity.

For layer \(l\):

$$
\boxed{x_l=H_l([x_0,x_1,\dots,x_{l-1}])}
$$

where:

$$
[\cdot]
$$

means concatenation.

The DenseNet paper connects every layer to every subsequent layer, creating \(L(L+1)/2\) direct connections in an \(L\)-layer network. ([arXiv][8])

---

## 3. How?

```text
x0 ───────────────┐
 │                │
 ▼                │
x1 ───────────┐   │
 │            │   │
 ▼            ▼   ▼
x2 ────────────────
 │
 ▼
x3
```

Every layer receives previous feature maps.

---

## 4. Advantage

DenseNet encourages:

* feature reuse
* strong feature propagation
* gradient flow
* parameter efficiency

The paper reports reduced parameter requirements and strong performance across several recognition benchmarks. ([arXiv][8])

---

## 5. Trade-off

This is the key:

```text
Feature reuse ↑
Connections ↑↑
Memory access ↑
Feature-map storage ↑
```

Concatenating many previous feature maps can create substantial **memory and implementation overhead**, even if the parameter count is attractive.

---

## 6. What came later?

**ResNeXt** asked a different question:

> Instead of making the network denser, can we increase the number of parallel transformations?

---

# 9. 🟧 ResNeXt — CARDINALITY

## 1. Problem

ResNet mainly gives us three obvious capacity dimensions:

* depth
* width
* number of transformations

ResNeXt focused on the third.

---

## 2. Innovation

### Cardinality

Instead of one transformation:

```text
Input
  ↓
Transformation
  ↓
Output
```

use multiple grouped transformations:

```text
             Input
                │
      ┌─────────┼─────────┐
      ↓         ↓         ↓
   Group 1   Group 2   Group 3
      ↓         ↓         ↓
      └─────────┼─────────┘
                ↓
             Output
```

The ResNeXt paper introduced **cardinality** as the size of this set of transformations and found it to be an important dimension of model capacity. ([arXiv][9])

---

## 3. How?

Grouped convolution divides channels into groups.

Each group performs its own transformation.

Then:

$$
\boxed{\text{Aggregate}}
$$

---

## 4. Advantage

It provides another way to increase representational capacity without simply making the network deeper or wider.

The authors found increasing cardinality could be more effective than increasing depth or width under comparable complexity in their experiments. ([arXiv][9])

---

## 5. Trade-off

Grouped operations introduce:

* more architectural complexity
* dependence on efficient hardware kernels
* less direct cross-group interaction

So:

$$
\text{Groups} \uparrow
$$

doesn't automatically mean:

$$
\text{real-world latency} \downarrow
$$

---

## 6. What came later?

This line of thinking around efficient grouped/decomposed convolution connects conceptually to **MobileNet/Xception**, which pushed computational efficiency much further through depthwise separable convolution.

---

![Image](https://images.openai.com/static-rsc-4/yurtvM6jwre8uxm5eKe2oNah63JGrLQh9cU9kLPmb9C8r9PAPFabCj2p7HINXWCwt4fy59gql8pFvm740laVRNaok8Zii0ljec88q2JCxB3uKANLageoWa3QqToS4-WEVfEYPI_rGGLGY1WTbb9Bp8fMbLQs7sFZCu0MU5F40HLGqOj4q09wWHIITsIZO9rM?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/clqSWJv67spcM8UlmNt6F1Nz0WJ6ghCRQoZms2mZehJHBBsq2wkFGV0vbP3wcnKh0XAO10x-eotd3RmM0C6cjbFSZKXjdOQt99aVDIKSPpBdvy3mzdj6gFj-ICR2XJG1g2rnFYen6rN2eg_G8X6lbBRcBr7CBzV4QMeuDhJDUVPhYb1EaJMSbMjitBi2fxQn?purpose=fullsize)

## 1. Problem

Standard CNNs are expensive.

Especially on:

* smartphones
* embedded devices
* edge hardware
* low-power systems

The question:

> **Can we drastically reduce computation while retaining useful accuracy?**

---

## 2. Innovation

### Depthwise separable convolution

Split:

$$
\text{Standard Conv}
$$

into:

```text
Depthwise Conv
      ↓
Pointwise 1×1 Conv
```

---

## 3. How?

### Standard convolution

One operation simultaneously performs spatial and channel mixing.

### Depthwise separable

```text
Input
  ↓
Depthwise Conv
  ↓
Spatial filtering
  ↓
1×1 Pointwise Conv
  ↓
Channel mixing
```

This drastically reduces computation.

The MobileNet paper specifically designed lightweight networks using depthwise separable convolutions and introduced width and resolution multipliers to trade accuracy against latency/resource requirements. ([arXiv][10])

---

## 4. Advantage

Huge reduction in:

* computation
* parameters
* model size
* suitability cost for mobile deployment

---

## 5. Trade-off

The efficiency comes from imposing more structure on how channels interact.

That can create:

* accuracy trade-offs
* hardware-specific performance differences
* sensitivity to implementation efficiency

---

## 6. What came later?

**Xception** pushed the depthwise-separable-convolution idea further.

---

# 11. 🟫 Xception — EXTREME DEPTHWISE SEPARABILITY

## 1. Problem

Inception had multiple branches.

Xception asked:

> **Can the Inception idea be interpreted as depthwise separable convolution and simplified further?**

---

## 2. Innovation

Xception replaces Inception-style modules with **depthwise separable convolutions**.

The paper interprets depthwise separable convolution as an extreme form of the Inception concept, with a very large number of independent spatial transformations followed by pointwise channel mixing. ([arXiv][11])

---

## 3. How?

```text
Input
  ↓
Depthwise Conv
  ↓
Pointwise 1×1 Conv
  ↓
Output
```

Conceptually:

```text
Spatial mixing
      +
Channel mixing
      ↓
Separated
```

---

## 4. Advantage

Better parameter utilization.

The Xception paper reported slightly better ImageNet performance than Inception V3 with approximately the **same number of parameters**, and substantially better results on its larger dataset experiment. ([arXiv][11])

---

## 5. Trade-off

Depthwise separable convolution:

* relies heavily on efficient implementation
* can have hardware-specific latency behavior
* introduces stronger factorization assumptions

---

## 6. What came later?

The efficiency branch eventually led toward architectures such as **EfficientNet**, which attacked efficiency from a different angle:

> Instead of only making individual operations cheaper, **how should the entire network be scaled?**

## 1. Problem

Suppose you have a CNN.

How do you make it larger?

You could increase:

### Depth

$$
\text{more layers}
$$

### Width

$$
\text{more channels}
$$

### Resolution

$$
\text{larger input}
$$

But which should you increase?

---

## 2. Innovation

EfficientNet introduced **compound scaling**.

Instead of arbitrarily scaling one dimension, scale:

$$
\boxed{\text{depth + width + resolution}}
$$

in a coordinated way.

The paper defines:

$$
d=\alpha^\phi
$$

$$
w=\beta^\phi
$$

$$
r=\gamma^\phi
$$

subject to a resource-balancing constraint. ([Proceedings of Machine Learning Research][12])

---

## 3. How?

Think:

```text
          Model scaling
               │
      ┌────────┼────────┐
      ↓        ↓        ↓
    Depth    Width   Resolution
      │        │        │
      └────────┼────────┘
               ↓
       Compound scaling
```

Instead of:

```text
Depth × 3
```

alone:

```text
Depth ↑
Width ↑
Resolution ↑
```

in a coordinated manner.

---

## 4. Advantage

Much better accuracy-efficiency trade-offs.

The paper reports EfficientNet-B7 at around **84.3–84.4% ImageNet top-1**, depending on the reported evaluation setup, while being substantially smaller/faster than the compared prior ConvNets. ([arXiv][13])

---

## 5. Trade-off

The model design and scaling procedure become more sophisticated.

Higher-resolution/larger models can also become expensive in:

* memory
* training time
* inference time

And the ideal scaling coefficients depend on the underlying architecture and resource regime.

---

## 6. What came later?

**RegNet** asked an even more fundamental question:

> Rather than manually choosing individual networks, can we design the **space of networks** systematically?

---

# 13. 🟦 RegNet — DESIGN THE DESIGN SPACE

## 1. Problem

CNN architecture design had become a giant collection of manually designed networks.

The problem wasn't simply:

> “Which model should I build?”

It became:

> **“How should we define the space from which good models can be generated?”**

---

## 2. Innovation

RegNet proposed designing a **network design space** rather than focusing only on individual network instances.

Its central parameterization produces regular networks where widths can be described by a **quantized linear function** across depth. ([arXiv][14])

---

## 3. How?

Conceptually:

```text
Traditional:

Design Network A
Design Network B
Design Network C
Design Network D
...


RegNet:

Define a design space
        ↓
Generate many networks
        ↓
Analyze the population
        ↓
Identify good regions
        ↓
Choose models for resource regimes
```

---

## 4. Advantage

More systematic architecture design.

The RegNet paper reports simple network families working across a broad range of FLOP regimes, and reports faster GPU inference than the EfficientNet models compared under its experimental setup. ([arXiv][14])

---

## 5. Trade-off

You now have another layer of complexity:

```text
Architecture design
        +
Design-space engineering
        +
Search/evaluation
```

It also doesn't mean one RegNet configuration is optimal for every hardware platform.

---

## 6. What came later?

**ConvNeXt** returned to a fundamental question:

> What if we take the accumulated lessons from modern vision architectures and apply them systematically to a pure CNN?

## 1. Problem

By the Transformer era, Vision Transformers were becoming extremely strong.

The question became:

> **Are CNNs fundamentally limited, or had CNN architecture simply stopped evolving as aggressively?**

---

## 2. Innovation

Start from ResNet and progressively modernize it.

Major changes include:

```text
BatchNorm
   ↓
LayerNorm

ReLU
   ↓
GELU

3×3 Conv
   ↓
7×7 Depthwise Conv

Traditional bottleneck
   ↓
Inverted bottleneck

Old stem
   ↓
Patchify-style stem

Old macro design
   ↓
Modernized stage design
```

The ConvNeXt paper explicitly describes this progressive modernization toward the design of Vision Transformers while retaining a pure ConvNet. ([arXiv][15])

---

## 3. How?

Simplified:

```text
Input
  ↓
7×7 Depthwise Conv
  ↓
LayerNorm
  ↓
1×1 / Linear
  ↓
GELU
  ↓
1×1 / Linear
  ↓
Residual
```

---

## 4. Advantage

It demonstrated that a modern pure ConvNet could remain highly competitive with contemporary Transformer-based vision architectures.

The ConvNeXt paper reports up to **87.8% ImageNet top-1** for its largest model and strong COCO/ADE20K results in its experimental comparisons. ([arXiv][15])

---

## 5. Trade-off

ConvNeXt is still a large modern backbone family.

Larger variants can have substantial:

* parameter counts
* FLOPs
* memory requirements
* training costs

And importantly:

> **Theoretical FLOPs ≠ actual latency.**

Hardware, kernels, memory bandwidth and implementation all matter.

This is where the roadmap connects into:

```text
CNN
  ↘
   Modern Vision
  ↗
Transformer
```

ConvNeXt demonstrated that CNNs could be modernized rather than simply abandoned.

# 17. 📊 NOW — HOW DO WE EVALUATE ARCHITECTURES?

Your syllabus gives:

* Parameters
* FLOPs
* Accuracy
* Model size
* Memory
* Latency
* Training complexity
* Deployment suitability

These are **not interchangeable**.

A model can have:

```text
low parameters
+
high latency
```

or:

```text
high FLOPs
+
good GPU throughput
```

or:

```text
high accuracy
+
terrible edge deployment
```

That's why real architecture evaluation is multi-dimensional.

---

# 18. 🧮 1. PARAMETERS

Parameters are the learned values:

$$
W=\{w_1,w_2,\dots,w_N\}
$$

So:

$$
\boxed{\text{Parameters}=N}
$$

Example:

```text
Model A → 5M parameters
Model B → 100M parameters
```

Model B generally has a larger learned state.

But:

> **More parameters does NOT automatically mean better accuracy.**

EfficientNet and DenseNet are classic examples of architectures designed to obtain strong performance without simply maximizing parameter count. ([arXiv][8])

---

# 19. 💻 2. FLOPs

FLOPs ≈ amount of floating-point computation.

For a standard convolution:

$$
\text{FLOPs}
\propto
H\times W\times C_{in}\times C_{out}\times K^2
$$

So increasing:

```text
resolution ↑
channels ↑
kernel size ↑
layers ↑
```

can increase computation.

### Example

If spatial resolution doubles:

$$
H,W \rightarrow 2H,2W
$$

then the number of spatial positions becomes:

$$
4HW
$$

So convolutional compute can increase by approximately:

$$
\boxed{4\times}
$$

for the same channel/kernel configuration.

EfficientNet explicitly discusses how depth, width and resolution affect FLOPs. ([Proceedings of Machine Learning Research][12])

---

# 20. 💾 3. MODEL SIZE

A rough estimate:

$$
\text{Model size}
\approx
\text{number of parameters}
\times
\text{bytes per parameter}
$$

For FP32:

$$
4\text{ bytes/parameter}
$$

So:

### 100M parameters

$$
100M\times4
$$

$$
=\boxed{400\text{ MB}}
$$

approximately, just for the weights.

With FP16:

$$
100M\times2
=
\boxed{200\text{ MB}}
$$

This matters enormously for:

* mobile deployment
* edge devices
* embedded systems
* memory-limited inference

---

# 21. 🧠 4. MEMORY

Memory is **not just model size**.

You have:

```text
Memory
├── Model weights
├── Activations
├── Gradients
├── Optimizer states
└── Temporary buffers
```

During training:

$$
\boxed{\text{Memory} \gg \text{just model weights}}
$$

For example, Adam stores additional optimizer state.

So a model with relatively modest parameter count can still have substantial training memory requirements because of activations and optimizer states.

---

# 22. ⚡ 5. LATENCY

Latency means:

> **How long does one inference take?**

For example:

```text
Model A → 5 ms
Model B → 25 ms
```

Model A has lower latency in that particular setup.

But latency depends on:

* CPU/GPU/NPU
* batch size
* framework
* kernel implementation
* precision
* memory bandwidth
* input resolution
* compiler/runtime

Therefore:

$$
\boxed{\text{FLOPs} \neq \text{Latency}}
$$

This is **VERY important for industry**.

---

# 23. 🏋️ 6. TRAINING COMPLEXITY

Training complexity includes more than FLOPs.

Consider:

### VGG

Very straightforward architecture:

```text
Conv
Conv
Pool
Conv
Conv
Pool
...
```

Easy to understand.

---

### Inception

More complicated:

```text
branch
branch
branch
branch
↓
concat
```

More architectural complexity.

---

### DenseNet

Many feature connections:

```text
x0 → x1 → x2 → x3
 ↘──────↗
```

Can create memory-access challenges.

---

### EfficientNet

Requires coordinated scaling.

---

### RegNet

Requires thinking in terms of design spaces.

---

### ConvNeXt

Uses a comparatively clean block, but modern training recipes and large-scale training are important to reproduce its intended performance.

So:

$$
\boxed{\text{Training complexity ≠ FLOPs alone}}
$$

---

# 24. 📱 7. DEPLOYMENT SUITABILITY

This is where an **AI/ML engineer** needs to think differently.

Suppose:

```text
Model A
90% accuracy
500M parameters
100 GFLOPs
```

and:

```text
Model B
88% accuracy
10M parameters
2 GFLOPs
```

For a smartphone:

Model B may fit the hardware constraints much more easily.

For a high-end server:

Model A may be feasible.

So the correct question is not:

> "Which architecture is best?"

It is:

> **"Which architecture satisfies the requirements of my deployment environment?"**

You need the **application requirements**.

For:

Accuracy and robustness may dominate.

Memory + latency + power become critical.

Throughput may matter more than single-request latency.

This is why architecture evaluation is an **engineering optimization problem**, not a leaderboard.

AlexNet demonstrated that deep CNNs could scale to large datasets with GPUs and modern training techniques. VGG explored depth using simple 3×3 convolutions. Inception improved computational utilization through multi-scale parallel branches. ResNet solved the optimization problem associated with very deep networks through residual connections. DenseNet focused on feature reuse and connectivity, while ResNeXt introduced cardinality through grouped transformations. MobileNet and Xception focused on computational efficiency using depthwise separable convolutions. EfficientNet addressed systematic scaling of depth, width and resolution, RegNet addressed systematic network design spaces, and ConvNeXt revisited the CNN design using modern Transformer-era principles while remaining a pure ConvNet.**

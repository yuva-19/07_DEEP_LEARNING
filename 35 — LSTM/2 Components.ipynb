## 1. Cell State — `Cₜ` 🧠💾

The **cell state** is the main pathway for carrying information across time.

```text
Cₜ₋₁ ───────────────────────────► Cₜ
```

It can carry information forward while the gates control what gets:

* 🗑️ removed
* ➕ added
* ➡️ carried forward

Think of it as the LSTM's **long-term memory pathway**.

## 2. Hidden State — `hₜ` 📤

The **hidden state** represents the information exposed by the LSTM at the current time step.

It is used to:

* produce the current output
* pass information to the next time step

So an LSTM has **two states**:

```text
Cell state   → Cₜ
Hidden state → hₜ
```

A useful distinction:

> **`Cₜ` carries long-term information, while `hₜ` represents the current exposed/working state.**

---

# 3. Forget Gate — `fₜ` 🗑️

The forget gate decides:

> **"What information from the previous cell state should I keep or forget?"**

It receives:

```text
xₜ + hₜ₋₁
```

and produces values between `0` and `1` using the sigmoid function:

$$
\boxed{
f_t=\sigma(W_f[h_{t-1},x_t]+b_f)
}
$$

Interpretation:

```text
0 → forget
1 → keep
```

Example:

```text
Previous memory:
[Important] [Old] [Important] [Irrelevant]

Forget gate:
[   1.0   ] [0.2] [   1.0   ] [  0.0  ]
```

So the gate controls how much of each part of the previous memory survives.

# 4. Input Gate — `iₜ` ➕

The input gate decides:

> **"What new information should I store?"**

It also uses:

```text
xₜ + hₜ₋₁
```

and uses a sigmoid:

$$
\boxed{
i_t=\sigma(W_i[h_{t-1},x_t]+b_i)
}
$$

Its values range between:

```text
0 ─────────────── 1
│                 │
Don't allow       Allow
much              much
```

### ⚠️ Important

The input gate **doesn't contain the new information itself**.

It controls **how much candidate information** should be allowed into the cell state.

---

# 5. Candidate State — `C̃ₜ` 📝

The **candidate state** contains potential new information that could be added to the cell state.

It answers:

> **"What new information could I remember?"**

It is commonly calculated as:

$$
\boxed{
\tilde{C}_t=
\tanh(W_c[h_{t-1},x_t]+b_c)
}
$$

Its values approximately range from:

$$
-1 \rightarrow +1
$$

Think of it as:

```text
Candidate state
      ↓
"Here is some new information
that might be useful."
```

Then:

```text
Candidate information
          ↓
     Input Gate
          ↓
How much should be stored?
          ↓
      Cell State
```

### 🔑 Important distinction

**Candidate state** → proposes new information.

**Input gate** → controls how much of that information is stored.

---

# 6. Updating the Cell State 🔥

Now the forget gate, input gate, and candidate state work together.

The cell-state equation is:

$$
\boxed{
C_t=f_t\odot C_{t-1}
+i_t\odot\tilde{C}_t
}
$$

There are two parts.

### Part 1 — Keep old information

$$
f_t\odot C_{t-1}
$$

The forget gate decides how much previous memory survives.

### Part 2 — Add new information

$$
i_t\odot\tilde{C}_t
$$

The input gate decides how much candidate information gets added.

So:

```text
Previous Cell State
        │
        ▼
   Forget Gate
        │
        ▼
  Keep useful old
    information
        │
        ├───────────┐
                    │
Candidate ──► Input Gate
                    │
                    ▼
              New information
                    │
                    ▼
                   Cₜ
```

# 7. Output Gate — `oₜ` 👀

After obtaining the new cell state `Cₜ`, the LSTM needs to decide:

> **"What information should I expose as the hidden state?"**

This is the job of the output gate.

$$
\boxed{
o_t=\sigma(W_o[h_{t-1},x_t]+b_o)
}
$$

The hidden state is then:

$$
\boxed{
h_t=o_t\odot\tanh(C_t)
}
$$

Flow:

```text
Cₜ
 │
 ▼
tanh
 │
 ▼
 × ◄──── Output Gate
 │
 ▼
hₜ
```

# 8. Complete LSTM Information Flow 🧠

Now connect everything:

```text
             hₜ₋₁ + xₜ
                  │
       ┌──────────┼──────────┐
       ▼          ▼          ▼
   Forget       Input      Output
    Gate         Gate        Gate
       │          │
       │      Candidate
       │          │
       ▼          ▼
Cₜ₋₁ ───────► Cell State ───────► Cₜ
                                    │
                                    ▼
                              Output Gate
                                    │
                                    ▼
                                   hₜ
```

The logical sequence is:

```text
1️⃣ Decide what old information to forget
              ↓
2️⃣ Create candidate new information
              ↓
3️⃣ Decide how much new information to store
              ↓
4️⃣ Update the cell state
              ↓
5️⃣ Decide what information to expose
              ↓
6️⃣ Produce the hidden state
```

# 🧠 Six Components — Quick Revision

| Component           | Symbol | What does it do?                          |
| ------------------- | -----: | ----------------------------------------- |
| **Cell State**      |   `Cₜ` | Carries long-term information             |
| **Hidden State**    |   `hₜ` | Current exposed/working state             |
| **Forget Gate**     |   `fₜ` | Controls what old information is retained |
| **Input Gate**      |   `iₜ` | Controls what new information is stored   |
| **Candidate State** |  `C̃ₜ` | Provides candidate new information        |
| **Output Gate**     |   `oₜ` | Controls what becomes the hidden state    |

---

# 🔥 Easiest Way to Remember

Think of an LSTM as a **smart memory manager**:

```text
              OLD MEMORY
                  │
                  ▼
          🗑️ FORGET GATE
          "What should go?"
                  │
                  ▼
             CELL STATE
                  ▲
                  │
          ➕ INPUT GATE
          "What should I add?"
                  ▲
                  │
           📝 CANDIDATE
           "New information"
                  
                  │
                  ▼
           👀 OUTPUT GATE
           "What should I expose?"
                  │
                  ▼
             HIDDEN STATE
```

### The three questions:

**Forget Gate:**
🗑️ *What should I forget?*

**Input Gate + Candidate:**
💾 *What should I remember?*

**Output Gate:**
👀 *What should I output?*

---

# ⭐ Two Equations to Lock In

### Cell-state update

$$
\boxed{
C_t=f_t\odot C_{t-1}
+i_t\odot\tilde{C}_t
}
$$

### Hidden-state update

$$
\boxed{
h_t=o_t\odot\tanh(C_t)
}
$$

These are the **two most important equations from the component-level architecture**.

```text
Cₜ₋₁
  │
  ▼
🗑️ Forget ──────┐
                 │
                 ▼
              Cₜ₋₁ (filtered)
                 +
                 │
📝 Candidate → ➕ Input Gate
                 │
                 ▼
                Cₜ
                 │
                 ▼
             👀 Output Gate
                 │
                 ▼
                hₜ
```

### 🔑 One-line summary

> **LSTM controls long-term information using the cell state and three gates: forget what is unnecessary, add useful new information, and expose the relevant information through the hidden state.** 🧠🔥

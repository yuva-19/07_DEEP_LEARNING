# 🧠 Module 35 — LSTM Mathematics — Notes

This is the **mathematical core of LSTM**. The main thing to understand is how each gate is calculated and how the gates work together to update memory. 🔥

---

## 1. Common Input to the Gates

At time step `t`, the LSTM has:

```text
Current input        → xₜ
Previous hidden state → hₜ₋₁
```

These are concatenated:

$$
\boxed{[h_{t-1},x_t]}
$$

This combined vector is used by the forget, input, candidate, and output calculations.

```text id="g3qz5d"
hₜ₋₁ ──┐
        ├──► [ concatenate ] ──► LSTM gates
xₜ ────┘
```

![Image](https://images.openai.com/static-rsc-4/mnLNkbmBvT4enM5v1SvSdPoB5p_jfbt0fWL7shbqAjuKG7uDov1PcyAoTCyo2rVFmrpcXXhXLY7Cq1zbtrcP2k98VFf205EDI3rKs0y71d8DBfGqPAqMv4aulPPD3wdiDhL5-5_WLG6HEVAIkJyjcBltjmnaguNgATUBDMWf_8HpdBiTmth_ymdcYh_m7cmg?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/UUfPV8LbhUokYSg6lttSaC8n--NSwl2hW1KQisGygZccfwYeTbpR5i0SguKwiVm_a5gC537aVhCnKlCRtMlsMnDIaU5j1Q0YWujAmHB1vhg16FyDj3HCy9yZcGi1wV4pGkBCnNnTf62gl13ZR-MFmbDiErBvTr78aOEPdHMNIpS5ejK7W95JqRAUHHejxhkq?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/9AMBUDhrXheFByghR_Z8cL_1pff7Ep7OETipje6NBxynUtHk_2xlul5K4bz0m-AQNdy18HH9oBId7uQrACCU10RJr6eK4g_dDSaSL2pcgm8bqeGE2lBKcLS2B_ndMK2jOhKigJ2v7fBrfUjNEatN_tKn5WVb88pbOdK7TcSHJ2NY0ZdtrRFvBAYqEsVEkWD2?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/e7FNDM_0xLLj3ss5oC6lxjc2hcOs6iI7ecCeMeVIBzCq8oJhFK56F0WOtNUiSOwN1xF21z5--SMwsc5AnjZG5geF3c4Tccs8qUEvF4cTtlQJj5xmaQfD5b2iZMuEqA7KhxQNpfi4qkuf2G09apR6uSQAKbUKhJhuKhAiGACWCQ4Wy1lEWpGAmd4xzAUPj-SF?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/ocifJzurp2pU5KlLv3RADQIAguBiWUNTmcAPTi2yPu-EoZpDrbvqI3lpemZCdU1C1pVZeTfRW36IArnqdtUDQs7TS2E546oBqsJTEdB6_0c0BwiYxR_LN8XIS4-Xd_Cl88d6Kb39iU0AayrVTRGQ8YXb3ALF-_bFTk0P2kwcjAK0p8KQ98vzYZ5blLqJ3qse?purpose=fullsize)

---

# 2. Forget Gate — `fₜ` 🗑️

The forget gate decides:

> **What portion of the previous cell state should be retained?**

$$
\boxed{
f_t=\sigma(W_f[h_{t-1},x_t]+b_f)
}
$$

The sigmoid function produces values between `0` and `1`.

```text id="jz2k3k"
0 ─────────────── 1
│                 │
Forget            Keep
```

So:

* `0` → forget that information
* `1` → keep that information

The forget gate acts on the previous cell state:

$$
f_t\odot C_{t-1}
$$

---

# 3. Input Gate — `iₜ` ➕

The input gate decides:

> **How much new information should be added to the cell state?**

$$
\boxed{
i_t=\sigma(W_i[h_{t-1},x_t]+b_i)
}
$$

Again, sigmoid gives values between `0` and `1`.

```text id="o0g5ub"
iₜ
│
├── 0 → allow very little
│
└── 1 → allow a lot
```

⚠️ The input gate does **not** contain the new information itself.

It controls how much of the **candidate state** gets added.

---

# 4. Candidate State — `C̃ₜ` 📝

The candidate state represents **potential new information** that could be stored.

$$
\boxed{
\tilde C_t=
\tanh(W_c[h_{t-1},x_t]+b_c)
}
$$

The `tanh` output lies between:

$$
-1 \leq \tilde C_t \leq 1
$$

Think of it as:

> 📝 **"Here is the new information that I think might be useful."**

Then the input gate decides how much of it should actually be stored.

```text id="e9w3oh"
Candidate C̃ₜ
     │
     │ "What should I add?"
     ▼
Input gate iₜ
     │
     │ "How much?"
     ▼
New memory
```

---

# 5. Input Gate × Candidate State

These two work together:

$$
\boxed{
i_t\odot\tilde C_t
}
$$

`⊙` means **element-wise multiplication**.

Example:

$$
i_t=[0.8,0.2,1.0]
$$

$$
\tilde C_t=[0.5,0.9,-0.3]
$$

Then:

$$
i_t\odot\tilde C_t
=
[0.4,0.18,-0.3]
$$

So the input gate controls how much candidate information actually enters memory.

---

# 6. Cell-State Update 🧠

Now the old memory and new memory are combined.

$$
\boxed{
C_t=
f_t\odot C_{t-1}
+
i_t\odot\tilde C_t
}
$$

This equation has **two parts**.

### 🗑️ Old information

$$
f_t\odot C_{t-1}
$$

The forget gate determines how much old information remains.

### ➕ New information

$$
i_t\odot\tilde C_t
$$

The input gate determines how much candidate information is added.

Therefore:

```text id="7q5m9m"
Previous cell state Cₜ₋₁
          │
          ▼
     Forget gate fₜ
          │
          ▼
    Retained memory
          │
          ├───────────┐
                      │
Candidate C̃ₜ ──► Input gate iₜ
                      │
                      ▼
                New information
                      │
                      ▼
                     Cₜ
```

![Image](https://images.openai.com/static-rsc-4/7BOL33RlcaDR8_t_aBMxRvdSU0Gjvy3BYOuYgjdZOWc1s_K9Ia0CGcSNM5y6iMC_4bicr9nUhS4gCZbcBV7HOHgxihdDM__XpFhGdOWhl21t_FVdGBGaV4gwNGHGjMxj3EFGfyTFKOhjudxg5WDhCNkxR0wVe9cT0Zx5ALPCIyhjo--VJcH_3EdlHr17EePn?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/aN1b6PidtrPhRRkeOql4oKF3tpHGD7Sg_qcXRDsDPUsMFb4QEVVYoK80LpSJyjEWsY6HYGQi_QC8n5Ov6y0raHl8ZmEbxScOczW2X5jxCnwUL_BZdLAVMosf-kC6yqLQUTeJlhOUGDVv5cdwzFnwvmLRgFkce0BSSOf-PXnwot7m-qoM_byCLLqzeyyIw6Yl?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/XzXjGfeD5_I2Yt4VPhxaluuGZtVeTWR6yyE7TnQuPbvdG2bvxyfImdC5cqH99G0wzYaxpXOYB5D8rqW6M5XqVNxzRVc7bkegRkms7NcdX6l5F6t_1YbOaBHQ6gRcyZBQh0Dx4e97kBM8ug1PYT2cO4yxL4hMnCcGLmc0ce_HvDN9lcRrov8uws1LGc95vMnE?purpose=fullsize)

---

# 7. Output Gate — `oₜ` 👀

The output gate decides:

> **What portion of the current cell state should be exposed?**

$$
\boxed{
o_t=\sigma(W_o[h_{t-1},x_t]+b_o)
}
$$

Again:

$$
0\leq o_t\leq1
$$

The output gate controls the information that becomes the hidden state.

---

# 8. Hidden-State Equation 📤

The final hidden state is:

$$
\boxed{
h_t=o_t\odot\tanh(C_t)
}
$$

First:

$$
\tanh(C_t)
$$

Then the output gate controls it:

$$
o_t\odot\tanh(C_t)
$$

```text id="e2eg0k"
        Cₜ
         │
         ▼
       tanh
         │
         │ ×
         ▲
         │
        oₜ
         │
         ▼
        hₜ
```

# 9. All Six LSTM Equations 🔥

### ① Forget Gate

$$
\boxed{
f_t=\sigma(W_f[h_{t-1},x_t]+b_f)
}
$$

### ② Input Gate

$$
\boxed{
i_t=\sigma(W_i[h_{t-1},x_t]+b_i)
}
$$

### ③ Candidate State

$$
\boxed{
\tilde C_t=
\tanh(W_c[h_{t-1},x_t]+b_c)
}
$$

### ④ Cell State

$$
\boxed{
C_t=
f_t\odot C_{t-1}
+
i_t\odot\tilde C_t
}
$$

### ⑤ Output Gate

$$
\boxed{
o_t=\sigma(W_o[h_{t-1},x_t]+b_o)
}
$$

### ⑥ Hidden State

$$
\boxed{
h_t=o_t\odot\tanh(C_t)
}
$$

# 🧠 10. The Complete LSTM Calculation

```text id="4pl6uq"
              hₜ₋₁ + xₜ
                   │
       ┌───────────┼────────────┐
       ▼           ▼            ▼
   Forget Gate  Input Gate  Output Gate
       fₜ          iₜ           oₜ
                    │
                    ▼
              Candidate C̃ₜ
                    │
                    ▼
       ┌───────────────────────┐
       │                       │
 Cₜ₋₁ × fₜ  +  C̃ₜ × iₜ
       │                       │
       └───────────┬───────────┘
                   ▼
                  Cₜ
                   │
                   ▼
                tanh(Cₜ)
                   │
                   × ◄──── oₜ
                   │
                   ▼
                  hₜ
```

---

# 11. Why Sigmoid for Gates? 🤔

The three gates use **sigmoid**:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

because:

$$
\boxed{0\leq\sigma(z)\leq1}
$$

This makes sigmoid useful as a **control mechanism**:

```text id="wmk8ju"
0 → block
0.5 → partially allow
1 → allow
```

Therefore:

```text
Forget Gate → sigmoid
Input Gate  → sigmoid
Output Gate → sigmoid
```

---

# 12. Why `tanh` for Candidate?

The candidate state uses:

$$
\tanh(z)
$$

which produces values between:

$$
-1 \text{ and } +1
$$

This allows the candidate to represent signed information.

```text id="6r3jif"
-1 ←──── 0 ────→ +1
```

So:

```text
Candidate State → tanh
```

The hidden state also uses `tanh(Cₜ)` before applying the output gate.

# 🔥 13. Don't Memorize — Understand the Story

### Step 1 — Forget

$$
f_t=\sigma(...)
$$

🗑️ **What old information should I keep?**

### Step 2 — Candidate

$$
\tilde C_t=\tanh(...)
$$

📝 **What new information could I store?**

### Step 3 — Input

$$
i_t=\sigma(...)
$$

➕ **How much of that new information should I store?**

### Step 4 — Update memory

$$
C_t=f_t\odot C_{t-1}+i_t\odot\tilde C_t
$$

🧠 **Update the cell state.**

### Step 5 — Output

$$
o_t=\sigma(...)
$$

👀 **What information should I expose?**

### Step 6 — Hidden state

$$
h_t=o_t\odot\tanh(C_t)
$$

📤 **Produce the current hidden state.**

# ⭐ Must-Know Equations

Keep these six equations together:

$$
\boxed{f_t=\sigma(W_f[h_{t-1},x_t]+b_f)}
$$

$$
\boxed{i_t=\sigma(W_i[h_{t-1},x_t]+b_i)}
$$

$$
\boxed{\tilde C_t=\tanh(W_c[h_{t-1},x_t]+b_c)}
$$

$$
\boxed{C_t=f_t\odot C_{t-1}+i_t\odot\tilde C_t}
$$

$$
\boxed{o_t=\sigma(W_o[h_{t-1},x_t]+b_o)}
$$

$$
\boxed{h_t=o_t\odot\tanh(C_t)}
$$

### 🧠 The easiest memory trick

**Forget → Candidate → Input → Cell Update → Output → Hidden**

```text id="o1xj4m"
🗑️ Forget
    ↓
📝 Candidate
    ↓
➕ Input
    ↓
🧠 Cell State
    ↓
👀 Output
    ↓
📤 Hidden State
```

> **LSTM learns what to forget, what new information to store, and what information to expose at every time step.** 🔥

---

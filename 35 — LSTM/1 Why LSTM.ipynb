# 🧠 Module 35 — LSTM: Why LSTM? — Notes

This section is the **motivation for LSTM**. The main question is:

> **Why wasn't the normal RNN enough?**

---

## 1. The Problem with Vanilla RNNs 🔄

A normal RNN carries information through the hidden state:

```text
x₁ → h₁ → h₂ → h₃ → h₄ → ... → hₜ
```

The idea is that earlier information should continue influencing later predictions.

But when the sequence becomes long, this becomes difficult.

```text
Important information
        ↓
h₁ → h₂ → h₃ → h₄ → ... → h₅₀
                              ↓
                         prediction
```

The information has to survive through many recurrent transformations.

---

## 2. Vanishing Gradient 📉

During training, RNNs use **Backpropagation Through Time (BPTT)**.

The gradient travels backward:

```text
h₅₀ → h₄₉ → h₄₈ → ... → h₂ → h₁
```

Because gradients are repeatedly multiplied, they can become extremely small:

```text
1
↓
0.5
↓
0.25
↓
0.125
↓
0.0625
↓
...
↓
≈ 0
```

This is the **vanishing-gradient problem**.

### Why is it important?

The early time steps receive a very weak learning signal.

Therefore, the RNN struggles to learn relationships between information that is **far apart in the sequence**.

$$
\boxed{\text{Vanishing gradients} \rightarrow
\text{difficulty learning long-term dependencies}}
$$

# 3. Long-Term Dependency 🧠⏳

A long-term dependency occurs when **information from an earlier part of a sequence is needed much later**.

For example:

> **"I grew up in France. ... [many words] ... I speak fluent ____."**

The important relationship is:

```text
France
  ↓
many time steps
  ↓
French
```

The RNN needs to preserve the information **"France"** for a long time.

```text
France
  ↓
h₁ → h₂ → h₃ → ... → h₅₀
                         ↓
                      French
```

A vanilla RNN can struggle to maintain this information over such a long distance.

# 4. Why Can't Normal RNN Just Remember Everything?

The vanilla RNN mainly relies on its hidden state:

```text
h₁ → h₂ → h₃ → h₄ → ... → hₜ
```

At every step, information is transformed and compressed into the next hidden state.

But not every piece of information is equally important.

For example:

```text
Information A → important later
Information B → irrelevant later
Information C → important later
Information D → irrelevant later
```

A basic RNN does not have an explicit mechanism to independently control:

* 🗑️ What information to **forget**
* 💾 What information to **store**
* 👀 What information to **output**

This is the key motivation for LSTM.

---

# 5. LSTM — Long Short-Term Memory 🚀

**LSTM** stands for:

> **Long Short-Term Memory**

LSTM introduces an additional state called the:

$$
\boxed{\text{Cell State } C_t}
$$

So unlike a vanilla RNN, an LSTM has two important states:

```text
Hidden state → hₜ
Cell state   → Cₜ
```

Conceptually:

```text
        Cell State
Cₜ₋₁ ───────────────► Cₜ
          │
          │
xₜ ────► LSTM ─────► hₜ
hₜ₋₁ ───►
```

# 6. The Big Idea Behind LSTM 💡

LSTM provides a more controlled mechanism for managing information over time.

Instead of simply passing information through the hidden state, LSTM uses a **cell state and gates**.

Conceptually:

```text
Previous memory
      ↓
┌─────────────┐
│   LSTM      │
│             │
│ Forget?     │
│ Store?      │
│ Output?     │
└─────────────┘
      ↓
Updated memory
```

The gates learn how information should flow.

---

# 7. Three Main Gates 🚪

LSTM has three major gates.

### 🔴 Forget Gate

Controls what information from the previous cell state should be discarded.

> **"What should I forget?"**

---

### 🟢 Input Gate

Controls what new information should be stored.

> **"What new information should I remember?"**

---

### 🔵 Output Gate

Controls what information should be exposed as the hidden state/output.

> **"What information should I output right now?"**

Conceptually:

```text
Previous memory
      │
      ▼
 Forget Gate
      │
      ▼
 Cell State ◄── Input Gate
      │
      ▼
 Output Gate
      │
      ▼
    hₜ
```

---

# 8. RNN vs LSTM 🔥

### Vanilla RNN

```text
xₜ + hₜ₋₁
     ↓
   RNN
     ↓
    hₜ
```

### LSTM

```text
             Cₜ₋₁
               │
               ▼
        ┌─────────────┐
xₜ ────►│    LSTM     │────► Cₜ
hₜ₋₁ ──►│             │
        └──────┬──────┘
               │
               ▼
              hₜ
```

The major architectural addition is:

$$
\boxed{\text{Cell State + Gates}}
$$

---

# 9. Why LSTM Helps 🧠

The purpose of the LSTM design is to make information flow through the sequence more controlled.

Instead of treating all information in the same way, the network can learn to:

```text
🗑️ Forget irrelevant information
        +
💾 Keep important information
        +
➕ Add useful new information
        +
👀 Expose relevant information
```

This makes LSTM better suited to handling **long-term dependencies** than a basic vanilla RNN.

---

# 🔥 Complete RNN → LSTM Motivation

Remember this chain:

```text
Vanilla RNN
     ↓
Hidden state carries information
     ↓
Long sequence
     ↓
BPTT through many time steps
     ↓
Vanishing gradients
     ↓
Difficulty learning long-term dependencies
     ↓
        LSTM
     ↓
Cell State + Gates
     ↓
Controlled information flow
     ↓
Better handling of long-term dependencies
```

# 🧠 Must Remember

### Why LSTM?

> **LSTM was designed to address the difficulty vanilla RNNs have in learning and preserving long-term dependencies, particularly because of the vanishing-gradient problem.**

### Vanilla RNN

$$
\boxed{\text{Hidden State}}
$$

### LSTM

$$
\boxed{\text{Hidden State + Cell State + Gates}}
$$

### Three gates

| Gate           | Main role                            |
| -------------- | ------------------------------------ |
| 🔴 Forget Gate | Decide what to discard               |
| 🟢 Input Gate  | Decide what new information to store |
| 🔵 Output Gate | Decide what information to expose    |

### ⭐ Core idea

```text
RNN:
"What information do I carry?"

LSTM:
"What should I forget,
what should I remember,
and what should I output?"
```

# 1️⃣ Problem VGG Addressed

After AlexNet showed that deep CNNs could work extremely well on large-scale image recognition, an important question remained:

> **Can we systematically improve CNN performance by making the network deeper while keeping its architecture simple?**

Earlier CNNs used different kernel sizes and architectural choices.

VGG investigated a much more systematic approach:

```text
Make CNN deeper
       +
Keep architecture simple
       +
Use small 3×3 convolutions
```

The VGG work studied networks with **increasing depth**, including VGG-16 and VGG-19.

---

# 2️⃣ Main Idea 🔥🔥🔥

The central idea is:

> **Stack multiple small 3×3 convolution layers instead of using larger convolution filters, and build a deeper network using repeated blocks.**

Instead of something like:

```text id="8l5g1u"
Large convolution
      ↓
Pooling
```

VGG commonly uses:

```text id="z0g2ve"
3×3 Conv
   ↓
ReLU
   ↓
3×3 Conv
   ↓
ReLU
   ↓
Pooling
```

Then repeat this pattern.

---

# 3️⃣ Why 3×3 Convolutions? 🔥

This is the **most important VGG concept**.

Imagine two consecutive 3×3 convolutions:

```text id="g6s3ai"
3×3 Conv
   ↓
3×3 Conv
```

The effective receptive field becomes approximately:

$$
5\times5
$$

So two 3×3 layers can cover the same spatial region as a 5×5 operation, while introducing an extra nonlinear activation between them.

```text id="7s8v4f"
5×5 Conv

        vs

3×3 Conv
   ↓
ReLU
   ↓
3×3 Conv
```

With the stacked version, you get:

* multiple nonlinearities
* more depth
* hierarchical feature transformation
* relatively small kernels

This became a major design principle of VGG.

The classic VGG-16 has:

```text
13 convolutional layers
+
3 fully connected layers
=
16 learnable layers
```

Hence:

$$
\boxed{\text{VGG-16}}
$$

This is a very important architectural pattern.

Start:

```text id="z9cm1j"
224 × 224 × 3
```

After Block 1:

```text id="7xoylq"
112 × 112 × 64
```

After Block 2:

```text id="pr0v2w"

```

After Block 3:

```text id="0i7ztr"
28 × 28 × 256
```

After Block 4:

```text id="wq1d0n"

```

After Block 5:

```text id="d2b17u"
7 × 7 × 512
```

So there is an important pattern:

```text id="p5qk0w"
Spatial size ↓
       +
Channels ↑
```

Visually:

```text id="xqrx6e"
224×224×3
     ↓
112×112×64
     ↓
56×56×128
     ↓
28×28×256
     ↓
14×14×512
     ↓
7×7×512
```

🔥 This is a pattern you will see in many CNN architectures.

# 7️⃣ Why Does VGG Increase the Number of Channels?

Early layers process simple features:

```text id="8u7plw"
Edges
Lines
Corners
```

Later layers need richer representations:

```text id="2jrl0s"
Textures
Shapes
Object parts
Complex patterns
```

Therefore VGG progressively increases the number of feature maps:

```text id="vwnq4h"
64
 ↓
128
 ↓
256
 ↓
512
```

At the same time, pooling reduces spatial dimensions.

So the network gradually changes from:

```text id="5v3q7a"
Large spatial representation
+
Few channels
```

to:

```text id="x0h8cg"
Small spatial representation
+
Many channels
```

---

# 8️⃣ Important Innovation 🔥🔥🔥

The most important VGG innovation is:

## **Small 3×3 convolution filters + increased network depth**

Instead of relying on large convolution filters, VGG showed the effectiveness of repeatedly applying **3×3 convolutions**.

For example:

```text id="xj14by"
3×3
 ↓
3×3
 ↓
3×3
```

This allows the network to build increasingly complex representations while maintaining a simple, repeatable architecture.

---

# 9️⃣ Why Stacking 3×3 Layers Is Powerful

Let's understand this numerically.

One 3×3 convolution:

```text id="jv5dcm"
Receptive field = 3×3
```

Two stacked 3×3 convolutions:

```text id="p2t7cj"
3×3
 ↓
3×3

Effective RF = 5×5
```

Three:

```text id="7un9as"
3×3
 ↓
3×3
 ↓
3×3

Effective RF = 7×7
```

So:

$$
3\times3 + 3\times3 \rightarrow 5\times5
$$

and:

$$
3\times3 + 3\times3 + 3\times3
\rightarrow 7\times7
$$

But now we have **multiple nonlinear transformations** between them.

That's one of the important reasons stacked small kernels are powerful.

---

# 🔟 VGG Architecture Philosophy

VGG is almost like a recipe:

```text id="7ld3k4"
Conv 3×3
Conv 3×3
Pool
        ↓
Conv 3×3
Conv 3×3
Pool
        ↓
Conv 3×3
Conv 3×3
Conv 3×3
Pool
        ↓
...
```

This makes VGG:

### Simple

The architecture follows repeated patterns.

### Deep

More convolution layers are stacked.

### Hierarchical

Features become increasingly complex.

### Easy to understand

The architecture is much more systematic than many complicated architectures.

# 1️⃣1️⃣ VGG vs AlexNet

This comparison is **very important** because VGG came after AlexNet.

|                 | AlexNet                    | VGG                                |
| --------------- | -------------------------- | ---------------------------------- |
| Main goal       | Large-scale CNN success    | Study deeper CNNs systematically   |
| Architecture    | More varied                | Very systematic                    |
| Common kernels  | 11×11, 5×5, 3×3            | Mostly **3×3**                     |
| Depth           | 8 learnable layers         | VGG-16 = 16                        |
| Main idea       | Deep CNN + modern training | **Small kernels + depth**          |
| Structure       | Less uniform               | Repeated blocks                    |
| Main limitation | —                          | Large parameter/computational cost |

---

# 1️⃣2️⃣ VGG's Major Limitation ⚠️

VGG is conceptually beautiful, but it is **expensive**.

The fully connected layers contain a huge number of parameters.

For example, after the convolutional portion:

```text id="b4pk9e"
7 × 7 × 512
```

Flatten:

$$
7\times7\times512=25,088
$$

Then connecting this to:

```text id="l6x1pr"
4096 neurons
```

requires:

$$
25,088\times4,096
$$

weights — over **100 million weights** just for that connection, before biases.

So VGG can have a very large parameter count.

This motivated later architectures to focus more heavily on **computational efficiency**.

## 4. Important Innovation

> **Extensive use of small 3×3 convolutions in a deeper, highly systematic architecture.**

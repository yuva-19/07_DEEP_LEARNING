## 1. What is a Bidirectional LSTM?

A **Bidirectional LSTM (BiLSTM)** consists of **two separate LSTMs** processing the same sequence in opposite directions:

* ➡️ Forward LSTM
* ⬅️ Backward LSTM

Their hidden representations are then combined.

$$
\boxed{\text{BiLSTM}=\text{Forward LSTM}+\text{Backward LSTM}}
$$

The main advantage is that the representation at a position can use:

$$
\boxed{\text{Past context + Future context}}
$$

---

# 2. Forward LSTM ➡️

The forward LSTM processes:

$$
x_1\rightarrow x_2\rightarrow\cdots\rightarrow x_T
$$

Its hidden state is:

$$
\boxed{
\overrightarrow{h_t}
=
LSTM(x_t,\overrightarrow{h_{t-1}},\overrightarrow{C_{t-1}})
}
$$

It captures information from the **past/left-side context**.

```text id="6f5g3r"
x₁ → x₂ → x₃ → x₄
↓    ↓    ↓    ↓
h⃗₁  h⃗₂  h⃗₃  h⃗₄
```

The forward LSTM maintains its own:

* hidden state \(\overrightarrow{h_t}\)
* cell state \(\overrightarrow{C_t}\)

---

# 3. Backward LSTM ⬅️

The backward LSTM processes the sequence in reverse:

$$
x_T\rightarrow x_{T-1}\rightarrow\cdots\rightarrow x_1
$$

Its hidden state is:

$$
\boxed{
\overleftarrow{h_t}
=
LSTM(x_t,\overleftarrow{h_{t+1}},\overleftarrow{C_{t+1}})
}
$$

It captures information from the **future/right-side context**.

```text id="m9j1m5"
x₁ ← x₂ ← x₃ ← x₄
↓    ↓    ↓    ↓
h⃖₁  h⃖₂  h⃖₃  h⃖₄
```

The backward LSTM has its **own** cell and hidden states.

---

# 4. Combining the Representations 🔀

At each time step \(t\), we have:

$$
\overrightarrow{h_t}
$$

from the forward LSTM and:

$$
\overleftarrow{h_t}
$$

from the backward LSTM.

They are usually concatenated:

$$
\boxed{
h_t=
[\overrightarrow{h_t};
\overleftarrow{h_t}]
}
$$

```text id="4z2b7c"
Forward LSTM
     │
     ▼
  h⃗ₜ ─────┐
           ├──► CONCATENATE ──► hₜ
  h⃖ₜ ─────┘
     ▲
     │
Backward LSTM
```

Therefore:

$$
\boxed{
h_t=\text{past context}+\text{future context}
}
$$

---

# 5. Complete Architecture

![Image](https://images.openai.com/static-rsc-4/aB7kEm2NJIRi40WHds5IOuZCENx_9YWgMByooPndsW1Dmbnw440grCq9uDc1IPdN62KimP_uGfelWMx0S2QCdG5c_UTuaY9JfknfWxLQopSIhHkKHwGTf1eTa3-N12sNgPSzhJGwuEeIQhe4b3vExNnjA4zhzeOMqAqXe-VumSXJQOLQOWS5FaLivi8MJMyH?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/C-YBsgZJnDYb7tT3GUJ9Y0qvmPYOA_XIpsaXb04Vzv58-xAliQqe6CMsvlehh1NCzvJtSburY16qYbQ0O8AKzAbpHQ1y--Q_XWu8EuPzmQ9DYLtOCU8WY01z6PQq80JYiQkCwtKVoo4_Cn6GfuNuMFJ6GJbSEUPPdyZ5ppbhoacoy0rsmy6qT8VuGOct-AB_?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/9Z8HXx-DwzMrvu-Kujoh9jQYIWrrX0viP1YxsfPjMSE_oAF276yV1sSjKfDOtUqP4wH69D3TnJSrZyeIjwKj-YnA-9wTxELBj6gUsNjlkiVWP9Ge5ZHiPguFmRuJyq0w84bIqxth-k03N-whh8wi2fayWAuyfnDV2GzA1UcTCTs2zuXZ1xkfDj2e1-Od1fKC?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/4e9-__YPXuOvMBf7pP3sCkXTyaB0AWHOhuPZ6uxhC1vOKtzVVXtKbKSRTMSiaLEhPYumrUiLwPB-0wNbp7MFgVaqBh1wIf2n3igcUUzvgVL4bQTonztOMRVTr_vOPaI-YHraVH78VRYwo6LiFUm8UcrgImaMO19RoeyzsFLddRWvRHMBMedK56kjh2LeFaiu?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/2HBUYrak3E_oYSGUAQpmNAuK6YxZqs6sp1UqfxB07w_0RI7Gg9jrAynctQIXoWiK4YZPKsDDqpx7hOjQJcsHJYFxpGgA8IUHZKT_5gkVQWYozDYDI3iriPl-wEVcFmmAU3R8F-lXowrvduFTbjGFRrQxCp60g-Z9yBvFEEmou-RMWs-q3Z4Mf65gDO2KLosv?purpose=fullsize)

```text id="q9e1py"
                 FORWARD LSTM
              ─────────────────►

x₁ ───────── x₂ ───────── x₃ ───────── x₄
 │            │            │            │
 ▼            ▼            ▼            ▼
h⃗₁           h⃗₂           h⃗₃           h⃗₄
 │            │            │            │
 └──────┐     └──────┐     └──────┐     │
        ▼            ▼            ▼     ▼
      CONCAT       CONCAT       CONCAT CONCAT
        │            │            │     │
        ▼            ▼            ▼     ▼
       h₁           h₂           h₃    h₄
        ▲            ▲            ▲     ▲
        │            │            │     │
      h⃖₁           h⃖₂           h⃖₃   h⃖₄
        ▲            ▲            ▲     ▲
        │            │            │     │
x₁ ───────── x₂ ───────── x₃ ───────── x₄
                 ◄─────────────────
                    BACKWARD LSTM
```

# 6. What Happens Inside Each LSTM?

An important point:

> BiLSTM does **not** introduce a new type of LSTM cell.

It simply uses **two independent LSTMs**.

Each direction has its own:

* Forget gate
* Input gate
* Candidate state
* Cell state
* Output gate
* Hidden state

So conceptually:

```text id="h9s0je"
Forward LSTM
    ↓
f⃗ₜ, i⃗ₜ, C⃗ₜ, o⃗ₜ, h⃗ₜ


Backward LSTM
    ↓
f⃖ₜ, i⃖ₜ, C⃖ₜ, o⃖ₜ, h⃖ₜ
```

The two LSTMs have **separate parameters**.

---

# 7. Why Is BiLSTM Useful?

Consider:

> **"I went to the bank yesterday."**

To understand **bank**, information from the left can be useful:

```text id="s75m2h"
I → went → to → bank
```

But information from the right can also be useful:

```text id="i0sq3b"
bank → yesterday
```

BiLSTM can therefore build a representation using both sides:

```text id="ym5qmx"
       LEFT CONTEXT       RIGHT CONTEXT
             ↓                  ↓
        "I went to"        "yesterday"
              \                /
               \              /
                    bank
                      ↓
              richer representation
```

# 8. BiRNN vs BiLSTM

| Feature                       | BiRNN          | BiLSTM            |
| ----------------------------- | -------------- | ----------------- |
| Forward network               | RNN            | LSTM              |
| Backward network              | RNN            | LSTM              |
| Future context                | ✅              | ✅                 |
| Long-term memory mechanism    | Basic RNN      | LSTM cell + gates |
| Long-term dependency handling | More difficult | Better            |
| Complexity                    | Lower          | Higher            |
| Parameters                    | Lower          | Higher            |

The key idea:

$$
\boxed{
\text{BiLSTM}
=
\text{Bidirectional processing}
+
\text{LSTM memory mechanism}
}
$$

# 9. Applications 🎯

BiLSTM is useful when the **complete sequence is available**.

Common applications include:

* 📝 Named Entity Recognition
* 🏷️ Part-of-Speech Tagging
* 📄 Sequence labeling
* 🗣️ Speech processing
* 🔤 Natural Language Processing
* 🧬 Biological sequence analysis
* Text classification

---

# ⚠️ 10. Important Limitation

BiLSTM needs access to the **future part of the sequence**.

Therefore, it is not naturally suitable for strict **real-time causal prediction**, where future inputs are unavailable.

```text id="2g7v5f"
Current time
     │
     ▼
Future data doesn't exist yet
     │
     ▼
Backward processing unavailable
```

So:

> **BiLSTM is particularly useful for offline/full-sequence processing rather than strictly causal real-time prediction.**

---

# 🧠 11. Mathematical Summary

### Forward LSTM

$$
\boxed{
\overrightarrow{h_t}
=
LSTM(x_t,\overrightarrow{h_{t-1}},\overrightarrow{C_{t-1}})
}
$$

### Backward LSTM

$$
\boxed{
\overleftarrow{h_t}
=
LSTM(x_t,\overleftarrow{h_{t+1}},\overleftarrow{C_{t+1}})
}
$$

### Combined representation

$$
\boxed{
h_t=
[\overrightarrow{h_t};
\overleftarrow{h_t}]
}
$$

---

# 🔥 12. One-Minute Revision

```text id="1f7t7y"
                 BiLSTM
                   │
          ┌────────┴────────┐
          ▼                 ▼
   Forward LSTM       Backward LSTM
       ➡️                  ⬅️
          │                 │
          ▼                 ▼
       h⃗ₜ                h⃖ₜ
          │                 │
          └───────┬─────────┘
                  ▼
             CONCATENATE
                  │
                  ▼
                 hₜ
```

### Remember:

$$
\boxed{\text{Forward LSTM} \rightarrow \text{past context}}
$$

$$
\boxed{\text{Backward LSTM} \rightarrow \text{future context}}
$$

$$
\boxed{\text{BiLSTM} \rightarrow \text{past + future context}}
$$

## ⭐ Core takeaway

> **A Bidirectional LSTM uses two independent LSTMs running in opposite directions and combines their hidden states, allowing each sequence position to use both past and future context while benefiting from LSTM's long-term memory mechanism.** 🔥

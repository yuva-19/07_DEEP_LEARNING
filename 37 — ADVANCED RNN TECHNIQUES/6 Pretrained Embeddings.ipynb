# 🔥 Pretrained Embeddings

Bro, this is an important NLP concept because instead of training word representations **from scratch**, we can start with embeddings that were already learned from a huge text corpus.

---

## 1. What are Pretrained Embeddings?

A **pretrained embedding** is a vector representation of a word that has already been learned from a large dataset.

Instead of starting with:

```text
"king" → random vector
"queen" → random vector
"man" → random vector
"woman" → random vector
```

we use embeddings that were trained beforehand:

```text
"king"  → [0.21, -0.45, 0.78, ...]
"queen" → [0.19, -0.41, 0.81, ...]
"man"   → [0.11, -0.32, 0.55, ...]
"woman" → [0.13, -0.30, 0.59, ...]
```

The vectors contain **learned semantic information** about words.

# 2. Why Do We Need Pretrained Embeddings?

Suppose we are building a sentiment classifier.

Our dataset might contain only:

```text
50,000 sentences
```

If we train embeddings from scratch, the model has to learn:

```text
word → vector representation
```

**and**

```text
vector representations → sentiment
```

at the same time.

That can be difficult when the dataset is small.

Instead, we can use an embedding trained on a massive corpus:

```text
Huge text corpus
      ↓
Pretraining
      ↓
Word Embeddings
      ↓
Our NLP task
      ↓
Sentiment / Classification / etc.
```

So the model starts with useful linguistic knowledge.

---

# 3. Famous Pretrained Embeddings

Some classic pretrained embedding methods are:

| Embedding    | Main idea                                 |
| ------------ | ----------------------------------------- |
| **Word2Vec** | Learns word relationships from context    |
| **GloVe**    | Uses global word co-occurrence statistics |
| **FastText** | Represents words using character n-grams  |

### Word2Vec

Word2Vec learns embeddings based on the words surrounding a target word.

For example:

```text
The cat sat on the mat
        ↑
     context
```

Words appearing in similar contexts tend to obtain similar vectors.

---

### GloVe

**GloVe = Global Vectors for Word Representation**

It uses statistics describing how frequently words co-occur across a large corpus.

For example:

```text
        cat   dog   car   engine
cat      —    high   low    low
dog     high   —     low    low
car      low  low     —    high
engine   low  low    high    —
```

These co-occurrence patterns help create meaningful vectors.

---

### FastText

FastText goes one step further by representing words using **subword information**.

For example:

```text
playing
```

can be represented using character pieces such as:

```text
play
lay
ayi
yin
ing
...
```

This makes FastText particularly useful for:

* rare words
* unknown words
* morphologically rich languages

# 4. How Are Pretrained Embeddings Used?

Suppose our vocabulary is:

```text
["cat", "dog", "car", "king", "queen"]
```

A pretrained embedding might give each word a 300-dimensional vector:

```text
cat   → [0.12, -0.34, 0.51, ...]  ← 300 values
dog   → [0.15, -0.31, 0.48, ...]
car   → [0.72,  0.18, -0.22, ...]
king  → [0.41, -0.11, 0.63, ...]
queen → [0.39, -0.09, 0.67, ...]
```

Then our neural network uses these vectors as input.

```text
Text
 ↓
Tokenization
 ↓
Words
 ↓
Pretrained Embedding
 ↓
Dense Vectors
 ↓
RNN / LSTM / GRU / CNN / Transformer
 ↓
Output
```

# 5. Pretrained Embeddings + RNN

This connects directly to what you've been learning. 🔥

Suppose:

```text
"I love this movie"
```

After tokenization:

```text
I
love
this
movie
```

Each word gets converted into a pretrained vector:

```text
I       → vector
love    → vector
this    → vector
movie   → vector
```

Then:

```text
vectors
   ↓
Embedding sequence
   ↓
LSTM / GRU
   ↓
Hidden states
   ↓
Classifier
   ↓
Positive
```

So the RNN doesn't receive raw words.

It receives **numerical vector representations**.

# 6. Frozen vs Trainable Embeddings

This is an **important interview concept**.

There are two common approaches.

### 🔒 Frozen Embeddings

The pretrained vectors are used as they are.

```text
Pretrained Embedding
        ↓
     FROZEN
        ↓
      LSTM
        ↓
     Output
```

The embedding weights are **not updated** during training.

### Why?

You want to preserve the pretrained representation and reduce the number of parameters being trained.

---

### 🔓 Fine-tuned Embeddings

The pretrained vectors are used as the **initialization**, but training is allowed to modify them.

```text
Pretrained Embedding
        ↓
   Initial weights
        ↓
      Training
        ↓
  Updated Embedding
```

This allows the embeddings to adapt to the specific task.

For example:

```text
General language embedding
          ↓
     Sentiment task
          ↓
Task-specific embedding
```

---

# 7. Frozen vs Fine-Tuned

| Feature                    | Frozen       | Fine-tuned                          |
| -------------------------- | ------------ | ----------------------------------- |
| Starting point             | Pretrained   | Pretrained                          |
| Embedding updated?         | ❌ No         | ✅ Yes                               |
| Training parameters        | Fewer        | More                                |
| Task adaptation            | Limited      | Greater                             |
| Risk of overfitting        | Lower        | Higher                              |
| Useful with small datasets | Often useful | Can be useful with careful training |

The choice depends on the dataset, task, and training setup.

---

# 8. Why Are Pretrained Embeddings Powerful?

Imagine we want to train a model on:

```text
"That film was fantastic"
```

A randomly initialized embedding initially knows nothing about:

```text
fantastic
```

But a pretrained embedding may already encode relationships such as:

```text
fantastic ≈ excellent
fantastic ≈ wonderful
fantastic ≈ amazing
```

So our downstream model starts with a much more informative representation.

---

# 9. Important Limitation ⚠️

Classic pretrained embeddings such as Word2Vec and GloVe generally assign **one fixed vector per word**.

For example:

```text
bank → one vector
```

But:

```text
I went to the bank.
```

and

```text
I sat beside the river bank.
```

have different meanings of "bank".

A single static embedding struggles to represent this contextual difference.

This limitation motivated **contextual embeddings** and later models such as:

* ELMo
* BERT
* GPT-style models

These produce representations influenced by the surrounding context.

```text
Static embedding:

bank → [same vector]

Contextual representation:

"bank" + surrounding words
          ↓
     context-aware vector
```

# 🧠 Interview-Level Summary

> **Pretrained embeddings are vector representations of words learned beforehand from large text corpora and reused in downstream NLP models. They provide useful semantic information before task-specific training begins. They can either be frozen or fine-tuned during downstream training.**

### Remember this chain:

```text
Large Corpus
     ↓
Pretraining
     ↓
Word Embeddings
     ↓
Our NLP Dataset
     ↓
Embedding Layer
     ↓
RNN / LSTM / GRU
     ↓
Task Output
```

And the key distinction:

```text
Word2Vec / GloVe / FastText
        ↓
Mostly STATIC embeddings

ELMo / BERT / GPT
        ↓
CONTEXTUAL representations
```

🔥 **Core memory:**
**Pretrained = already learned representation → reuse it instead of starting from random vectors.**

## 1. What is a Bidirectional RNN?

A normal RNN processes a sequence in **one direction**:

$$
x_1\rightarrow x_2\rightarrow x_3\rightarrow\cdots\rightarrow x_T
$$

A **Bidirectional RNN (BiRNN)** processes the sequence in **both directions**:

* ➡️ Forward direction
* ⬅️ Backward direction

The two hidden representations are then combined.

$$
\boxed{\text{BiRNN}=\text{Forward RNN}+\text{Backward RNN}}
$$

---

# 2. Forward Direction ➡️

The forward RNN processes the sequence from beginning to end:

$$
x_1\rightarrow x_2\rightarrow\cdots\rightarrow x_T
$$

Its hidden state is:

$$
\boxed{
\overrightarrow{h_t}
=
f(x_t,\overrightarrow{h_{t-1}})
}
$$

At time \(t\), it contains information from the sequence's **past/left-side context**.

```text
x₁ → x₂ → x₃ → x₄
↓    ↓    ↓    ↓
h⃗₁  h⃗₂  h⃗₃  h⃗₄
```

---

# 3. Backward Direction ⬅️

The backward RNN processes the same sequence from end to beginning:

$$
x_T\rightarrow x_{T-1}\rightarrow\cdots\rightarrow x_1
$$

Its hidden state is:

$$
\boxed{
\overleftarrow{h_t}
=
f(x_t,\overleftarrow{h_{t+1}})
}
$$

It captures information from the **future/right-side context**.

```text
x₁ ← x₂ ← x₃ ← x₄
↓    ↓    ↓    ↓
h⃖₁  h⃖₂  h⃖₃  h⃖₄
```

---

# 4. Combined Representation 🔀

At each time step, there are two hidden states:

$$
\overrightarrow{h_t}
$$

and

$$
\overleftarrow{h_t}
$$

They are usually **concatenated**:

$$
\boxed{
h_t=
[\overrightarrow{h_t};
\overleftarrow{h_t}]
}
$$

So the final representation contains:

$$
\boxed{\text{Past context + Future context}}
$$

# 5. Complete Flow

```text
                 Forward RNN
              ───────────────►

x₁ ───────── x₂ ───────── x₃ ───────── x₄
 │           │           │           │
 ▼           ▼           ▼           ▼
h⃗₁          h⃗₂          h⃗₃          h⃗₄
 │           │           │           │
 └──────┐    └──────┐    └──────┐    │
        ▼           ▼           ▼    ▼
      CONCAT      CONCAT      CONCAT CONCAT
        │           │           │    │
        ▼           ▼           ▼    ▼
       h₁          h₂          h₃   h₄
        ▲           ▲           ▲    ▲
        │           │           │    │
      h⃖₁          h⃖₂          h⃖₃  h⃖₄
        ▲           ▲           ▲    ▲
        │           │           │    │
x₁ ───────── x₂ ───────── x₃ ───────── x₄
              ◄───────────────
                 Backward RNN
```

At every position:

$$
\boxed{
h_t=[\overrightarrow{h_t};\overleftarrow{h_t}]
}
$$

---

# 6. Why Use Both Directions?

Consider:

> **"I went to the bank yesterday."**

To understand **"bank"**, the model can use:

### Forward context

```text
I → went → to → bank
```

### Backward context

```text
yesterday ← bank ← to ← went ← I
```

Therefore:

```text
        Left context       Right context
              ↓                  ↓
          "I went to"       "yesterday"
                 \             /
                  \           /
                    "bank"
                       ↓
              richer representation
```

The word's representation can therefore use information from **both sides**.

---

# 7. Mathematical Summary

### Forward:

$$
\boxed{
\overrightarrow{h_t}
=
f(x_t,\overrightarrow{h_{t-1}})
}
$$

### Backward:

$$
\boxed{
\overleftarrow{h_t}
=
f(x_t,\overleftarrow{h_{t+1}})
}
$$

### Combined:

$$
\boxed{
h_t=
[\overrightarrow{h_t};
\overleftarrow{h_t}]
}
$$

### Output:

$$
y_t=g(W_yh_t+b_y)
$$

# 8. Normal RNN vs Bidirectional RNN

| Feature               | Normal RNN   | Bidirectional RNN  |
| --------------------- | ------------ | ------------------ |
| Directions            | One          | Two                |
| Past context          | ✅            | ✅                  |
| Future context        | ❌            | ✅                  |
| Hidden representation | One          | Forward + backward |
| Context               | Past-focused | Past + future      |
| Parameters            | Lower        | Higher             |
| Computation           | Lower        | Higher             |

A BiRNN effectively contains **two recurrent networks**, so it generally requires more computation and parameters than a comparable one-directional RNN.

---

# 9. Applications 🎯

Bidirectional RNNs are useful when the **complete sequence is available**.

Common applications include:

* 📝 Named Entity Recognition
* 🏷️ Part-of-Speech Tagging
* 📄 Sequence labeling
* 📝 Text processing
* 🗣️ Speech processing when the full sequence is available
* 🧬 Biological sequence analysis

---

# ⚠️ Important Limitation

A Bidirectional RNN needs access to future inputs.

Therefore, it is not naturally suitable for **strict real-time causal prediction**, where future data does not yet exist.

```text
Current time
     ↓
Future data unavailable
     ↓
Cannot use backward context
```

So:

> **BiRNN is useful when the entire sequence is available, but not naturally suited to causal real-time prediction.**

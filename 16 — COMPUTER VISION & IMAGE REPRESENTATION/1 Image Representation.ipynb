> **Goal:** Understand how a real image becomes numerical data that a CNN can process.

The complete chain is:

```text
🖼️ Image
   ↓
Pixels
   ↓
Color / intensity values
   ↓
Channels
   ↓
Height × Width
   ↓
Image Tensor
   ↓
N × C × H × W
   ↓
CNN
```

# 2. 🔥🔥🔥 Pixel

A **pixel** is one individual element/location in a digital image.

For a grayscale image, a pixel can be represented by **one intensity value**.

For RGB, a pixel contains **three color values**:

```text
Pixel = [R, G, B]
```

For example:

```text
[255, 0, 0]
```

represents pure red in the conventional 8-bit RGB representation.

Stanford notes that RGB values are commonly represented from `0` to `255`. ([Stanford University][1])

### Remember

```text
Grayscale pixel → 1 value

RGB pixel       → 3 values
```

---

# 3. 🖤 Grayscale Image

A grayscale image contains only **brightness/intensity** information.

Therefore:

```text
1 pixel → 1 value
```

A grayscale image can be represented as:

$$
\boxed{H\times W}
$$

Example:

```text
28 × 28
```

means:

```text
Height = 28
Width  = 28
```

Total pixels:

$$
28\times28=784
$$

### Example

```text
┌────┬────┬────┬────┐
│  0 │ 50 │100 │150 │
├────┼────┼────┼────┤
│ 20 │ 70 │120 │180 │
├────┼────┼────┼────┤
│ 40 │ 90 │140 │220 │
└────┴────┴────┴────┘
```

For a typical 8-bit grayscale representation:

```text
0   → black
255 → white
```

# 4. 🌈 RGB Image

RGB stands for:

| Channel | Meaning |
| ------- | ------- |
| **R**   | Red     |
| **G**   | Green   |
| **B**   | Blue    |

Each pixel has three values.

Example:

```text
Pixel = [120, 40, 200]

R = 120
G = 40
B = 200
```

The three values combine to determine the pixel's color.

Stanford's RGB explanation describes every color as a combination of red, green and blue values. ([Stanford University][1])

# 5. 🔥🔥🔥 Channels

This is **one of the most important CNN concepts**.

A **channel** is a 2D plane containing one type of information across the image.

For an RGB image:

```text
                 RGB IMAGE
                     │
          ┌──────────┼──────────┐
          ↓          ↓          ↓
         🔴          🟢         🔵
        RED        GREEN       BLUE
          │          │          │
          ↓          ↓          ↓
        H × W      H × W      H × W
```

So an RGB image can be viewed as **three 2D matrices stacked together**.

# 6. 🧠 RGB as Three Matrices

Suppose an image is:

```text
4 × 4 RGB
```

We can think of it as:

### 🔴 Red

```text
10  20  30  40
15  25  35  45
20  30  40  50
25  35  45  55
```

### 🟢 Green

```text
50  60  70  80
55  65  75  85
60  70  80  90
65  75  85  95
```

### 🔵 Blue

```text
100 110 120 130
105 115 125 135
110 120 130 140
115 125 135 145
```

Each channel:

$$
4\times4
$$

Together:

$$
\boxed{3\times4\times4}
$$

Therefore:

```text
Channels = 3
Height   = 4
Width    = 4
```

# 7. 📐 Height and Width

These describe the **spatial dimensions** of an image.

```text
              WIDTH
        ───────────────→

       ┌────────────────┐
       │                │
       │                │
HEIGHT │     IMAGE      │
  ↑    │                │
  │    │                │
  ↓    └────────────────┘
```

If an image is:

```text
224 × 128
```

then:

```text
Height = 224
Width  = 128
```

⚠️ Don't confuse:

```text
Height / Width → spatial dimensions

Channels       → color/information dimensions
```

# 8. 🔥🔥🔥 Image Tensor

A **tensor** is a multidimensional numerical array.

In Deep Learning, images are commonly stored as tensors.

For one RGB image:

$$
\boxed{C\times H\times W}
$$

For example:

```text
3 × 224 × 224
```

means:

```text
3   → channels
224 → height
224 → width
```

PyTorch's `PILToTensor` documentation explicitly converts an image with height `H`, width `W`, and `C` channels into a tensor with shape:

$$
(C,H,W)
$$

([PyTorch Documentation][2])

# 9. 🔄 HWC vs CHW

You will encounter two common ways of writing image dimensions.

### HWC

```text
H × W × C
```

Example:

```text
224 × 224 × 3
```

### CHW

```text
C × H × W
```

Example:

```text
3 × 224 × 224
```

They describe the **same three dimensions**, but in a different order.

```text
HWC
 │ │ │
 │ │ └── Channels
 │ └──── Width
 └────── Height


CHW
 │ │ │
 │ │ └── Width
 │ └──── Height
 └────── Channels
```

# 10. 🔥🔥🔥 PyTorch Convention — NCHW

For CNNs, the really important shape is the **batch shape**.

PyTorch commonly represents a batch of images as:

$$
\boxed{N\times C\times H\times W}
$$

where:

| Symbol | Meaning          |
| ------ | ---------------- |
| `N`    | Number of images |
| `C`    | Channels         |
| `H`    | Height           |
| `W`    | Width            |

PyTorch documentation uses NCHW as the logical dimension convention for image tensors. ([PyTorch][3])

# 15. 🧠 Why This Matters for CNNs

A CNN receives something like:

```text
N × C × H × W
```

and operates on its spatial structure.

For example:

```text
32 × 3 × 224 × 224
```

The CNN knows:

```text
32 images
 ↓
each has 3 channels
 ↓
each channel is 224 × 224
```

This spatial organization is fundamental to how convolutional networks process images.

# ⚠️ Common Mistakes

### ❌ 1. RGB has one value per pixel

Wrong:

```text
RGB → 1 value
```

Correct:

```text
RGB → 3 values
```

---

### ❌ 2. Channel = height

No.

```text
C → channels
H → height
W → width
```

---

### ❌ 3. Forgetting batch dimension

One image:

```text
3 × 224 × 224
```

Batch:

```text
32 × 3 × 224 × 224
```

---

### ❌ 4. Confusing HWC and CHW

```text
HWC = 224 × 224 × 3

CHW = 3 × 224 × 224
```

Same logical dimensions, different ordering.

---

### ❌ 5. Confusing tensor shape with image resolution

```text
224 × 224
```

describes spatial resolution.

```text
3 × 224 × 224
```

describes a single RGB tensor in CHW order.

# 🎯 Interview Questions

### Q1. What is a pixel?

> The smallest spatial element of a digital image, represented by numerical intensity or color information.

### Q2. What is a channel?

> A 2D plane representing one type of image information, such as the red, green, or blue component of an RGB image.

### Q3. How is an RGB image represented?

> It contains three channels: red, green, and blue.

### Q4. What does `3 × 224 × 224` mean?

> A single RGB image with 3 channels, height 224, and width 224, using CHW ordering.

### Q5. What does `32 × 3 × 224 × 224` mean?

> A batch of 32 RGB images, each having height and width 224.

# 📊 Quick Comparison

| Representation  | Shape           | Values per pixel |

| RGB, HWC        | `H × W × 3`     |                3 |
| RGB, CHW        | `3 × H × W`     |                3 |
| RGB batch, NCHW | `N × 3 × H × W` |                3 |

```text
PIXEL
↓
Smallest image element


GRAYSCALE
↓
1 value / pixel
↓
H × W


RGB
↓
3 values / pixel
↓
R + G + B
↓
3 channels


CHANNEL
↓
2D plane of information


IMAGE TENSOR
↓
C × H × W


BATCH
↓
N × C × H × W
```

$$
\boxed{\text{CNN input} = N\times C\times H\times W}

$$

And the conceptual pipeline:

> **Image → Pixels → Channels → Tensor → Batch → CNN** 🧠🔥

# 1. 🎯 What is EfficientNet?

EfficientNet is a family of CNN architectures introduced by **Mingxing Tan and Quoc V. Le**.

Its main contribution is a systematic way to scale a CNN along three dimensions:

```text
                    CNN
                     │
       ┌─────────────┼─────────────┐
       ↓             ↓             ↓
     DEPTH         WIDTH       RESOLUTION
       │             │             │
   More layers   More channels   Bigger input
```

Instead of increasing only one of these, EfficientNet increases **all three in a balanced way**.

This is called:

$$
\boxed{\text{Compound Scaling}}
$$

# 2. 🧠 The Scaling Problem

Imagine you already have a CNN.

Then you receive:

```text
More GPU
More memory
More computation
More training budget
```

You want a larger model.

What should you do?

### Option 1

Increase depth:

```text
More layers
```

### Option 2

Increase width:

```text
More channels
```

### Option 3

Increase resolution:

```text
Larger input image
```

### Option 4

Increase all three.

But the question becomes:

> **How much should each dimension increase?**

That's the **model scaling problem**.

---

# 3. 🧱 Depth

Depth means approximately:

> **How many layers/blocks the network contains.**

Example:

```text
Input
 ↓
Conv
 ↓
Block
 ↓
Block
 ↓
Block
 ↓
Output
```

Increasing depth:

```text
Input
 ↓
Conv
 ↓
Block
 ↓
Block
 ↓
Block
 ↓
Block
 ↓
Block
 ↓
Block
 ↓
Output
```

So:

$$
\boxed{\text{Depth}=\text{number of layers/blocks}}
$$

### Why increase it?

More depth allows the network to perform more hierarchical transformations.

```text
Early layers
   ↓
Edges
   ↓
Textures
   ↓
Parts
   ↓
Objects
```

But simply adding layers isn't necessarily the best use of additional computation.

---

# 4. 📐 Width

Width means:

> **Number of channels/features in a layer.**

Example:

```text
64 channels
      ↓
128 channels
      ↓
256 channels
```

A wider network can represent more feature types simultaneously.

Conceptually:

```text
Channel 1 → edge
Channel 2 → texture
Channel 3 → curve
Channel 4 → shape
...
```

More channels provide more representational capacity.

Therefore:

$$
\boxed{\text{Width}=\text{number of channels}}
$$

---

# 5. 🖼️ Resolution

Resolution means the spatial size of the input.

Examples:

```text
224 × 224
384 × 384
512 × 512
```

Higher resolution provides more spatial information.

Imagine a small object occupying:

```text
3 × 3 pixels
```

in a low-resolution image.

At a higher resolution, the same object could occupy many more pixels, giving the network more information about its:

* edges
* shape
* texture
* fine details

Therefore:

$$
\boxed{\text{Resolution}=\text{input spatial size}}
$$

Think of them as three independent ways of making a CNN bigger:

```text
                 MAKE CNN BIGGER
                       │
        ┌──────────────┼──────────────┐
        ↓              ↓              ↓
      DEPTH          WIDTH       RESOLUTION
        │              │              │
        ↓              ↓              ↓
   More layers    More channels   Bigger image
```

> **Don't treat these three dimensions independently.**

---

# 7. ❌ Why Scaling Only One Dimension Is Limited

### Only depth

```text
Depth ↑
Width ─
Resolution ─
```

You get a deeper model, but its feature capacity and input detail aren't increasing proportionally.

---

### Only width

```text
Depth ─
Width ↑
Resolution ─
```

You get more channels, but relatively fewer processing stages.

---

### Only resolution

```text
Depth ─
Width ─
Resolution ↑
```

You give the model more detailed input, but don't proportionally increase its ability to process that information.

---

### EfficientNet

```text
Depth       ↑
Width       ↑
Resolution  ↑
```

together.

That's the core idea.

---

# 8. 💡 Compound Scaling

EfficientNet introduces a single overall scaling coefficient:

$$
\boxed{\phi}
$$

Think of \(\phi\) as:

> **"How much larger should my model become?"**

Then depth, width and resolution are scaled according to fixed coefficients.

$$
\boxed{d=\alpha^\phi}
$$

$$
\boxed{w=\beta^\phi}
$$

$$
\boxed{r=\gamma^\phi}
$$

where:

| Symbol     | Meaning                      |
| ---------- | ---------------------------- |
| \(d\)      | Depth scaling                |
| \(w\)      | Width scaling                |
| \(r\)      | Resolution scaling           |
| \(\phi\)   | Overall compound coefficient |
| \(\alpha\) | Depth coefficient            |
| \(\beta\)  | Width coefficient            |
| \(\gamma\) | Resolution coefficient       |

# 9. 🧠 What Does φ Actually Do?

Imagine:

$$
\phi=0
$$

Then:

$$
d=\alpha^0=1
$$

$$
w=\beta^0=1
$$

$$
r=\gamma^0=1
$$

So you have your **baseline model**.

Now increase:

$$
\phi
$$

and:

```text
φ ↑
 │
 ├── Depth ↑
 ├── Width ↑
 └── Resolution ↑
```

Therefore:

$$
\boxed{\phi=\text{overall model scaling knob}}
$$

---

# 10. 🔢 α, β and γ

These determine **how the extra capacity is distributed**.

### α — depth

$$
d=\alpha^\phi
$$

### β — width

$$
w=\beta^\phi
$$

### γ — resolution

$$
r=\gamma^\phi
$$

Think:

```text
                    φ
                    │
          ┌─────────┼─────────┐
          ↓         ↓         ↓
          α         β         γ
          ↓         ↓         ↓
        Depth      Width   Resolution
```

So:

> **φ = how much to scale**

while:

> **α, β, γ = how to distribute the scaling**

---

# 11. ⚙️ Why the Famous Constraint?

The EfficientNet paper uses the approximate computational relationship:

$$
\text{FLOPs}\propto d\,w^2\,r^2
$$

Why?

### Depth

If the number of layers doubles:

$$
D\rightarrow2D
$$

roughly:

$$
2\times
$$

computation.

### Width

For convolution, both input and output channels matter.

So:

$$
W\rightarrow2W
$$

approximately gives:

$$
2^2=4\times
$$

compute.

### Resolution

If both height and width double:

$$
H\rightarrow2H
$$

$$
W\rightarrow2W
$$

number of pixels becomes:

$$
4\times
$$

So:

$$
\boxed{\text{FLOPs}\propto d\,w^2\,r^2}
$$

---

# 12. 🔥 EfficientNet's Constraint

Substitute:

$$
d=\alpha^\phi
$$

$$
w=\beta^\phi
$$

$$
r=\gamma^\phi
$$

into:

$$
d\,w^2\,r^2
$$

giving:

$$
\text{FLOPs}
\propto
\alpha^\phi
\beta^{2\phi}
\gamma^{2\phi}
$$

Therefore:

$$
\text{FLOPs}
\propto
(\alpha\beta^2\gamma^2)^\phi
$$

EfficientNet chooses the coefficients under:

$$
\boxed{\alpha\beta^2\gamma^2\approx2}
$$

Therefore:

$$
\boxed{\text{FLOPs}\propto2^\phi}
$$

approximately.

This is one of the **most important equations in EfficientNet**.

---

# 13. 🧮 Original Coefficients

The EfficientNet paper found approximately:

$$
\boxed{\alpha=1.2}
$$

$$
\boxed{\beta=1.1}
$$

$$
\boxed{\gamma=1.15}
$$

through a small grid search under the scaling constraint.

Let's verify:

$$
1.2\times1.1^2\times1.15^2
$$

$$
=1.2\times1.21\times1.3225
$$

$$
\approx1.92
$$

which is close to:

$$
2
$$

Hence the approximation.

---

# 14. 🧮 Numerical Example — φ = 1

Take:

$$
\phi=1
$$

### Depth

$$
d=1.2^1=1.2
$$

### Width

$$
w=1.1^1=1.1
$$

### Resolution

$$
r=1.15^1=1.15
$$

So approximately:

```text
Depth       ×1.20
Width       ×1.10
Resolution  ×1.15
```

Compute multiplier:

$$
1.2\times1.1^2\times1.15^2
\approx1.92
$$

≈ **2× compute**.

---

# 15. 🧮 Numerical Example — φ = 2

Now:

$$
\phi=2
$$

### Depth

$$
1.2^2=1.44
$$

### Width

$$
1.1^2=1.21
$$

### Resolution

$$
1.15^2=1.3225
$$

Therefore:

```text
Depth       ×1.44
Width       ×1.21
Resolution  ×1.3225
```

The approximate computational increase is:

$$
2^2=4
$$

so roughly:

$$
\boxed{4\times\text{ compute}}
$$

---

# 16. 🔥 The Important Insight From That Example

Notice what EfficientNet **didn't** do.

It didn't say:

```text
Depth       ×4
Width       ×4
Resolution  ×4
```

Instead:

```text
Depth       ×1.44
Width       ×1.21
Resolution  ×1.32
```

Together they produce roughly a **4× computational budget**.

That's why it's called:

$$
\boxed{\text{Compound Scaling}}
$$

The three dimensions are scaled **together but unequally**.

---

# 17. 🧠 EfficientNet-B0

Compound scaling requires a starting model.

That's:

$$
\boxed{\text{EfficientNet-B0}}
$$

B0 is the **baseline architecture**.

The authors used neural architecture search to construct the baseline, then applied compound scaling to generate larger EfficientNet models.

Conceptually:

```text
                 EfficientNet-B0
                    Baseline
                       │
                       ▼
              Compound Scaling
                       │
           ┌───────────┼───────────┐
           ↓           ↓           ↓
         Depth       Width     Resolution
```

# 23. 🧠 Why Compound Scaling Makes Sense

Imagine we increase only resolution:

```text
224×224
     ↓
448×448
```

Now the network receives much more detail.

But:

```text
Depth = same
Width = same
```

The network's capacity hasn't increased proportionally.

Instead EfficientNet says:

```text
Resolution ↑
     +
Width ↑
     +
Depth ↑
```

So:

```text
More information
      +
More feature capacity
      +
More processing stages
```

This creates a more balanced scaling strategy.

---

# 24. ⚠️ Important: Compound Scaling ≠ Equal Scaling

This is a common misconception.

EfficientNet does **not** mean:

```text
Depth ×1.2
Width ×1.2
Resolution ×1.2
```

Instead:

$$
d=\alpha^\phi
$$

$$
w=\beta^\phi
$$

$$
r=\gamma^\phi
$$

with different coefficients:

$$
\alpha\neq\beta\neq\gamma
$$

approximately:

$$
1.2,\;1.1,\;1.15
$$

So the model grows in a **balanced but unequal** way.

# 27. 🎯 Why αβ²γ² ≈ 2?

Because:

$$
\text{FLOPs}\propto d\,w^2\,r^2
$$

and therefore:

$$

\propto

\boxed{\text{FLOPs}\approx2^\phi}
$$

So:

```text
φ + 1
   ↓
approximately 2× computational budget
```

# 28. 🎤 Interview Answer — "What is EfficientNet?"

> **EfficientNet is a CNN family based on a baseline network and a compound scaling strategy. Its main idea is to scale depth, width, and input resolution together rather than scaling only one dimension. The scaling is controlled by a compound coefficient φ using \(d=\alpha^\phi\), \(w=\beta^\phi\), and \(r=\gamma^\phi\), with the approximate constraint \(\alpha\beta^2\gamma^2\approx2\).**

🔥 That's the answer I'd want you to remember.

---

# 29. 🎤 Interview Answer — "What is compound scaling?"

> **Compound scaling is a systematic method for increasing CNN capacity by jointly scaling depth, width, and resolution using one coefficient. Instead of independently guessing how many layers, channels, or input pixels to add, fixed scaling coefficients determine how additional computational resources are distributed among the three dimensions.**

---

# 30. 🎤 Interview Answer — "Why not just make the network deeper?"

Because:

> A CNN's capacity depends on more than depth. Increasing only depth, only width, or only resolution can lead to an unbalanced allocation of computation. EfficientNet therefore scales all three dimensions together.

---

# 31. 🎤 Interview Answer — "What is B0?"

> **EfficientNet-B0 is the baseline architecture from which the larger EfficientNet variants are obtained using compound scaling. The B0 architecture itself was obtained through neural architecture search and uses MBConv-style blocks with squeeze-and-excitation.**

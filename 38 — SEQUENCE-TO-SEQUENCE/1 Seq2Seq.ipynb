## 1. What is Seq2Seq?

**Sequence-to-Sequence (Seq2Seq)** is a neural network architecture that maps one sequence to another sequence.

$$
\boxed{\text{Input Sequence} \rightarrow \text{Output Sequence}}
$$

The input and output sequences **can have different lengths**.

### Example — Machine Translation

```text
Input:
"I love machine learning"

             ↓
          Seq2Seq
             ↓

Output:
"J'aime l'apprentissage automatique"
```

So:

```text
Input length  ≠  Output length
```

This is one of the main reasons Seq2Seq architectures are useful.

# 2. Why Do We Need Seq2Seq?

A normal **many-to-one** model can take a sequence and produce one output:

```text
"I really enjoyed this movie"
             ↓
            RNN
             ↓
          Positive
```

But some tasks require a **sequence as output**.

Example:

```text
"I love AI"
     ↓
Translation model
     ↓
"J'aime l'IA"
```

We therefore need:

```text
Multiple inputs
      ↓
   Model
      ↓
Multiple outputs
```

The output length may also be different from the input length.

---

# 3. Basic Seq2Seq Architecture

The classic Seq2Seq architecture contains two major components:

```text
              Seq2Seq
                 │
        ┌────────┴────────┐
        ↓                 ↓
     Encoder           Decoder
        │                 │
        └── Context ──────┘
                          ↓
                    Output Sequence
```

### Encoder

Reads and processes the input sequence.

### Decoder

Generates the output sequence step by step.

### Core idea

$$
\boxed{\text{Encoder} \rightarrow \text{Context} \rightarrow \text{Decoder}}
$$

---

# 4. Encoder

The encoder processes the input sequence **one element at a time**.

Suppose:

```text
"I love AI"
```

After tokenization:

$$
x_1=I,\quad x_2=love,\quad x_3=AI
$$

The encoder processes them sequentially:

```text
        h₀
        ↓
x₁ → [RNN] → h₁
              ↓
x₂ → [RNN] → h₂
              ↓
x₃ → [RNN] → h₃
```

The encoder's job is to extract information from the entire input sequence.

---

# 5. Context Vector

In the **classic Seq2Seq architecture**, the encoder's final hidden state is used as the context vector.

$$
\boxed{c=h_T}
$$

where:

* \(c\) = context vector
* \(h_T\) = final encoder hidden state
* \(T\) = final input time step

So:

```text
Input sequence
      ↓
   Encoder
      ↓
Final hidden state
      ↓
 Context vector
      ↓
   Decoder
```

### 🧠 Remember:

> **Context vector = information from the encoder passed to the decoder.**

---

# 6. Decoder

The decoder generates the output sequence **one token at a time**.

For example:

```text
<START>
   ↓
J'aime
   ↓
le
   ↓
cinéma
   ↓
<END>
```

Conceptually:

```text
       Context
          ↓
      [Decoder]
          ↓
        y₁
          ↓
      [Decoder]
          ↓
        y₂
          ↓
      [Decoder]
          ↓
        y₃
          ↓
       <END>
```

The decoder is **autoregressive**.

That means its previous output can influence the next output.

---

# 7. Complete Encoder → Decoder Flow

Example:

```text
Input:
"I love AI"
```

### Encoder

```text
I ─────→ h₁
         ↓
love ──→ h₂
         ↓
AI ────→ h₃
         ↓
      Context
```

### Decoder

```text
Context
   ↓
<START>
   ↓
J'aime
   ↓
l'
   ↓
IA
   ↓
<END>
```

Complete:

```text
"I love AI"
     │
     ▼
┌───────────┐
│  Encoder  │
└─────┬─────┘
      │
      ▼
 Context Vector
      │
      ▼
┌───────────┐
│  Decoder  │
└─────┬─────┘
      │
      ▼
"J'aime l'IA"
```

# 8. Mathematical View

### Encoder

At every time step:

$$
\boxed{h_t=f(x_t,h_{t-1})}
$$

After processing the complete input:

$$
\boxed{c=h_T}
$$

### Decoder

The decoder maintains its own state and generates outputs:

$$
\boxed{s_t=f(y_{t-1},s_{t-1},c)}
$$

Output:

$$
\boxed{y_t=g(s_t)}
$$

Overall:

$$
\boxed{
x_1,x_2,\ldots,x_T
\rightarrow
c
\rightarrow
y_1,y_2,\ldots,y_{T'}
}
$$

Importantly:

$$
\boxed{T\neq T'}
$$

---

# 9. `<START>` and `<END>` Tokens

The decoder needs to know when generation begins and ends.

### `<START>`

Tells the decoder:

> Start generating.

### `<END>`

Tells the decoder:

> The sequence is complete.

Example:

```text
<START>
   ↓
I
   ↓
love
   ↓
AI
   ↓
<END>
```

Without such boundary information, determining when to start and stop generation becomes difficult.

---

# 10. Training with Teacher Forcing

Suppose the target sequence is:

```text
<START> I love AI <END>
```

During training:

```text
Decoder input       Target

<START>       →       I
I             →       love
love          →       AI
AI            →       <END>
```

This is **teacher forcing**.

Instead of always feeding the decoder's own previous prediction back into itself, we provide the **ground-truth previous token** during training.

### Connection to the previous module 🔥

```text
Teacher forcing
      ↓
Ground-truth previous output
      ↓
Next decoder input
```

During inference, however, the true previous token is unavailable, so the model must use its own previous prediction.

# 11. Seq2Seq with RNN, LSTM, or GRU

Seq2Seq is an **architecture**, not a requirement to use only vanilla RNNs.

The encoder and decoder can use:

* RNN
* LSTM
* GRU

A common architecture:

```text
Input
  ↓
Embedding
  ↓
LSTM Encoder
  ↓
Context
  ↓
LSTM Decoder
  ↓
Output
```

For example:

```text
English sentence
       ↓
    Embedding
       ↓
  LSTM Encoder
       ↓
 Context Vector
       ↓
  LSTM Decoder
       ↓
French sentence
```

# 12. Major Limitation — Information Bottleneck ⚠️

The original Seq2Seq architecture compresses the entire input sequence into **one fixed-size context vector**.

```text
Long input sequence
       ↓
   ┌─────────┐
   │ Encoder │
   └────┬────┘
        ↓
 ONE context vector
        ↓
   ┌─────────┐
   │ Decoder │
   └─────────┘
```

For short sequences, this can work reasonably well.

For very long sequences, it becomes difficult for one fixed-size vector to preserve all the information needed by the decoder.

This creates an **information bottleneck**.

### Example

```text
Very long sentence
       ↓
 ┌────────────────┐
 │ Encoder        │
 └───────┬────────┘
         ↓
    ONE vector
         ↓
 ┌────────────────┐
 │ Decoder        │
 └────────────────┘
```

This limitation motivated the development of **attention mechanisms**, which allow the decoder to access information from different encoder states instead of relying only on one fixed context vector.

# 13. Seq2Seq Applications

Seq2Seq is particularly useful for tasks where:

$$
\text{Sequence} \rightarrow \text{Sequence}
$$

Examples include:

### 🌍 Machine Translation

```text
English sentence
      ↓
   Seq2Seq
      ↓
French sentence
```

### 🤖 Chatbots

```text
User message
      ↓
   Seq2Seq
      ↓
Bot response
```

### 📝 Text Summarization

```text
Long document
      ↓
   Seq2Seq
      ↓
Short summary
```

# 14. Seq2Seq vs Many-to-One

| Feature       | Many-to-One        | Seq2Seq           |
| ------------- | ------------------ | ----------------- |
| Input         | Sequence           | Sequence          |
| Output        | Single output      | Sequence          |
| Output length | Fixed/single       | Can vary          |
| Example       | Sentiment analysis | Translation       |
| Architecture  | RNN/LSTM/GRU       | Encoder + Decoder |

Example:

```text
Many-to-One:

"I love this movie"
        ↓
     Positive
```

```text
Seq2Seq:

"I love this movie"
        ↓
   "J'aime ce film"
```

# 🧠 15. Core Architecture to Memorize

```text
              INPUT SEQUENCE
                    │
                    ▼
             ┌────────────┐
             │  ENCODER   │
             └─────┬──────┘
                   │
                   ▼
            CONTEXT VECTOR
                   │
                   ▼
             ┌────────────┐
             │  DECODER   │
             └─────┬──────┘
                   │
                   ▼
             OUTPUT SEQUENCE
```

### The three words to remember:

> **Encoder → Context → Decoder**

# 🎯 Interview Questions

### Q1. What is Seq2Seq?

A Seq2Seq model maps an input sequence to an output sequence, where the input and output can have different lengths.

### Q2. What are the main components?

**Encoder and Decoder.**

### Q3. What is the context vector?

In the classic architecture, it is the encoder's final hidden-state representation passed to the decoder.

$$
c=h_T
$$

### Q4. Why are `<START>` and `<END>` used?

They indicate the beginning and end of the generated sequence.

### Q5. What is the major limitation of basic Seq2Seq?

The entire input must be compressed into a fixed-size context vector, creating an information bottleneck, especially for long sequences.

### Q6. What technique helped address this limitation?

**Attention mechanisms.**

### 🧠 One-line revision

$$
\boxed{
\text{Seq2Seq = Encoder + Context + Decoder}
}
$$

**Encoder reads → Context carries information → Decoder generates.** 🚀

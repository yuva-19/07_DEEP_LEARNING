# 1. What is a Decoder?

The **decoder** is the component of a Seq2Seq model responsible for generating the **output sequence one token at a time**.

Overall:

```text id="5p8z3q"
Input Sequence
      ↓
   Encoder
      ↓
 Context
      ↓
   Decoder
      ↓
Output Sequence
```

### Core idea

> **Encoder = reads and represents** 📖
> **Decoder = generates** ✍️

![Image](https://images.openai.com/static-rsc-4/fh5PvKAwk_5mT0by066kxKyYj-a2JDGADtrdBWgVFApSp9HM17K9aWVj8C_oqBHMeB_zIDbgelfDp1ywqX2GwG03E-15p770CmEfsANbbGkEIFheCnQORjptCQ_qgJXCHle761A3K33aZQG_U0Tm0ox4khD4WOjJmbtvIUDknDazC1ydBaDNBIJYBMYWKGfj?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/tSV_ogTjhQF7tBljNKUVwuFi_Z-J50ah7uvsFvYyVRpW3N-UKFuB3qcdi0FjlL_cSNcRmbZp3see575j-VUGKLxzvguW8_-p-QYsaMMciW5uwU6hCciYLIhrNG8QsCFbcJs2GP6W5707SmJJd4Q4bK4tzIw6Otfvmpncr9fom16Mi2x_z-OgpBb3AGx2I4z7?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/Tk7zp23Vd99L7vmsWSUhFaenn1rb-sX82flE-tWANoNgbdSz-y_-JjXkD6_rrpqBCbiUIuXcZwO4LW5x6lucOQ4u1QnSEAkURZEMVTqEgcH8M_Mod-bhFMS9FwWRyHp3Pm2IlsVj4fo3bzaYFM1y-uDItDm4ll_VDAEvjDZUnoDCj9Tg7uiFFME9amfhNhxO?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/sSl-48g2HbnA5yqUi8wpeMwgufUtgRbU3HSU6bsRw4cXnhwqpO1DBZ8u8aZ7NpcJw42wFG515qsOofeeEwgk9af54nJG1UG1mbszIquFmPEPGB7eRxS0vlFLcNHCwOaBkysCh5MPZcZPd2UlG6iI0XOe3R9Rw9WDlaOhv-YXdLQBOZgsIgJcmqm0JnPqO3vC?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/EWuFDWbvji2b47oxoUXMfH_Bvjz_Z-tAudtv8KfQCike-WQZKpdkpt8yjgVdad80P4WW3QwS7a-x0SeqY8gbd0iceeBzVv3be8k0ZnsRH8TcRF7eTFNizpNRsd0WlWwcIgKguGfCgP-X4szWgDf9bO8-gLI1UY4yPHCg4I7MfKOGUZaq_vpd5j8QpVw5dTRK?purpose=fullsize)

---

# 2. Why Do We Need a Decoder?

The encoder converts the input sequence into a learned representation.

The decoder converts that information into the desired output sequence.

Example:

```text id="5t8c1m"
"I love AI"
     ↓
  Encoder
     ↓
 Context
     ↓
  Decoder
     ↓
"J'aime l'IA"
```

The decoder doesn't normally generate the entire sequence simultaneously.

Instead:

$$
\boxed{y_1\rightarrow y_2\rightarrow y_3\rightarrow\cdots\rightarrow<END>}
$$

---

# 3. Basic Decoder Architecture

```text id="9m4f2k"
             Context
                ↓
          ┌───────────┐
          │  Decoder  │
          └─────┬─────┘
                ↓
               y₁
                ↓
          ┌───────────┐
          │  Decoder  │
          └─────┬─────┘
                ↓
               y₂
                ↓
          ┌───────────┐
          │  Decoder  │
          └─────┬─────┘
                ↓
               y₃
                ↓
              <END>
```

The decoder repeatedly generates the next token.

---

# 4. `<START>` Token

The decoder needs to know when generation should begin.

Therefore, a special token is used:

$$
\boxed{<START>}
$$

Example:

```text id="8s3j7n"
<START>
   ↓
J'aime
   ↓
l'
   ↓
IA
   ↓
<END>
```

The `<START>` token is typically the first decoder input.

---

# 5. `<END>` Token

The decoder also needs to know when generation is complete.

A special:

$$
\boxed{<END>}
$$

token is used.

Example:

```text id="0n6y4x"
<START>
   ↓
I
   ↓
love
   ↓
AI
   ↓
<END>
```

When:

$$
y_t=<END>
$$

the decoder stops generating.

---

# 6. Decoder State

A basic RNN decoder maintains a hidden state.

$$
\boxed{s_t=f(s_{t-1},y_{t-1},c)}
$$

where:

* \(s_t\) = current decoder hidden state
* \(s_{t-1}\) = previous decoder hidden state
* \(y_{t-1}\) = previous token
* \(c\) = encoder context

The decoder then produces the next output:

$$
\boxed{y_t=g(s_t)}
$$

Conceptually:

```text id="c8g2k4"
Previous token
      +
Previous state
      +
Context
      ↓
   Decoder
      ↓
 New state
      ↓
Next token
```

---

# 7. Context Vector

In the **classic Seq2Seq architecture**:

$$
\boxed{c=h_T}
$$

where \(h_T\) is the final encoder hidden state.

The decoder uses this context to generate the output.

```text id="6y1v8q"
             Encoder
                ↓
             h₁...hₜ
                ↓
          Context c=hₜ
                ↓
             Decoder
                ↓
         Output sequence
```

---

# 8. Step-by-Step Decoder

Suppose the desired output is:

```text id="3p6z9r"
"I love AI"
```

### Step 1

```text id="q1m8c4"
<START>
   ↓
Decoder
   ↓
I
```

### Step 2

```text id="h4k2s8"
I
↓
Decoder
↓
love
```

### Step 3

```text id="j8n5w1"
love
  ↓
Decoder
  ↓
AI
```

### Step 4

```text id="p7x3m6"
AI
 ↓
Decoder
 ↓
<END>
```

Complete:

```text id="v5r2n9"
<START>
   ↓
   I
   ↓
 love
   ↓
   AI
   ↓
 <END>
```

---

# 9. Autoregressive Decoder 🔥

A decoder is **autoregressive** when previously generated tokens are used to generate future tokens.

Conceptually:

$$
\boxed{
P(y_t|y_1,\ldots,y_{t-1},c)
}
$$

So:

```text id="3d9v6k"
<START>
   ↓
 y₁
   ↓
 y₂
   ↓
 y₃
   ↓
 ...
```

Each generated token can influence the next one.

### Memory trick

> **Autoregressive = previous output helps generate the next output.**

---

# 10. What Does the Decoder Use?

At each time step, the decoder uses:

### ① Previous token

What token came before?

### ② Previous hidden state

What information has the decoder already accumulated?

### ③ Encoder information

What does the input sequence represent?

Therefore:

$$
\boxed{
\text{Previous Token}
+
\text{Previous State}
+
\text{Context}
\rightarrow
\text{Decoder}
}
$$

---

# 11. Teacher Forcing During Training

This connects directly to the previous **Teacher Forcing** topic.

Suppose the correct target is:

```text id="8k2p6q"
<START> I love AI <END>
```

During training:

```text id="f3w9m2"
Decoder Input       Target

<START>        →      I
I              →      love
love           →      AI
AI             →      <END>
```

The **ground-truth previous token** is supplied to the decoder.

### Training

```text id="6c4x8v"
Ground-truth previous token
             ↓
          Decoder
             ↓
       Next prediction
```

This is **teacher forcing**.

---

# 12. Decoder During Inference

During inference, the correct target sequence isn't available.

Therefore, the decoder uses its **own previous prediction**.

```text id="7n1m5r"
<START>
   ↓
Predict "I"
   ↓
"I" becomes next input
   ↓
Predict "love"
   ↓
"love" becomes next input
   ↓
Predict "AI"
   ↓
"AI" becomes next input
   ↓
Predict <END>
```

### Key difference

| Training                    | Inference                   |
| --------------------------- | --------------------------- |
| Ground-truth previous token | Model's previous prediction |
| Teacher forcing may be used | No ground truth available   |
| Faster/stable training      | Actual generation           |

This difference is related to **exposure bias**.

---

# 13. Predicting the Next Token

The decoder usually produces a probability distribution over the vocabulary.

Example:

```text id="2k7p4m"
I       → 0.72
love    → 0.08
like    → 0.05
AI      → 0.03
movie   → 0.02
...
```

The model assigns a probability to each possible token.

A simple strategy is **greedy decoding**:

$$
\boxed{
y_t=\arg\max_y P(y|y_{<t},c)
}
$$

Meaning:

> Choose the token with the highest predicted probability.

---

# 14. Decoder + Softmax

The typical prediction pipeline is:

```text id="5m8q2v"
Decoder hidden state
        ↓
   Linear layer
        ↓
      Logits
        ↓
     Softmax
        ↓
Probability distribution
        ↓
   Next token
```

Mathematically:

$$
z_t=W_{hy}s_t+b_y
$$

Then:

$$
\boxed{P(y_t)=Softmax(z_t)}
$$

The Softmax converts the logits into probabilities over the vocabulary.

---

# 15. LSTM Decoder

The decoder can use an **LSTM**.

```text id="7x4p9n"
Previous token
      ↓
   Embedding
      ↓
  LSTM Decoder
      ↓
 Hidden state
      ↓
    Linear
      ↓
   Softmax
      ↓
 Next token
```

The LSTM maintains:

$$
\boxed{h_t,C_t}
$$

So it has both:

* hidden state
* cell state

---

# 16. GRU Decoder

A GRU can also be used.

```text id="1k6m3q"
Previous token
      ↓
   Embedding
      ↓
   GRU Decoder
      ↓
 Hidden state
      ↓
    Linear
      ↓
   Softmax
      ↓
 Next token
```

The GRU maintains:

$$
\boxed{h_t}
$$

without a separate cell state.


![Image](https://images.openai.com/static-rsc-4/hGtSvbLHVKPys0tjI8NzKg0WWml7-7ApUSXGVxr3GerORl15MJrg_VIjFoekcPtxTLtW97qHgClow-PZsDTPn__qkoN6_i1mP4UD5Q6xUQSb7Ed7C_6H4At5uS45Pmg4fIIeL3z1foUq2ogKyh1um7182C-8FwLUD0RcTGEbFqJ_xSCoTaemRZL-AryjJGSu?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/UgMmEjzc3M8YrtjpdmyuKXOS58CKdRjoVyTbAZzNoCSrPpH9y8oBz_YzCV-TS2Qtrzk4xGGYhoMUjw30XH_zGrawWIHQgC2IL_wv4ANr4eYs_TQF5k2Mkr3H5F7GbTsdhlrGpKnXGlkMBDBOHZPtUml5JWJfXifK1FuW8iQNBKxkIWdhxpWZktcnAPicRHQx?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/6643hm33uGDCQ-3xYVzOEfH09-dXAQIC8jU02x_Bd0kA6Ya-PLDH5jln3Cp-GVWe1IwRffX5T9S-q-3jXPPm87T1dj4u_QLYzVdhnKJvube46nWOB_7CXN0rW3DKGmMVKRQgYS9_CZcQTsshjJpmw4j8JIkQZMwR4_s3dnMyV2MKaNeYzzPrQDJNYGr4pVaZ?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/DOixkVz1CxclvU4KM9VRZCosahK962slecSX-_PUK73WssH3wADZ03AH8AscVDBHW0ljgpA6N7x54dwXjSTinR4Ed6oe-pbQ7Zh0w3zOzZ8mc4hdSkyItCbWizezZWa5RAVLWkJDtr4C-awS8RBn5ocWmo5Pfz7x32rvYdTB9pw6tr37AuJEL3_katShXSgZ?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/rRMSrJvNtuLAzt60GxpsCJ1R6MgdbcQ2pdcazXWMef0oiMq9F7oO3N1kb9Iudk_peFLu0yXfEaj-4XDk0eSJzOlHWhKEsWfqx85D6CVgpnlJPPO3p9GammYgFoAM27RwjqOsLZ2mbkK-gI0IssTjJm9mqg1uPyf4NZrvWq4UW6dM9Qou4B_Sffrv2qiI8HUZ?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/s3uscL6Op-tHzmMHh104HNMFlb9tEUo6QT4v8hjLYZmfcCXF-AFndaXuTmX2fUDIip8byxjSXiqJJuWwX7U2ExIFURJCCDxtSzLtVQxnNaH42gssZhv5xP-Oj72sBgi8QldANvWoGc3pApWLXosKRdO8JPVPMnraRhScg26TlbXsNLtfdx93XoBMwvnoatwV?purpose=fullsize)

---

# 18. Encoder vs Decoder

| Encoder                         | Decoder                          |
| ------------------------------- | -------------------------------- |
| Reads input                     | Generates output                 |
| Processes input sequence        | Generates target sequence        |
| Produces hidden representations | Produces output probabilities    |
| Creates/produces context        | Uses context                     |
| Input is known                  | Output is generated step-by-step |
| Runs over input tokens          | Runs until `<END>`               |

### 🧠 Easy memory trick

```text id="x7c3m9"
📖 ENCODER = READ

✍️ DECODER = WRITE
```

# 19. Decoder's Main Limitation

The classic decoder relies heavily on the context:

$$
c=h_T
$$

For long input sequences, the encoder may struggle to compress all relevant information into one fixed-size vector.

```text id="v6n2k8"
Long Input
    ↓
 Encoder
    ↓
ONE Context Vector
    ↓
 Decoder
```

This is the **information bottleneck** problem.

It motivated the development of **attention mechanisms**, where the decoder can use multiple encoder hidden states instead of relying only on the final state.

![Image](https://images.openai.com/static-rsc-4/YgwHDb4qjdhqOcGO8cNql5xKeZwsgEH3Z6kLPPG4zuGCbVjjqq50rimYrYVhsV9p8IBiLwgvsnnP_AiG3CPIgIzEvdkykSjtYl3IMXj9VYbttu9Cq3cYQ8PYePqK30fEE2LYfSW1h_o5zy5Dwa6WLrMs86x1G0ARDQtyY5KWSnGlA0V5iXScFhGBGn1Aa0qj?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/XZpOrI7YK7McH_eI54qbSMaolsv8HetW_RlpVgpBpOQN1GXCuwJMllTZelphJHiVGU2Plf_kM9l8HyvXr9fjR9Sp25Q9FU_guAiR4XUFhTXJYDdZAVxRu_ybaO_m6pDamr-w1fhDUdyvkqu4e-a1MNz_bAxsfrWrBz8DVeZj4lLogPFX1bCSS7EZHR4l6HaK?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/TpLWN4ptq9NF9kAvbs1T-P4mbrhDbmobmJQ5KjSEsl7eLsAQIjieXrUc1tlNSlNhPzyXVzNouLN6cxe-ux4AW2JVHMv3_Mzrpeauh2g85XnKhohmFyyhPiNMv7VrzTNaEfKL5_QcGSCUptU2vcjHoRvZP8kdugtpu_L6G1QF2MH-zhdKofW-sg8j7g9J4qP6?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/hkG9eKvYsjrNhC-wTeZafBuh_mk94W0PyvfCzrnhk70i4WzN9BIXAB_jIcs3Gvp5p6jqNS31M7CsvKC9KCwLwNSQdWynm02QgK6AIHxCjPfxNXZwqSoARCP2B5SVIz8Wyvz4r_t6GWTYlFYwy7rD_TQJRfoNok_bBFpAguGgPu8V-KkrPq6PSr6lr3g6kist?purpose=fullsize)

![Image](https://images.openai.com/static-rsc-4/yjCEHjE-hZNrQ9EYRpPTWrQTrHRCXuU7SbLavwj0T0G_8k3wP7mLQCWU90fpYKzLa2SpM47vrlTpcRY8wJ42eaR8AdMZHx5Tr6C6onCMupD3Z4zko8-8crQ3ffp3t-KhfgXjI43R8b6dttxCYsP-bo-a4ag7jHJELWqoS0SSnNim6Kw5Ibagd1Qs7FgmMa2m?purpose=fullsize)

---

# 🧮 20. Mathematical Summary

### Decoder state

$$
\boxed{s_t=f(s_{t-1},y_{t-1},c)}
$$

### Output

$$
\boxed{y_t=g(s_t)}
$$

Often:

$$
z_t=W_{hy}s_t+b_y
$$

$$
\boxed{P(y_t)=Softmax(z_t)}
$$

Generation continues:

$$
y_1,y_2,\ldots,y_{T'}
$$

until:

$$
\boxed{y_t=<END>}
$$

---

# 🎯 21. Interview Questions

### Q1. What is the role of the decoder?

The decoder generates the target sequence one token at a time using the encoder's information and its previous state/output.

### Q2. What is an autoregressive decoder?

A decoder where previously generated tokens are used as inputs for generating subsequent tokens.

### Q3. What is teacher forcing?

During training, the ground-truth previous output token is supplied as the decoder's next input.

### Q4. What happens during inference?

The decoder uses its own previous prediction as the next input.

### Q5. How is the next token predicted?

Typically:

$$
\text{Hidden State}
\rightarrow
\text{Linear}
\rightarrow
\text{Softmax}
\rightarrow
\text{Next Token}
$$

### Q6. When does generation stop?

When the decoder generates:

$$
\boxed{<END>}
$$

---

# 🔥 Final Memory Map

```text id="8p3m6x"
                     ENCODER
                        │
                        ▼
                   Context c
                        │
                        ▼
                  ┌───────────┐
                  │  DECODER  │
                  └─────┬─────┘
                        │
                     <START>
                        │
                        ▼
                       y₁
                        │
                        ▼
                       y₂
                        │
                        ▼
                       y₃
                        │
                        ▼
                      <END>
```

### 🧠 Remember these **5 points**:

1. **Decoder generates the output sequence.**
2. **It generates one token at a time.**
3. **It uses context + previous token + previous state.**
4. **It is typically autoregressive.**
5. **Generation stops at `<END>`.**

$$
\boxed{
\text{Context + Previous Token + Previous State}
\rightarrow
\text{Decoder}
\rightarrow
\text{Next Token}
}
$$

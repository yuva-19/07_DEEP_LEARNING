## 1. What is a Context Vector?

A **context vector** is a learned numerical representation of the input sequence that is passed from the **encoder to the decoder**.

In the classic Seq2Seq architecture:

$$
\boxed{c=h_T}
$$

where:

* \(c\) = context vector
* \(h_T\) = final encoder hidden state
* \(T\) = final input time step

### Basic flow

```text
Input Sequence
      ↓
   Encoder
      ↓
h₁ → h₂ → h₃ → ... → hₜ
                       ↓
                 Context Vector
                       ↓
                    Decoder
```

# 2. Why Is It Called "Context"?

The vector is intended to contain information about the **input sequence as a whole**.

Example:

```text
"I love machine learning"
          ↓
       Encoder
          ↓
   h₁ → h₂ → h₃ → h₄
                     ↓
                Context c
                     ↓
                  Decoder
```

The final hidden state has accumulated information through the previous time steps.

Therefore:

$$
\boxed{c=h_4}
$$

---

# 3. How Is the Context Vector Created?

Suppose:

```text
"I love AI"
```

The encoder processes the sequence:

### Step 1

$$
h_1=f(x_1,h_0)
$$

### Step 2

$$
h_2=f(x_2,h_1)
$$

### Step 3

$$
h_3=f(x_3,h_2)
$$

Therefore:

$$
\boxed{c=h_3}
$$

Complete flow:

```text
"I" ──→ h₁
          ↓
"love" → h₂
          ↓
"AI" ──→ h₃
          ↓
       Context
          ↓
       Decoder
```

---

# 4. Context Vector Is a Numerical Vector

The context vector is **not a word or sentence**.

It is a learned numerical vector.

For example:

$$
c=
[0.21,-0.43,0.17,0.82,-0.11,\ldots]
$$

If the encoder hidden size is 512:

$$
\boxed{c\in\mathbb{R}^{512}}
$$

The exact dimension depends on the architecture.

---

# 5. What Information Does It Carry?

The context vector contains information accumulated while the encoder processes the input.

Example:

```text
"The cat is sleeping"
          ↓
       Encoder
          ↓
   Context Vector
          ↓
       Decoder
```

Conceptually, it represents information about the processed sequence that the decoder can use to generate the output.

---

# 6. Context Connects Encoder and Decoder

The context vector acts as the communication bridge:

```text
             ENCODER
                │
                ▼
          Context Vector
                │
                ▼
             DECODER
```

### 🧠 Easy memory:

> **Context Vector = Encoder's information passed to the Decoder.**

---

# 7. Context Vector with RNN

For an RNN encoder:

$$
\boxed{
h_t=\tanh(W_{xh}x_t+W_{hh}h_{t-1}+b_h)
}
$$

After the final input:

$$
\boxed{c=h_T}
$$

Flow:

```text
x₁ → h₁
       ↓
x₂ → h₂
       ↓
x₃ → h₃
       ↓
...
       ↓
xₜ → hₜ = c
           ↓
        Decoder
```

---

# 8. Context Vector with LSTM

An LSTM maintains:

$$
h_t
$$

and

$$
C_t
$$

At the final time step:

$$
h_T,\ C_T
$$

can be passed to the decoder.

```text
        LSTM Encoder
             │
       ┌─────┴─────┐
       ↓           ↓
      hₜ          Cₜ
       │           │
       └─────┬─────┘
             ↓
          Decoder
```

### Important distinction

| Encoder | Final information |
| ------- | ----------------- |
| RNN     | \(h_T\)           |
| GRU     | \(h_T\)           |
| LSTM    | \(h_T,\ C_T\)     |

---

# 9. Fixed-Size Information Bottleneck ⚠️

This is the **most important limitation** of the classic context-vector approach.

The entire input sequence is compressed into **one fixed-size representation**.

```text
Long Input
    ↓
  Encoder
    ↓
h₁ → h₂ → h₃ → ... → hₜ
                         ↓
                    ONE vector
                         ↓
                      Decoder
```

For short sequences, this can work reasonably well.

For very long sequences, preserving all useful information in one fixed-size vector can become difficult.

This is called the:

$$
\boxed{\text{Information Bottleneck}}
$$

---

# 10. Why Long Sequences Are Difficult

Suppose the input contains 100 tokens:

```text
x₁ x₂ x₃ ... x₁₀₀
```

The encoder produces:

```text
h₁ h₂ h₃ ... h₁₀₀
```

But the classic architecture ultimately uses:

```text
h₁₀₀
  ↓
Context
  ↓
Decoder
```

So:

$$
\boxed{
\text{Many input states}
\rightarrow
\text{One fixed-size vector}
}
$$

This can create an information bottleneck.

---

# 11. Connection to Attention 🔥

The limitation of the fixed context vector motivated **attention mechanisms**.

### Classic Seq2Seq

The decoder mainly receives:

$$
\boxed{c=h_T}
$$

```text
h₁  h₂  h₃  h₄  h₅
             │
             ▼
            h₅
             │
             ▼
          Decoder
```

### Attention-based Seq2Seq

The decoder can use information from multiple encoder states:

```text
h₁   h₂   h₃   h₄   h₅
 \    |    |    |    /
       Attention
           ↓
        Decoder
```

# 12. Context Vector vs Encoder Hidden States

The encoder produces multiple hidden states:

$$
\boxed{h_1,h_2,\ldots,h_T}
$$

In classic Seq2Seq:

$$
\boxed{c=h_T}
$$

So:

```text
Encoder hidden states

h₁   h₂   h₃   h₄   h₅
│    │    │    │    │
└────┴────┴────┴────┘
          ↓
       Context
          ↓
       Decoder
```

With attention, multiple encoder states can be accessed by the decoder instead of relying only on \(h_T\).

---

# 13. Context Vector vs Word Embedding

Don't confuse these two. ⚠️

### Word Embedding

Represents an individual token:

```text
"cat"
  ↓
Embedding vector
```

### Context Vector

Represents information encoded from the input sequence:

```text
"The cat is sleeping"
          ↓
       Encoder
          ↓
    Context vector
```

| Word Embedding               | Context Vector                          |
| ---------------------------- | --------------------------------------- |
| Represents a token           | Represents encoded sequence information |
| Produced by embedding layer  | Produced by encoder                     |
| Used as input representation | Passed toward decoder                   |
| Example: `"cat"` vector      | Representation of input sequence        |

# 🧮 14. Mathematical Summary

### Encoder:

$$
\boxed{h_t=f(x_t,h_{t-1})}
$$

### Hidden states:

$$
\boxed{h_1,h_2,\ldots,h_T}
$$

### Classic context vector:

$$
\boxed{c=h_T}
$$

### Decoder:

$$
\boxed{s_t=f(s_{t-1},y_{t-1},c)}
$$

Complete:

$$
\boxed{
x_1,\ldots,x_T
\rightarrow
h_1,\ldots,h_T
\rightarrow
c=h_T
\rightarrow
\text{Decoder}
}
$$

# 🎯 15. Interview Questions

### Q1. What is a context vector?

A learned numerical representation of the input sequence passed from the encoder to the decoder.

### Q2. What is the classic context-vector equation?

$$
\boxed{c=h_T}
$$

### Q3. Why is it important?

It provides the decoder with information about the encoded input sequence.

### Q4. What is its major limitation?

The entire input sequence is compressed into a fixed-size representation, creating an information bottleneck for long sequences.

### Q5. What helped address this limitation?

**Attention mechanisms.**

### Q6. Is a context vector the same as a word embedding?

**No.**

A word embedding represents a token, while a context vector represents information encoded from the input sequence.

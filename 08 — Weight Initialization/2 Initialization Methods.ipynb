# 1. 🔥🔥🔥 Why Do We Need Specialized Initialization?

From **8.1**, we learned:

$$
W=\text{small random values}
$$

breaks symmetry.

But simply saying "random" isn't enough.

If weights are **too small**:

```text
Small weights
     ↓
Small activations
     ↓
Small gradients
     ↓
Possible vanishing-gradient problem
```

If weights are **too large**:

```text
Large weights
     ↓
Large activations
     ↓
Large gradients
     ↓
Possible exploding-gradient problem
```

Therefore, we want:

$$
\boxed{\text{Randomness + Appropriate Scale}}
$$

This motivates **Xavier/Glorot** and **He/Kaiming** initialization.

---

# 2. 🔥🔥🔥 Xavier / Glorot Initialization

**Xavier initialization**, also called **Glorot initialization**, chooses weight scales using both:

* $n_{in}$ = number of input units
* $n_{out}$ = number of output units

Its goal is to keep the scale of activations and gradients reasonably controlled across layers.

It is commonly associated with:

$$
\boxed{\text{Tanh / Sigmoid}}
$$

---

# 3. 🔥🔥 Xavier Normal

Xavier normal samples weights from a normal distribution:

$$
W\sim\mathcal{N}(0,\sigma^2)
$$

where:

$$
\sigma^2=
\frac{2}{n_{in}+n_{out}}
$$

Therefore:

$$
\boxed{
\sigma=
\sqrt{\frac{2}{n_{in}+n_{out}}}
}
$$

So the variance is:

$$
\boxed{
Var(W)=\frac{2}{n_{in}+n_{out}}
}
$$

---

# 4. 🔥🔥 Xavier Uniform

Xavier can also use a uniform distribution:

$$
W\sim U(-a,a)
$$

where:

$$
\boxed{
a=
\sqrt{\frac{6}{n_{in}+n_{out}}}
}
$$

Therefore:

$$
W\sim
U
\left(
-\sqrt{\frac{6}{n_{in}+n_{out}}},
\sqrt{\frac{6}{n_{in}+n_{out}}}
\right)
$$

So:

```text
Xavier / Glorot
├── Xavier Normal
└── Xavier Uniform
```

---

# 5. 🔥🔥🔥 He / Kaiming Initialization

**He initialization**, also called **Kaiming initialization**, was designed particularly for networks using **ReLU-like activations**.

Basic association:

$$
\boxed{\text{ReLU} \rightarrow \text{He}}
$$

Why a different initialization?

ReLU is:

$$
f(x)=\max(0,x)
$$

It sets negative values to zero.

Therefore, the signal propagation behavior differs from activations such as tanh.

He initialization adjusts the weight scale accordingly.

---

# 6. 🔥🔥 He Normal

He normal uses:

$$
W\sim\mathcal{N}(0,\sigma^2)
$$

where:

$$
\sigma^2=
\frac{2}{n_{in}}
$$

Therefore:

$$
\boxed{
Var(W)=\frac{2}{n_{in}}
}
$$

and:

$$
\boxed{
\sigma=
\sqrt{\frac{2}{n_{in}}}
}
$$

---

# 7. 🔥🔥 He Uniform

He uniform uses:

$$
W\sim U(-a,a)
$$

where:

$$
\boxed{
a=
\sqrt{\frac{6}{n_{in}}}
}
$$

Therefore:

$$
W\sim
U
\left(
-\sqrt{\frac{6}{n_{in}}},
\sqrt{\frac{6}{n_{in}}}
\right)
$$

So:

```text
He / Kaiming
├── He Normal
└── He Uniform
```

---

# 8. 🔥🔥🔥 Xavier vs He

This comparison is **must know**.

|                  | Xavier / Glorot                   | He / Kaiming              |
| ---------------- | --------------------------------- | ------------------------- |
| Main association | Tanh / Sigmoid                    | ReLU / ReLU-like          |
| Uses             | $n_{in},n_{out}$                  | Mainly $n_{in}$           |
| Normal variance  | $\frac{2}{n_{in}+n_{out}}$        | $\frac{2}{n_{in}}$        |
| Uniform limit    | $\sqrt{\frac{6}{n_{in}+n_{out}}}$ | $\sqrt{\frac{6}{n_{in}}}$ |
| Main goal        | Maintain reasonable signal scale  | Account for ReLU behavior |

### 🧠 Memory trick

> **Xavier → Tanh/Sigmoid**
> **He → ReLU**

---

# 9. 🔥🔥🔥 Normal vs Uniform

Both Xavier and He can use either a **normal** or **uniform** distribution.

## Normal distribution

Weights follow:

$$
W\sim\mathcal{N}(0,\sigma^2)
$$

The distribution is centered around zero.

```text id="xw7z5u"
Frequency
   ↑
   │          /\
   │         /  \
   │        /    \
   │_______/______\_______→ Weight
                 0
```

Most values are concentrated around the mean.

---

## Uniform distribution

Weights follow:

$$
W\sim U(-a,a)
$$

Values are sampled across a specified interval.

```text id="1m2f7j"
Frequency
   ↑
   │      ┌───────────┐
   │      │           │
   │______└───────────┘______→ Weight
        -a      0      +a
```

Values within the interval have equal probability density.

---

# 10. 🔥🔥 Normal vs Uniform — Practical Difference

|                 | Normal           | Uniform                    |
| --------------- | ---------------- | -------------------------- |
| Distribution    | Bell-shaped      | Flat within interval       |
| Center          | Usually around 0 | Interval centered around 0 |
| Parameters      | Mean + variance  | Lower + upper bounds       |
| Used by Xavier? | ✅                | ✅                          |
| Used by He?     | ✅                | ✅                          |

### Important

The difference is primarily **how the random values are distributed**.

The initialization method still determines the appropriate scale.

---

# 11. 🔥🔥🔥 Initialization Family Tree

Memorize this structure:

```text id="y0k3nz"
Initialization Methods
│
├── Xavier / Glorot
│   │
│   ├── Normal
│   └── Uniform
│
└── He / Kaiming
    │
    ├── Normal
    └── Uniform
```

So:

> **Xavier Normal and Xavier Uniform are both Xavier initialization.**

> **He Normal and He Uniform are both He initialization.**

---

# 12. 🔥🔥 Numerical Example — Xavier

Suppose:

$$
n_{in}=100
$$

and:

$$
n_{out}=50
$$

### Xavier Normal

Variance:

$$
Var(W)
=
\frac{2}{100+50}
$$

$$
=
\frac{2}{150}
\approx0.0133
$$

Standard deviation:

$$
\sigma
=
\sqrt{0.0133}
\approx0.115
$$

---

### Xavier Uniform

$$
a=
\sqrt{\frac{6}{100+50}}
$$

$$
a=
\sqrt{\frac{6}{150}}
$$

$$
a=\sqrt{0.04}
$$

$$
a=0.2
$$

Therefore:

$$
\boxed{W\sim U(-0.2,0.2)}
$$

---

# 13. 🔥🔥 Numerical Example — He

Suppose:

$$
n_{in}=100
$$

### He Normal

$$
Var(W)
=
\frac{2}{100}
=
0.02
$$

Therefore:

$$
\sigma=\sqrt{0.02}\approx0.141
$$

---

### He Uniform

$$
a=
\sqrt{\frac{6}{100}}
$$

$$
a=
\sqrt{0.06}
\approx0.245
$$

Therefore:

$$
\boxed{W\sim U(-0.245,0.245)}
$$

---

# 14. 🔥🔥🔥 Practical Selection

### ReLU hidden layers

A common choice:

$$
\boxed{\text{He / Kaiming}}
$$

Example:

```text id="gibz7s"
Input
  ↓
Dense + ReLU
  ↓
Dense + ReLU
  ↓
Dense + ReLU
  ↓
Output
```

He initialization is well suited to the ReLU hidden layers.

---

### Tanh hidden layers

A common choice:

$$
\boxed{\text{Xavier / Glorot}}
$$

Example:

```text id="6m7t8x"
Input
  ↓
Dense + Tanh
  ↓
Dense + Tanh
  ↓
Output
```

---

# 15. 🔥🔥 Important Nuance

Don't treat the following as an absolute law:

> "Xavier can ONLY be used with tanh."

> "He can ONLY be used with ReLU."

That's too rigid.

Initialization choice depends on:

* activation
* architecture
* layer type
* framework
* desired signal propagation

But for your current syllabus, the practical rule is:

$$
\boxed{\text{Tanh → Xavier}}
$$

$$
\boxed{\text{ReLU → He}}
$$

---

# 16. 🔥🔥 PyTorch Implementation

PyTorch provides these initialization methods directly.

### Xavier Uniform

```python
torch.nn.init.xavier_uniform_(layer.weight)
```

### Xavier Normal

```python
torch.nn.init.xavier_normal_(layer.weight)
```

### He / Kaiming Uniform

```python
torch.nn.init.kaiming_uniform_(
    layer.weight,
    nonlinearity="relu"
)
```

### He / Kaiming Normal

```python
torch.nn.init.kaiming_normal_(
    layer.weight,
    nonlinearity="relu"
)
```

In practical deep-learning work, you normally **don't manually generate every random weight**. Framework initialization utilities handle this.

# 1. Why RMSProp?

### Vanilla SGD

$$
\theta_t=\theta_{t-1}-\eta g_t
$$

Problem:

> Same base learning rate is used for all parameters.

---

### Momentum

$$
v_t=\beta v_{t-1}+g_t
$$

Problem addressed:

> Use previous directional information to reduce oscillations and accelerate consistent movement.

---

### AdaGrad

$$
G_t=G_{t-1}+g_t^2
$$

Problem addressed:

> Give different parameters different effective learning rates.

But AdaGrad has a major problem:

$$
G_t
$$

**continually increases** because all historical squared gradients are permanently accumulated.

Therefore:

$$
\eta_{\text{effective}}
=
\frac{\eta}{\sqrt{G_t}+\epsilon}
$$

continually decreases.

Eventually the learning rate can become **too small**.

---

# 2. RMSProp's Solution 💡

Instead of:

$$
G_t=G_{t-1}+g_t^2
$$

RMSProp uses a **decaying/exponential moving average**:

$$
\boxed{
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
}
$$

Old information gradually loses influence.

Then:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}{\sqrt{s_t}+\epsilon}g_t
}
$$

So:

```text
AdaGrad
   ↓
Remember ALL squared gradients
   ↓
Learning rate can become tiny


RMSProp
   ↓
Remember recent squared gradients more
   ↓
Forget old information gradually
   ↓
More stable adaptive learning rate
```

---

# 3. Core RMSProp Equations 🔥🔥🔥

Let:

$$
g_t=\nabla_\theta L_t
$$

### Squared-gradient moving average

$$
\boxed{
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
}
$$

### Parameter update

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}{\sqrt{s_t}+\epsilon}g_t
}
$$

### Effective learning rate

$$
\boxed{
\eta_{\text{effective}}
=
\frac{\eta}{\sqrt{s_t}+\epsilon}
}
$$

---

# 4. Meaning of Each Term

| Symbol     | Meaning                             |
| ---------- | ----------------------------------- |
| $g_t$      | Current gradient                    |
| $g_t^2$    | Squared current gradient            |
| $s_t$      | Moving average of squared gradients |
| $\beta$    | Decay/memory factor                 |
| $\eta$     | Base learning rate                  |
| $\epsilon$ | Numerical stability constant        |
| $\theta$   | Model parameter                     |

A common decay value is around:

$$
\beta=0.9
$$

The exact value is a hyperparameter and can vary by implementation.

---

# 5. What Does $\beta$ Do? 🔥🔥

Consider:

$$
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
$$

If:

$$
\beta=0.9
$$

then:

$$
s_t
=
0.9s_{t-1}
+
0.1g_t^2
$$

So the previous moving average receives a large weight, while the newest squared gradient receives a smaller weight.

### Larger $\beta$

→ longer memory
→ slower adaptation to recent changes

### Smaller $\beta$

→ shorter memory
→ greater emphasis on recent gradients

---

# 6. Why "RMSProp"?

RMS = **Root Mean Square**.

The optimizer maintains a moving average of squared gradients:

$$
s_t
$$

and uses:

$$
\sqrt{s_t}
$$

in the denominator.

Hence the name:

> **Root Mean Square Propagation**

The square root gives a scale based on the recent magnitude of the gradients.

---

# 7. Effective Learning Rate 🔥🔥🔥

RMSProp can be understood most easily through:

$$
\eta_{\text{effective}}
=
\frac{\eta}{\sqrt{s_t}+\epsilon}
$$

### Large recent gradients

$$
s_t\uparrow
$$

Therefore:

$$
\eta_{\text{effective}}\downarrow
$$

### Small recent gradients

$$
s_t\downarrow
$$

Therefore:

$$
\eta_{\text{effective}}\uparrow
$$

So RMSProp automatically adjusts the step size **for each parameter**.

# 8. Numerical Example 🧮

Ignore $\epsilon$ for simplicity.

Suppose:

$$
\beta=0.9
$$

and:

$$
s_0=0
$$

Current gradient:

$$
g_1=2
$$

Then:

$$
g_1^2=4
$$

Therefore:

$$
s_1
=
0.9(0)+0.1(4)
$$

$$
\boxed{s_1=0.4}
$$

---

### Second step

Suppose:

$$
g_2=2
$$

Then:

$$
s_2
=
0.9(0.4)+0.1(4)
$$

$$
s_2=0.36+0.4
$$

$$
\boxed{s_2=0.76}
$$

---

### Third step

$$
s_3
=
0.9(0.76)+0.1(4)
$$

$$
s_3=0.684+0.4
$$

$$
\boxed{s_3=1.084}
$$

Notice:

> $s_t$ is increasing toward a stable region rather than accumulating indefinitely like AdaGrad.

---

# 9. Why Doesn't RMSProp Grow Forever?

Suppose the gradient magnitude stays approximately constant:

$$
g_t^2=c
$$

Then:

$$
s_t
=
\beta s_{t-1}
+
(1-\beta)c
$$

Eventually:

$$
s_t\approx c
$$

because old values are continuously discounted.

This is fundamentally different from:

$$
G_t=g_1^2+g_2^2+\cdots+g_t^2
$$

used by AdaGrad.

# 10. AdaGrad vs RMSProp 🔥🔥🔥

This is probably the **most important comparison** in this block.

| Feature                              | AdaGrad        | RMSProp                     |
| ------------------------------------ | -------------- | --------------------------- |
| Tracks squared gradients             | ✅              | ✅                           |
| Uses gradient history                | ✅              | ✅                           |
| Historical calculation               | Cumulative sum | Exponential moving average  |
| Old gradients forgotten?             | ❌              | Gradually                   |
| Effective LR can shrink indefinitely | ✅ Potentially  | Much less likely            |
| Parameter-specific LR                | ✅              | ✅                           |
| Main improvement                     | Adaptive LR    | Adaptive LR + fading memory |

### The one sentence to memorize:

> **AdaGrad accumulates all squared gradients; RMSProp uses a decaying average of squared gradients.**

# 12. Momentum vs RMSProp 🔥🔥

Don't confuse these two.

### Momentum

Tracks gradient **direction/history**:

$$
v_t=\beta v_{t-1}+g_t
$$

Its purpose is primarily:

> **Build velocity and smooth optimization.**

### RMSProp

Tracks gradient **magnitude/history**:

$$
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
$$

Its purpose is primarily:

> **Adapt the learning rate.**

```text
Momentum
   ↓
Gradient history
   ↓
Velocity
   ↓
Direction / smoother movement


RMSProp
   ↓
Squared-gradient history
   ↓
Scale
   ↓
Adaptive learning rate
```

This distinction becomes extremely useful when learning **Adam**.

---

# 13. Parameter-Specific Adaptation

Suppose:

```text
w₁ → large recent gradients
w₂ → small recent gradients
```

RMSProp maintains separate states:

$$
s_{w_1}
$$

and

$$
s_{w_2}
$$

If:

$$
s_{w_1}>s_{w_2}
$$

then:

$$
\frac{\eta}{\sqrt{s_{w_1}}+\epsilon}
<
\frac{\eta}{\sqrt{s_{w_2}}+\epsilon}
$$

Therefore:

```text
w₁ → smaller effective step
w₂ → larger effective step
```

This happens **automatically**.

# 15. RMSProp and Changing Gradient Statistics

Another advantage of the moving average is that the optimizer can **adapt when gradient behavior changes**.

For example:

```text
Early training
     ↓
Large gradients
     ↓
Large s


Later training
     ↓
Smaller gradients
     ↓
s gradually adjusts downward
```

Because old information fades, the optimizer isn't permanently dominated by gradients from the beginning of training.

# 16. PyTorch Implementation 🔥🔥🔥

PyTorch provides RMSProp directly:

```python
optimizer = torch.optim.RMSprop(
    model.parameters(),
    lr=0.001,
    alpha=0.99
)
```

Training loop:

```python
optimizer.zero_grad()

predictions = model(x_batch)

loss = criterion(predictions, y_batch)

loss.backward()

optimizer.step()
```

The important change is inside:

```python
optimizer.step()
```

The optimizer maintains the moving-average state internally.

# 18. Common Mistakes ⚠️

### ❌ "RMSProp is the same as AdaGrad."

No.

The difference is:

$$
\text{AdaGrad}=\text{cumulative sum}
$$

$$
\text{RMSProp}=\text{decaying average}
$$

---

### ❌ "RMSProp completely forgets old gradients."

No.

Old information **gradually loses influence**.

---

### ❌ "RMSProp is Momentum."

No.

Momentum tracks directional information.

RMSProp tracks squared-gradient magnitude to scale updates.

---

### ❌ "RMSProp uses one effective learning rate for the entire model."

No.

Each parameter has its own moving-average state.

Therefore effective learning rates can differ.

---

### ❌ "A higher `alpha` means more importance to the newest gradient."

Actually, in the common PyTorch formulation:

$$
s_t=\alpha s_{t-1}+(1-\alpha)g_t^2
$$

a higher `alpha` means **more weight on the previous moving average** and slower adaptation to new gradients.

# 🎯 Interview Questions

### Q1. What problem does RMSProp solve?

It addresses AdaGrad's tendency to continually shrink the effective learning rate by using a decaying average of squared gradients.

### Q2. What does RMSProp maintain?

An exponential moving average of squared gradients.

### Q3. What does `alpha`/$\beta$ control?

How much historical squared-gradient information is retained.

### Q4. Why square the gradients?

To track magnitude without positive and negative gradients cancelling.

### Q5. Why use the square root?

To obtain an RMS-like scale for adaptive normalization.

### Q6. Does RMSProp have parameter-specific learning rates?

Yes. Each parameter has its own accumulated state.

### Q7. Biggest difference between AdaGrad and RMSProp?

> **AdaGrad remembers all squared gradients; RMSProp uses a fading memory.**

```text
                    OPTIMIZATION

                         SGD
                          │
                          │ Current gradient
                          ▼
                     MOMENTUM
                          │
                          │ + direction history
                          ▼
                      ADAGRAD
                          │
                          │ + adaptive LR
                          │
                          │ Problem:
                          │ LR may shrink too much
                          ▼
                     RMSPROP
                          │
                          │ + decaying squared-gradient
                          │   average
                          ▼
                        ADAM
                          │
                          │ Momentum-like first moment
                          │ +
                          │ RMSProp-like second moment
                          ▼
                 Modern Deep Learning
```

This progression is much more useful than memorizing each optimizer independently.

## 1. What is Momentum? 🔥🔥🔥

**Momentum is an improvement to vanilla SGD.**

Vanilla SGD uses mainly the **current gradient**:

$$
\theta_t
=
\theta_{t-1}
-
\eta\nabla_\theta L_t
$$

Momentum additionally keeps information about **previous updates**.

> **SGD reacts to the current gradient; Momentum remembers previous movement.**

---

# 2. Why Do We Need Momentum?

Vanilla SGD can:

* zig-zag during optimization
* oscillate in steep directions
* move slowly when gradients consistently point in one direction

Momentum addresses this by accumulating a **velocity**.

# 3. Core Equations 🔥🔥🔥

Momentum introduces velocity $v$.

### Velocity update

$$
\boxed{
v_t
=
\beta v_{t-1}
+
\nabla_\theta L_t
}
$$

### Parameter update

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\eta v_t
}
$$

Where:

| Symbol              | Meaning              |
| ------------------- | -------------------- |
| $\theta$            | Model parameter      |
| $v_t$               | Momentum/velocity    |
| $\nabla_\theta L_t$ | Current gradient     |
| $\eta$              | Learning rate        |
| $\beta$             | Momentum coefficient |

A common value is:

$$
\beta=0.9
$$

---

# 4. What Does $\beta$ Do? 🔥🔥🔥

$\beta$ controls **how much previous velocity is remembered**.

$$
v_t=\beta v_{t-1}+g_t
$$

where:

$$
g_t=\nabla_\theta L_t
$$

### If $\beta=0$

$$
v_t=g_t
$$

There is no momentum.

Therefore:

> **$\beta=0$ → vanilla SGD**

### If $\beta=0.9$

A large portion of previous velocity is retained.

So the optimizer has a **decaying memory** of previous gradients.

---

# 5. Why Is It Called "Momentum"?

Think about a ball rolling downhill. 🏀

Without momentum:

```text
gradient → step
gradient → step
gradient → step
```

With momentum:

```text
gradient
   ↓
velocity builds
   ↓
gradient
   ↓
velocity builds more
   ↓
faster movement
```

The optimizer develops something similar to **inertia**.

---

# 6. Momentum Accumulates Previous Gradients

Start with:

$$
v_t=\beta v_{t-1}+g_t
$$

Expanding:

$$
v_t
=
g_t
+
\beta g_{t-1}
+
\beta^2g_{t-2}
+
\beta^3g_{t-3}
+\cdots
$$

Therefore:

* Current gradient → strongest influence
* Previous gradient → $\beta$ times its influence
* Older gradients → progressively smaller influence

For $\beta=0.9$:

```text
Current gradient       → × 1
Previous gradient      → × 0.9
Two steps ago          → × 0.9²
Three steps ago        → × 0.9³
...
```

So momentum behaves like a **decaying memory of past gradients**.

---

# 7. Numerical Example 🔥

Suppose:

$$
\beta=0.9
$$

$$
\eta=0.1
$$

and initially:

$$
v_0=0
$$

### Step 1

Current gradient:

$$
g_1=5
$$

Velocity:

$$
v_1=0.9(0)+5
$$

$$
v_1=5
$$

Parameter update:

$$
\theta_1
=
\theta_0-0.1(5)
$$

$$
\boxed{\theta_1=\theta_0-0.5}
$$

---

### Step 2

Suppose:

$$
g_2=5
$$

Then:

$$
v_2
=
0.9(5)+5
$$

$$
\boxed{v_2=9.5}
$$

Notice:

```text
Current gradient = 5
Velocity         = 9.5
```

Why?

Because velocity contains information from the **previous gradient**.

Then:

$$
\theta_2
=
\theta_1-0.1(9.5)
$$

$$
\boxed{\theta_2=\theta_1-0.95}
$$

The consistent direction has caused the movement to build up.

---

# 8. Momentum Reduces Zig-Zagging 🔥🔥

Consider a narrow valley in a loss landscape.

Vanilla SGD may behave like:

```text
↗ ↘ ↗ ↘ ↗ ↘
```

because gradients can change direction rapidly across the steep sides.

Momentum smooths this behavior:

```text
      ↘
       ↘
        ↘
         ↘
          ↓
        minimum
```

### Why?

If gradients repeatedly point in opposite directions, their contributions can partially cancel.

If gradients repeatedly point in the **same direction**, their contributions accumulate.

Therefore:

> **Oscillations are damped while consistent movement is reinforced.**

---

# 9. SGD vs Momentum

| Feature                            | Vanilla SGD             | SGD + Momentum |
| ---------------------------------- | ----------------------- | -------------- |
| Current gradient                   | ✅                       | ✅              |
| Previous gradient information      | ❌                       | ✅              |
| Velocity/state                     | ❌                       | ✅              |
| Can reduce oscillation             | Limited                 | ✅              |
| Can accelerate consistent movement | Limited                 | ✅              |
| Extra hyperparameter               | No momentum coefficient | $\beta$        |

### Simplest distinction

$$
\boxed{\text{SGD = current gradient}}
$$

$$
\boxed{\text{Momentum = current gradient + history}}
$$

---

# 10. Learning Rate vs Momentum 🔥🔥🔥

Do **not** confuse these.

### Learning rate

$$
\eta
$$

Controls:

> **How large the update is.**

### Momentum

$$
\beta
$$

Controls:

> **How much previous velocity is remembered.**

Think:

```text
Learning rate
     ↓
"How big should my step be?"

Momentum
     ↓
"How much should I remember about my previous movement?"
```

# 11. PyTorch Implementation 🔥🔥🔥

### Vanilla SGD

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

### SGD + Momentum

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01,
    momentum=0.9
)
```

The training loop remains essentially the same:

```python
optimizer.zero_grad()

predictions = model(x_batch)

loss = criterion(predictions, y_batch)

loss.backward()

optimizer.step()
```

The difference is in what happens internally during:

```python
optimizer.step()
```

# 12. Advantages of Momentum 🔥🔥

### 1. Faster movement in consistent directions

Repeated gradients reinforce each other.

### 2. Reduced oscillation

Opposing gradient components can partially cancel.

### 3. Smoother optimization

The update is influenced by historical gradient information instead of only the current mini-batch.

### 4. Better behavior in narrow valleys

Momentum can reduce repeated side-to-side bouncing.

---

# 13. Limitations

Momentum isn't automatically optimal.

It introduces another hyperparameter:

$$
\beta
$$

And training can still be unstable if the learning rate is inappropriate.

Momentum is also not the end of the optimizer story.

Later optimizers such as **Adam** combine ideas related to momentum with adaptive learning rates.

---

# 14. Common Mistakes ⚠️

### ❌ "Momentum increases the learning rate."

Not exactly.

Momentum introduces accumulated velocity.

---

### ❌ "Momentum ignores the current gradient."

No.

It uses both:

$$
\text{previous velocity}
+
\text{current gradient}
$$

---

### ❌ "Momentum always makes training faster."

Not guaranteed.

It is designed to improve optimization behavior, but its effectiveness depends on the problem and hyperparameters.

---

### ❌ "$\beta$ is the learning rate."

No.

$$
\eta=\text{learning rate}
$$

$$
\beta=\text{momentum coefficient}
$$

---

# 15. Interview Questions 🎯

### Q1. What problem does momentum solve?

It can reduce oscillations and accelerate optimization in directions where gradients consistently point the same way.

### Q2. What does momentum remember?

Previous gradient/update information through a velocity term.

### Q3. What does $\beta$ control?

The amount of previous velocity retained.

### Q4. What happens when $\beta=0$?

The momentum contribution disappears, giving vanilla SGD.

### Q5. Why does momentum help in narrow valleys?

It can suppress oscillations in the steep direction while accumulating movement along the consistent direction.

### Q6. Is momentum the same as increasing learning rate?

No.

* $\eta$ controls update scale.
* $\beta$ controls memory of previous movement.

# 🧠 FINAL CHEAT SHEET

### Vanilla SGD

$$
\theta_t
=
\theta_{t-1}
-
\eta g_t
$$

### Momentum

$$
\boxed{
v_t=\beta v_{t-1}+g_t
}
$$

$$
\boxed{
\theta_t=\theta_{t-1}-\eta v_t
}
$$

where:

$$
g_t=\nabla_\theta L_t
$$

### Remember this:

```text
SGD
 ↓
Current gradient

Momentum
 ↓
Current gradient
      +
Previous movement
      ↓
Accumulated velocity
      ↓
Parameter update
```

### 🔥 One-line memory trick

> **SGD asks "which way now?" — Momentum asks "which way now, and where have I already been moving?"**

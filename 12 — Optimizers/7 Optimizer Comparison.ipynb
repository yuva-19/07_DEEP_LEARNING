# 1. 🔥 What Is an Optimizer?

During training:

$$
\text{Input}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\boxed{\text{Optimizer}}
\rightarrow
\text{Updated Parameters}
$$

The **loss function tells us how wrong the model is**.

The **gradient tells us which direction increases the loss**.

The **optimizer decides how to use that gradient to update the parameters**.

General update:

$$
\boxed{
\theta_{t+1}
=
\theta_t-\text{update}
}
$$

For basic gradient descent:

$$
\boxed{
\theta_{t+1}
=
\theta_t-\eta\nabla_\theta L
}
$$

where:

* $\theta$ = model parameters
* $\eta$ = learning rate
* $\nabla_\theta L$ = gradient

# 3. 🔥 12.1 SGD

## Core idea

SGD updates parameters using the current gradient.

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\eta g_t
}
$$

where:

$$
g_t=\nabla_\theta L_B
$$

for the current mini-batch $B$.

### PyTorch

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

PyTorch's optimizer interface updates parameters after gradients have been computed with `backward()`. ([PyTorch Docs][1])

---

## Learning rate

The learning rate controls the **size of the step**.

```text
Too small
    ↓
tiny updates
    ↓
slow training

Good LR
    ↓
useful progress

Too large
    ↓
overshooting
    ↓
oscillation / divergence
```

### Important distinction

> **Gradient = direction/information**

> **Learning rate = step-size scale**

---

# 4. 🔥 Batch GD vs SGD vs Mini-Batch SGD

### Batch Gradient Descent

Uses the entire dataset:

$$
g=\frac{1}{N}\sum_{i=1}^{N}\nabla_\theta L_i
$$

### Stochastic Gradient Descent

Uses one example at a time.

### Mini-Batch SGD

Uses a small batch:

$$
g_B
=
\frac{1}{B}
\sum_{i\in B}
\nabla_\theta L_i
$$

Modern deep-learning training generally uses **mini-batches**, even though the optimizer is often called "SGD."

---

# 5. 🔥 12.2 Momentum

Vanilla SGD can zig-zag, especially in narrow valleys.

Momentum introduces memory.

$$
\boxed{
v_t
=
\beta v_{t-1}
+
g_t
}
$$

Then:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\eta v_t
}
$$

Typical:

$$
\beta=0.9
$$

---

## Intuition

Without momentum:

```text
gradient
   ↓
move
   ↓
new gradient
   ↓
change direction
```

With momentum:

```text
current gradient
       +
previous movement
       ↓
smoothed direction
       ↓
parameter update
```

## What does $\beta$ do?

$$
v_t=\beta v_{t-1}+g_t
$$

Higher $\beta$:

> Remember more of the past.

Lower $\beta$:

> React more strongly to recent gradients.

# 6. 🔥 12.3 AdaGrad

**AdaGrad = Adaptive Gradient.**

Its main idea:

> Give different parameters different effective learning rates.

It accumulates squared gradients:

$$
\boxed{
G_t=G_{t-1}+g_t^2
}
$$

Then:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}
{\sqrt{G_t}+\epsilon}
g_t
}
$$

Therefore:

$$
\eta_{\text{effective}}
=
\frac{\eta}
{\sqrt{G_t}+\epsilon}
$$

---

## Intuition

If a parameter has received large gradients historically:

$$
G_t\uparrow
$$

therefore:

$$
\eta_{\text{effective}}\downarrow
$$

If its accumulated gradients are relatively small:

$$
\eta_{\text{effective}}
$$

remains relatively larger.

### Why square the gradient?

Because:

$$
g^2\geq0
$$

so positive and negative gradients don't cancel when accumulating magnitude.

---

## ❌ AdaGrad's problem

The accumulation never forgets:

$$
G_t
=
g_1^2+g_2^2+\cdots+g_t^2
$$

Therefore $G_t$ keeps increasing.

Eventually:

$$
\eta_{\text{effective}}\rightarrow\text{very small}
$$

Training can become extremely slow.

---

# 7. 🔥 12.4 RMSProp

RMSProp addresses AdaGrad's continuously shrinking learning rate.

Instead of:

$$
G_t=G_{t-1}+g_t^2
$$

it uses an exponential moving average:

$$
\boxed{
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
}
$$

Then:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}
{\sqrt{s_t}+\epsilon}
g_t
}
$$

---

## AdaGrad vs RMSProp

|                      | AdaGrad           | RMSProp             |
| -------------------- | ----------------- | ------------------- |
| Gradient history     | Entire history    | Recent history      |
| Accumulation         | Sum               | Moving average      |
| Old gradients        | Never forgotten   | Gradually forgotten |
| LR shrinking problem | Can become severe | Controlled          |

### Memory trick

> **AdaGrad remembers everything.**

> **RMSProp gradually forgets old gradients.**

---

## First moment

$$
\boxed{
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
}
$$

This tracks a smoothed version of the gradient.

Think:

> **direction / momentum**

---

## Second moment

$$
\boxed{
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
}
$$

This tracks the magnitude of squared gradients.

Think:

> **adaptive scaling**

---

# 9. 🔥 Why Bias Correction?

Initially:

$$
m_0=0
$$

and

$$
v_0=0
$$

Because these start at zero, early estimates are biased toward zero.

So Adam corrects them:

$$
\boxed{
\hat m_t
=
\frac{m_t}{1-\beta_1^t}
}
$$

$$
\boxed{
\hat v_t
=
\frac{v_t}{1-\beta_2^t}
}
$$

---

# 10. 🔥 Adam Update

Final Adam update:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}
}
$$

Typical values:

$$
\beta_1=0.9
$$

$$
\beta_2=0.999
$$

$$
\epsilon\approx10^{-8}
$$

A common starting learning rate is:

$$
\eta=0.001
$$

but this is **not a universal rule**.

---

## Adam mental model

```text
             Gradient
                │
        ┌───────┴────────┐
        ↓                ↓
 First moment       Second moment
   m_t                v_t
        │                │
        ↓                ↓
   Direction       Scale / normalize
        │                │
        └───────┬────────┘
                ↓
         Adam update
```

---

# 11. 🔥 Adam = Momentum + RMSProp?

For learning purposes:

$$
\boxed{
\text{Adam}
\approx
\text{Momentum}
+
\text{RMSProp}
}
$$

More precisely:

* $m_t$ → Momentum-like first moment
* $v_t$ → RMSProp-like squared-gradient tracking
* bias correction → corrects initial moment estimates

So don't memorize Adam as an isolated formula.

Understand its **components**.

# 12. 🔥 12.6 AdamW

AdamW introduces:

> **Decoupled weight decay**

The key difference is how weight decay is handled.

### Adam with coupled L2-style regularization

Conceptually:

$$
g_t
=
\nabla L+\lambda\theta
$$

The combined quantity is then processed by Adam.

### AdamW

The loss gradient goes through Adam:

$$
\nabla L
\rightarrow
\text{Adam}
$$

while weight decay separately shrinks the parameters.

The compact update is:

$$
\boxed{
\theta_t
=
(1-\eta\lambda)\theta_{t-1}
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}
}
$$

PyTorch documents AdamW as Adam where weight decay does not accumulate into the momentum or variance terms. ([PyTorch Docs][2])

---

## Why "decoupled"?

```text
Adam
─────────────────
Loss gradient
      +
regularization
      ↓
    Adam
      ↓
   update


AdamW
─────────────────
Loss gradient
      ↓
    Adam
      ↓
 Adam update

Weights
   ↓
direct decay
```

# 13. 🔥 L2 Regularization vs Weight Decay

For SGD, the relationship is straightforward.

L2 objective:

$$
J(\theta)
=
L(\theta)
+
\frac{\lambda}{2}\|\theta\|^2
$$

Gradient:

$$
\nabla J
=
\nabla L+\lambda\theta
$$

SGD:

$$
\theta_t
=
\theta_{t-1}
-
\eta(\nabla L+\lambda\theta)
$$

which gives:

$$
\theta_t
=
(1-\eta\lambda)\theta_{t-1}
-
\eta\nabla L
$$

So under standard SGD, L2 regularization and weight decay are equivalent in this formulation.

For adaptive optimizers such as Adam, this equivalence does not generally hold, motivating AdamW's decoupled formulation. ([PyTorch Docs][2])

# 16. 🔥 Convergence

Convergence means the optimization process is approaching a stable low-loss region.

Different optimizers can travel through the loss landscape differently.

For example:

```text
             narrow valley

        ╲                ╱
         ╲              ╱
          ╲            ╱
           ╲__________╱
                 ↑
              minimum
```

### SGD

Can zig-zag across the valley.

### Momentum

Uses accumulated direction and can reduce some zig-zagging.

### Adaptive methods

Adjust parameter-wise scaling based on gradient history.

An illustrative trajectory comparison shows visibly different paths for SGD, Momentum, AdaGrad, RMSProp and Adam on a quadratic objective.

**Important:** A smoother/faster trajectory on one toy landscape does **not** mean that optimizer universally performs better on every neural network.

# 17. 🔥 Learning Rate

The learning rate is one of the most important hyperparameters.

### Too small

$$
\eta\downarrow
$$

→ tiny updates
→ slow training

### Too large

$$
\eta\uparrow
$$

→ overshooting
→ oscillation
→ possible divergence

### Appropriate

→ stable progress

---

## Global vs effective learning rate

SGD:

$$
\text{step}\approx\eta g
$$

Adaptive optimizers:

$$
\text{step}\approx
\frac{\eta}{\text{gradient-history scale}}
g
$$

Therefore different parameters can have different **effective step sizes**.

Don't memorize this as a rigid "best optimizer" ranking. Optimizer choice depends on the model, data, training setup and hyperparameter tuning.

### SGD

Use when:

* you want a simple baseline
* you want direct optimization behavior
* you are willing to tune learning-rate schedules

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

---

### SGD + Momentum

Useful when:

* vanilla SGD is oscillating
* you want accumulated direction
* you want a simple optimizer with momentum

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01,
    momentum=0.9
)
```

---

### AdaGrad

Particularly useful for:

* sparse features
* sparse data
* situations where infrequent features should receive relatively larger effective learning rates

Its major trade-off is continually decreasing learning rates.

---

### RMSProp

Useful when:

* adaptive learning rates are useful
* you want recent gradient history
* gradient scales change over training

---

### Adam

Useful when:

* you want adaptive optimization
* you want momentum + adaptive scaling
* you want a strong general-purpose starting point

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

---

### AdamW

Useful when:

* you want Adam's adaptive optimization
* you also want decoupled weight decay
* you're training modern deep-learning models where weight decay is part of the training setup

```python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=0.01
)
```

Current PyTorch documentation explicitly supports SGD, Adam, AdamW, RMSprop, Adagrad and several other optimizers; its documentation also identifies AdamW's decoupled weight-decay behavior. ([PyTorch Docs][1])

# 22. 🎤 Interview / GATE Questions

### Q1. What does an optimizer do?

It uses gradients to update model parameters in order to minimize the objective.

---

### Q2. Why is Momentum better than basic SGD in some situations?

It incorporates past gradient information, which can reduce oscillations and accelerate movement in consistent directions.

---

### Q3. What is AdaGrad's main disadvantage?

Its accumulated squared-gradient term can continually grow, causing effective learning rates to become excessively small.

---

### Q4. How does RMSProp solve this?

It uses an exponentially weighted moving average of squared gradients instead of accumulating them forever.

---

### Q5. What are Adam's two moments?

$$
m_t
$$

First moment → gradient/direction information.

$$
v_t
$$

Second raw moment → squared-gradient magnitude.

---

### Q6. Why does Adam use bias correction?

Because $m_t$ and $v_t$ are initialized at zero and are therefore biased toward zero during early steps.

---

### Q7. What is AdamW?

$$
\boxed{
\text{AdamW}
=
\text{Adam}
+
\text{decoupled weight decay}
}
$$

---

### Q8. What does "decoupled" mean?

Weight decay is applied directly to the parameters instead of being included in the gradient that Adam adaptively scales.

---

### Q9. Does AdamW use momentum?

Yes. Its first moment $m_t$ is momentum-like.

---

### Q10. Does AdamW use adaptive learning rates?

Yes.

# 23. ⚠️ Common Mistakes

### ❌ "Learning rate and gradient are the same."

No.

$$
\text{gradient}=\text{direction/information}
$$

$$
\text{learning rate}=\text{step-size scale}
$$

---

### ❌ "Adam always converges faster than SGD."

Not universally.

Optimizer performance depends on architecture, data and hyperparameters.

---

### ❌ "AdaGrad and RMSProp are the same."

No.

```text
AdaGrad
→ cumulative squared gradients

RMSProp
→ exponential moving average
```

---

### ❌ "AdamW is a completely different optimizer."

No.

AdamW retains Adam's first and second moments.

The major change is **decoupled weight decay**.

---

### ❌ "Weight decay means dropout."

No.

```text
Weight decay → shrinks parameters

Dropout → randomly masks activations
```

---

### ❌ "Gradient clipping is weight decay."

No.

```text
Gradient clipping
→ limits gradient magnitude

Weight decay
→ shrinks parameters
```

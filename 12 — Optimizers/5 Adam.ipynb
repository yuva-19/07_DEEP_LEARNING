# 1. Why Adam? 🔥🔥🔥

We've now seen the evolution:

```text id="6bq8zv"
SGD
 ↓
Current gradient
 ↓
Problem: noisy / oscillatory updates
        │
        ▼
Momentum
 ↓
Gradient history
 ↓
Problem: no adaptive parameter-wise scaling
        │
        ▼
AdaGrad
 ↓
Accumulated squared gradients
 ↓
Problem: learning rate can become too small
        │
        ▼
RMSProp
 ↓
Decaying squared-gradient average
        │
        ▼
Adam
 ↓
First moment + Second moment
```

### The big idea

Adam simultaneously keeps track of:

1. **Direction/history of gradients**
2. **Magnitude/history of gradients**

---

# 2. What Does Adam Mean?

**Adam = Adaptive Moment Estimation**

It maintains two moving averages.

### First moment

$$
m_t
$$

This tracks the average of the **gradient**.

Think:

> **"Which direction has the gradient generally been pointing?"**

### Second raw moment

$$
v_t
$$

This tracks the average of the **squared gradient**.

Think:

> **"How large have the gradients been?"**

So:

```text id="4r9y2t"
                 Gradient gₜ
                     │
             ┌───────┴───────┐
             ↓               ↓
       First moment     Second moment
             ↓               ↓
            mₜ              vₜ
             ↓               ↓
        Direction          Scale
             │               │
             └───────┬───────┘
                     ↓
                   ADAM
                     ↓
              Parameter update
```

---

# 3. First Moment — Momentum-Like Part 🔥🔥🔥

Adam calculates:

$$
\boxed{
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
}
$$

where:

* $g_t$ = current gradient
* $m_t$ = first moment estimate
* $\beta_1$ = decay factor

This is an **exponential moving average of the gradient**.

It is conceptually similar to Momentum.

### Therefore:

$$
\boxed{
m_t\rightarrow\text{direction/history}
}
$$

---

# 4. Second Moment — RMSProp-Like Part 🔥🔥🔥

Adam also calculates:

$$
\boxed{
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
}
$$

where:

* $v_t$ = second raw moment estimate
* $\beta_2$ = decay factor
* $g_t^2$ = element-wise squared gradient

This is the RMSProp-like part.

### Therefore:

$$
\boxed{
v_t\rightarrow\text{gradient magnitude/scale history}
}
$$

---

# 5. The Most Important Adam Concept

Remember these two questions:

### $m_t$

> **Which direction should I move?**

### $v_t$

> **How should I scale that movement?**

Then:

$$
\boxed{
\text{Adam}
=
\text{direction/history}
+
\text{adaptive scaling}
}
$$

---

# 6. Why Square the Gradient?

Adam uses:

$$
g_t^2
$$

in the second moment.

Suppose gradients are:

$$
+5,\ -5
$$

If we simply added them:

$$
5+(-5)=0
$$

The magnitudes would cancel.

But:

$$
5^2+(-5)^2
=
25+25
=
50
$$

So squaring allows Adam to track **gradient magnitude without sign cancellation**.

---

# 7. Bias Correction 🔥🔥🔥

This is one of the most important mathematical parts of Adam.

Initially:

$$
m_0=0
$$

$$
v_0=0
$$

Because both moving averages start at zero, their early values are biased toward zero.

Adam therefore applies **bias correction**.

### Corrected first moment

$$
\boxed{
\hat m_t
=
\frac{m_t}{1-\beta_1^t}
}
$$

### Corrected second moment

$$
\boxed{
\hat v_t
=
\frac{v_t}{1-\beta_2^t}
}
$$

---

# 8. Why Is Bias Correction Needed?

Suppose:

$$
\beta_1=0.9
$$

and:

$$
m_0=0
$$

Current gradient:

$$
g_1=5
$$

Then:

$$
m_1
=
0.9(0)+0.1(5)
$$

$$
m_1=0.5
$$

But the gradient is actually:

$$
5
$$

The estimate is initially pulled toward zero because the moving average started from zero.

Bias correction:

$$
\hat m_1
=
\frac{0.5}{1-0.9}
$$

$$
=
\frac{0.5}{0.1}
$$

$$
\boxed{\hat m_1=5}
$$

So the correction compensates for the initialization bias.

---

# 9. Final Adam Update 🔥🔥🔥

After calculating the two moving averages and correcting their bias:

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}
}
$$

This is the equation you should **definitely know**.

# 10. Complete Adam Algorithm

At every optimization step:

### Step 1 — Calculate gradient

$$
g_t=\nabla_\theta L_t
$$

### Step 2 — Update first moment

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

### Step 3 — Update second moment

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

### Step 4 — Bias correction

$$
\hat m_t
=
\frac{m_t}{1-\beta_1^t}
$$

$$
\hat v_t
=
\frac{v_t}{1-\beta_2^t}
$$

### Step 5 — Update parameters

$$
\theta_t
=
\theta_{t-1}
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}
$$

---

# 11. Typical Hyperparameters

Common Adam defaults are approximately:

$$
\boxed{\beta_1=0.9}
$$

$$
\boxed{\beta_2=0.999}
$$

and:

$$
\epsilon\approx10^{-8}
$$

A commonly used starting learning rate is:

$$
\eta=0.001
$$

But **$0.001$ is not a universal rule**. The appropriate learning rate depends on the model and training problem.

The original Adam paper gives $\beta_1=0.9$, $\beta_2=0.999$, and $\epsilon=10^{-8}$ as default values in its experiments. ([DOI][2])

---

# 12. What Do $\beta_1$ and $\beta_2$ Mean?

### $\beta_1$

Controls the memory of the **gradient**:

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

Think:

> **How much directional history should I retain?**

---

### $\beta_2$

Controls the memory of the **squared gradient**:

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

Think:

> **How much gradient-scale history should I retain?**

### Memory trick 🧠

$$
\boxed{
\beta_1\rightarrow\text{direction}
}
$$

$$
\boxed{
\beta_2\rightarrow\text{magnitude}
}
$$

---

# 13. Simple Numerical Example

Suppose:

$$
g_1=2
$$

$$
\beta_1=0.9
$$

$$
\beta_2=0.999
$$

Initially:

$$
m_0=0
$$

$$
v_0=0
$$

### First moment

$$
m_1
=
0.9(0)+0.1(2)
$$

$$
m_1=0.2
$$

### Second moment

$$
v_1
=
0.999(0)+0.001(2^2)
$$

$$
v_1=0.004
$$

---

### Bias correction

First moment:

$$
\hat m_1
=
\frac{0.2}{1-0.9}
=
2
$$

Second moment:

$$
\hat v_1
=
\frac{0.004}{1-0.999}
=
4
$$

Therefore:

$$
\sqrt{\hat v_1}=2
$$

So the update is approximately:

$$
\theta_1
=
\theta_0
-
\eta\frac{2}{2+\epsilon}
$$

Approximately:

$$
\boxed{
\theta_1\approx\theta_0-\eta
}
$$

# 14. Adam vs SGD 🔥🔥

### SGD

$$
\theta_t
=
\theta_{t-1}-\eta g_t
$$

### Adam

$$
\theta_t
=
\theta_{t-1}
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}
$$

| Feature          | SGD | Adam |
| ---------------- | --- | ---- |
| Current gradient | ✅   | ✅    |
| Gradient history | ❌   | ✅    |
| Adaptive scaling | ❌   | ✅    |
| First moment     | ❌   | ✅    |
| Second moment    | ❌   | ✅    |
| Bias correction  | ❌   | ✅    |

# 15. Adam vs Momentum 🔥🔥🔥

### Momentum

$$
v_t=\beta v_{t-1}+g_t
$$

Primarily gives:

> **Directional memory / velocity**

### Adam

$$
m_t=\beta_1m_{t-1}+(1-\beta_1)g_t
$$

and:

$$
v_t=\beta_2v_{t-1}+(1-\beta_2)g_t^2
$$

So Adam adds adaptive scaling based on squared-gradient history.

```text id="h8r4v3"
Momentum
    ↓
Gradient history
    ↓
Direction / velocity


Adam
    ↓
Gradient history
    ↓
Direction

    +

Squared-gradient history
    ↓
Adaptive scale
```

---

# 16. Adam vs RMSProp 🔥🔥🔥

This comparison is **very important**.

### RMSProp

Tracks:

$$
s_t
=
\beta s_{t-1}
+
(1-\beta)g_t^2
$$

So it primarily tracks the **second moment**.

### Adam

Tracks both:

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

and:

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

So:

```text id="p1dd4j"
RMSProp
   ↓
Second moment
   ↓
Adaptive scaling


Adam
   ↓
First moment
   +
Second moment
   ↓
Direction + adaptive scaling
```

---

# 17. Adam vs AdaGrad

### AdaGrad

$$
G_t=G_{t-1}+g_t^2
$$

Cumulative history.

### Adam

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

Decaying history.

Plus:

$$
m_t
$$

for gradient direction.

Therefore Adam inherits the useful idea of adaptive scaling without simply accumulating squared gradients forever.

# 18. Why Adam Is Popular

The original Adam paper highlights several practical properties, including computational efficiency, relatively low memory requirements, suitability for large problems, and handling of noisy or sparse gradients. ([arXiv][1])

From a learning perspective, the important reasons are:

* Momentum-like gradient smoothing
* Adaptive parameter-wise scaling
* Decaying gradient statistics
* Bias correction
* Usually a convenient starting optimizer
* Works well across many deep-learning problems

But remember:

> **Popular does not mean universally optimal.**

Optimizer choice can depend on the architecture, task, and training objective.

# 20. PyTorch Implementation 🔥🔥🔥

```python
optimizer.zero_grad()

predictions = model(x_batch)

loss = criterion(predictions, y_batch)

loss.backward()

optimizer.step()
```

# 21. Common Mistakes ⚠️

### ❌ "Adam is literally Momentum + RMSProp."

It's a useful **mental model**, but technically Adam has its own formulation combining first- and second-moment estimates with bias correction.

---

### ❌ "Adam doesn't need a learning rate."

It does.

$$
\eta
$$

is still a key hyperparameter.

---

### ❌ "Adam automatically finds the perfect learning rate."

No.

Adam adapts the scaling of individual parameter updates, but the base learning rate still matters.

---

### ❌ "$\beta_1$ and $\beta_2$ do the same thing."

No.

$$
\beta_1
\rightarrow
\text{first moment}
$$

$$
\beta_2
\rightarrow
\text{second moment}
$$

---

### ❌ "Second moment = variance."

Technically, Adam tracks the **second raw moment**:

$$
E[g^2]
$$

not the centered variance:

$$
E[g^2]-E[g]^2
$$

# 22. Adam Mental Model 🧠

Imagine driving down a mountain.

### First moment

Looks at your recent movement:

> **"Which way have I generally been heading?"**

### Second moment

Looks at how strong the slopes/gradients have been:

> **"How aggressive should my step be?"**

### Adam

Combines both:

```text id="4o9z6q"
Recent gradient history
        ↓
   First moment
        ↓
   Direction


Recent squared-gradient history
        ↓
  Second moment
        ↓
    Scaling

        ↓
      ADAM
        ↓
Adaptive parameter update
```

---

# 23. Complete Optimizer Evolution 🔥🔥🔥

This is probably the **single most useful revision diagram** for Module 12:

```text id="f5h4l9"
                    OPTIMIZERS
                        │
                        ▼
                       SGD
                        │
                 Current gradient
                        │
                        ▼
                    MOMENTUM
                        │
              + directional history
                        │
                        ▼
                    ADAGRAD
                        │
            + adaptive learning rate
                        │
              Problem: LR shrinks
                        │
                        ▼
                    RMSPROP
                        │
          + decaying squared-gradient
                moving average
                        │
                        ▼
                      ADAM
                        │
             ┌──────────┴──────────┐
             ▼                     ▼
      First moment           Second moment
       (gradient)          (squared gradient)
             │                     │
             ▼                     ▼
        Direction               Scale
             └──────────┬──────────┘
                        ▼
                 Bias correction
                        ▼
                Parameter update
```

$$

## 🧠 The 5 Optimizers in One Table

| Optimizer    | Main idea                      | Memory                    |
| ------------ | ------------------------------ | ------------------------- |
| **SGD**      | Current gradient               | None                      |
| **Momentum** | Accumulate direction           | Gradient                  |
| **AdaGrad**  | Adaptive LR                    | Cumulative $g^2$          |
| **RMSProp**  | Adaptive LR with fading memory | EMA of $g^2$              |
| **Adam**     | Direction + adaptive scaling   | EMA of $g$ + EMA of $g^2$ |

### 🔑 Ultimate memory trick

> **SGD → where now?**
> **Momentum → where have I been going?**
> **AdaGrad → how large have gradients been?**
> **RMSProp → how large have recent gradients been?**
> **Adam → where have I been going + how large have recent gradients been?**

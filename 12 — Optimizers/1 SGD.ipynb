# 📚 MODULE 12.1 — SGD — Revision Notes

## 1. What is an Optimizer? 🔥🔥🔥

An **optimizer** updates the model's learnable parameters (weights and biases) so that the **loss decreases**.

The basic update rule is:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
$$

Where:

| Symbol            | Meaning                           |
| ----------------- | --------------------------------- |
| $\theta$          | Model parameter                   |
| $L$               | Loss                              |
| $\nabla_\theta L$ | Gradient of loss w.r.t. parameter |
| $\eta$            | Learning rate                     |

### Remember

> **Backpropagation calculates the gradient; the optimizer uses it to update parameters.**

---

# 2. Vanilla SGD 🔥🔥🔥

**SGD = Stochastic Gradient Descent**

The basic idea is:

1. Take a training sample / mini-batch
2. Perform forward propagation
3. Calculate loss
4. Perform backpropagation
5. Obtain gradients
6. Update parameters

$$
\boxed{
\theta\leftarrow\theta-\eta\nabla_\theta L_B
}
$$

Here $L_B$ is the loss calculated from the current mini-batch.

# 3. Why the Negative Gradient? 🔥🔥

The gradient points in the direction of **increasing loss**.

Therefore, we move in the opposite direction:

$$
-\nabla_\theta L
$$

Example:

$$
w=2
$$

$$
\frac{\partial L}{\partial w}=4
$$

$$
\eta=0.1
$$

Then:

$$
w_{\text{new}}
=
2-(0.1)(4)
$$

$$
\boxed{w_{\text{new}}=1.6}
$$

If the gradient is negative:

$$
\frac{\partial L}{\partial w}=-4
$$

then:

$$
w_{\text{new}}
=
2-(0.1)(-4)
=
2.4
$$

So:

```text
Positive gradient  → parameter decreases
Negative gradient  → parameter increases
```

# 4. Why "Stochastic"? 🔥🔥

Suppose your dataset contains:

$$
100,000
$$

samples.

Instead of calculating the gradient using all 100,000 samples for every update, we can use a smaller batch.

Example:

```text
100,000 samples
       ↓
 ┌────────────┐
 │ Batch 1    │ → Update
 ├────────────┤
 │ Batch 2    │ → Update
 ├────────────┤
 │ Batch 3    │ → Update
 ├────────────┤
 │ ...        │
 └────────────┘
```

The gradient calculated from a mini-batch is an **estimate** of the gradient from the complete dataset.

Therefore, successive updates can contain some noise.

---

# 5. Batch GD vs SGD vs Mini-Batch SGD

| Method                 | Samples used per update |   Updates |
| ---------------------- | ----------------------: | --------: |
| Batch Gradient Descent |          Entire dataset |       Few |
| SGD                    |              One sample | Very many |
| Mini-Batch SGD         |             Small batch |      Many |

### Important practical point

In modern deep learning, people often casually say **"SGD"** while actually using **mini-batch SGD**.

For example:

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

The DataLoader typically supplies mini-batches.

# 6. Learning Rate $\eta$ 🔥🔥🔥

The **learning rate determines the size of each parameter update.**

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla_\theta L
$$

Think of it as:

> **Gradient = direction**
> **Learning rate = step size**

## Learning Rate Too Small

```text
minimum
   ↑
  ●
   ↑
  ●
   ↑
  ●
   ↑
  ●
```

Updates are tiny.

### Problems

* Training becomes slow
* Requires many iterations/epochs
* Model may appear to learn extremely slowly

---

## Learning Rate Too Large

```text
       ●
          ↘
             ●
          ↙
       ●
          ↘
             ●
```

The optimizer can overshoot the minimum.

### Problems

* Oscillation
* Unstable loss
* Divergence
* Potential numerical instability

---

## Suitable Learning Rate

```text
●
 \
  ●
   \
    ●
     \
      ●
       ↓
    minimum
```

The optimizer makes meaningful progress toward a lower-loss region.

# 7. SGD Training Process 🔥🔥🔥

```text
Mini-batch
    ↓
Forward Pass
    ↓
Predictions
    ↓
Loss
    ↓
Backpropagation
    ↓
Gradients
    ↓
SGD
    ↓
Parameter Update
    ↓
Next Mini-batch
```

Across many epochs, this process gradually adjusts the model.

---

# 8. PyTorch Implementation 🔥🔥🔥

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

for x_batch, y_batch in train_loader:

    optimizer.zero_grad()

    predictions = model(x_batch)

    loss = criterion(predictions, y_batch)

    loss.backward()

    optimizer.step()
```

### What each line does

**`optimizer.zero_grad()`**

Clears gradients from the previous update.

**`model(x_batch)`**

Performs the forward pass.

**`criterion(...)`**

Calculates the loss.

**`loss.backward()`**

Calculates gradients through backpropagation.

**`optimizer.step()`**

Uses the gradients to update the parameters.

# 9. Mini-Batch Example 🔥

Suppose:

* Dataset = 10,000 samples
* Batch size = 100

Then:

$$
\text{Updates per epoch}
=
\frac{10000}{100}
=
100
$$

Therefore:

```text
1 epoch
   ↓
100 mini-batches
   ↓
100 parameter updates
```

More generally:

$$
\text{steps per epoch}
\approx
\left\lceil\frac{N}{B}\right\rceil
$$

where:

* $N$ = number of training samples
* $B$ = batch size

# 10. Why SGD Updates Look Noisy

Because each mini-batch contains only part of the dataset, its gradient isn't exactly the full-dataset gradient.

Therefore:

```text
Ideal smooth path:

       ↓
      ↓
     ↓
    ↓
   minimum


Mini-batch SGD:

      ↘
    ↙
       ↘
     ↙
        ↘
       minimum
```

This noise is a natural consequence of mini-batch estimation.

# 11. SGD Mental Model 🧠

Imagine walking down a mountain in fog.

| ----------------- | ------------------ |
| Loss              | Height             |
| Gradient          | Direction uphill   |
| Negative gradient | Direction downhill |

The basic SGD process:

> **Find the slope → turn downhill → take a step → repeat.**

# 12. Critical Relationship

This is one of the most important training concepts to remember:

```text
              LOSS
                ↓
         BACKPROPAGATION
                ↓
             GRADIENT
       "Which direction?"
                ↓
            OPTIMIZER
        "How should I move?"
                ↓
        LEARNING RATE
          "How far?"
                ↓
        PARAMETER UPDATE
                ↓
          NEW PARAMETERS
```

### Core equation

$$
\boxed{
\theta\leftarrow\theta-\eta\nabla_\theta L
}
$$

# 13. Common Mistakes ⚠️

### ❌ Mistake 1: Thinking optimizer calculates gradients

No.

```text
Backpropagation → calculates gradients
Optimizer       → uses gradients to update parameters
```

---

### ❌ Mistake 2: Thinking learning rate determines direction

No.

The **gradient determines the direction**.

The learning rate determines **how far to move**.

---

### ❌ Mistake 3: Thinking larger learning rate is always better

No.

Too large → overshooting/instability.

Too small → extremely slow learning.

---

### ❌ Mistake 4: Confusing `eval()` with gradient disabling

`model.eval()` controls behavior of layers such as **Dropout and BatchNorm**.

It does **not** itself disable gradient calculation.

For inference, we commonly use:

```python
model.eval()

with torch.no_grad():
    predictions = model(x)
```

# 🎯 Interview / Exam Points

### Q1. What does SGD do?

SGD updates model parameters using the gradient of the loss computed from a sample or mini-batch.

### Q2. What does the learning rate control?

The magnitude of parameter updates.

### Q3. What happens when learning rate is too small?

Training becomes very slow.

### Q4. What happens when learning rate is too large?

Updates can overshoot, oscillate, or diverge.

### Q5. What does `loss.backward()` do?

Computes gradients using backpropagation.

### Q6. What does `optimizer.step()` do?

Updates the model parameters using those gradients.

### Q7. Why does SGD have noisy updates?

Because the gradient is estimated from only a sample/mini-batch rather than the entire dataset.

### Remember:

* **SGD** → gradient-based parameter updates
* **Gradient** → direction of increasing loss
* **Negative gradient** → direction toward decreasing loss
* **Learning rate** → update size
* **Small LR** → slow learning
* **Large LR** → unstable/overshooting
* **Mini-batch** → practical modern training
* **Backpropagation** → computes gradients
* **Optimizer** → updates parameters
* **`loss.backward()`** → gradients
* **`optimizer.step()`** → parameter update
* **`zero_grad()`** → clears previous gradients

### 🔥 One sentence to remember

> **SGD repeatedly uses the gradient from a sample/mini-batch to move model parameters opposite the gradient, while the learning rate controls how large each movement is.**

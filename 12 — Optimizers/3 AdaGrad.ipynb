# 📚 MODULE 12.3 — AdaGrad — Revision Notes

**Priority: 🔥🔥 Important**

AdaGrad is an important step in the evolution from **fixed learning rates → adaptive optimizers**.

> **Core idea:** AdaGrad adapts the effective learning rate separately for each parameter based on the history of its gradients.

The original AdaGrad method was introduced by Duchi, Hazan, and Singer in their work on adaptive subgradient methods. The paper specifically discusses adapting learning rates using information from previously observed gradients and highlights sparse/infrequently occurring features. ([JMLR][1])

---

# 1. Why AdaGrad?

### Vanilla SGD

SGD uses:

$$
\theta_t
=
\theta_{t-1}
-
\eta g_t
$$

The same base learning rate $\eta$ is applied to every parameter.

But different parameters may have very different gradient histories.

For example:

```text
Parameter w₁
Large gradients → frequently

Parameter w₂
Small gradients → infrequently
```

Using exactly the same effective step size for both may not be ideal.

### AdaGrad's idea 💡

> **Adjust the effective learning rate according to each parameter's historical gradient magnitude.**

# 2. AdaGrad = Adaptive Gradient

**AdaGrad** stands for **Adaptive Gradient**.

Its central mechanism is:

```text
Current gradient
      ↓
Look at historical squared gradients
      ↓
Calculate accumulated gradient magnitude
      ↓
Scale current gradient
      ↓
Parameter-specific update
```

So unlike vanilla SGD:

$$
\text{same effective LR}
$$

AdaGrad produces:

$$
\text{different effective LR for different parameters}
$$

---

# 3. Core Equations 🔥🔥🔥

Let:

$$
g_t=\nabla_\theta L_t
$$

### Step 1 — Accumulate squared gradients

$$
\boxed{
G_t=G_{t-1}+g_t^2
}
$$

### Step 2 — Update parameters

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}{\sqrt{G_t}+\epsilon}g_t
}
$$

where:

| Symbol     | Meaning                            |
| ---------- | ---------------------------------- |
| $\theta_t$ | Parameter at step $t$              |
| $g_t$      | Current gradient                   |
| $G_t$      | Accumulated squared gradients      |
| $\eta$     | Base learning rate                 |
| $\epsilon$ | Small numerical-stability constant |

Typically:

$$
\epsilon\approx10^{-8}
$$

---

# 4. Effective Learning Rate 🔥🔥🔥

The most useful way to understand AdaGrad is to rewrite the update:

$$
\theta_t
=
\theta_{t-1}
-
\underbrace{
\frac{\eta}{\sqrt{G_t}+\epsilon}
}_{\text{effective learning rate}}
g_t
$$

Therefore:

$$
\boxed{
\eta_{\text{effective}}
=
\frac{\eta}{\sqrt{G_t}+\epsilon}
}
$$

### If accumulated gradients are large:

$$
G_t\uparrow
$$

then:

$$
\eta_{\text{effective}}\downarrow
$$

### If accumulated gradients are small:

$$
G_t\downarrow
$$

then the effective learning rate remains relatively larger.

---

# 5. Why Square the Gradients?

AdaGrad accumulates:

$$
g_t^2
$$

rather than simply $g_t$.

Why?

Because gradients can have different signs.

Suppose:

$$
g_1=+5
$$

and:

$$
g_2=-5
$$

If we simply added them:

$$
5+(-5)=0
$$

The history would incorrectly appear to have zero magnitude.

Instead:

$$
5^2+(-5)^2
=
25+25
=
50
$$

So squaring prevents positive and negative gradients from cancelling.

### Mental model

> **Squared gradients measure accumulated gradient magnitude rather than direction.**

---

# 6. Parameter-Specific Learning Rates 🔥🔥🔥

Suppose we have two parameters:

```text
w₁ → large historical gradients
w₂ → small historical gradients
```

For $w_1$:

$$
G_1\text{ becomes large}
$$

Therefore:

$$
\frac{\eta}{\sqrt{G_1}}\text{ becomes smaller}
$$

For $w_2$:

$$
G_2\text{ remains smaller}
$$

Therefore:

$$
\frac{\eta}{\sqrt{G_2}}\text{ remains relatively larger}
$$

So:

```text
w₁ → large gradient history → smaller effective LR

w₂ → small gradient history → larger effective LR
```

This parameter-wise adaptation is the key feature of AdaGrad. The original paper describes the method as giving frequently occurring features lower learning rates and infrequent features higher learning rates. ([JMLR][2])

---

# 7. Numerical Example 🧮

Suppose:

$$
\eta=0.1
$$

Initially:

$$
G_0=0
$$

Current gradient:

$$
g_1=2
$$

### Step 1 — Accumulate

$$
G_1=0+2^2
$$

$$
G_1=4
$$

### Step 2 — Effective learning rate

Ignoring $\epsilon$ for simplicity:

$$
\eta_{\text{effective}}
=
\frac{0.1}{\sqrt{4}}
$$

$$
=
\frac{0.1}{2}
$$

$$
=0.05
$$

### Step 3 — Parameter update

$$
\theta_1
=
\theta_0-(0.05)(2)
$$

$$
\boxed{
\theta_1=\theta_0-0.1
}
$$

---

# 8. What Happens Over Multiple Steps?

Suppose:

$$
g=2
$$

at every step.

Then:

### Step 1

$$
G_1=4
$$

### Step 2

$$
G_2=8
$$

### Step 3

$$
G_3=12
$$

### Step 4

$$
G_4=16
$$

So:

```text
Accumulated squared gradients

G
│
│              /
│           /
│        /
│     /
│  /
└──────────────────→ time
```

As $G_t$ grows:

$$
\frac{\eta}{\sqrt{G_t}+\epsilon}
$$

gets smaller.

Therefore:

> **AdaGrad naturally decreases its effective learning rate over time.**

# 10. AdaGrad vs SGD 🔥🔥🔥

### SGD

$$
\boxed{
\theta_t=\theta_{t-1}-\eta g_t
}
$$

### AdaGrad

$$
\boxed{
\theta_t
=
\theta_{t-1}
-
\frac{\eta}{\sqrt{G_t}+\epsilon}g_t
}
$$

| Feature                                 | SGD | AdaGrad |
| --------------------------------------- | --- | ------- |
| Current gradient                        | ✅   | ✅       |
| Gradient history                        | ❌   | ✅       |
| Squared-gradient accumulation           | ❌   | ✅       |
| Parameter-specific effective LR         | ❌   | ✅       |
| Effective LR automatically decreases    | ❌   | ✅       |
| Particularly useful for sparse features | —   | ✅       |

# 11. AdaGrad vs Momentum 🔥🔥

These optimizers modify SGD in **different ways**.

### Momentum

Momentum maintains accumulated directional information:

$$
v_t=\beta v_{t-1}+g_t
$$

Then:

$$
\theta_t=\theta_{t-1}-\eta v_t
$$

**Main idea:**

> Remember previous movement.

---

### AdaGrad

AdaGrad accumulates squared gradient magnitudes:

$$
G_t=G_{t-1}+g_t^2
$$

Then scales the gradient:

$$
\theta_t
=
\theta_{t-1}
-
\frac{\eta}{\sqrt{G_t}+\epsilon}g_t
$$

**Main idea:**

> Adapt the effective learning rate using gradient history.

### Memory trick 🧠

```text
Momentum
   ↓
"What direction have I been moving?"

AdaGrad
   ↓
"How large have my gradients historically been?"
```

---

# 12. The Major Weakness of AdaGrad ⚠️🔥🔥🔥

This is the **most important limitation** to remember.

AdaGrad uses:

$$
G_t
=
g_1^2+g_2^2+g_3^2+\cdots+g_t^2
$$

Notice:

> **It keeps accumulating forever.**

It doesn't forget old gradients.

Therefore:

$$
G_t\uparrow
$$

and consequently:

$$
\frac{\eta}{\sqrt{G_t}+\epsilon}\downarrow
$$

continuously.

Eventually the effective learning rate can become **too small**.

```text
Effective LR

large
  ↓
  ↓
medium
  ↓
small
  ↓
very small
  ↓
almost zero
```

### Consequence

Training may become extremely slow or effectively stop making meaningful progress.

---

# 13. Why RMSprop Came Next

This limitation leads directly to the next optimizer.

### AdaGrad

```text
All historical squared gradients
          ↓
     accumulate forever
          ↓
       G becomes huge
          ↓
Effective LR can become tiny
```

### RMSprop

Instead of accumulating every squared gradient forever, RMSprop uses a **decaying average** of squared gradients.

Conceptually:

```text
AdaGrad
↓
Remember everything


RMSprop
↓
Remember recent history more strongly
↓
Gradually forget old history
```

This is one of the key steps in the evolution of adaptive optimizers.

---

# 14. AdaGrad and Sparse Features 🔥🔥

AdaGrad is particularly associated with **sparse features**.

Imagine a model with many features:

```text
Feature A → appears frequently
Feature B → appears frequently
Feature C → appears rarely
Feature D → appears rarely
```

Frequently occurring features accumulate gradient history more rapidly.

Infrequent features accumulate less.

AdaGrad therefore tends to produce:

```text
Frequent feature
→ larger accumulated G
→ smaller effective LR


Infrequent feature
→ smaller accumulated G
→ larger effective LR
```

This property was a major motivation in the original AdaGrad work, which explicitly discussed highly predictive but rarely observed features. ([JMLR][1])

# 15. PyTorch Implementation 🔥

PyTorch provides AdaGrad directly:

```python
optimizer = torch.optim.Adagrad(
    model.parameters(),
    lr=0.01
)
```

Training remains familiar:

```python
optimizer.zero_grad()

predictions = model(x_batch)

loss = criterion(predictions, y_batch)

loss.backward()

optimizer.step()
```

The optimizer maintains its accumulated state internally.

# 16. Common Mistakes ⚠️

### ❌ Mistake 1: "AdaGrad increases the learning rate."

No.

As accumulated squared gradients increase:

$$
\eta_{\text{effective}}\downarrow
$$

---

### ❌ Mistake 2: "AdaGrad uses momentum."

Not as its core mechanism.

Momentum accumulates **directional information**.

AdaGrad accumulates **squared gradient magnitudes**.

---

### ❌ Mistake 3: "Every parameter has the exact same learning rate."

No.

The base $\eta$ may be shared, but:

$$
G_t
$$

is parameter-specific.

Therefore the **effective learning rates differ**.

---

### ❌ Mistake 4: "AdaGrad forgets old gradients."

No.

This is actually its major weakness.

$$
G_t=G_{t-1}+g_t^2
$$

keeps accumulating.

# 🎯 Interview Questions

### Q1. What is AdaGrad?

An adaptive optimization algorithm that adjusts each parameter's effective learning rate using its historical squared gradients.

### Q2. Why square the gradients?

To accumulate gradient magnitude without positive and negative gradients cancelling.

### Q3. What happens when $G_t$ becomes large?

The effective learning rate becomes smaller.

### Q4. Why does AdaGrad provide parameter-specific learning rates?

Because each parameter has its own accumulated squared-gradient history.

### Q5. What is AdaGrad's main limitation?

Its accumulated squared gradients continually grow, potentially shrinking the effective learning rate too much.

### Q6. What type of problems historically motivated AdaGrad?

Problems with sparse and unevenly occurring features.

### Q7. What comes after AdaGrad conceptually?

**RMSprop**, which replaces the forever-growing accumulation with a decaying average of squared gradients.

* Uses **squared gradients**

* Major weakness → **learning rate can shrink excessively**

# 📚 10.1 Why Regularization — Revision Notes

## 🔥 Core Idea

The goal of a neural network is **not** to memorize the training data.

The real goal is:

> **Learn patterns from training data that also work well on unseen data.**

This ability is called **generalization**.

---

## 1. Overfitting

**Overfitting** happens when a model learns the training data too specifically, including noise or accidental patterns.

### Typical behavior

|                     | Training Data | Unseen/Validation Data |
| ------------------- | ------------- | ---------------------- |
| Underfitting        | Poor          | Poor                   |
| Good generalization | Good          | Good                   |
| Overfitting         | Very good     | Poor                   |

### Example

Suppose we're training a model to recognize cats.

A good model learns:

> "Cats generally have certain visual patterns."

An overfitted model may effectively memorize:

> "This particular training image has these exact pixels."

Then:

* Training accuracy → very high
* Validation/test accuracy → much lower

That is **poor generalization**.

# 2. Generalization

**Generalization** = the ability of a trained model to perform well on **data it has never seen before**.

This is what we actually care about in deployment.

### Example

Training:

> 100,000 images

Testing:

> 10,000 completely unseen images

If the model performs well on both, it has learned useful patterns rather than simply memorizing the training examples.

# 3. Training Loss vs Validation Loss

This is one of the **most important ways to recognize overfitting**.

During training:

```text
Loss
 ↑
 │\
 │ \
 │  \       Training loss
 │   \____________
 │
 │      /\
 │     /  \       Validation loss
 │____/    \____
 └────────────────→ Epoch
             ↑
        Overfitting begins
```

Initially:

* Training loss decreases
* Validation loss decreases

The model is learning useful patterns.

Eventually:

* Training loss continues decreasing
* Validation loss starts increasing

➡️ **Overfitting has begun.**

### Important point

The model with the **lowest training loss** isn't necessarily the model with the best generalization.

---

# 4. Why Does Overfitting Happen?

Common causes:

### ① Model is too flexible

A very large/deep network can learn extremely complicated patterns.

```text
Small model
    ↓
Limited patterns

Large model
    ↓
More expressive
    ↓
Can learn useful patterns
BUT
    ↓
Can also memorize noise
```

### ② Too little training data

If there aren't enough examples, the model has fewer examples from which to learn the true underlying pattern.

### ③ Noisy data

The dataset may contain:

* Incorrect labels
* Measurement errors
* Random variations
* Outliers

A sufficiently powerful model may learn these as if they were meaningful.

### ④ Training for too long

Even if the model initially generalizes well, continuing training can eventually cause it to fit the training data too specifically.

---

# 5. Generalization Gap

A useful way to think about overfitting is the difference between training and unseen-data performance.

For loss:

$$
\text{Generalization Gap}
=
L_{unseen}-L_{train}
$$

A large positive gap can indicate that the model performs substantially worse on unseen data than training data.

### Example

```text
Training loss   = 0.05
Validation loss = 0.80
```

Large gap → possible overfitting.

But:

> ⚠️ The gap alone isn't a complete diagnosis. Dataset differences, noise, distribution shift, and evaluation setup also matter.

---

# 6. So What Is Regularization?

**Regularization** is a collection of techniques used to reduce overfitting and improve generalization.

The basic idea:

> **Don't allow the model to become unnecessarily dependent on overly complicated solutions.**

Instead of minimizing only the training loss:

$$
J(\theta)=L(\theta)
$$

we add a regularization term:

$$
J(\theta)=L(\theta)+\lambda R(\theta)
$$

Where:

* $L(\theta)$ = original training loss
* $R(\theta)$ = regularization penalty
* $\lambda$ = regularization strength
* $\theta$ = model parameters

---

# 7. What Does $\lambda$ Do?

$\lambda$ controls **how strongly regularization affects training**.

### Small $\lambda$

```text
Training loss
     ↓
Regularization effect
     ↓
Weak
```

The model has more freedom → potentially more overfitting.

### Large $\lambda$

```text
Training loss
     ↓
Strong regularization
     ↓
Model becomes more constrained
```

Too much regularization can cause **underfitting**.

So conceptually:

$$
\lambda \uparrow
\Rightarrow
\text{stronger regularization}
$$

But:

$$
\lambda \text{ too large}
\Rightarrow
\text{possible underfitting}
$$

---

# 8. Important Insight: Regularization Can Increase Training Loss

This seems strange at first. 😵

Suppose:

### Without regularization

$$
L_{train}=0.02
$$

### With regularization

$$
L_{train}=0.10
$$

You might think:

> "The second model is worse!"

Not necessarily.

Suppose validation loss is:

```text
Without regularization → 0.70
With regularization    → 0.25
```

The regularized model has **higher training loss but better unseen-data performance**.

That's exactly the point.

> 🎯 We don't optimize solely for memorizing the training set. We care about generalization.

---

# 9. Regularization and Bias–Variance

This connects directly to your earlier **overfitting/underfitting** concepts.

```text
                 Model complexity
                       →
      Low                         High

   Underfitting                Overfitting
      ↑                            ↑
 High bias                   High variance
```

Regularization generally pushes the model toward a **less flexible solution**.

Conceptually:

$$
\text{Regularization}
\rightarrow
\text{less variance}
\rightarrow
\text{better generalization}
$$

But excessive regularization can increase bias:

$$
\text{Too much regularization}
\rightarrow
\text{underfitting}
$$

# 10. Major Regularization Techniques

You will study these individually later.

| Technique             | Basic idea                                                                |
| --------------------- | ------------------------------------------------------------------------- |
| **L1 regularization** | Penalize absolute weights                                                 |
| **L2 regularization** | Penalize squared weights                                                  |
| **Weight decay**      | Shrink weights during optimization                                        |
| **Dropout**           | Randomly deactivate neurons during training                               |
| **Early stopping**    | Stop training before overfitting                                          |
| **Data augmentation** | Create varied training examples                                           |
| **Normalization**     | Stabilize/condition training; can sometimes provide a regularizing effect |

### Don't memorize all the details yet.

For this block, the important idea is:

> **Regularization is a broad family of methods for improving generalization and reducing overfitting.**

# 11. L1 and L2 — Preview

### L1

$$
J(\theta)
=
L(\theta)
+
\lambda\sum_i|\theta_i|
$$

Penalizes the **absolute value** of weights.

It can encourage some weights to become exactly zero.

---

### L2

$$
J(\theta)
=
L(\theta)
+
\lambda\sum_i\theta_i^2
$$

Penalizes **large weights more strongly**.

It tends to keep weights smaller rather than allowing very large values.

# 12. The Big Picture 🧠

Think of training like this:

```text
              Training Data
                    │
                    ▼
              Neural Network
                    │
                    ▼
              Training Loss
                    │
                    ├───────────────┐
                    │               │
                    ▼               ▼
             Learn patterns    Regularization
                                    │
                                    ▼
                             Restrict excessive
                              model complexity
                                    │
                                    ▼
                           Better generalization
```

The overall objective becomes:

$$
\boxed{
\text{Training Objective}
=
\text{Data Loss}
+
\text{Regularization Penalty}
}
$$

# 🔥 Interview / Exam Points

### Q1. What is overfitting?

A model performs very well on training data but poorly on unseen data because it has learned overly specific patterns/noise.

### Q2. What is generalization?

The ability of a model to perform well on unseen data.

### Q3. Why do we use regularization?

To reduce overfitting and improve generalization.

### Q4. Does regularization always reduce training loss?

**No.** It can increase training loss while improving validation/test performance.

### Q5. What happens if regularization is too strong?

The model can become too constrained and **underfit**.

### Q6. What does $\lambda$ control?

The strength of the regularization penalty.

---

# 🧠 Final Cheat Sheet

```text
OVERFITTING
    ↓
Excellent training performance
Poor unseen-data performance
    ↓
Poor generalization

GENERALIZATION
    ↓
Good performance on unseen data

REGULARIZATION
    ↓
Adds a constraint/penalty
    ↓
Reduces excessive fitting
    ↓
Usually improves generalization

Objective:

J(θ) = L(θ) + λR(θ)

λ small → weak regularization
λ large → strong regularization
λ too large → underfitting
```

### ⭐ One sentence to remember

> **Regularization deliberately sacrifices some training-set fit to obtain a model that generalizes better to unseen data.**

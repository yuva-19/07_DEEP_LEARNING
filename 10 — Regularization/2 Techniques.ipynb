# 1. L1 Regularization 🔥🔥

## Core Concept

L1 regularization adds a penalty proportional to the **absolute value of the weights**.

$$
J(\theta)
=
L(\theta)
+
\lambda\sum_i |w_i|
$$

Where:

* $L(\theta)$ = original training loss
* $w_i$ = model weights
* $\lambda$ = regularization strength

The model therefore minimizes:

> **Data error + penalty for having weights**

---

## Why L1?

L1 has a special property:

> **It can push some weights exactly to zero.**

Example:

```text
Before L1:

w₁ = 2.5
w₂ = 0.8
w₃ = 0.03
w₄ = -1.7


After strong L1:

w₁ = 2.1
w₂ = 0
w₃ = 0
w₄ = -1.3
```

A weight of zero means that connection contributes nothing.

So L1 encourages a **sparse model**.

Google's ML glossary describes L1 as a regularization method that can drive weights of irrelevant or barely relevant features exactly to zero. ([Google for Developers][1])

---

## Why does L1 create zeros?

For:

$$
R(w)=|w|
$$

the derivative away from zero is:

$$
\frac{d|w|}{dw}
=
\begin{cases}
+1 & w>0\\
-1 & w<0
\end{cases}
$$

So L1 applies a roughly constant pull toward zero.

At $w=0$, $|w|$ has a sharp corner, handled using a **subgradient**.

### Mental model

```text
Large positive weight
        ↓
Pull toward 0
        ↓
Small weight
        ↓
Can reach exactly 0
```

### ⭐ Remember

> **L1 → sparsity → some weights become zero**

# 2. L2 Regularization 🔥🔥🔥

L2 adds a penalty proportional to the **square of the weights**.

$$
J(\theta)
=
L(\theta)
+
\lambda\sum_i w_i^2
$$

Sometimes you'll see:

$$
J(\theta)
=
L(\theta)
+
\frac{\lambda}{2}\sum_iw_i^2
$$

The $\frac12$ is just a convention that makes the derivative cleaner.

---

## Why square the weights?

Look at the penalty:

| Weight | $w^2$ |
| -----: | ----: |
|    0.1 |  0.01 |
|      1 |     1 |
|      2 |     4 |
|      5 |    25 |
|     10 |   100 |

Large weights become **very expensive**.

Therefore L2 encourages the network to avoid unnecessarily large weights.

---

## Derivative

For:

$$
R(w)=w^2
$$

we have:

$$
\frac{dR}{dw}=2w
$$

So the larger the weight, the stronger the regularization gradient.

```text
Large weight
     ↓
Large L2 gradient
     ↓
Strong pull toward zero

Small weight
     ↓
Small L2 gradient
     ↓
Weak pull toward zero
```

### Important distinction

L2 generally makes weights **small**, but unlike L1 it does not normally drive them exactly to zero. ([Google for Developers][2])

# 4. Weight Decay 🔥🔥🔥

Weight decay is another way of encouraging weights to become smaller.

The basic idea is:

> **Shrink the weights a little during every optimization step.**

A simplified weight-decay update is:

$$
w
\leftarrow
(1-\eta\lambda)w-\eta g
$$

where:

* $g$ = gradient from the original loss
* $\eta$ = learning rate
* $\lambda$ = weight-decay coefficient

The term:

$$
(1-\eta\lambda)w
$$

shrinks the existing weight.

---

## Example

Suppose:

$$
w=10
$$

and:

$$
\eta\lambda=0.01
$$

Then:

$$
w_{new}=0.99(10)=9.9
$$

So:

```text
10
 ↓
9.9
 ↓
9.801
 ↓
...
```

The weight is continually pulled toward zero.

---

# 5. L2 Regularization vs Weight Decay ⚠️🔥

This is a common source of confusion.

They are **closely related**, but don't blindly say they're always identical.

### L2 regularization

You modify the objective:

$$
J=L+\frac{\lambda}{2}\sum_iw_i^2
$$

Therefore:

$$
\frac{\partial J}{\partial w}
=
\frac{\partial L}{\partial w}
+
\lambda w
$$

The regularization term becomes part of the gradient.

---

### Weight decay

You directly shrink the parameters:

$$
w
\leftarrow
(1-\eta\lambda)w-\eta g
$$

So the shrinkage is applied separately from the loss gradient.

---

## When are they equivalent?

For standard **SGD**, L2 regularization and weight decay are mathematically equivalent up to the parameterization of the coefficient.

But with adaptive optimizers such as **Adam**, they are not generally equivalent. This distinction motivated **AdamW**, which uses decoupled weight decay. ([arXiv][3])

### Therefore:

```text
SGD
 │
 └── L2 ≈ Weight Decay

Adam
 │
 ├── L2 regularization
 │
 └── Weight decay
       ↓
     not generally equivalent

AdamW
 │
 └── decoupled weight decay
```

### ⭐ Practical takeaway

If you see:

```python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)
```

that `weight_decay` is implementing **decoupled weight decay**, not simply "add an L2 term to the loss." ([arXiv][3])

# 7. Why Does Dropout Help?

Without dropout, a network might become overly dependent on particular neurons.

For example:

```text
Neuron A ──┐
Neuron B ──┼──→ Output
Neuron C ──┤
Neuron D ──┘
```

The network could develop strong dependencies such as:

> "If neuron A detects this pattern, neuron B can rely heavily on it."

Dropout repeatedly removes some of these shortcuts.

Therefore:

> **The network is encouraged to learn representations that remain useful even when some neurons are unavailable.**

---

# 8. Dropout Probability

Suppose:

$$
p=0.2
$$

This means:

> **20% dropout probability**

or approximately 20% of the activations are dropped during training.

The keep probability is:

$$
q=1-p
$$

So:

$$
q=0.8
$$

| Dropout $p$ | Keep probability |
| ----------: | ---------------: |
|         0.1 |              0.9 |
|         0.2 |              0.8 |
|         0.5 |              0.5 |

### ⚠️ Important

**Dropout probability ≠ keep probability.**

If:

```text
p = 0.3
```

you drop 30% and keep approximately 70%.

---

# 9. Inverted Dropout

Modern implementations usually use **inverted dropout**.

Suppose:

$$
p=0.2
$$

Then:

$$
q=1-p=0.8
$$

If an activation survives, it is scaled by:

$$
\frac{1}{q}
=
\frac{1}{0.8}
=
1.25
$$

Example:

```text
Before:
[2, 4, 6, 8]

Mask:
[1, 0, 1, 1]

After dropping:
[2, 0, 6, 8]

After scaling:
[2.5, 0, 7.5, 10]
```

The scaling helps preserve the expected activation magnitude. ([TensorFlow][4])

---

# 10. Dropout: Training vs Inference 🔥🔥🔥

This is **very important**.

### Training

```text
Input
  ↓
Layer
  ↓
Randomly drop activations
  ↓
Output
```

### Inference

```text
Input
  ↓
Layer
  ↓
ALL activations used
  ↓
Output
```

So:

> **Dropout is active during training and disabled during inference.**

In PyTorch:

```python
model.train()
```

→ dropout active

```python
model.eval()
```

→ dropout disabled

---

# 11. Dropout and Ensemble Intuition

A useful mental model:

```text
Training step 1 → Subnetwork A
Training step 2 → Subnetwork B
Training step 3 → Subnetwork C
Training step 4 → Subnetwork D
...
```

The same large network is effectively trained under many different neuron configurations.

This discourages excessive reliance on a particular path through the network.

### ⚠️ Don't interpret this too literally

Dropout isn't literally training thousands of completely independent models and averaging them. It's a useful intuition for why dropout can improve robustness.

# 12. Early Stopping 🔥🔥🔥

**Stop training when validation performance stops improving.**

| ----: | ------------: | --------------: |
|     1 |          0.80 |            0.82 |
|     2 |          0.60 |            0.64 |
|     3 |          0.45 |            0.50 |
|     4 |          0.34 |            0.39 |
|     5 |          0.25 |            0.31 |
|     6 |          0.18 |      **0.30 ⭐** |
|     7 |          0.13 |            0.34 |
|     8 |          0.09 |            0.40 |
|     9 |          0.06 |            0.48 |

Training loss keeps improving.

But validation loss starts increasing.

That is a classic overfitting signal.

# 13. Why Does Early Stopping Work?

Think about the training process:

```text
Early epochs
     ↓
Model learning useful patterns
     ↓
Validation improves
     ↓
Eventually
     ↓
Model starts fitting training-specific details
     ↓
Validation worsens
     ↓
STOP
```

So we don't allow the model to keep optimizing training loss indefinitely.

---

# 14. Patience 🔥

We usually don't stop after **one** bad epoch.

Why?

Because validation loss can fluctuate.

Example:

```text
Epoch 10 → 0.30
Epoch 11 → 0.31
Epoch 12 → 0.29
```

If we stopped immediately at epoch 11, we'd have stopped unnecessarily.

Instead we use **patience**.

Example:

$$
\text{patience}=3
$$

Meaning:

> Allow several consecutive epochs without meaningful improvement before stopping.

---

# 15. Restore Best Weights

Suppose:

```text
Epoch 10 → validation loss = 0.25 ⭐
Epoch 11 → 0.27
Epoch 12 → 0.30
Epoch 13 → 0.34
```

If training stops at epoch 13, we don't necessarily want the parameters from epoch 13.

We want the **best parameters**, from epoch 10.

Conceptually:

```text
Train
  ↓
Monitor validation
  ↓
Save best weights
  ↓
Validation stops improving
  ↓
STOP
  ↓
Restore best weights
```

This is why "best checkpoint" logic is extremely useful in practical training.

# 16. Comparing All Five 🔥🔥🔥

| Technique          | What is changed?            | Main effect                                         |
| ------------------ | --------------------------- | --------------------------------------------------- |
| **L1**             | Loss/objective              | Encourages sparse weights                           |
| **L2**             | Loss/objective              | Discourages large weights                           |
| **Weight decay**   | Parameter update            | Shrinks weights                                     |
| **Dropout**        | Activations during training | Reduces reliance on specific neurons                |
| **Early stopping** | Training duration           | Prevents continued fitting after validation worsens |

---

# 17. The Most Important Difference

Think of the five techniques as five different "controls":

```text
L1
 ↓
"Remove unnecessary connections."


L2
 ↓
"Don't make weights too large."


Weight Decay
 ↓
"Keep shrinking weights."


Dropout
 ↓
"Don't depend on specific neurons."


Early Stopping
 ↓
"Don't train beyond the useful point."
```

# 20. Common Mistakes ⚠️

### ❌ "L1 and L2 both make weights zero."

Not quite.

**L1 can produce exact zeros.**

L2 generally produces small but nonzero weights.

---

### ❌ "Dropout permanently removes neurons."

No.

Dropout randomly deactivates activations **during training**.

---

### ❌ "Dropout is active during inference."

No.

It is disabled during normal inference/evaluation.

---

### ❌ "Higher dropout is always better."

No.

Too much dropout can cause underfitting.

---

### ❌ "Early stopping watches training loss."

The important signal is generally **validation performance**, because the goal is generalization.

---

### ❌ "Stop immediately after one validation-loss increase."

Not necessarily.

Use **patience** because validation metrics can fluctuate.

---

### ❌ "L2 and weight decay are always identical."

They are equivalent under standard SGD formulations, but not generally under adaptive optimizers such as Adam. AdamW uses decoupled weight decay. ([arXiv][3])

### Q1. Which regularization encourages sparse weights?

**L1**

### Q2. Which penalizes squared weights?

**L2**

### Q3. Which can drive weights exactly to zero?

**L1**

### Q4. What is the primary effect of L2?

**Shrink large weights.**

### Q5. What does dropout do?

Randomly sets activations to zero during training.

### Q6. Is dropout used during inference?

**No.**

### Q7. What is early stopping?

Stopping training when validation performance stops improving.

### Q8. Why is patience used?

To tolerate temporary fluctuations in validation performance.

### Q9. What does AdamW use?

**Decoupled weight decay.**

## ⭐ Five one-line memories

> **L1 → Zero** 🎯

> **L2 → Small** 📉

> **Weight decay → Shrink** 🔽

> **Dropout → Randomly switch off** 🎲

> **Early stopping → Stop before overfitting** 🛑

### The overall picture

$$
\boxed{
\text{Regularization}
\rightarrow
\text{reduce overfitting}
\rightarrow
\text{improve generalization}
}
$$

And remember the key principle from the previous block:

$$
\boxed{
\text{Training Objective}
=
\text{Data Loss}
+
\text{Regularization}
}
$$

The important nuance is that **better regularization does not necessarily mean lower training loss**. The actual objective is good performance on unseen data.

# 📚 10.4 Early Stopping — Complete Revision Notes

Early stopping is a **regularization technique** that prevents a model from continuing to fit the training data after its performance on unseen validation data stops improving.

The three things to remember are:

> **Validation Loss → Patience → Best Checkpoint**

---

# 1. Core Idea 🔥🔥🔥

During training, we usually observe:

* **Training loss** → how well the model fits training data
* **Validation loss** → how well the model generalizes to unseen validation data

A typical pattern is:

```text
Loss
 ↑
 │\
 │ \
 │  \        Training loss
 │   \______________
 │
 │       \___
 │           \__
 │              \___
 │                  \__
 │       Validation loss
 │
 └────────────────────────→ Epoch
                 ↑
            Best point
```

Initially, both losses improve.

Eventually:

```text
Training loss     ↓ ↓ ↓ ↓ ↓
Validation loss   ↓ ↓ ↑ ↑ ↑
                      ↑
                  Overfitting
```

**Early stopping says:**

> "Once validation performance stops improving for long enough, stop training."

# 2. Validation Loss 🔥🔥🔥

## What is it?

**Validation loss** is the loss calculated on a separate validation dataset that is not used to update the model's parameters.

Training:

$$
\text{Training data}
\rightarrow
\text{Forward}
\rightarrow
\text{Loss}
\rightarrow
\text{Backprop}
\rightarrow
\text{Update weights}
$$

Validation:

$$
\text{Validation data}
\rightarrow
\text{Forward}
\rightarrow
\text{Validation loss}
$$

The validation loss is used to **evaluate** the current model, not to directly update its weights.

---

# 3. Typical Training Curve

Suppose we get:

| Epoch | Training Loss | Validation Loss |
| ----: | ------------: | --------------: |
|     1 |          0.90 |            0.95 |
|     2 |          0.70 |            0.76 |
|     3 |          0.52 |            0.59 |
|     4 |          0.39 |            0.45 |
|     5 |          0.30 |      **0.35 ⭐** |
|     6 |          0.22 |            0.37 |
|     7 |          0.16 |            0.42 |
|     8 |          0.11 |            0.50 |

Notice:

### Training loss

$$
0.90\rightarrow0.70\rightarrow0.52\rightarrow...\rightarrow0.11
$$

continues decreasing.

### Validation loss

$$
0.95\rightarrow0.76\rightarrow0.59\rightarrow...\rightarrow0.35
$$

then starts increasing.

So:

$$
\boxed{\text{Epoch 5 = best validation performance}}
$$

This is the important point.

**The lowest training loss is NOT necessarily the best model.**

# 4. Why Validation Loss Is Useful

The purpose of training is not:

> "Get the smallest possible training loss."

The real goal is:

> **Perform well on unseen data.**

So validation performance gives us a signal about generalization.

```text
Training loss ↓
       │
       ▼
Model keeps fitting training data

Validation loss ↓
       │
       ▼
Generalization improving

Validation loss ↑
while training loss ↓
       │
       ▼
Possible overfitting
```

# 5. Patience 🔥🔥🔥

Real validation loss isn't perfectly smooth.

It can fluctuate:

```text
Epoch 10 → 0.30
Epoch 11 → 0.32
Epoch 12 → 0.29
Epoch 13 → 0.28
```

If we stopped immediately after epoch 11:

❌ We would have stopped too early.

So we introduce **patience**.

---

# 6. What Is Patience?

**Patience = number of consecutive epochs allowed without sufficient improvement before training stops.**

Suppose:

$$
\text{patience}=3
$$

And the best validation loss so far is:

$$
0.25
$$

Then:

```text
Epoch 20 → 0.25 ⭐ best
Epoch 21 → 0.27
Epoch 22 → 0.28
Epoch 23 → 0.29
```

There have been 3 consecutive epochs without improvement.

➡️ **Stop training.**

Keras defines `patience` as the number of epochs with no improvement after which training is stopped. ([Keras][1])

---

# 7. Why Patience Matters

Without patience:

```text
One bad epoch
      ↓
STOP
```

With patience:

```text
One bad epoch
      ↓
Wait

Another bad epoch
      ↓
Wait

Another bad epoch
      ↓
STOP
```

So patience protects against temporary fluctuations.

### 🧠 Easy memory

> **Patience = "Don't panic after one bad epoch."** 😄

---

# 8. `min_delta` 🔥

Another useful parameter is **minimum improvement**, often called `min_delta`.

Suppose:

$$
\text{min\_delta}=0.001
$$

An improvement must be at least $0.001$ to count as meaningful.

Suppose current best:

$$
0.300
$$

New validation loss:

$$
0.2995
$$

Improvement:

$$
0.300-0.2995=0.0005
$$

That's less than:

$$
0.001
$$

So it may be treated as **no meaningful improvement**.

But:

$$
0.300-0.298=0.002
$$

is sufficient.

Keras defines `min_delta` as the minimum change required for an improvement to count. ([Keras][1])

---

# 9. Best Checkpoint 🔥🔥🔥

Now the most practical part.

A **checkpoint** is a saved snapshot of the model's state.

Suppose:

```text
Epoch 1 → Val loss 0.80
Epoch 2 → Val loss 0.60
Epoch 3 → Val loss 0.45
Epoch 4 → Val loss 0.35 ⭐
Epoch 5 → Val loss 0.39
Epoch 6 → Val loss 0.44
```

At epoch 4:

```text
Validation improved
       ↓
Save checkpoint
```

So we preserve the model from epoch 4.

---

# 10. Why Save the Best Checkpoint?

Because the **final model may not be the best model**.

Example:

```text
Best model
Epoch 4
Validation loss = 0.35

Final model
Epoch 10
Validation loss = 0.62
```

If we simply use the final model:

❌ We lose the better model from epoch 4.

Instead:

```text
Training
   ↓
Evaluate validation
   ↓
Better?
   ↓
YES → Save checkpoint
   ↓
Continue
   ↓
No improvement
   ↓
Patience exceeded
   ↓
STOP
   ↓
Use best checkpoint
```

---

# 11. Best Checkpoint vs Early Stopping

These are related but **not exactly the same thing**.

### Early stopping

Answers:

> **"When should I stop training?"**

### Checkpointing

Answers:

> **"Which model state should I save/use?"**

Together:

```text
Early stopping
      +
Best checkpoint
      ↓
Stop at the appropriate time
AND
keep the best model
```

This distinction is important in real training pipelines.

---

# 12. `restore_best_weights`

Many frameworks provide an option to restore the weights from the best epoch.

For example, Keras provides:

```python
restore_best_weights=True
```

This means that after early stopping, the model's weights are restored to the epoch with the best monitored metric. ([Keras][1])

Conceptually:

```text
Epoch 10 → Best ⭐
Epoch 11 → Worse
Epoch 12 → Worse
Epoch 13 → Worse
        ↓
      STOP
        ↓
Restore Epoch 10 weights
```

---

# 13. Checkpoint File vs Restoring Weights

There are two related approaches.

### Approach 1 — Keep best weights in memory

```text
Best epoch
   ↓
Save weights internally
   ↓
Training stops
   ↓
Restore best weights
```

### Approach 2 — Save checkpoint to disk

```text
Best epoch
   ↓
Save model checkpoint
   ↓
Training continues
   ↓
Training stops
   ↓
Load checkpoint
```

The second approach is particularly useful for long training runs, because you have a physical saved model state that can survive interruptions.

Keras provides `ModelCheckpoint` with options such as `save_best_only=True` for this workflow. ([Keras][2])

---

# 14. Complete Workflow 🔥🔥🔥

```text
                 START TRAINING
                       │
                       ▼
                  Train Epoch
                       │
                       ▼
              Calculate Validation Loss
                       │
                       ▼
             Is it better than before?
                  /            \
                YES             NO
                 │               │
                 ▼               ▼
          Save Best Model    Increase
          Reset Patience     Patience Counter
                 │               │
                 │               ▼
                 │        Patience exceeded?
                 │             /      \
                 │           NO        YES
                 │           │          │
                 └───────────┘          ▼
                                      STOP
                                        │
                                        ▼
                              Restore Best Checkpoint
```

---

# 15. Simple Pseudocode

```python
best_val_loss = float("inf")
patience_counter = 0

for epoch in range(max_epochs):

    train_one_epoch(model)

    val_loss = validate(model)

    if val_loss < best_val_loss:
        best_val_loss = val_loss

        save_checkpoint(model)

        patience_counter = 0

    else:
        patience_counter += 1

    if patience_counter >= patience:
        break

load_best_checkpoint(model)
```

The logic is:

```text
Better?
 ↓
YES → save + reset patience

NO
 ↓
increase patience

Patience exceeded?
 ↓
YES → stop
```

---

# 16. What Goes Inside a Checkpoint?

A checkpoint can contain model parameters:

```python
model.state_dict()
```

But for **resuming training**, you may also save:

```python
checkpoint = {
    "epoch": epoch,
    "model_state": model.state_dict(),
    "optimizer_state": optimizer.state_dict(),
    "best_val_loss": best_val_loss
}
```

Why save optimizer state?

Optimizers such as Adam maintain internal state used during optimization. If you want to resume training properly, preserving that state can matter.

# 17. Maximum Epochs vs Early Stopping

These are different.

Suppose:

```python
max_epochs = 100
patience = 5
```

This means:

> Train for **at most** 100 epochs, but stop earlier if the monitored validation metric fails to improve for the patience period.

Example:

```text
Maximum possible:
100 epochs

Actual:
37 epochs

Reason:
Validation stopped improving
```

So:

$$
\boxed{
\text{max epochs = upper limit}
}
$$

while:

$$
\boxed{
\text{early stopping = stopping rule}
}
$$

# 18. What Metric Can We Monitor?

Validation loss is common:

```python
monitor = "val_loss"
```

But early stopping can monitor other metrics too.

### Classification

* `val_loss`
* `val_accuracy`
* `val_f1`
* `val_auc`

### Regression

* `val_loss`
* `val_mae`
* `val_rmse`

The important thing is to choose a validation metric that reflects what you care about.

Keras supports monitoring a chosen metric and automatically determining whether it should increase or decrease in `mode="auto"`; `mode="min"` and `"max"` can also be specified explicitly. ([Keras][1])

---

# 19. Minimizing vs Maximizing ⚠️

For **loss**, lower is better:

$$
\boxed{\text{minimize}}
$$

For metrics such as accuracy, higher is better:

$$
\boxed{\text{maximize}}
$$

Example:

### Validation loss

```text
0.50 → 0.40 → 0.30
```

Improvement = **decrease**.

### Validation accuracy

```text
80% → 85% → 90%
```

Improvement = **increase**.

This sounds obvious, but it matters when configuring callbacks.

# 20. Early Stopping as Regularization

Why is early stopping considered regularization?

Because it limits how much the model can fit the training data.

Without early stopping:

```text
Useful learning
      ↓
More learning
      ↓
Training-specific patterns
      ↓
Memorization
      ↓
Overfitting
```

With early stopping:

```text
Useful learning
      ↓
Validation stops improving
      ↓
STOP
```

Therefore:

$$
\boxed{
\text{Early Stopping}
\rightarrow
\text{limit excessive fitting}
\rightarrow
\text{reduce overfitting}
}
$$

# 21. Early Stopping vs Dropout

You just learned Dropout in 10.3, so connect them:

|                      | Dropout                        | Early Stopping                      |
| -------------------- | ------------------------------ | ----------------------------------- |
| What does it affect? | Activations                    | Training duration                   |
| During training      | Randomly drops activations     | Monitors validation                 |
| Main idea            | Don't rely on specific neurons | Don't continue past useful training |
| Inference effect     | Dropout disabled               | No special layer behavior           |
| Main hyperparameter  | Dropout rate                   | Patience                            |
| Goal                 | Reduce overfitting             | Reduce overfitting                  |

### Memory trick 🧠

> **Dropout controls the network while learning.**

> **Early stopping controls how long it learns.**

# 22. Common Mistakes ⚠️

### ❌ Mistake 1: Monitor training loss

Training loss can keep decreasing while generalization gets worse.

Use a **validation metric** for the stopping decision.

---

### ❌ Mistake 2: Stop after one bad epoch

Validation metrics fluctuate.

Use patience.

---

### ❌ Mistake 3: Use the final model automatically

The final model may be worse than an earlier checkpoint.

Save the best checkpoint.

---

### ❌ Mistake 4: Confuse patience with number of epochs

If:

```text
patience = 5
```

it does **not** mean train for five epochs.

It means:

> Allow five consecutive epochs without sufficient improvement before stopping.

---

### ❌ Mistake 5: Use the test set for early stopping

The test set should generally be reserved for **final evaluation**.

Use validation data for model-selection decisions such as early stopping.
